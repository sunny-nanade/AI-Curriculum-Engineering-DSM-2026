### Procedure

### Why Hammer Test?
- **Broadband**: Impulse contains all frequencies
- **Quick**: One hit gives full response
- **Portable**: No need for shaker or fixtures
- **Standard**: Used in aerospace, automotive, civil

### Impulse Modeling
- Short duration pulse (1-5 ms typical)
- Integrated force = momentum change
- Approximated as initial velocity

## Part 3: Frequency Domain Analysis

### FFT Processing
1. Apply FFT to acceleration data
2. Identify peaks in magnitude spectrum
3. Each peak corresponds to natural frequency

### Peak Picking Algorithm
- Find local maxima in FFT magnitude
- Filter by prominence (reject noise)
- Extract frequency at each peak

### Half-Power Method for Damping
At resonance peak:
- $f_r$ = resonance frequency
- $f_1, f_2$ = frequencies where amplitude = peak/sqrt(2)
- $\\zeta \\approx \\frac{f_2 - f_1}{2f_r}$ (bandwidth method)

## Part 5: Mode Shape Identification

### Procedure
1. Extract FFT amplitude at each natural frequency
2. For each mode, amplitude ratio = mode shape ratio
3. Normalize mode shapes
4. Compare with theoretical prediction

### Modal Assurance Criterion (MAC)
Correlation between measured and theoretical mode shapes:
$$\\text{MAC}(\\phi_i, \\psi_i) = \\frac{|\\phi_i^T \\psi_i|^2}{(\\phi_i^T \\phi_i)(\\psi_i^T \\psi_i)}$$

- MAC = 1: Perfect match
- MAC > 0.9: Good correlation
- MAC < 0.8: Poor correlation

## Summary: Modal Testing Results

### Extracted Parameters

**Natural Frequencies**:
- Extracted from FFT peak picking
- Typical accuracy: < 1% error with good data
- Limited by frequency resolution (sampling rate / duration)

**Damping Ratios**:
- Method 1: Logarithmic decrement from decay
- Method 2: Half-power bandwidth from FRF
- Typical accuracy: 5-20% error (harder to measure)
- Affected by: noise, nonlinearity, joint damping

**Mode Shapes**:
- Extracted from FRF amplitude ratios
- Validated with MAC (Modal Assurance Criterion)
- MAC > 0.9 indicates good correlation
- Requires multi-point measurements

### Key Takeaways

1. **Hammer test is practical**: Quick, portable, broadband
2. **FFT reveals all modes**: Peaks at natural frequencies
3. **Multiple methods for damping**: Cross-validate results
4. **MAC validates modes**: Quantifies correlation
5. **Measurement matters**: Noise, resolution, duration affect accuracy
