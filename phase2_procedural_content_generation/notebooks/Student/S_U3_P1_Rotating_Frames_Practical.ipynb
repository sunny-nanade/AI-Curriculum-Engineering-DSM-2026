# Part 2: Coriolis Effect Demonstration (30 minutes)

**Goal**: Isolate and visualize the Coriolis acceleration term.

## The Coriolis Term

From Lecture 4:
$$\vec{a}_{\text{Coriolis}} = 2\vec{\Omega} \times \vec{v}_{\text{rel}}$$

**Key properties**:
1. Perpendicular to **both** Ω and v_rel
2. Magnitude proportional to speed in rotating frame
3. Deflects moving objects (hurricanes, projectiles, etc.)

---

## Task 2.1: Coriolis Deflection

Object moves radially outward on rotating platform. 

**No tangential velocity initially** - but Coriolis causes deflection!

---

```
# Coriolis demonstration
omega_cor = 1.0 # rad/s
v0_cor = np.array([3.0, 0.0]) # Pure radial motion
r0_cor = np.array([0.5, 0.0]) # Start near center

print("Coriolis Effect Demonstration")
print("="*60)
print(f"Platform rotation: Ω = {omega_cor} rad/s (CCW)")
print(f"Initial position: r = {r0_cor}")
print(f"Initial velocity: v_rel = {v0_cor} (purely radial)")
print()
print("Expected: Coriolis deflection to the RIGHT (in rotating frame)")
print(f"a_Coriolis = 2·Ω·v_rel = 2×{omega_cor}×{v0_cor[0]} = {2*omega_cor*v0_cor[0]} m/s²")
print()

anim_cor = animate_coriolis_deflection(
 omega=omega_cor,
 v0=v0_cor,
 r0=r0_cor,
 duration=4.0,
 fps=30
)

print("\n Animation complete!")
print("\nOBSERVATIONS:")
print("1. In FIXED frame: Object moves in STRAIGHT line (no external force)")
print("2. In ROTATING frame: Object curves to the RIGHT")
print("3. This is the CORIOLIS EFFECT!")
```

## Task 2.2: Quantitative Analysis

Calculate the Coriolis acceleration and compare to simulation.

---

```
# Analytical calculation of Coriolis acceleration

# Given parameters from simulation above
Omega_vec = np.array([0, 0, omega_cor]) # Rotation about Z-axis
v_rel_vec = np.array([v0_cor[0], v0_cor[1], 0]) # Velocity in XY plane

# Coriolis acceleration: a_C = 2 * Omega × v_rel
a_Coriolis = 2 * np.cross(Omega_vec, v_rel_vec)

print("Coriolis Acceleration Calculation")
print("="*60)
print(f"Ω = {Omega_vec} rad/s")
print(f"v_rel = {v_rel_vec} m/s")
print()
print(f"a_Coriolis = 2·Ω × v_rel")
print(f"= {a_Coriolis} m/s²")
print()
print(f"Magnitude: |a_C| = {np.linalg.norm(a_Coriolis[:2]):.2f} m/s²")
print(f"Direction: {np.degrees(np.arctan2(a_Coriolis[1], a_Coriolis[0])):.1f}° from +X")
print()
print("This acceleration is PERPENDICULAR to v_rel!")
print("→ Causes curved trajectory in rotating frame")
print("→ But in fixed frame: a = 0 (straight line motion)")
```

# Part 3: Foucault Pendulum Simulation (30 minutes)

**Goal**: See Earth's rotation through pendulum precession.

## The Foucault Pendulum

**Historical context**: 1851, Léon Foucault demonstrated Earth's rotation using a long pendulum.

**Physics**:
- Pendulum swings in a fixed plane (inertial frame)
- Earth rotates beneath it
- From Earth's surface: plane appears to precess!

**Precession period**:
$$T_{\text{precession}} = \frac{24 \text{ hours}}{\sin(\text{latitude})}$$

**Coriolis is responsible!**

---

## Task 3.1: Simple Pendulum Simulation

Simulate a pendulum including Coriolis effect at different latitudes.

---

```
def foucault_pendulum(latitude_deg, duration=100, length=10.0):
 """
 Simulate Foucault pendulum at given latitude.
 
 Parameters:
 - latitude_deg: Latitude in degrees (0=equator, 90=north pole)
 - duration: Simulation duration in seconds
 - length: Pendulum length in meters
 """
 # Earth's rotation rate
 Omega_Earth = 2 * np.pi / (24 * 3600) # rad/s
 
 # Effective Coriolis parameter at latitude
 lat_rad = np.radians(latitude_deg)
 f = 2 * Omega_Earth * np.sin(lat_rad) # Coriolis parameter
 
 # Pendulum natural frequency
 g = 9.81 # m/s²
 omega0 = np.sqrt(g / length) # rad/s
 
 # Equations of motion (small angle approximation)
 def derivatives(state, t):
 x, y, vx, vy = state
 
 # Acceleration: pendulum restoring + Coriolis
 ax = -omega0**2 * x + f * vy # Coriolis: f·vy
 ay = -omega0**2 * y - f * vx # Coriolis: -f·vx
 
 return [vx, vy, ax, ay]
 
 # Initial conditions: displaced in X direction
 x0 = 0.5 # m
 y0 = 0.0
 vx0 = 0.0
 vy0 = 0.0
 state0 = [x0, y0, vx0, vy0]
 
 # Time array
 t = np.linspace(0, duration, 2000)
 
 # Solve ODE
 solution = odeint(derivatives, state0, t)
 x, y, vx, vy = solution.T
 
 # Calculate precession period
 if np.abs(np.sin(lat_rad)) > 0.01:
 T_precession = (24 * 3600) / np.sin(lat_rad) # seconds
 else:
 T_precession = np.inf
 
 # Plot
 fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))
 
 # Left: Trajectory
 ax1.plot(x, y, 'b-', alpha=0.6, linewidth=1)
 ax1.plot(x[0], y[0], 'go', markersize=10, label='Start')
 ax1.plot(x[-1], y[-1], 'ro', markersize=10, label='End')
 ax1.set_xlabel('X position (m)')
 ax1.set_ylabel('Y position (m)')
 ax1.set_title(f'Foucault Pendulum at {latitude_deg}°N')
 ax1.grid(True, alpha=0.3)
 ax1.axis('equal')
 ax1.legend()
 
 # Right: Position vs time
 ax2.plot(t, x, 'r-', label='X position', linewidth=1.5)
 ax2.plot(t, y, 'b-', label='Y position', linewidth=1.5)
 ax2.set_xlabel('Time (s)')
 ax2.set_ylabel('Position (m)')
 ax2.set_title('Position Components vs Time')
 ax2.grid(True, alpha=0.3)
 ax2.legend()
 
 plt.tight_layout()
 plt.show()
 
 # Print results
 print(f"\nFoucault Pendulum Simulation")
 print("="*60)
 print(f"Latitude: {latitude_deg}°N")
 print(f"Pendulum length: {length} m")
 print(f"Simulation duration: {duration} s ({duration/60:.1f} minutes)")
 print()
 print(f"Earth rotation: Ω = {Omega_Earth*1e6:.3f} μrad/s")
 print(f"Coriolis parameter: f = {f*1e6:.3f} μrad/s")
 print(f"Pendulum frequency: ω₀ = {omega0:.3f} rad/s (T = {2*np.pi/omega0:.2f} s)")
 print()
 if T_precession < 1e6:
 print(f"Precession period: {T_precession/3600:.1f} hours")
 print(f"(Theoretical: 24h/sin({latitude_deg}°) = {24/np.sin(lat_rad):.1f} hours)")
 else:
 print("No precession (at equator, sin(0°) = 0)")
 print("="*60)

# Run simulation at North Pole (maximum effect)
print("Simulation 1: North Pole (90°N)")
foucault_pendulum(latitude_deg=90, duration=200, length=10.0)
```
