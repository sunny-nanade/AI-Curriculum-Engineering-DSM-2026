## SETUP: Import Libraries and Review Kinematics

**Connection to Lectures**: In all three lectures, you used Python libraries (NumPy, Matplotlib) for computation and visualization. We continue using the same tools.

**New Tool**: **SymPy** - Symbolic mathematics library for automatic calculus

### Why SymPy?

In **Lecture 2**, you calculated derivatives by hand:
- Given: x(t) = 2t² 
- Derivative: v(t) = dx/dt = 4t (by power rule)
- Second derivative: a(t) = dv/dt = 4 (constant)

For **complex motions** (spirals, Lissajous curves, cardioids), hand calculation becomes:
- **Tedious**: Many terms, chain rule, product rule, trig identities
- **Error-prone**: Easy to make algebraic mistakes
- **Time-consuming**: Takes focus away from understanding the physics

**SymPy automates this**:
- You define x(t) and y(t) symbolically (as formulas with variable t)
- SymPy computes vx(t) = dx/dt automatically (exact symbolic result)
- SymPy computes ax(t) = d²x/dt² automatically
- You can then substitute numerical values to plot

**Pedagogical benefit**: Focus on **interpreting** the motion (what does the velocity mean physically?) rather than algebraic manipulation.

**Preparation for Unit 2**: Lagrangian mechanics (Unit 2-3) requires symbolic derivatives of kinetic and potential energy. SymPy will be essential there.

### Example 1: Uniform Circular Motion

#### Mathematical Description
Position equations:
- x(t) = R cos(ωt)
- y(t) = R sin(ωt)

Where:
- **R** = radius (meters)
- **ω** = angular velocity (radians/second)
- **t** = time (seconds)

#### Connection to Lecture 1
This is **identical** to Homework Problem 2. You calculated the position vector for a particle moving in a circle:
- Given: R = 5m, θ = 30°
- Found: r = x i + y j = 4.33i + 2.5j m

Here, we make θ time-dependent: θ(t) = ωt (angle increases linearly with time).

#### Connection to Lecture 3 (Coordinate Systems)
**Cartesian representation** (x, y):
- Requires **two equations** (one for x, one for y)
- Both involve trigonometric functions

**Polar representation** (r, θ):
- r(t) = R (constant radius)
- θ(t) = ωt (linear in time)
- Simpler! Only one variable (θ) changes.

**Conclusion**: Circular motion is naturally described in **polar coordinates**. This illustrates L3 Homework Problem 4 (system selection depends on problem geometry).

#### Physical Examples
- Satellite in circular orbit around Earth
- Point on rim of rotating wheel (view from axis)
- Carousel or merry-go-round ride
- Centrifuge sample holder
- Electron orbiting nucleus (classical Bohr model)

````python
# Example 1: Circular Motion
# This extends L1 Homework Problem 2 by parameterizing with time

# Parameters
R = 2.0 # radius in meters
omega = 1.0 # angular velocity in rad/s
period = 2*np.pi/omega # time for one complete revolution
t_vals = np.linspace(0, period, 100) # 100 points over one revolution

# Position equations (from L1 Homework Problem 2)
x_circle = R * np.cos(omega * t_vals)
y_circle = R * np.sin(omega * t_vals)

# Visualization
plt.figure()
plt.plot(x_circle, y_circle, 'b-', linewidth=2, label='Circular path')
plt.plot(x_circle[0], y_circle[0], 'go', markersize=10, label='Start (t=0)')
plt.plot(x_circle[-1], y_circle[-1], 'ro', markersize=10, label=f'End (t={period:.2f}s)')
plt.plot(0, 0, 'k+', markersize=15, label='Center (origin)')
plt.xlabel('x (m)')
plt.ylabel('y (m)')
plt.title(f'Example 1: Uniform Circular Motion (R={R}m, ω={omega} rad/s)')
plt.axis('equal') # Equal scaling on both axes (circle looks circular)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

# Calculate motion properties
speed = R * omega # For circular motion: |v| = Rω (from L2)
circumference = 2 * np.pi * R

print("Motion Properties:")
print(f"Radius: {R} m")
print(f"Angular velocity: {omega} rad/s = {omega*180/np.pi:.1f} deg/s")
print(f"Period (one revolution): {period:.2f} seconds")
print(f"Speed (constant): {speed:.2f} m/s")
print(f"Circumference: {circumference:.2f} m")
print()
print("Coordinate System Analysis:")
print("Cartesian: x(t) = 2*cos(t), y(t) = 2*sin(t) [two equations]")
print("Polar: r(t) = 2 (constant), θ(t) = t [simpler!]")
print("Best choice: POLAR coordinates (from L3)")
````

### Example 2: Spiral Motion (Archimedes Spiral)

#### Mathematical Description
Polar form:
- r(t) = a + bt (radius increases linearly with time)
- θ(t) = ωt (angle increases linearly with time)

Cartesian form (converted from polar):
- x(t) = (a + bt) cos(ωt)
- y(t) = (a + bt) sin(ωt)

Where:
- **a** = initial radius (meters)
- **b** = radius growth rate (meters/second)
- **ω** = angular velocity (radians/second)

#### Connection to Lecture 3
Spiral motion demonstrates **when polar coordinates simplify the problem**:
- Polar: r = a + bt (linear), θ = ωt (linear) - Both are simple!
- Cartesian: x and y involve products of linear and trig functions - More complex!

This is the converse of circular motion:
- **Circle**: r = constant (simple in polar)
- **Spiral**: r = f(t) (still relatively simple in polar)

#### Physical Examples
- Spiral galaxy arm rotation
- Coiled spring being stretched
- Hurricane or tornado air flow pattern
- Vinyl record groove (viewed from top)
- Nautilus shell growth pattern
- Spider web construction

````python
# Example 2: Spiral Motion (Archimedes Spiral)
# Extends L3 coordinate system concepts: polar → Cartesian conversion

# Parameters
a = 0.5 # initial radius (m)
b = 0.2 # radius growth rate (m/s)
omega = 1.0 # angular velocity (rad/s)
t_vals = np.linspace(0, 4*np.pi/omega, 200) # 2 full rotations

# Position in polar coordinates (natural representation)
r_spiral = a + b * t_vals
theta_spiral = omega * t_vals

# Convert to Cartesian for plotting (using L3 conversion formulas)
# x = r*cos(θ), y = r*sin(θ)
x_spiral = r_spiral * np.cos(theta_spiral)
y_spiral = r_spiral * np.sin(theta_spiral)

# Visualization
plt.figure()
plt.plot(x_spiral, y_spiral, 'purple', linewidth=2, label='Spiral path')
plt.plot(x_spiral[0], y_spiral[0], 'go', markersize=10, label='Start')
plt.plot(x_spiral[-1], y_spiral[-1], 'ro', markersize=10, label='End')
plt.plot(0, 0, 'k+', markersize=15, label='Center')
plt.xlabel('x (m)')
plt.ylabel('y (m)')
plt.title(f'Example 2: Archimedes Spiral (r = {a} + {b}t, θ = {omega}t)')
plt.axis('equal')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

# Motion properties
print("Motion Properties:")
print(f"Initial radius: {a} m")
print(f"Final radius: {r_spiral[-1]:.2f} m")
print(f"Radius growth rate: {b} m/s")
print(f"Total rotations: {t_vals[-1]/(2*np.pi):.2f}")
print(f"Duration: {t_vals[-1]:.2f} seconds")
print()
print("Coordinate System Analysis:")
print("Polar: r(t) = 0.5 + 0.2t, θ(t) = t [both linear - simple!]")
print("Cartesian: x(t) = (0.5+0.2t)*cos(t), y(t) = (0.5+0.2t)*sin(t) [product - complex!]")
print("Best choice: POLAR coordinates (from L3 guidance)")
````

### Example 3: Lissajous Curve (Figure-8 Family)

#### Mathematical Description
Position equations:
- x(t) = A sin(ωₓt + φ)
- y(t) = B sin(ωᵧt)

Where:
- **A, B** = amplitudes in x and y directions (meters)
- **ωₓ, ωᵧ** = angular frequencies (radians/second)
- **φ** = phase shift between x and y oscillations (radians)

#### Connection to Lecture 1
Lissajous curves are **parametric equations** (r(t) = x(t)i + y(t)j) where **both components oscillate sinusoidally**.

Compare to circular motion (Example 1):
- **Circle**: ωₓ = ωᵧ, φ = π/2 (90°), A = B
- **Lissajous**: ωₓ ≠ ωᵧ (different frequencies create complex patterns)

#### Frequency Ratio Effects
The shape depends on the ratio ωₓ/ωᵧ:
- **ωₓ/ωᵧ = 1:1** → Circle or ellipse (depends on φ and A, B)
- **ωₓ/ωᵧ = 1:2** → Figure-8 pattern
- **ωₓ/ωᵧ = 2:3** → Complex closed curve (shown below)
- **ωₓ/ωᵧ = irrational** → Never closes (dense filling of region)

#### Connection to Lecture 3
**Best coordinate system**: CARTESIAN
- x and y oscillate independently
- No natural center or axis of symmetry
- Polar coordinates would be unnecessarily complex

This demonstrates L3 Homework Problem 4: coordinate choice depends on problem structure, not just geometry.

#### Physical Examples
- Oscilloscope display of two AC signals
- Double pendulum trace (certain modes)
- Harmonograph drawing machine
- Two-dimensional harmonic oscillator (quantum mechanics)
- Charged particle in crossed electric/magnetic fields

````python
# Example 3: Lissajous Curve (2:3 frequency ratio)
# Demonstrates Cartesian coordinates for independent x-y oscillations

# Parameters
A, B = 2.0, 1.5 # amplitudes (m)
omega_x, omega_y = 2.0, 3.0 # frequency ratio 2:3
phi = 0 # phase shift (try changing to np.pi/2 for different shape!)
t_vals = np.linspace(0, 2*np.pi, 500) # enough points for smooth curve

# Position equations (parametric form from L1)
x_liss = A * np.sin(omega_x * t_vals + phi)
y_liss = B * np.sin(omega_y * t_vals)

# Visualization
plt.figure()
plt.plot(x_liss, y_liss, 'orange', linewidth=2, label='Lissajous curve')
plt.plot(x_liss[0], y_liss[0], 'go', markersize=10, label='Start')
plt.xlabel('x (m)')
plt.ylabel('y (m)')
plt.title(f'Example 3: Lissajous Curve (ωₓ:ωᵧ = {int(omega_x)}:{int(omega_y)})')
plt.axis('equal')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

# Motion properties
print("Motion Properties:")
print(f"X amplitude: {A} m")
print(f"Y amplitude: {B} m")
print(f"Frequency ratio: ωₓ/ωᵧ = {omega_x}/{omega_y} = {omega_x/omega_y:.2f}")
print(f"Phase shift: {phi} radians = {phi*180/np.pi:.0f} degrees")
print()
print("Coordinate System Analysis:")
print("Cartesian: x(t) = 2*sin(2t), y(t) = 1.5*sin(3t) [natural representation]")
print("Polar: Would require complex r(θ) relationship [not natural]")
print("Best choice: CARTESIAN (independent oscillations in x and y)")
print()
print("Extension idea: Try changing phi to np.pi/2 and rerun to see how phase affects shape!")
````

### Example 4: Elliptical Motion

#### Mathematical Description
Position equations:
- x(t) = a cos(ωt)
- y(t) = b sin(ωt)

Where:
- **a** = semi-major axis (larger radius, meters)
- **b** = semi-minor axis (smaller radius, meters)
- **ω** = angular velocity (radians/second)

#### Connection to Lecture 1 Homework
Elliptical motion is a **generalization of circular motion**:
- **Circle** (L1 HW Problem 2): a = b = R (equal radii)
- **Ellipse**: a ≠ b (different radii in x and y)

If you set a = b in the code below, you'll get circular motion (Example 1).

#### Eccentricity
The "flatness" of an ellipse is measured by **eccentricity** e:
$$e = \sqrt{1 - \left(\frac{b}{a}\right)^2}$$

- e = 0: Perfect circle (b = a)
- 0 < e < 1: Ellipse (the example below has e ≈ 0.866)
- e = 1: Parabola (not closed)
- e > 1: Hyperbola (not closed)

#### Connection to Lecture 3
**Best coordinate system**: CARTESIAN
- Although similar to circular motion, ellipse has **no single center-distance** (r varies)
- Cartesian separates the two axes cleanly: x-direction has radius a, y-direction has radius b
- Polar would require complex r(θ) formula

This shows that **geometry alone doesn't determine coordinate choice** - ellipse is curved like circle, but Cartesian is still better!

#### Physical Examples
- Planetary orbits (Kepler's first law: planets move in ellipses with Sun at one focus)
- Satellite orbits (elliptical transfer orbits, e.g., Hohmann transfer)
- Particle in anisotropic harmonic potential (different spring constants in x and y)
- Elliptical exercise machine path

````python
# Example 4: Elliptical Motion
# Generalizes L1 Homework Problem 2 (circle) to ellipse

# Parameters
a, b = 3.0, 1.5 # semi-major and semi-minor axes (m)
omega = 1.0 # angular velocity (rad/s)
t_vals = np.linspace(0, 2*np.pi/omega, 100)

# Position equations
x_ellipse = a * np.cos(omega * t_vals)
y_ellipse = b * np.sin(omega * t_vals)

# Calculate eccentricity
eccentricity = np.sqrt(1 - (b/a)**2)

# Visualization
plt.figure()
plt.plot(x_ellipse, y_ellipse, 'green', linewidth=2, label='Elliptical path')
plt.plot(x_ellipse[0], y_ellipse[0], 'go', markersize=10, label='Start')
plt.plot(0, 0, 'k+', markersize=15, label='Center')
# Draw axes for reference
plt.plot([-a, a], [0, 0], 'k--', alpha=0.3, label='Major axis')
plt.plot([0, 0], [-b, b], 'k--', alpha=0.3, label='Minor axis')
plt.xlabel('x (m)')
plt.ylabel('y (m)')
plt.title(f'Example 4: Elliptical Motion (a={a}m, b={b}m, e={eccentricity:.3f})')
plt.axis('equal')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

# Motion properties
period = 2*np.pi/omega
print("Motion Properties:")
print(f"Semi-major axis (a): {a} m")
print(f"Semi-minor axis (b): {b} m")
print(f"Eccentricity (e): {eccentricity:.3f}")
print(f"Period: {period:.2f} seconds")
print(f"Angular velocity: {omega} rad/s")
print()
print("Coordinate System Analysis:")
print("Cartesian: x(t) = 3*cos(t), y(t) = 1.5*sin(t) [clean separation of axes]")
print("Polar: r(θ) = ab/sqrt((b*cos(θ))² + (a*sin(θ))²) [complex!]")
print("Best choice: CARTESIAN (despite curved path)")
print()
print("Try this: Set a = b = 2.0 and rerun. You'll get circular motion (Example 1)!")
````

### Example 5: Cardioid (Heart-Shaped Curve)

#### Mathematical Description
Polar form:
- r(θ) = a(1 + cos(θ))
- θ = parameter (not necessarily time)

For animation, we make θ time-dependent: θ(t) = ωt

Cartesian form (converted from polar using L3 formulas):
- x(t) = a(1 + cos(ωt)) cos(ωt)
- y(t) = a(1 + cos(ωt)) sin(ωt)

Where:
- **a** = scale parameter (meters)
- **ω** = angular velocity (radians/second)

#### Connection to Lecture 3
Cardioid is a **polar curve** (naturally expressed as r(θ)):
- Polar: r = a(1 + cos(θ)) - Simple!
- Cartesian: Complex expressions involving products

This demonstrates L3 principle: **If curve is naturally described in one coordinate system, use that system.**

#### Special Properties
- **Cusp at origin**: r = 0 when θ = π (point touches origin)
- **Maximum distance**: r = 2a when θ = 0
- **Symmetry**: Symmetric about x-axis

#### Physical Examples
- Microphone polar pickup pattern (cardioid microphone)
- Particle trajectory in certain force fields
- Water fountain spray pattern
- Curve traced by point on circle rolling around another circle of same size (epicycloid special case)

#### Extension Challenge
What coordinate system would you use for **velocity** calculation? (Think about this before Part 3!)

````python
# Example 5: Cardioid (Heart-Shaped Curve)
# Demonstrates polar → Cartesian conversion from L3

# Parameters
a = 1.0 # scale parameter (m)
omega = 1.0 # angular velocity (rad/s)
theta = np.linspace(0, 2*np.pi, 500) # full rotation

# Position in polar form (natural representation)
r_cardioid = a * (1 + np.cos(theta))

# Convert to Cartesian using L3 conversion formulas
# x = r*cos(θ), y = r*sin(θ)
x_cardioid = r_cardioid * np.cos(theta)
y_cardioid = r_cardioid * np.sin(theta)

# Visualization
plt.figure()
plt.plot(x_cardioid, y_cardioid, 'red', linewidth=2, label='Cardioid')
plt.plot(x_cardioid[0], y_cardioid[0], 'go', markersize=10, label='Start (θ=0)')
plt.plot(0, 0, 'k+', markersize=15, label='Origin (cusp point)')
plt.xlabel('x (m)')
plt.ylabel('y (m)')
plt.title(f'Example 5: Cardioid (r = {a}(1 + cos(θ)))')
plt.axis('equal')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

# Motion properties
print("Motion Properties:")
print(f"Scale parameter (a): {a} m")
print(f"Maximum radius: {2*a} m (at θ=0)")
print(f"Minimum radius: 0 m (at θ=π, cusp at origin)")
print()
print("Coordinate System Analysis:")
print("Polar: r(θ) = 1*(1 + cos(θ)) [simple algebraic form]")
print("Cartesian: x = [1 + cos(θ)]*cos(θ), y = [1 + cos(θ)]*sin(θ) [product form]")
print("Best choice: POLAR (curve defined naturally in polar)")
print()
print("Fun fact: This is the polar pattern of a cardioid microphone!")
print("Challenge: How would you calculate velocity? (Hint: Consider L2 + L3 together)")
````

### Teacher Example: Cycloid (Rolling Wheel Trajectory)

**Physical Motivation**: 
When a wheel rolls along a straight line without slipping, any point on the wheel's rim traces a curve called a **cycloid**. This is not just a mathematical curiosity - it appears throughout engineering:

- **Brachistochrone problem**: Cycloid is the fastest descent curve (particle slides from A to B in minimum time)
- **Tautochrone problem**: Cycloid is the isochronous curve (pendulum period independent of amplitude)
- **Gear teeth**: Cycloidal gear profiles for smooth meshing
- **Architecture**: Cycloidal arches distribute loads efficiently

**Mathematical Description**:

For a wheel of radius R rolling with angular velocity ω:

$$x(t) = R(\omega t - \sin(\omega t))$$
$$y(t) = R(1 - \cos(\omega t))$$

Where:
- **R** = wheel radius (meters)
- **ω** = rolling angular velocity (radians/second)
- **t** = time (seconds)

**Derivation** (optional, for curious students):
- Wheel center moves horizontally: x_center = Rωt
- Point on rim rotates relative to center: Δx = -R*sin(ωt), Δy = -R*cos(ωt)
- Total position: x = x_center + Δx, y = R + Δy
- Simplify: y(t) = R - R*cos(ωt) = R(1 - cos(ωt))

**Special Features**:
- **Cusps**: y = 0 when ωt = 2πn (point touches ground, velocity = 0)
- **Maximum height**: y = 2R (top of wheel)
- **Periodicity**: Repeats every T = 2π/ω seconds (one wheel rotation)
- **Distance per cycle**: Δx = 2πR (wheel circumference)

**Connection to Lecture 3**:
- Best coordinate system: CARTESIAN (x and y have different functional forms)
- Polar would be complex (r varies non-uniformly with θ)

## PART 3: ANALYTICAL DERIVATIVES (15 minutes)

### Purpose
Calculate velocity v(t) and acceleration a(t) **symbolically** using SymPy. This extends Lecture 2 where you calculated derivatives by hand.

### Connection to Lecture 2

**Lecture 2 Homework Problem 1**:
- Given: r(t) = (2t²)i + (3t)j
- Calculate: v(t) = dr/dt = (4t)i + (3)j (by hand using power rule)
- Calculate: a(t) = dv/dt = (4)i + (0)j (constant acceleration)

**Extension here**:
- Your motion equations may be much more complex (cycloid, spiral, Lissajous)
- Hand calculation would require: chain rule, product rule, trig identities, simplification
- SymPy does this automatically - you focus on **interpreting** results

### From L2: The Kinematic Relationships

**Position**: r(t) = x(t)i + y(t)j

**Velocity** (first derivative of position):
$$\vec{v}(t) = \frac{d\vec{r}}{dt} = \frac{dx}{dt}\mathbf{i} + \frac{dy}{dt}\mathbf{j}$$

**Acceleration** (second derivative of position, or first derivative of velocity):
$$\vec{a}(t) = \frac{d\vec{v}}{dt} = \frac{d^2\vec{r}}{dt^2} = \frac{d^2x}{dt^2}\mathbf{i} + \frac{d^2y}{dt^2}\mathbf{j}$$

### Why Symbolic Differentiation?

**Advantages over numerical differentiation** (which you'll use in Part 4):
1. **Exact**: No approximation errors
2. **Analytical**: Get formula, not just numbers
3. **Insight**: Can see functional form (e.g., "acceleration is sinusoidal")
4. **Simplification**: SymPy can apply trig identities automatically

**Preparation for Unit 2-3**:
- Lagrangian mechanics requires ∂L/∂q and ∂L/∂q̇ (partial derivatives)
- SymPy will compute these automatically from kinetic and potential energy expressions
- This practical builds your skill with symbolic computation tools

### Teaching Note
**Common student question**: "Why learn to differentiate by hand if computers can do it?"

**Answer**: 
- You need to **understand** what a derivative means (rate of change)
- Hand calculation teaches the **process** and **rules** (power rule, chain rule, etc.)
- Once you understand, using tools is efficient and reduces errors
- Analogy: You learn arithmetic by hand, then use calculators for complex calculations

### Step 1: Define Symbolic Variables and Position Functions

In SymPy, we first declare which symbols are variables. For kinematics:
- **t** = time (independent variable)
- Parameters like R, ω are also symbols (but we'll substitute numbers later)

**Important**: Tell SymPy that variables are **real** and **positive** (if applicable). This helps simplification.

````python
# STEP 1: Define symbolic variables

# Time variable (independent variable)
t = sp.Symbol('t', real=True, positive=True)

# Parameters for cycloid (will substitute numerical values later)
R = sp.Symbol('R', real=True, positive=True) # wheel radius
omega = sp.Symbol('omega', real=True, positive=True) # angular velocity

print("Symbolic variables defined:")
print(f"t = {t} (time)")
print(f"R = {R} (radius parameter)")
print(f"omega = {omega} (angular velocity parameter)")
print()

# STEP 2: Define position equations symbolically
# For cycloid: x(t) = R(ωt - sin(ωt)), y(t) = R(1 - cos(ωt))

x_sym = R * (omega * t - sp.sin(omega * t))
y_sym = R * (1 - sp.cos(omega * t))

print("Position equations (symbolic):")
print(f"x(t) = {x_sym}")
print(f"y(t) = {y_sym}")
print()
print("These are exact symbolic expressions, not numerical approximations.")
print("SymPy will differentiate these algebraically, just like you would by hand.")
print()
````

### Step 2: Calculate Velocity (First Derivative)

Velocity is the time derivative of position: v = dr/dt

Component-wise:
- vx(t) = dx/dt
- vy(t) = dy/dt

In SymPy, we use `sp.diff(expression, variable)` to differentiate.

````python
# Calculate velocity components by differentiating position
# From L2: v = dr/dt, so vx = dx/dt and vy = dy/dt

vx_sym = sp.diff(x_sym, t)
vy_sym = sp.diff(y_sym, t)

print("="*70)
print("VELOCITY CALCULATION (Analytical)")
print("="*70)
print()
print("Derivative rules applied:")
print("d/dt[R(ωt - sin(ωt))] = R(ω - ω*cos(ωt)) = Rω(1 - cos(ωt))")
print("d/dt[R(1 - cos(ωt))] = R*ω*sin(ωt)")
print()
print("Results:")
print(f"vx(t) = {vx_sym}")
print(f"vy(t) = {vy_sym}")
print()

# Simplify expressions (SymPy can apply trig identities)
vx_simplified = sp.simplify(vx_sym)
vy_simplified = sp.simplify(vy_sym)

print("Simplified:")
print(f"vx(t) = {vx_simplified}")
print(f"vy(t) = {vy_simplified}")
print()

# Physical interpretation
print("Physical Interpretation:")
print("vx depends on (1 - cos(ωt)): zero when ωt = 0, 2π, 4π, ... (at cusps)")
print("vy depends on sin(ωt): zero at same times (cusps)")
print("Therefore: Speed |v| = 0 at cusps (particle momentarily at rest)")
print()

# Calculate speed magnitude symbolically
speed_sym = sp.sqrt(vx_sym**2 + vy_sym**2)
speed_simplified = sp.simplify(speed_sym)

print(f"Speed magnitude: |v| = sqrt(vx² + vy²) = {speed_simplified}")
print()
print("Note: SymPy might not simplify to the simplest form.")
print("Known result for cycloid: |v| = 2Rω*|sin(ωt/2)|")
print()
````

### Step 3: Calculate Acceleration (Second Derivative)

Acceleration is the time derivative of velocity: a = dv/dt

Or equivalently, the second derivative of position: a = d²r/dt²

Component-wise:
- ax(t) = dvx/dt = d²x/dt²
- ay(t) = dvy/dt = d²y/dt²

We can differentiate either the position equations twice, or the velocity equations once. Results are identical.

````python
# Calculate acceleration by differentiating velocity
# From L2: a = dv/dt, so ax = dvx/dt and ay = dvy/dt

ax_sym = sp.diff(vx_sym, t)
ay_sym = sp.diff(vy_sym, t)

print("="*70)
print("ACCELERATION CALCULATION (Analytical)")
print("="*70)
print()
print("Derivative rules applied:")
print("d/dt[Rω(1 - cos(ωt))] = Rω²*sin(ωt)")
print("d/dt[Rω*sin(ωt)] = Rω²*cos(ωt)")
print()
print("Results:")
print(f"ax(t) = {ax_sym}")
print(f"ay(t) = {ay_sym}")
print()

# Simplify
ax_simplified = sp.simplify(ax_sym)
ay_simplified = sp.simplify(ay_sym)

print("Simplified:")
print(f"ax(t) = {ax_simplified}")
print(f"ay(t) = {ay_simplified}")
print()

# Verify second derivative (should match)
ax_from_x = sp.diff(x_sym, t, 2) # Second derivative of x
ay_from_y = sp.diff(y_sym, t, 2) # Second derivative of y

print("Verification (differentiate position twice):")
print(f"d²x/dt² = {ax_from_x}")
print(f"d²y/dt² = {ay_from_y}")
print()
print(f"Matches dvx/dt? {sp.simplify(ax_from_x - ax_sym) == 0}")
print(f"Matches dvy/dt? {sp.simplify(ay_from_y - ay_sym) == 0}")
print()

# Physical interpretation
print("Physical Interpretation:")
print("Acceleration components are sinusoidal (oscillatory)")
print("Magnitude of acceleration: |a| = Rω² (constant magnitude!)")
print("Direction: Points toward instantaneous center of curvature")
print("At cusps: ax=0, ay=Rω² (pure vertical acceleration)")
print()

# Calculate acceleration magnitude
accel_mag_sym = sp.sqrt(ax_sym**2 + ay_sym**2)
accel_mag_simplified = sp.simplify(accel_mag_sym)

print(f"Acceleration magnitude: |a| = {accel_mag_simplified}")
print()

# Trig identity: sin²(x) + cos²(x) = 1
print("Using identity sin²(ωt) + cos²(ωt) = 1:")
print(f"|a| = Rω²*sqrt(sin²(ωt) + cos²(ωt)) = Rω²*sqrt(1) = Rω²")
print()
print("Result: Constant acceleration magnitude (like circular motion!)")
print()
````

### Step 4: Convert Symbolic Expressions to Numerical Functions

Symbolic expressions are great for insight, but we need **numbers** to plot and compare.

SymPy's `lambdify` function converts symbolic expressions to numerical Python functions:
- Input: Symbolic expression with parameters
- Substitute: Parameter values (R=1.0, ω=2.0)
- Output: Fast numpy function that accepts arrays

This is similar to Lecture 2 where you had a formula (like v = 4t) and evaluated it at specific times.

````python
# Convert symbolic expressions to numerical functions
# This allows us to plot them

# Parameter values (from Part 2)
R_val = 1.0 # meters
omega_val = 2.0 # rad/s

print("Converting symbolic expressions to numerical functions...")
print(f"Parameter values: R = {R_val} m, ω = {omega_val} rad/s")
print()

# Substitute parameter values into symbolic expressions
# Then convert to numpy functions using lambdify

x_func = sp.lambdify(t, x_sym.subs([(R, R_val), (omega, omega_val)]), 'numpy')
y_func = sp.lambdify(t, y_sym.subs([(R, R_val), (omega, omega_val)]), 'numpy')
vx_func = sp.lambdify(t, vx_sym.subs([(R, R_val), (omega, omega_val)]), 'numpy')
vy_func = sp.lambdify(t, vy_sym.subs([(R, R_val), (omega, omega_val)]), 'numpy')
ax_func = sp.lambdify(t, ax_sym.subs([(R, R_val), (omega, omega_val)]), 'numpy')
ay_func = sp.lambdify(t, ay_sym.subs([(R, R_val), (omega, omega_val)]), 'numpy')

print("Numerical functions created:")
print("x_func(t), y_func(t) - position components")
print("vx_func(t), vy_func(t) - velocity components")
print("ax_func(t), ay_func(t) - acceleration components")
print()

# Test at a specific time
t_test = 1.0 # seconds
print(f"Test evaluation at t = {t_test} s:")
print(f"x({t_test}) = {x_func(t_test):.4f} m")
print(f"y({t_test}) = {y_func(t_test):.4f} m")
print(f"vx({t_test}) = {vx_func(t_test):.4f} m/s")
print(f"vy({t_test}) = {vy_func(t_test):.4f} m/s")
print(f"ax({t_test}) = {ax_func(t_test):.4f} m/s²")
print(f"ay({t_test}) = {ay_func(t_test):.4f} m/s²")
print()

# Evaluate over full time range (from Part 2)
# These are the ANALYTICAL values (exact from calculus)
vx_analytical = vx_func(t_custom)
vy_analytical = vy_func(t_custom)
ax_analytical = ax_func(t_custom)
ay_analytical = ay_func(t_custom)

print(f"Evaluated over {len(t_custom)} time points from Part 2")
print("These are the EXACT values from symbolic differentiation.")
print("In Part 4, we'll compare these to numerical approximations.")
print()
print("Part 3 complete: Analytical derivatives calculated successfully!")
````

## PART 4: NUMERICAL VERIFICATION (15 minutes)

### Purpose
Compare analytical derivatives (from Part 3) with numerical derivatives (from finite differences). This verifies correctness and builds confidence in numerical methods.

### Connection to Previous Learning

**From Lecture 2**:
- You learned numerical differentiation using finite differences
- Formula: f'(t) ≈ [f(t+Δt) - f(t-Δt)] / (2Δt) (central difference)
- Used when you have data points but no equation

**From Practical 1 Station 2**:
- You calculated velocity and acceleration numerically from **experimental video data**
- No analytical solution available (real data doesn't have equations!)
- Numerical was the ONLY method

**Extension here**:
- Now you have BOTH analytical (Part 3) and can calculate numerical
- Compare them to verify analytical derivatives are correct
- Understand error sources in numerical methods

### Why Compare Both Methods?

**Pedagogical value**:
1. **Verification**: If numerical ≈ analytical, both are correct
2. **Error analysis**: Quantify numerical approximation error
3. **Confidence building**: Shows numerical methods work (important for real data)
4. **Method selection**: Learn when to use each approach

**Practical engineering**:
- Analytical: Use when you have equations (design, theoretical analysis)
- Numerical: Use when you have data (experiments, measurements, sensors)
- Both: Use for verification and validation

### Expected Results

**Errors should be small** (typically < 0.001):
- Velocity error: ~10⁻⁴ to 10⁻⁶
- Acceleration error: ~10⁻³ to 10⁻⁵ (second derivative is less accurate)

**If errors are large** (> 0.01), possible causes:
- Time step Δt too large (need more points)
- Edge effects (first/last points less accurate)
- Wrong formula used

---

````python
# Calculate numerical derivatives using finite differences
# From L2: Use central difference for better accuracy

# Time step (spacing between points)
dt = t_custom[1] - t_custom[0]
print(f"Time step: Δt = {dt:.6f} seconds")
print(f"Number of points: {len(t_custom)}")
print()

# Numerical first derivatives (velocity)
# np.gradient uses central difference: f'(i) ≈ [f(i+1) - f(i-1)] / (2Δt)
# At edges, uses forward/backward difference

vx_numerical = np.gradient(x_custom, dt)
vy_numerical = np.gradient(y_custom, dt)

print("Velocity calculated numerically using np.gradient()")
print("Method: Central difference (except at edges)")
print("Formula: f'(t) ≈ [f(t+Δt) - f(t-Δt)] / (2Δt)")
print()

# Numerical second derivatives (acceleration)
# Apply gradient twice: first to get velocity, then to velocity

ax_numerical = np.gradient(vx_numerical, dt)
ay_numerical = np.gradient(vy_numerical, dt)

print("Acceleration calculated numerically")
print("Method: Gradient of velocity (or second gradient of position)")
print("Note: Second derivatives are less accurate than first derivatives")
print()

# Alternative: Could also do ax = np.gradient(np.gradient(x_custom, dt), dt)
# Results are identical

print("Numerical derivatives complete!")
print("Now comparing to analytical results from Part 3...")
print()
````

````python
# Compare analytical vs numerical derivatives
# Create 4-panel plot showing both methods

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Panel 1: X-Velocity
axes[0, 0].plot(t_custom, vx_analytical, 'b-', linewidth=2, label='Analytical (SymPy)')
axes[0, 0].plot(t_custom, vx_numerical, 'r--', linewidth=1.5, alpha=0.7, label='Numerical (np.gradient)')
axes[0, 0].set_xlabel('Time (s)')
axes[0, 0].set_ylabel('vx (m/s)')
axes[0, 0].set_title('X-Velocity Comparison')
axes[0, 0].grid(True, alpha=0.3)
axes[0, 0].legend()

# Panel 2: Y-Velocity
axes[0, 1].plot(t_custom, vy_analytical, 'b-', linewidth=2, label='Analytical (SymPy)')
axes[0, 1].plot(t_custom, vy_numerical, 'r--', linewidth=1.5, alpha=0.7, label='Numerical (np.gradient)')
axes[0, 1].set_xlabel('Time (s)')
axes[0, 1].set_ylabel('vy (m/s)')
axes[0, 1].set_title('Y-Velocity Comparison')
axes[0, 1].grid(True, alpha=0.3)
axes[0, 1].legend()

# Panel 3: X-Acceleration
axes[1, 0].plot(t_custom, ax_analytical, 'b-', linewidth=2, label='Analytical (SymPy)')
axes[1, 0].plot(t_custom, ax_numerical, 'r--', linewidth=1.5, alpha=0.7, label='Numerical (np.gradient)')
axes[1, 0].set_xlabel('Time (s)')
axes[1, 0].set_ylabel('ax (m/s²)')
axes[1, 0].set_title('X-Acceleration Comparison')
axes[1, 0].grid(True, alpha=0.3)
axes[1, 0].legend()

# Panel 4: Y-Acceleration
axes[1, 1].plot(t_custom, ay_analytical, 'b-', linewidth=2, label='Analytical (SymPy)')
axes[1, 1].plot(t_custom, ay_numerical, 'r--', linewidth=1.5, alpha=0.7, label='Numerical (np.gradient)')
axes[1, 1].set_xlabel('Time (s)')
axes[1, 1].set_ylabel('ay (m/s²)')
axes[1, 1].set_title('Y-Acceleration Comparison')
axes[1, 1].grid(True, alpha=0.3)
axes[1, 1].legend()

plt.tight_layout()
plt.show()

print("Observation: The curves should overlap almost perfectly!")
print("Red dashed line (numerical) should be barely visible under blue solid (analytical).")
print()
````

````python
# Quantitative error analysis
# Calculate maximum and RMS errors

# Error = Analytical - Numerical (difference)
error_vx = vx_analytical - vx_numerical
error_vy = vy_analytical - vy_numerical
error_ax = ax_analytical - ax_numerical
error_ay = ay_analytical - ay_numerical

# Maximum absolute error
max_error_vx = np.max(np.abs(error_vx))
max_error_vy = np.max(np.abs(error_vy))
max_error_ax = np.max(np.abs(error_ax))
max_error_ay = np.max(np.abs(error_ay))

# RMS (root mean square) error - overall error magnitude
rms_error_vx = np.sqrt(np.mean(error_vx**2))
rms_error_vy = np.sqrt(np.mean(error_vy**2))
rms_error_ax = np.sqrt(np.mean(error_ax**2))
rms_error_ay = np.sqrt(np.mean(error_ay**2))

print("="*70)
print("ERROR ANALYSIS: Analytical vs Numerical")
print("="*70)
print()
print("Maximum Absolute Errors:")
print(f"Velocity X: {max_error_vx:.6f} m/s")
print(f"Velocity Y: {max_error_vy:.6f} m/s")
print(f"Acceleration X: {max_error_ax:.6f} m/s²")
print(f"Acceleration Y: {max_error_ay:.6f} m/s²")
print()
print("RMS (Root Mean Square) Errors:")
print(f"Velocity X: {rms_error_vx:.6f} m/s")
print(f"Velocity Y: {rms_error_vy:.6f} m/s")
print(f"Acceleration X: {rms_error_ax:.6f} m/s²")
print(f"Acceleration Y: {rms_error_ay:.6f} m/s²")
print()

# Interpret errors
print("Interpretation:")
if max_error_vx < 0.001 and max_error_vy < 0.001:
 print("Velocity errors are EXCELLENT (<< 0.1%)")
elif max_error_vx < 0.01 and max_error_vy < 0.01:
 print("Velocity errors are GOOD (< 1%)")
else:
 print("Velocity errors are LARGE - check time step or calculation")

if max_error_ax < 0.01 and max_error_ay < 0.01:
 print("Acceleration errors are EXCELLENT (second derivatives are harder!)")
elif max_error_ax < 0.1 and max_error_ay < 0.1:
 print("Acceleration errors are ACCEPTABLE (second derivatives have larger errors)")
else:
 print("Acceleration errors are LARGE - consider increasing resolution")

print()
print("Conclusion: Numerical methods accurately approximate analytical derivatives!")
print("This validates both:")
print("1. Our analytical derivatives are correct (SymPy worked)")
print("2. Numerical differentiation is reliable (np.gradient is accurate)")
print()
````

### Animation Components

**Four visual elements**:
1. **Full trajectory** (gray, thin): Complete path for reference
2. **Moving point** (large blue circle): Current position
3. **Trail** (thick blue line): Recent history (last 20% of motion)
4. **Velocity vector** (red arrow): Direction and speed

### Why Animate?

**Pedagogical value**:
- **Intuition**: See how motion evolves over time
- **Vector understanding**: Velocity tangent to path becomes obvious
- **Speed variation**: Arrow length shows speed changes

**Engineering value**:

### Technical Details

**`FuncAnimation` parameters**:
- `fig`: The figure to animate
- `func`: Update function called each frame
- `frames`: Number of frames (we use number of time points)
- `interval`: Milliseconds between frames (20 ms → 50 fps)
- `blit=True`: Faster rendering (only redraw changed elements)
- `repeat=True`: Loop animation continuously

**Update function**:
- Called once per frame
- Updates plot elements (point position, trail, arrow)
- Returns tuple of modified artists

````python
# Animation using FuncAnimation
# From L2: Visualize velocity vectors dynamically

from matplotlib.animation import FuncAnimation

# Setup figure and axis
fig, ax = plt.subplots(figsize=(10, 6))

# Plot full trajectory (static, for reference)
ax.plot(x_custom, y_custom, 'gray', linewidth=1, alpha=0.5, label='Full trajectory')

# Initialize animated elements (will be updated each frame)
point, = ax.plot([], [], 'bo', markersize=12, label='Current position')
trail, = ax.plot([], [], 'b-', linewidth=3, alpha=0.7, label='Trail (recent path)')

# Velocity arrow (quiver)
# quiver(x, y, vx, vy, ...) draws arrow from (x,y) with components (vx, vy)
# We'll update the arrow position and components each frame
velocity_arrow = ax.quiver([], [], [], [], 
 color='red', scale=10, width=0.005, 
 label='Velocity vector')

# Axis setup
ax.set_xlim(np.min(x_custom) - 0.5, np.max(x_custom) + 0.5)
ax.set_ylim(np.min(y_custom) - 0.5, np.max(y_custom) + 0.5)
ax.set_xlabel('x (m)', fontsize=12)
ax.set_ylabel('y (m)', fontsize=12)
ax.set_title('Custom Motion Animation with Velocity Vector', fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend(loc='upper right')
ax.set_aspect('equal', 'box')

# Trail length: Show last 20% of motion (adjust for desired trail length)
trail_length = int(0.2 * len(t_custom))

# Animation update function
def update(frame):
 """
 Called for each frame of animation.
 Updates: point position, trail, velocity arrow.
 """
 # Current position
 point.set_data([x_custom[frame]], [y_custom[frame]])
 
 # Trail: Show recent history
 trail_start = max(0, frame - trail_length)
 trail.set_data(x_custom[trail_start:frame+1], y_custom[trail_start:frame+1])
 
 # Update velocity arrow
 # Remove old arrow and create new one
 global velocity_arrow
 velocity_arrow.remove()
 velocity_arrow = ax.quiver(x_custom[frame], y_custom[frame],
 vx_analytical[frame], vy_analytical[frame],
 color='red', scale=10, width=0.005)
 
 return point, trail, velocity_arrow

# Create animation
# frames: Number of animation frames (one per time point)
# interval: Milliseconds between frames (20 ms = 50 fps)
# blit: Faster rendering (True recommended for smooth animation)
# repeat: Loop animation continuously

print("Creating animation...")
print(f"Total frames: {len(t_custom)}")
print(f"Frame rate: 50 fps (interval=20ms)")
print(f"Animation duration: {len(t_custom) * 0.02:.1f} seconds per loop")
print(f"Trail length: {trail_length} points ({trail_length/len(t_custom)*100:.0f}% of path)")
print()

anim = FuncAnimation(fig, update, frames=len(t_custom), 
 interval=20, blit=False, repeat=True)

plt.show()

print("Animation complete!")
print()
print("Observations:")
print("- Velocity arrow is always TANGENT to trajectory (direction of motion)")
print("- Arrow length varies with SPEED (longer = faster)")
print("- Trail shows recent path (helps see where motion has been)")
print("- For cycloid: Arrow shrinks to zero at cusps (motion stops momentarily)")
print()
````

````python
# Continue hardware analysis: Sensors, Control, Mechanical

# ==== 2. SENSOR REQUIREMENTS ====
print("2. SENSOR REQUIREMENTS")
print("(Position feedback for closed-loop control)")
print()

# Position resolution requirement
# Rule of thumb: 10× finer than desired accuracy
desired_accuracy = 0.001 # 1mm positioning accuracy
required_resolution = desired_accuracy / 10

print(f"Desired position accuracy: {desired_accuracy*1000:.1f} mm")
print(f"Required sensor resolution: {required_resolution*1000:.2f} mm (0.1 mm)")
print()

# Encoder calculation
# For rotary encoder on motor shaft:
# Linear resolution = (2πr) / (steps_per_rev)

encoder_ppr = 2000 # pulses per revolution (typical high-res encoder)
linear_resolution = (2 * np.pi * pulley_radius) / encoder_ppr

print(f"SENSOR RECOMMENDATION:")
print(f"Type: Quadrature rotary encoder on motor shaft")
print(f"Resolution: {encoder_ppr} PPR (pulses per revolution)")
print(f"Linear resolution: {linear_resolution*1000:.3f} mm per pulse")
print(f"Assessment: EXCELLENT ({linear_resolution*1000:.3f} mm < {required_resolution*1000:.2f} mm required)")
print()

# Sample rate requirement
# Nyquist: Sample at 2× highest frequency
# Motion frequency: f = ω/(2π)
motion_frequency = float(omega_val) / (2 * np.pi)
min_sample_rate = 10 * motion_frequency # 10× for good margin

print(f"Motion frequency: {motion_frequency:.2f} Hz (ω={float(omega_val):.2f} rad/s)")
print(f"Minimum sample rate: {min_sample_rate:.0f} Hz (10× motion frequency)")
print(f"Typical encoder rate: 1-10 kHz (MORE than sufficient)")
print()
print()

# ==== 3. CONTROL SYSTEM ====
print("3. CONTROL SYSTEM")
print("(Real-time trajectory tracking)")
print()

control_frequency = 1000 # Hz (typical for motion control)
control_period = 1 / control_frequency

print(f"CONTROLLER RECOMMENDATION:")
print(f"Platform: Arduino Mega 2560 or Teensy 4.0")
print(f"Control loop frequency: {control_frequency} Hz ({control_period*1000:.1f} ms period)")
print(f"Algorithm: PID position control + trajectory interpolation")
print()
print(f"Control frequency margin: {control_frequency / motion_frequency:.0f}× motion frequency")
print(f"(Rule: Control should be 10-100× faster than motion)")
print(f"Assessment: GOOD (100× margin is excellent for smooth tracking)")
print()

# Memory requirement
num_points = len(t_custom)
bytes_per_point = 12 # 3 floats (x, y, t) × 4 bytes each
total_memory = num_points * bytes_per_point

print(f"Memory requirement:")
print(f"Trajectory points: {num_points}")
print(f"Storage per point: {bytes_per_point} bytes (x, y, t as floats)")
print(f"Total memory: {total_memory} bytes = {total_memory/1024:.1f} KB")
print(f"Arduino Mega RAM: 8 KB (sufficient for trajectory storage)")
print()
print()

# ==== 4. MECHANICAL CONSTRAINTS ====
print("4. MECHANICAL CONSTRAINTS")
print("(Physical limitations and challenges)")
print()

print(f"Workspace:")
print(f"Required: {x_range:.2f} m × {y_range:.2f} m")
print(f"Typical XY table: 0.5 m × 0.5 m (e.g., 3D printer bed)")
print(f"Assessment: FEASIBLE (motion fits standard workspace)")
print()

# Cycloid-specific challenge
print(f"CYCLOID-SPECIFIC CHALLENGE: Cusps (y=0 points)")
print(f"At cusps: velocity = 0, but acceleration ≠ 0")
print(f"Problem: Instantaneous direction reversal (infinite jerk!)")
print(f"Solution 1: Smooth cusps with small radius (r=1mm)")
print(f"Solution 2: Reduce speed near cusps (slow down before reversing)")
print(f"Solution 3: Accept positioning error at cusps (relax tolerance)")
print()

# Jerk calculation (third derivative)
# j = da/dt, at cusps this is theoretically infinite
jerk_x = np.gradient(ax_analytical, t_custom[1]-t_custom[0])
jerk_y = np.gradient(ay_analytical, t_custom[1]-t_custom[0])
max_jerk = np.max(np.sqrt(jerk_x**2 + jerk_y**2))

print(f"Jerk (da/dt) analysis:")
print(f"Maximum jerk: {max_jerk:.1f} m/s³")
print(f"Note: High jerk causes mechanical vibrations and wear")
print(f"Typical limit: < 100 m/s³ for smooth operation")
if max_jerk > 100:
 print(f"WARNING: Jerk exceeds recommended limit - expect vibrations!")
else:
 print(f"Assessment: Acceptable jerk level")
print()
print()

print("="*70)
print("OVERALL FEASIBILITY: BUILDABLE with standard mechatronics hardware")
print("="*70)
print()
print("Estimated cost: $500-800 (motors, drivers, controller, frame)")
print("Build time: 2-3 weeks (design, assembly, calibration, testing)")
print()
````
