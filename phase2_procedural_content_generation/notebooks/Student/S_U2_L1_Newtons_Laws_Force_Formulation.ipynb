## Newton's Three Laws of Motion

### Historical Context

**Sir Isaac Newton** (1643-1727) formulated these laws in **Philosophiæ Naturalis Principia Mathematica** (1687). They form the foundation of **classical mechanics**.

---

### Newton's First Law: Law of Inertia

**Statement**: 
> *An object at rest stays at rest, and an object in motion continues in motion with constant velocity, unless acted upon by an external force.*

**Mathematical Form**:
$$\boxed{\sum \vec{F} = 0 \quad \Rightarrow \quad \vec{v} = \text{constant}}$$

**Physical Meaning**:
- Objects resist changes to their motion (inertia)
- Velocity remains constant (including zero) when no net force acts
- **Requires inertial reference frame** (non-accelerating frame)

**Examples**:
- Hockey puck sliding on frictionless ice
- Spacecraft drifting in deep space
- Book resting on table (forces balanced)

---

### Newton's Second Law: F = ma

**Statement**: 
> *The acceleration of an object is proportional to the net force acting on it and inversely proportional to its mass.*

**Mathematical Form**:
$$\boxed{\sum \vec{F} = m \vec{a}}$$

**Expanded Form**:
$$\sum \vec{F} = m \frac{d\vec{v}}{dt} = m \frac{d^2\vec{r}}{dt^2}$$

**Physical Meaning**:
- Force **causes** acceleration
- More force → more acceleration (proportional)
- More mass → less acceleration (inversely proportional)

**Component Form** (Cartesian coordinates):
$$\sum F_x = m a_x$$
$$\sum F_y = m a_y$$
$$\sum F_z = m a_z$$

**Key Insight**: This is a **vector equation**! Must be applied in each coordinate direction.

---

### Newton's Third Law: Action-Reaction

**Statement**: 
> *For every action, there is an equal and opposite reaction.*

**Mathematical Form**:
$$\boxed{\vec{F}_{AB} = -\vec{F}_{BA}}$$

Where:
- $\vec{F}_{AB}$ = force exerted by object A on object B
- $\vec{F}_{BA}$ = force exerted by object B on object A

**Physical Meaning**:
- Forces always come in pairs
- Act on **different objects**
- Equal magnitude, opposite direction
- Simultaneous (occur at same instant)

**Examples**:
- You push wall → wall pushes you back
- Earth pulls apple → apple pulls Earth
- Rocket exhaust pushes back → gases push rocket forward

**Common Misconception**: "If forces are equal and opposite, why do things move?"
- **Answer**: Action-reaction forces act on **different objects**
- Net force on *each object* determines *that object's* motion

---

## Types of Forces

When solving dynamics problems, we encounter two main categories:

### 1. Applied Forces (Active Forces)

**Definition**: Forces that do work on the system and can be specified explicitly.

**Examples**:
- **Gravity**: $\vec{F}_g = -mg\hat{j}$ (downward)
- **Applied push/pull**: Specified force $\vec{F}_{\text{applied}}$
- **Spring force**: $\vec{F}_s = -k\vec{x}$ (Hooke's law)
- **Damping**: $\vec{F}_d = -c\vec{v}$ (viscous damping)
- **Thrust**: Rocket/motor forces

**Characteristics**:
- Can be written as explicit functions of position, velocity, time
- Do work on system (can add/remove energy)

---

### 2. Constraint Forces (Reaction Forces)

**Definition**: Forces that enforce geometric constraints without doing work.

**Examples**:
- **Normal force** $N$: Keeps object on surface
- **Tension** $T$: Keeps rope/cable taut
- **Friction** (special case): $f \leq \mu N$
 - Static friction: prevents motion
 - Kinetic friction: opposes motion

**Characteristics**:
- Magnitude typically **unknown** initially (must be solved for)
- Do zero work on ideal constraints (perpendicular to motion)
- Enforce conditions like "stays on surface" or "rope doesn't stretch"

---

### Force-Based Method: Procedure

**STEP 1**: Choose a coordinate system 
**STEP 2**: Draw free body diagram (FBD) showing ALL forces 
**STEP 3**: Apply $\sum \vec{F} = m\vec{a}$ in component form 
**STEP 4**: Solve the resulting equations for unknowns

**Challenge**: Must include constraint forces explicitly!
- This works great for simple systems
- Becomes tedious for complex systems with many constraints
- **Motivation for Lagrangian methods** (next lecture!)

# EXAMPLE 1: Block Sliding Down a Frictionless Incline

## Physical Situation

A wooden block of mass $m = 5.0$ kg rests on a smooth (frictionless) inclined plane. The plane makes an angle $\theta = 30°$ with the horizontal ground. At time $t = 0$, the block is released from rest and begins to slide down the incline.

**Scenario Details**:
- The incline is long enough that the block accelerates uniformly
- Air resistance is negligible
- The surface is perfectly smooth (no friction)
- We want to find the acceleration of the block and the normal force from the incline

**Physical Question**: 
1. What is the acceleration of the block as it slides down?
2. What is the magnitude of the normal force exerted by the incline on the block?
3. How far does the block slide in the first 2.0 seconds?

---

## Given Information

| Parameter | Symbol | Value | Units | Physical Meaning |
|-----------|--------|-------|-------|------------------|
| Mass of block | $m$ | 5.0 | kg | Inertial property (resistance to acceleration) |
| Incline angle | $\theta$ | 30° | degrees | Angle between incline surface and horizontal |
| Initial velocity | $v_0$ | 0 | m/s | Block starts from rest |
| Initial position | $s_0$ | 0 | m | Origin at starting point |
| Friction coefficient | $\mu$ | 0 | — | Smooth surface (frictionless) |
| Gravitational acceleration | $g$ | 9.8 | m/s² | Standard Earth gravity |

**Known Physics**:
- Newton's Second Law: $\sum \vec{F} = m\vec{a}$
- Weight force: $\vec{W} = m\vec{g}$ (magnitude $W = mg$)
- Normal force: perpendicular to surface, magnitude $N$ (unknown)

---

## Detailed Solution

### STEP 1: Understand the Physics

**Forces Acting on the Block**:

1. **Gravitational Force (Weight)**: $\vec{W} = m\vec{g}$
 - Magnitude: $W = mg = (5.0)(9.8) = 49$ N
 - Direction: Vertically downward (toward Earth's center)
 - Always present near Earth's surface

2. **Normal Force**: $\vec{N}$
 - Magnitude: $N$ (unknown, to be determined)
 - Direction: Perpendicular to the incline surface
 - Physical meaning: Contact force from surface preventing block from penetrating

3. **Friction Force**: NONE (frictionless surface)
 - This simplifies our analysis significantly

**Constraints**:
- Block **must** stay on the incline surface (cannot penetrate or lift off)
- This constraint is enforced by the normal force $N$
- Motion is restricted to along the incline (1D problem)

**Expected Behavior**:
- Block will accelerate down the incline
- Acceleration should be less than $g$ (since only component of weight acts along incline)
- Normal force should be less than weight (since not all weight is supported)

---

### STEP 2: Choose a Coordinate System

**Choice**: Tilted coordinate system aligned with the incline

**Axes Definition**:
- **x-axis**: Along the incline, positive pointing down the slope
- **y-axis**: Perpendicular to incline, positive pointing away from surface

**Justification**:
1. **Simplifies motion**: Block moves only in x-direction (y = constant)
2. **Simplifies constraint**: $a_y = 0$ automatically (stays on surface)
3. **Reduces complexity**: One equation of motion instead of two coupled equations

**Alternative** (not chosen): Horizontal-vertical (ground-fixed) axes
- Would require solving coupled x-y equations
- Both $a_x$ and $a_y$ would be non-zero
- More algebraic work for same result

**Angle Relationships**:
- Incline makes angle $\theta = 30°$ with horizontal
- Our x-axis makes angle $\theta$ with horizontal
- Our y-axis makes angle $(90° - \theta) = 60°$ with horizontal

---

### STEP 3: Draw Free Body Diagram (FBD)

**Free Body Diagram**: Shows block as a point mass with all forces acting on it.

**Force Analysis**:

| Force | Direction | Components |
|-------|-----------|------------|
| Weight $\vec{W}$ | Vertical down | $W_x = mg\sin\theta$ (along incline)<br>$W_y = -mg\cos\theta$ (into surface) |
| Normal $\vec{N}$ | Perpendicular to surface | $N_x = 0$ (perpendicular to x)<br>$N_y = +N$ (away from surface) |

**Why these components?**

For the weight vector $\vec{W} = mg$ (downward):
- Draw weight vector straight down
- Project onto x-axis (along incline): requires $\sin\theta$ because weight makes angle $(90° - \theta)$ with incline
- Project onto y-axis (perpendicular): requires $\cos\theta$
- Check: $W_x^2 + W_y^2 = (mg\sin\theta)^2 + (mg\cos\theta)^2 = (mg)^2(\sin^2\theta + \cos^2\theta) = (mg)^2$ 

**Visualization**:
```
 y (perpendicular to surface)
 ↑
 | N (normal force)
 | ↑
 | |
 | |
 _____|_____|_____ ← incline surface
 x → | m |
 /| |
 / | ↓ W (weight, vertical)
 / | mg
 /θ |
 /____|_______________ horizontal
```

---

### STEP 4: Write Equations of Motion

**Apply Newton's Second Law**: $\sum \vec{F} = m\vec{a}$

**In component form**:

**x-direction (along incline)**:
$$\sum F_x = m a_x$$

Forces in x-direction:
- Weight component: $W_x = mg\sin\theta$ (positive, down the slope)
- Normal component: $N_x = 0$ (perpendicular to x-axis)

Equation:
$$\boxed{mg\sin\theta = m a_x}$$

**y-direction (perpendicular to incline)**:
$$\sum F_y = m a_y$$

Forces in y-direction:
- Weight component: $W_y = -mg\cos\theta$ (negative, into surface)
- Normal force: $N_y = N$ (positive, away from surface)

Constraint: Block stays on surface, so $a_y = 0$

Equation:
$$\boxed{N - mg\cos\theta = 0}$$

**Two equations, two unknowns** ($a_x$ and $N$):
$$\begin{cases}
mg\sin\theta = m a_x \quad \text{(motion equation)} \\
N - mg\cos\theta = 0 \quad \text{(constraint equation)}
\end{cases}$$

---

### STEP 5: Solve Algebraically

**For acceleration** $a_x$:

From motion equation:
$$mg\sin\theta = m a_x$$

Divide both sides by $m$:
$$a_x = g\sin\theta$$

**Key insight**: Acceleration is independent of mass! (All objects slide down at same rate, as Galileo discovered)

**For normal force** $N$:

From constraint equation:
$$N - mg\cos\theta = 0$$

Solve for $N$:
$$N = mg\cos\theta$$

**Physical interpretation**:
- $N < mg$ (since $\cos\theta < 1$ for $\theta > 0°$)
- Normal force supports only the perpendicular component of weight
- Parallel component causes acceleration

---

### STEP 6: Calculate Numerically

Now substitute given values to get numerical answers.

**Given**:
- $m = 5.0$ kg
- $\theta = 30°$
- $g = 9.8$ m/s²

**Calculate acceleration**:
$$a_x = g\sin\theta = (9.8)(\sin 30°) = (9.8)(0.5) = 4.9 \text{ m/s}^2$$

**Calculate normal force**:
$$N = mg\cos\theta = (5.0)(9.8)(\cos 30°) = (5.0)(9.8)(0.866) = 42.4 \text{ N}$$

**Calculate distance after $t = 2.0$ s**:

Use kinematic equation (constant acceleration):
$$s = s_0 + v_0 t + \frac{1}{2}a_x t^2$$

With $s_0 = 0$ and $v_0 = 0$:
$$s = \frac{1}{2}a_x t^2 = \frac{1}{2}(4.9)(2.0)^2 = \frac{1}{2}(4.9)(4.0) = 9.8 \text{ m}$$

---

````python
# STEP 6: Numerical Calculations with Code

import numpy as np
import matplotlib.pyplot as plt

# Given values
m = 5.0 # mass (kg)
theta_deg = 30 # angle (degrees)
g = 9.8 # gravity (m/s^2)
t_final = 2.0 # time to calculate distance (s)

# Convert angle to radians
theta_rad = np.radians(theta_deg)

# Calculate acceleration
a_x = g * np.sin(theta_rad)

# Calculate normal force
N = m * g * np.cos(theta_rad)

# Calculate distance at t = 2.0 s
s = 0.5 * a_x * t_final**2

# Calculate velocity at t = 2.0 s
v = a_x * t_final

# Display results
print("="*70)
print("NUMERICAL RESULTS: Block on Inclined Plane")
print("="*70)
print(f"\nGiven:")
print(f"Mass: m = {m} kg")
print(f"Incline angle: θ = {theta_deg}°")
print(f"Gravity: g = {g} m/s²")
print(f"\nCalculated:")
print(f"Acceleration: a = {a_x:.2f} m/s²")
print(f"Normal force: N = {N:.2f} N")
print(f"\nAfter t = {t_final} s:")
print(f"Distance: s = {s:.2f} m")
print(f"Velocity: v = {v:.2f} m/s")
print("="*70)

# Verification checks
print(f"\n Check: a < g? {a_x} < {g} → {a_x < g}")
print(f"Check: N < mg? {N:.2f} < {m*g:.2f} → {N < m*g}")
print(f"Check: For θ=30°, sin(30°)={np.sin(theta_rad):.3f}, cos(30°)={np.cos(theta_rad):.3f}")
````

### STEP 7: Verify and Interpret Results

**Unit Verification**:

1. **Acceleration**: $a_x = 4.9$ m/s²
 - Units: $[a] = [g][\sin\theta] = (\text{m/s}^2)(\text{dimensionless}) = \text{m/s}^2$ 
 
2. **Normal force**: $N = 42.4$ N
 - Units: $[N] = [m][g][\cos\theta] = (\text{kg})(\text{m/s}^2)(\text{dimensionless}) = \text{N}$ 
 
3. **Distance**: $s = 9.8$ m
 - Units: $[s] = [a][t^2] = (\text{m/s}^2)(\text{s}^2) = \text{m}$ 

**Limiting Case Analysis**:

1. **As $\theta \to 0°$** (horizontal surface):
 - $\sin(0°) = 0 \Rightarrow a_x = 0$ (no acceleration on horizontal surface)
 - $\cos(0°) = 1 \Rightarrow N = mg$ (normal force equals full weight)

2. **As $\theta \to 90°$** (vertical drop):
 - $\sin(90°) = 1 \Rightarrow a_x = g$ (free fall!)
 - $\cos(90°) = 0 \Rightarrow N = 0$ (no contact with surface)

3. **For $\theta = 30°$**:
 - $a_x = 4.9$ m/s² $= g/2$ (exactly half of g, since $\sin 30° = 0.5$)
 - $N = 42.4$ N $< mg = 49$ N (normal force less than weight)

**Physical Interpretation**:

1. **Acceleration is independent of mass**: Heavier and lighter objects accelerate at the same rate (Galileo's insight!)

2. **Normal force is less than weight**: The incline only supports the perpendicular component of the weight. The parallel component is unbalanced and causes acceleration.

3. **Smooth vs. Real surfaces**: In reality, friction would reduce acceleration:
 - With friction: $a = g(\sin\theta - \mu\cos\theta)$
 - For $\theta = 30°$ and typical $\mu \approx 0.3$: $a \approx 2.4$ m/s² (about half our result)

4. **Energy perspective** (preview of next lecture):
 - Gravitational potential energy converts to kinetic energy
 - Energy conservation gives same result: $mgh = \frac{1}{2}mv^2$

---

**Answers**:
1. **Acceleration**: $a_x = 4.9$ m/s² down the incline
2. **Normal force**: $N = 42.4$ N perpendicular to surface
3. **Distance in 2.0 s**: $s = 9.8$ m along the incline

# EXAMPLE 2: Atwood Machine (Two Masses and a Pulley)

## Physical Situation

An **Atwood machine** consists of two masses connected by an inextensible (non-stretching) string that passes over an ideal (massless, frictionless) pulley. This classic apparatus was invented by George Atwood in 1784 to demonstrate Newton's laws and measure gravitational acceleration.

**Our System**:
- Mass 1: $m_1 = 3.0$ kg (heavier mass, will accelerate downward)
- Mass 2: $m_2 = 2.0$ kg (lighter mass, will accelerate upward)
- Connected by ideal rope over ideal pulley
- System released from rest at $t = 0$

**Physical Questions**:
1. What is the acceleration of each mass?
2. What is the tension in the rope?
3. How far does each mass move in 1.5 seconds?

**Why This Example Matters**:
- Demonstrates **constraint forces** (tension is unknown, must be solved)
- Shows how **connected systems** move together (same acceleration magnitude)
- Illustrates Newton's Third Law (tension pulls both masses equally)

---

## Given Information

| Parameter | Symbol | Value | Units | Physical Meaning |
|-----------|--------|-------|-------|------------------|
| Mass 1 (heavier) | $m_1$ | 3.0 | kg | Larger mass (will descend) |
| Mass 2 (lighter) | $m_2$ | 2.0 | kg | Smaller mass (will ascend) |
| Initial velocity (both) | $v_0$ | 0 | m/s | Released from rest |
| Rope type | — | Inextensible | — | Length constant, no stretching |
| Pulley | — | Ideal | — | Massless, frictionless |
| Gravitational acceleration | $g$ | 9.8 | m/s² | Standard Earth gravity |

**System Characteristics**:
- **Inextensible rope**: Both masses have same acceleration magnitude
- **Ideal pulley**: Tension same on both sides (no friction or inertia)
- **Constraint**: $|a_1| = |a_2| = a$ (call this $a$)

---

## Detailed Solution

### STEP 1: Understand the Physics

**Forces on Mass 1** ($m_1 = 3.0$ kg, heavier):
1. **Weight**: $W_1 = m_1 g$ downward
 - Magnitude: $(3.0)(9.8) = 29.4$ N
 - Direction: Downward (negative y-direction in our convention)

2. **Tension**: $T$ upward
 - Magnitude: $T$ (unknown, to be determined)
 - Direction: Upward (positive y-direction)
 - Physical meaning: Pull from rope

**Forces on Mass 2** ($m_2 = 2.0$ kg, lighter):
1. **Weight**: $W_2 = m_2 g$ downward
 - Magnitude: $(2.0)(9.8) = 19.6$ N
 - Direction: Downward

2. **Tension**: $T$ upward
 - Magnitude: $T$ (same as for $m_1$ due to ideal pulley)
 - Direction: Upward

**Constraints**:
1. **Rope inextensible**: If $m_1$ moves down by distance $d$, then $m_2$ moves up by distance $d$
 - Mathematically: $|a_1| = |a_2|$
 
2. **Ideal pulley**: Tension is the same on both sides
 - No friction losses
 - No pulley inertia

**Expected Behavior**:
- Heavier mass ($m_1$) will accelerate downward
- Lighter mass ($m_2$) will accelerate upward
- Acceleration magnitude less than $g$ (tension partially supports both masses)
- Tension must be between $m_2 g$ and $m_1 g$

---

### STEP 2: Choose a Coordinate System

**Choice**: Vertical axis for each mass, with consistent sign convention

**For Mass 1**:
- **Positive direction**: Downward (direction of expected motion)
- **Negative direction**: Upward

**For Mass 2**:
- **Positive direction**: Upward (direction of expected motion)
- **Negative direction**: Downward

**Justification**:
1. **Simplifies equations**: Acceleration $a$ is positive for both masses (moving in their positive directions)
2. **Constraint automatically satisfied**: $a_1 = a_2 = a$ (same positive value)
3. **Clear physical meaning**: Positive $a$ means masses accelerate as expected

**Alternative** (also valid): Use upward-positive for both masses
- Would require $a_1 = -a_2$ (opposite signs)
- More bookkeeping, same final result

**Key Point**: The choice of coordinate system doesn't change the physics, only the algebra!

---

### STEP 3: Draw Free Body Diagrams

We need **two separate FBDs**, one for each mass.

**Free Body Diagram for Mass 1** (heavier, $m_1 = 3.0$ kg):

```
 ↑ T (tension, upward)
 |
 |
 +-------+
 | |
 | m₁ | ← mass 1
 | |
 +-------+
 |
 ↓ W₁ = m₁g (weight, downward)
```

**Forces on $m_1$**:
- Tension $T$: upward (negative in our downward-positive convention)
- Weight $W_1 = m_1 g = 29.4$ N: downward (positive)

**Free Body Diagram for Mass 2** (lighter, $m_2 = 2.0$ kg):

```
 ↑ T (tension, upward)
 |
 |
 +-------+
 | |
 | m₂ | ← mass 2
 | |
 +-------+
 |
 ↓ W₂ = m₂g (weight, downward)
```

**Forces on $m_2$**:
- Tension $T$: upward (positive in our upward-positive convention)
- Weight $W_2 = m_2 g = 19.6$ N: downward (negative)

**Key Observations**:
1. Tension $T$ is the **same** for both masses (ideal pulley, Newton's 3rd Law)
2. Weight forces are **different** ($W_1 > W_2$)
3. Net force on system drives acceleration

---

### STEP 4: Write Equations of Motion

Apply Newton's Second Law to **each mass separately**: $\sum F = ma$

**For Mass 1** (downward-positive):

Sum of forces = mass × acceleration:
$$m_1 g - T = m_1 a$$

**Explanation**:
- $m_1 g$: Weight pulls downward (positive)
- $-T$: Tension pulls upward (negative)
- $m_1 a$: Mass accelerates downward with acceleration $a$

**For Mass 2** (upward-positive):

Sum of forces = mass × acceleration:
$$T - m_2 g = m_2 a$$

**Explanation**:
- $T$: Tension pulls upward (positive)
- $-m_2 g$: Weight pulls downward (negative)
- $m_2 a$: Mass accelerates upward with acceleration $a$

**System of Equations**:
$$\begin{cases}
m_1 g - T = m_1 a \quad \text{(equation 1: mass 1)} \\
T - m_2 g = m_2 a \quad \text{(equation 2: mass 2)}
\end{cases}$$

**Two equations, two unknowns** ($a$ and $T$)!

---

### STEP 5: Solve Algebraically

**Method**: Add the two equations to eliminate tension $T$.

**Add Equation 1 + Equation 2**:
$$(m_1 g - T) + (T - m_2 g) = m_1 a + m_2 a$$

Simplify (tensions cancel):
$$m_1 g - m_2 g = (m_1 + m_2) a$$

Factor:
$$(m_1 - m_2) g = (m_1 + m_2) a$$

**Solve for acceleration**:
$$\boxed{a = \frac{m_1 - m_2}{m_1 + m_2} g}$$

**Physical interpretation**:
- Acceleration depends on **difference** in masses (numerator)
- Normalized by **total mass** (denominator)
- If $m_1 = m_2$: $a = 0$ (balanced, no motion) 
- If $m_2 = 0$: $a = g$ (free fall) 

**Solve for tension** $T$:

From Equation 2: $T = m_2 a + m_2 g = m_2(a + g)$

Substitute $a = \frac{m_1 - m_2}{m_1 + m_2} g$:
$$T = m_2\left(\frac{m_1 - m_2}{m_1 + m_2} g + g\right)$$

Factor out $g$:
$$T = m_2 g \left(\frac{m_1 - m_2}{m_1 + m_2} + 1\right)$$

Common denominator:
$$T = m_2 g \left(\frac{m_1 - m_2 + m_1 + m_2}{m_1 + m_2}\right)$$

Simplify:
$$\boxed{T = \frac{2 m_1 m_2}{m_1 + m_2} g}$$

**Alternative form** (harmonic mean):
$$T = \frac{2}{\frac{1}{m_1} + \frac{1}{m_2}} g = 2 \left(\frac{1}{m_1} + \frac{1}{m_2}\right)^{-1} g$$

**Physical interpretation**:
- Tension is between $m_2 g$ and $m_1 g$ (as expected)
- If $m_1 = m_2 = m$: $T = mg$ (each mass fully supported) 
- If $m_1 \gg m_2$: $T \approx 2m_2 g$ (light mass pulled hard)

---

### STEP 6: Calculate Numerically

Substitute given values:
- $m_1 = 3.0$ kg
- $m_2 = 2.0$ kg
- $g = 9.8$ m/s²

**Calculate acceleration**:
$$a = \frac{m_1 - m_2}{m_1 + m_2} g = \frac{3.0 - 2.0}{3.0 + 2.0} (9.8) = \frac{1.0}{5.0} (9.8) = 0.2 \times 9.8 = 1.96 \text{ m/s}^2$$

**Calculate tension**:
$$T = \frac{2 m_1 m_2}{m_1 + m_2} g = \frac{2(3.0)(2.0)}{3.0 + 2.0} (9.8) = \frac{12.0}{5.0} (9.8) = 2.4 \times 9.8 = 23.52 \text{ N}$$

**Calculate distance after $t = 1.5$ s**:

Use kinematic equation (constant acceleration):
$$s = s_0 + v_0 t + \frac{1}{2} a t^2$$

With $s_0 = 0$ and $v_0 = 0$:
$$s = \frac{1}{2} a t^2 = \frac{1}{2} (1.96) (1.5)^2 = \frac{1}{2} (1.96) (2.25) = 2.205 \text{ m}$$

**Summary of Results**:
- Acceleration: $a = 1.96$ m/s² (both masses)
- Tension: $T = 23.52$ N
- Distance in 1.5 s: $s = 2.21$ m ($m_1$ down, $m_2$ up)

---

````python
# STEP 6: Numerical Calculations with Code

import numpy as np
import matplotlib.pyplot as plt

# Given values
m1 = 3.0 # heavier mass (kg)
m2 = 2.0 # lighter mass (kg)
g = 9.8 # gravity (m/s^2)
t_final = 1.5 # time for distance calculation (s)

# Calculate acceleration
a = ((m1 - m2) / (m1 + m2)) * g

# Calculate tension
T = (2 * m1 * m2 / (m1 + m2)) * g

# Calculate distance at t = 1.5 s
s = 0.5 * a * t_final**2

# Calculate velocity at t = 1.5 s
v = a * t_final

# Calculate weights for reference
W1 = m1 * g
W2 = m2 * g

# Display results
print("="*70)
print("NUMERICAL RESULTS: Atwood Machine")
print("="*70)
print(f"\nGiven:")
print(f"Mass 1 (heavier): m₁ = {m1} kg")
print(f"Mass 2 (lighter): m₂ = {m2} kg")
print(f"Gravity: g = {g} m/s²")
print(f"\nCalculated:")
print(f"Acceleration: a = {a:.2f} m/s²")
print(f"Tension: T = {T:.2f} N")
print(f"\nAfter t = {t_final} s:")
print(f"Distance: s = {s:.3f} m")
print(f"Velocity: v = {v:.2f} m/s")
print(f"\nReference:")
print(f"Weight 1: W₁ = {W1:.2f} N")
print(f"Weight 2: W₂ = {W2:.2f} N")
print("="*70)

# Verification checks
print(f"\n Check: a < g? {a:.2f} < {g} → {a < g}")
print(f"Check: W₂ < T < W₁? {W2:.2f} < {T:.2f} < {W1:.2f} → {W2 < T < W1}")
print(f"\n Special cases:")
print(f"If m₁ = m₂: a = 0 (no motion)")
print(f"Check: a = ({m1:.1f} - {m1:.1f})/({m1:.1f} + {m1:.1f}) × g = 0 ")
print(f"If m₂ = 0: a = g (free fall)")
print(f"Check: a = ({m1:.1f} - 0)/({m1:.1f} + 0) × g = g ")
````

### STEP 7: Verify and Interpret Results

**Unit Verification**:

1. **Acceleration**: $a = 1.96$ m/s²
 - Units: $[a] = \frac{[m_1 - m_2]}{[m_1 + m_2]} [g] = \frac{\text{kg}}{\text{kg}} (\text{m/s}^2) = \text{m/s}^2$ 
 
2. **Tension**: $T = 23.52$ N
 - Units: $[T] = \frac{[m_1][m_2]}{[m_1 + m_2]} [g] = \frac{\text{kg} \cdot \text{kg}}{\text{kg}} (\text{m/s}^2) = \text{kg} \cdot \text{m/s}^2 = \text{N}$ 
 
3. **Distance**: $s = 2.21$ m
 - Units: $[s] = [a][t^2] = (\text{m/s}^2)(\text{s}^2) = \text{m}$ 

**Limiting Case Analysis**:

1. **Equal masses** ($m_1 = m_2 = m$):
 - Acceleration: $a = \frac{m - m}{m + m} g = 0$ (system balanced)
 - Tension: $T = \frac{2m \cdot m}{m + m} g = mg$ (rope fully supports each mass)

2. **Very heavy mass 1** ($m_1 \gg m_2$):
 - Acceleration: $a \approx \frac{m_1}{m_1} g = g$ (approaches free fall)
 - Tension: $T \approx 2m_2 g$ (light mass pulled up hard)

3. **Mass 2 approaches zero** ($m_2 \to 0$):
 - Acceleration: $a = \frac{m_1 - 0}{m_1 + 0} g = g$ (free fall for $m_1$)
 - Tension: $T = 0$ (no rope needed)

4. **For our values** ($m_1 = 3$ kg, $m_2 = 2$ kg):
 - $a = 1.96$ m/s² = $g/5$ (20% of gravitational acceleration)
 - $W_2 = 19.6$ N $<$ $T = 23.52$ N $<$ $W_1 = 29.4$ N 

**Physical Interpretation**:

1. **Acceleration depends on mass difference**:
 - Small difference → small acceleration
 - Our case: $\frac{m_1 - m_2}{m_1 + m_2} = \frac{1}{5} = 20\%$ mass difference

2. **Tension is NOT the average weight**:
 - Average: $\frac{W_1 + W_2}{2} = 24.5$ N
 - Actual: $T = 23.52$ N (slightly less due to acceleration)

3. **Rope does work on both masses**:
 - On $m_1$: Negative work (opposes motion downward)
 - On $m_2$: Positive work (assists motion upward)
 - Net work by tension: Zero (internal force in isolated system)

4. **Energy perspective** (preview):
 - $m_1$ loses potential energy: $\Delta PE_1 = -m_1 g s$
 - $m_2$ gains potential energy: $\Delta PE_2 = +m_2 g s$
 - Net change converts to kinetic energy: $\frac{1}{2}(m_1 + m_2)v^2$
 - Check: $m_1 g s - m_2 g s = \frac{1}{2}(m_1 + m_2)v^2$
 - LHS: $(3.0)(9.8)(2.21) - (2.0)(9.8)(2.21) = 21.7$ J
 - RHS: $\frac{1}{2}(5.0)(2.94)^2 = 21.6$ J (matches within rounding)

---

**Answers**:
1. **Acceleration**: $a = 1.96$ m/s² (mass 1 downward, mass 2 upward)
2. **Tension**: $T = 23.52$ N (same throughout rope)
3. **Distance in 1.5 s**: $s = 2.21$ m (each mass moves this distance)

# Summary and Key Takeaways

## What We Learned Today

### 1. Newton's Three Laws
- **First Law** (Inertia): Objects maintain velocity unless forced to change
- **Second Law** (F = ma): Force causes acceleration, inversely proportional to mass
- **Third Law** (Action-Reaction): Forces come in equal-opposite pairs on different objects

### 2. Force-Based Formulation Procedure
1. Choose coordinate system
2. Draw free body diagram (all forces!)
3. Apply $\sum \vec{F} = m\vec{a}$ in components
4. Solve for unknowns

### 3. Types of Forces
- **Applied forces**: Gravity, springs, applied pushes (known)
- **Constraint forces**: Normal, tension (unknown, must solve)

### 4. Key Physics Insights
- Acceleration independent of mass (inclined plane)
- Constraint forces enforce geometric conditions (rope tension)
- Connected systems move with correlated accelerations
- Free body diagrams are essential for complex problems

---

## Limitations of Force-Based Methods

**Challenge**: Must explicitly include constraint forces

**Example**: Simple pendulum
- Forces: Tension $T$ (unknown) + Weight $mg$
- Equations become coupled and messy
- Tension does no work, yet we must solve for it!

**Preview of Next Lecture**: 
**Lagrangian Mechanics** (energy-based formulation)
- Uses kinetic and potential energy instead of forces
- Constraint forces automatically eliminated!
- Easier for complex systems with multiple constraints
- Same equations of motion, different (often simpler) derivation

**Spoiler**: For the simple pendulum:
- Newton's way: Messy coupled equations with tension
- Lagrange's way: One clean equation, no tension appears!
