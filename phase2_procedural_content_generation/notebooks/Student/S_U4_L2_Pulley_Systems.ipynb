---# Part 1: The Atwood Machine## What is an Atwood Machine?The **Atwood machine** is a classical physics problem:- Two masses connected by an inextensible rope- Rope passes over a massless, frictionless pulley- System moves under gravityNamed after **George Atwood** (1784) who used it to demonstrate Newton's laws.---## Constraint from Inextensible Rope**Key Insight**: The rope length is constant!Let:- $y_1(t)$ = position of mass 1 (measured downward from pulley)- $y_2(t)$ = position of mass 2 (measured downward from pulley)- $r$ = pulley radius- $\ell$ = total rope lengthTotal rope length:$$\ell = y_1 + y_2 + \pi r = \text{constant}$$**Differentiate once** (with respect to time):$$\frac{d\ell}{dt} = 0 \Rightarrow \dot{y}_1 + \dot{y}_2 = 0$$Therefore:$$v_1 = -v_2$$**Velocity constraint**: If mass 1 moves down at speed $v$, mass 2 moves up at speed $v$.**Differentiate again** (acceleration):$$a_1 = -a_2$$**Acceleration constraint**: Magnitudes equal, directions opposite.---## Free Body Diagrams### Mass 1 (assume moving down):- Weight: $m_1 g$ (downward)- Tension: $T$ (upward)Newton's 2nd law (taking down as positive):$$m_1 a_1 = m_1 g - T$$### Mass 2 (assume moving up):- Weight: $m_2 g$ (downward)- Tension: $T$ (upward)Newton's 2nd law (taking up as positive):$$m_2 a_2 = T - m_2 g$$### Using constraint $a_1 = -a_2 = a$ (defining $a$ as acceleration of $m_1$ downward):**Equation 1**: $m_1 a = m_1 g - T$ **Equation 2**: $m_2 (-a) = T - m_2 g$ → $-m_2 a = T - m_2 g$**Add equations** to eliminate $T$:$$m_1 a - m_2 a = m_1 g - m_2 g$$$$(m_1 - m_2)a = (m_1 - m_2)g$$**Solution for acceleration**:$$\boxed{a = \frac{(m_1 - m_2)g}{m_1 + m_2}}$$**Solution for tension** (substitute back):$$T = m_1 g - m_1 a = m_1 g - m_1 \frac{(m_1 - m_2)g}{m_1 + m_2}$$$$\boxed{T = \frac{2m_1 m_2 g}{m_1 + m_2}}$$---## 🤔 Special Cases (Physical Intuition)### Case 1: Equal masses ($m_1 = m_2 = m$)$$a = \frac{(m - m)g}{m + m} = 0$$$$T = \frac{2m \cdot m \cdot g}{m + m} = mg$$ **Makes sense**: System balanced, no acceleration. Tension equals weight.### Case 2: One mass much heavier ($m_1 \gg m_2$)$$a \approx \frac{m_1 g}{m_1} = g$$$$T \approx \frac{2m_1 m_2 g}{m_1} = 2m_2 g$$ **Makes sense**: Heavy mass falls freely (like $g$). Tension supports light mass (≈ $2m_2 g$).### Case 3: Very small difference ($m_1 = m + \Delta m$, $m_2 = m - \Delta m$, $\Delta m \ll m$)$$a \approx \frac{2\Delta m \cdot g}{2m} = \frac{\Delta m}{m} g$$ **Makes sense**: Small mass difference → small acceleration (Atwood used this to "slow down gravity"!)

---# Part 2: Mechanical Advantage## The Big Idea: Force-Velocity Tradeoff**Question**: Can we lift a heavy load with less force?**Answer**: YES! But there's a catch... ---## Mechanical Advantage (MA)**Definition**:$$MA = \frac{F_{\text{out}}}{F_{\text{in}}} = \frac{\text{Force on load}}{\text{Force applied by you}}$$**Key Principle**: Conservation of EnergyIn an **ideal system** (no friction):$$W_{\text{in}} = W_{\text{out}}$$$$F_{\text{in}} \cdot d_{\text{in}} = F_{\text{out}} \cdot d_{\text{out}}$$Therefore:$$\frac{F_{\text{out}}}{F_{\text{in}}} = \frac{d_{\text{in}}}{d_{\text{out}}}$$**The Tradeoff**:- **Higher MA** (lift heavier loads) → You pull **more distance**- **Lower MA** (less force multiplication) → You pull **less distance**---## Velocity Ratio (VR)Since work = force × distance, we can also think in terms of **power**:$$P_{\text{in}} = P_{\text{out}}$$$$F_{\text{in}} \cdot v_{\text{in}} = F_{\text{out}} \cdot v_{\text{out}}$$**Velocity Ratio**:$$VR = \frac{v_{\text{in}}}{v_{\text{out}}}$$For an **ideal system**:$$\boxed{MA = VR}$$You gain force at the expense of speed!---## Real Systems: EfficiencyReal pulleys have **friction** in bearings and rope bending.**Actual Mechanical Advantage** (AMA):$$AMA = \frac{\text{Actual output force}}{\text{Input force}}$$**Efficiency**:$$\eta = \frac{AMA}{IMA} = \frac{\text{Actual MA}}{\text{Ideal MA}}$$Typical values: $\eta \approx 0.90 - 0.95$ per pulleyFor **compound systems** (multiple pulleys):$$\eta_{\text{total}} = \eta_1 \cdot \eta_2 \cdot \eta_3 \cdots$$**Key insight**: Efficiency **decreases** with more pulleys!---## Example: Single Fixed Pulley```  ← Fixed pulley  / \\ / \\ (Pull) (Load)```**Analysis**:- Rope tension is same on both sides: $T = F_{\text{in}} = F_{\text{out}}$- **MA = 1** (no mechanical advantage!)- **VR = 1** (input and output move same distance)**Purpose**: Changes **direction** of force (useful!), but no force multiplication.---## Example: Single Movable Pulley``` ← Fixed pulley (changes direction)  | | Rope | ← Movable pulley (load attached here)  | Load  pulls here```**Analysis**:- Load is supported by **two rope segments**- Each segment has tension $T = F_{\text{in}}$- Total upward force on load: $2T = 2F_{\text{in}}$Force balance:$$2T = W \Rightarrow T = \frac{W}{2}$$**Mechanical Advantage**:$$\boxed{MA = 2}$$**Constraint equation**:- If load moves up by distance $d$, each rope segment shortens by $d$- Total rope pulled: $2d$- **VR = 2** (matches MA for ideal system)---## 🧮 General Rule for Simple SystemsFor **n supporting rope segments**:$$MA = n$$**Examples**:- 1 segment → MA = 1 (fixed pulley)- 2 segments → MA = 2 (one movable pulley)- 4 segments → MA = 4 (compound system with 2 movable pulleys)Let's simulate this!

![Pulley Systems — Mechanical Advantage Comparison](figs/U4_L2_pulley_comparison.png)*Pulley Systems — Mechanical Advantage Comparison*

````python
# Single Movable Pulley Simulation (with friction)
print("Single Movable Pulley Analysis")
print("=" * 60)

# System parameters
W = 500 # N (load weight)
mu = 0.05 # Friction coefficient (bearing friction)

# Case 1: Ideal system (no friction)
print("\\n Case 1: IDEAL System (no friction)")
MA_ideal = 2
F_in_ideal = W / MA_ideal
VR_ideal = 2
v_out_ideal = 1.0 # m/s (assume load moves up at 1 m/s)
v_in_ideal = VR_ideal * v_out_ideal

print(f" Load weight: W = {W} N")
print(f" Ideal MA: {MA_ideal}")
print(f" Required input force: F_in = {F_in_ideal:.2f} N")
print(f" Velocity ratio: VR = {VR_ideal}")
print(f" If load rises at {v_out_ideal} m/s:")
print(f" → Input rope must move at {v_in_ideal} m/s")

# Power analysis
P_in_ideal = F_in_ideal * v_in_ideal
P_out_ideal = W * v_out_ideal
print(f" Input power: P_in = {P_in_ideal:.2f} W")
print(f" Output power: P_out = {P_out_ideal:.2f} W")
print(f" Efficiency: η = {P_out_ideal/P_in_ideal * 100:.1f}%")

# Case 2: Real system (with friction)
print(f"\\n Case 2: REAL System (with friction μ = {mu:.2f})")

# Friction force on pulley
# Model: friction = μ × (load on pulley) = μ × W
F_friction = mu * W

# Actual force balance:
# 2*T = W + F_friction (need extra force to overcome friction)
F_in_real = (W + F_friction) / 2
MA_real = W / F_in_real

print(f" Friction force: F_friction = {F_friction:.2f} N")
print(f" Required input force: F_in = {F_in_real:.2f} N")
print(f" Actual MA: {MA_real:.3f}")
print(f" Efficiency: η = {MA_real/MA_ideal * 100:.2f}%")

# Energy analysis for 1 meter of lift
d_out = 1.0 # m
d_in = VR_ideal * d_out # m (2 meters of rope pulled)

W_in = F_in_real * d_in
W_out = W * d_out
W_lost = W_in - W_out

print(f"\\n Energy Analysis (lift load by {d_out} m):")
print(f" Input work: W_in = {W_in:.2f} J")
print(f" Output work: W_out = {W_out:.2f} J")
print(f" Energy lost to friction: {W_lost:.2f} J ({W_lost/W_in*100:.1f}% of input)")

# Summary table
print("\\n Summary: Ideal vs Real")
print(f"{'Parameter':<25} {'Ideal':<15} {'Real (μ=0.05)':<15}")
print(f"{'-'*55}")
print(f"{'Input force (N)':<25} {F_in_ideal:<15.2f} {F_in_real:<15.2f}")
print(f"{'Mechanical advantage':<25} {MA_ideal:<15.2f} {MA_real:<15.3f}")
print(f"{'Efficiency (%)':<25} {100:<15.1f} {MA_real/MA_ideal*100:<15.2f}")
print(f"{'Work per 1m lift (J)':<25} {W*d_out:<15.2f} {W_in:<15.2f}")

print("\\n Key Insight: Friction reduces MA and efficiency!")
````

## Block and Tackle: Multiple Pulleys Working TogetherA **block and tackle** is a system of multiple pulleys arranged to provide **high mechanical advantage**.### Structure:- **Fixed block**: Contains one or more fixed pulleys (changes direction)- **Moving block**: Contains one or more movable pulleys (attached to load)- **Rope**: Threaded through all pulleys in a continuous path---## General Formula**Count the rope segments supporting the moving block**:$$\boxed{MA = n}$$where $n$ = number of rope segments supporting the load.### Examples:#### 2-Pulley System (1 fixed + 1 movable):```Fixed:  / \\ / \\Moving: | | | Load Pull```- **n = 2** rope segments support load- **MA = 2**#### 4-Pulley System (2 fixed + 2 movable):```Fixed:  /| |\\ / | | \\Moving: | \\ / | Load Pull```- **n = 4** rope segments support load- **MA = 4**---## 🧮 Constraint EquationsFor a **4-pulley block and tackle**:**Rope length**:$$\ell = \ell_1 + \ell_2 + \ell_3 + \ell_4 + \text{(pulley circumferences)} = \text{const}$$If load moves up by $\Delta y$:- Each of the 4 supporting segments shortens by $\Delta y$- Total rope pulled in: $4 \Delta y$**Velocity relationship**:$$v_{\text{in}} = 4 \cdot v_{\text{load}}$$**Force relationship** (ideal):$$4 T = W \Rightarrow T = \frac{W}{4}$$Therefore **MA = 4** ---## Real Block and Tackle: Cumulative FrictionEach pulley introduces friction loss:$$\eta_{\text{per pulley}} \approx 0.95$$For **n pulleys**:$$\eta_{\text{total}} = (0.95)^n$$**Problem**: Efficiency drops fast!- 2 pulleys: $\eta = 0.90$ (90%)- 4 pulleys: $\eta = 0.81$ (81%)- 6 pulleys: $\eta = 0.74$ (74%)- 10 pulleys: $\eta = 0.60$ (60%)**Design Tradeoff**: More pulleys = higher MA but lower efficiency!Let's simulate a 4-pulley system:

````python
g = 9.81  # m/s²
# 4-Pulley Block and Tackle Simulation
print("4-Pulley Block and Tackle System")
print("=" * 60)

# System parameters
n_supporting_segments = 4 # MA = 4 (ideal)
load_mass = 200 # kg
load_weight = load_mass * g # N
mu_per_pulley = 0.05 # Friction per pulley
n_pulleys = 4 # Total number of pulleys in system

print(f"\nSystem Configuration:")
print(f" Number of supporting rope segments: {n_supporting_segments}")
print(f" Total number of pulleys: {n_pulleys}")
print(f" Load mass: {load_mass} kg")
print(f" Load weight: {load_weight:.2f} N")
print(f" Friction coefficient per pulley: μ = {mu_per_pulley}")

# Ideal case
MA_ideal = n_supporting_segments
F_in_ideal = load_weight / MA_ideal
VR = n_supporting_segments

print(f"\n IDEAL Analysis (no friction):")
print(f" Ideal MA: {MA_ideal}")
print(f" Required input force: F_in = {F_in_ideal:.2f} N")
print(f" Velocity ratio: VR = {VR}")
print(f" To lift load 1 m → pull {VR} m of rope")

# Real case with friction
# Model: each pulley reduces efficiency
eta_per_pulley = 1 - mu_per_pulley
eta_total = eta_per_pulley ** n_pulleys
MA_real = MA_ideal * eta_total
F_in_real = load_weight / MA_real

print(f"\n REAL Analysis (with friction):")
print(f" Efficiency per pulley: η_pulley = {eta_per_pulley:.3f}")
print(f" Total efficiency: η_total = {eta_per_pulley:.3f}^{n_pulleys} = {eta_total:.3f}")
print(f" Actual MA: {MA_real:.3f}")
print(f" Required input force: F_in = {F_in_real:.2f} N")
print(f" Extra force due to friction: {F_in_real - F_in_ideal:.2f} N ({(F_in_real/F_in_ideal - 1)*100:.1f}% increase)")

# Energy analysis for lifting 2 meters
h = 2.0 # m (lift height)
d_in = VR * h # Input rope distance

W_out_useful = load_weight * h # Useful work (lifting load)
W_in = F_in_real * d_in # Input work
W_lost_friction = W_in - W_out_useful # Energy lost to friction

print(f"\n Energy Analysis (lift load by {h} m):")
print(f" Input rope distance: {d_in:.1f} m")
print(f" Input work: W_in = {W_in:.2f} J")
print(f" Useful work (lifting): W_out = {W_out_useful:.2f} J")
print(f" Energy lost to friction: {W_lost_friction:.2f} J")
print(f" Percentage lost: {W_lost_friction/W_in * 100:.2f}%")

# Power analysis
v_load = 0.5 # m/s (load rising speed)
v_in = VR * v_load # Input rope speed

P_out = load_weight * v_load
P_in = F_in_real * v_in
P_lost = P_in - P_out

print(f"\n Power Analysis (load rising at {v_load} m/s):")
print(f" Input rope speed: {v_in} m/s")
print(f" Output power (lifting): P_out = {P_out:.2f} W")
print(f" Input power: P_in = {P_in:.2f} W")
print(f" Power lost to friction: {P_lost:.2f} W")
print(f" Efficiency check: η = P_out/P_in = {P_out/P_in:.3f} ")

# Comparison table
print(f"\n Summary Table:")
print(f"{'':<25} {'Ideal':<15} {'Real (4 pulleys)':<20}")
print(f"{'-'*60}")
print(f"{'Mechanical Advantage':<25} {MA_ideal:<15.1f} {MA_real:<20.3f}")
print(f"{'Input force (N)':<25} {F_in_ideal:<15.2f} {F_in_real:<20.2f}")
print(f"{'Efficiency (%)':<25} {100:<15.1f} {eta_total*100:<20.2f}")
print(f"{'Work to lift 2m (J)':<25} {W_out_useful:<15.2f} {W_in:<20.2f}")

print(f"\n Block and tackle provides high MA but with efficiency cost!")
````

Atwood Machine- **Constraint equation**: Inextensible rope → $v_1 = -v_2$ → $a_1 = -a_2$- **Acceleration**: $a = \frac{(m_1 - m_2)g}{m_1 + m_2}$- **Tension**: $T = \frac{2m_1 m_2 g}{m_1 + m_2}$- **Special cases**: Equal masses → no acceleration### 2. Mechanical Advantage- **Definition**: $MA = \frac{F_{\text{out}}}{F_{\text{in}}}$- **Ideal systems**: $MA = VR$ (velocity ratio)- **Energy principle**: $W_{\text{in}} = W_{\text{out}}$ → force-distance tradeoff- **Power principle**: $P_{\text{in}} = P_{\text{out}}$ → force-velocity tradeoff### 3.
