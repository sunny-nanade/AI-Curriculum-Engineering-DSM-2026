## Energy Methods in Dynamics

### Learning Objectives
1. Understand work done by forces
2. Apply work-energy theorem to solve dynamics problems
3. Distinguish conservative vs. non-conservative forces
4. Use potential energy for conservative systems
5. Analyze power and efficiency

### Context
**Alternative to force-based analysis**:
- Previous units: F=ma, torque equations
- **Energy methods**: Often simpler, especially for complex paths
- **Key advantage**: Scalar quantities (energy, work) vs. vector forces

**When to use energy methods**:
- Finding speeds without detailed trajectory
- Conservative forces (gravity, springs)
- Systems with constraints
- Efficiency and power calculations

### Key Concepts

**Work** (force through distance):
$$W = \int_{s_1}^{s_2} \mathbf{F} \cdot d\mathbf{s} = \int_{s_1}^{s_2} F \cos\theta \\, ds$$

**Kinetic Energy**:
$$T = \frac{1}{2}mv^2 \quad \text{(particle)}$$
$$T = \frac{1}{2}mv_{cm}^2 + \frac{1}{2}I\omega^2 \quad \text{(rigid body)}$$

**Work-Energy Theorem**:
$$W_{1\\to 2} = T_2 - T_1 = \Delta T$$

Work done by all forces equals change in kinetic energy!

**Power** (rate of work):
$$P = \frac{dW}{dt} = \mathbf{F} \cdot \mathbf{v}$$

## Part A: Work Done by Forces

### Definition of Work

**Work = Force component parallel to motion × distance**

$$W = \int \mathbf{F} \cdot d\mathbf{s} = \int F_s \\, ds$$

where $F_s = F\cos\theta$ is force component along path

### Sign Convention

- $W > 0$: Force aids motion (adds energy)
- $W < 0$: Force opposes motion (removes energy)
- $W = 0$: Force perpendicular to motion

### Common Forces

**1. Constant force**: $W = \mathbf{F} \cdot \Delta\mathbf{s} = F \Delta s \cos\theta$

**2. Gravity**:
$$W_g = -mg\Delta h = -mg(y_2 - y_1)$$
Negative when rising (opposes motion), positive when falling

**3. Spring force**: $F = -kx$
$$W_s = -\int_{x_1}^{x_2} kx \\, dx = -\frac{1}{2}k(x_2^2 - x_1^2)$$

**4. Friction** (constant): $W_f = -\mu_k N \cdot s$
Always negative (dissipates energy)

**5. Normal force**: $W_N = 0$
Perpendicular to motion, no work

### Work-Energy Theorem Derivation

Start with Newton's 2nd law: $F = ma = m\frac{dv}{dt}$

Multiply by $ds$:
$$F \\, ds = m\frac{dv}{dt} ds = m\frac{dv}{dt} v \\, dt = mv \\, dv$$

Integrate:
$$\int_{s_1}^{s_2} F \\, ds = \int_{v_1}^{v_2} mv \\, dv$$

$$W_{1\\to 2} = \frac{1}{2}mv_2^2 - \frac{1}{2}mv_1^2 = T_2 - T_1$$

### Key Insight

Work-energy theorem holds for **any path**!
- Don't need to know detailed trajectory
- Only need initial and final states
- Much simpler than solving F=ma along curved paths

```python
# Example 1: Block Sliding Down Incline

print("=" * 60)
print("Example 1: Block on Incline with Friction")
print("=" * 60)

# Given
m = 10.0  # kg
theta = np.radians(30)  # 30 degree incline
mu_k = 0.2  # coefficient of kinetic friction
s = 5.0  # m (distance traveled)
v1 = 0.0  # m/s (starts from rest)

g = 9.81  # m/s^2

print(f"\nGiven:")
print(f"Mass: m = {m} kg")
print(f"Incline angle: theta = {np.degrees(theta):.0f} degrees")
print(f"Friction coefficient: mu_k = {mu_k}")
print(f"Distance traveled: s = {s} m")
print(f"Initial velocity: v1 = {v1} m/s")

# Calculate work done by each force
# 1. Gravity (component parallel to incline)
W_gravity = m * g * np.sin(theta) * s

# 2. Normal force (perpendicular to motion)
W_normal = 0

# 3. Friction (opposes motion)
N = m * g * np.cos(theta)  # Normal force
W_friction = -mu_k * N * s

# Total work
W_total = W_gravity + W_normal + W_friction

print(f"\nWork Analysis:")
print(f"Normal force: N = {N:.2f} N")
print(f"Work by gravity: W_g = {W_gravity:.2f} J (positive: aids motion)")
print(f"Work by normal: W_N = {W_normal:.2f} J (perpendicular)")
print(f"Work by friction: W_f = {W_friction:.2f} J (negative: opposes)")
print(f"Total work: W_total = {W_total:.2f} J")

# Apply work-energy theorem
T1 = 0.5 * m * v1**2
T2 = T1 + W_total
v2 = np.sqrt(2 * T2 / m)

print(f"\nWork-Energy Theorem:")
print(f"Initial KE: T1 = {T1:.2f} J")
print(f"Final KE: T2 = {T2:.2f} J")
print(f"Final velocity: v2 = {v2:.2f} m/s")

# Verify with kinematics (for comparison)
a_net = g * np.sin(theta) - mu_k * g * np.cos(theta)
v2_kinematics = np.sqrt(v1**2 + 2 * a_net * s)

print(f"\nVerification (kinematics):")
print(f"Net acceleration: a = {a_net:.3f} m/s^2")
print(f"Final velocity: v2 = {v2_kinematics:.2f} m/s")
print(f"Difference: {abs(v2 - v2_kinematics):.6f} m/s (should be ~0)")

# Plotting
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Work breakdown
ax1 = axes[0]
works = [W_gravity, W_friction, W_total]
labels = ['Gravity', 'Friction', 'Total']
colors = ['green', 'red', 'blue']
x_pos = np.arange(len(labels))

bars = ax1.bar(x_pos, works, color=colors, alpha=0.7)
ax1.axhline(y=0, color='k', linestyle='-', linewidth=1)
ax1.set_ylabel('Work (J)', fontsize=11)
ax1.set_title('Work Done by Each Force', fontsize=12, fontweight='bold')
ax1.set_xticks(x_pos)
ax1.set_xticklabels(labels)
ax1.grid(True, alpha=0.3, axis='y')

# Add values on bars
for bar, work in zip(bars, works):
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2.,
             height,
             f'{work:.1f} J',
             ha='center',
             va='bottom' if height > 0 else 'top',
             fontsize=10,
             fontweight='bold')

# Energy flow
ax2 = axes[1]
energies = [T1, W_total, T2]
labels_e = ['Initial KE', 'Work Done', 'Final KE']
colors_e = ['blue', 'green', 'blue']
x_pos_e = np.arange(len(labels_e))

ax2.bar(x_pos_e, energies, color=colors_e, alpha=0.7)
ax2.set_ylabel('Energy (J)', fontsize=11)
ax2.set_title('Energy Balance: T1 + W = T2', fontsize=12, fontweight='bold')
ax2.set_xticks(x_pos_e)
ax2.set_xticklabels(labels_e)
ax2.grid(True, alpha=0.3, axis='y')

# Add equation annotation
ax2.text(0.5,
         max(energies)*0.9,
         f'{T1:.1f} + {W_total:.1f} = {T2:.1f}',
         ha='center',
         fontsize=12,
         bbox=dict(boxstyle='round',
                   facecolor='wheat',
                   alpha=0.5))

plt.tight_layout()
plt.show()

print(f"\nKey observations:")
print(f"1. Gravity does positive work (aids motion downhill)")
print(f"2. Friction does negative work (dissipates energy)")
print(f"3. Net work increases kinetic energy")
print(f"4. No need to track detailed path - only start and end!")

# TODO: Use these parameters in your solutions below
```

### What This Graph Tells You

**Work Breakdown (left):** Each force contributes differently to the total work:
- **Gravity** does **positive work** (aids motion down the incline) — it's the driving force
- **Friction** does **negative work** (opposes motion) — it dissipates energy as heat
- **Normal force** does **zero work** (perpendicular to displacement) — it constrains but doesn't energize

**Energy Flow (right):** Shows the transformation chain: gravitational PE converts to KE and heat. The work-energy theorem in action:
$$\Delta KE = W_{gravity} + W_{friction} = W_{net}$$

**Key Observations:**
- The block accelerates because $|W_{gravity}| > |W_{friction}|$ (net positive work)
- Friction converts mechanical energy irreversibly to thermal energy
- On a steeper incline: more gravity work, less friction (friction $\propto \cos\theta$)
- The normal force is the **constraint force** — it does no work but is essential for the problem

> **Student Takeaway:** The work-energy theorem connects forces to energy changes. Each force's work depends on its component ALONG the displacement direction. Perpendicular forces (like normal force) do zero work — they constrain motion without adding or removing energy.

## Part B: Conservative Forces & Potential Energy

### Conservative Forces

**Definition**: Work done is **path-independent**

$$\\oint \mathbf{F} \cdot d\mathbf{s} = 0$$

Work around closed loop = 0

**Examples**:
- Gravity: $W = mg\Delta h$ (depends only on height change)
- Springs: $W = -\frac{1}{2}k(x_2^2 - x_1^2)$ (depends only on displacement)
- Electrostatic: Coulomb force

### Non-Conservative Forces

**Definition**: Work done depends on path

**Examples**:
- Friction: longer path = more work dissipated
- Air resistance
- Viscous damping

### Potential Energy

For **conservative** forces only:
$$W_{1\\to 2} = -(V_2 - V_1) = -\Delta V$$

Work done = negative change in potential energy

**Common potential energies**:

1. **Gravity**: $V_g = mgh$ (h = height above reference)
 $$W_g = -\Delta V_g = -mg(h_2 - h_1)$$

2. **Spring**: $V_s = \frac{1}{2}kx^2$ (x = deformation from natural length)
 $$W_s = -\Delta V_s = -\frac{1}{2}k(x_2^2 - x_1^2)$$

### Conservation of Mechanical Energy

**If only conservative forces** (no friction, no damping):
$$T_1 + V_1 = T_2 + V_2 = \text{constant}$$

Total mechanical energy E = T + V is conserved!

**With non-conservative forces**:
$$T_1 + V_1 + W_{nc} = T_2 + V_2$$

where $W_{nc}$ = work by non-conservative forces

### Relationship: Force from Potential

**1D**: $F = -\frac{dV}{dx}$

**3D**: $\mathbf{F} = -\nabla V = -\left(\frac{\partial V}{\partial x}\hat{i} + \frac{\partial V}{\partial y}\hat{j} + \frac{\partial V}{\partial z}\hat{k}\right)$

Force points in direction of decreasing potential energy!

The pendulum below illustrates the most fundamental energy exchange in mechanics:
potential energy converting to kinetic energy and back. At the extremes, all energy is
potential; at the bottom, all is kinetic.

![Pendulum Energy Conversion — PE and KE Exchange](figs/U6_L1_pendulum_energy.png)

*Pendulum Energy Conversion — PE and KE Exchange*

> **Key Insight:** The pendulum is the canonical example of energy exchange. This back-and-forth
> between PE and KE is the essence of **all oscillatory systems** — springs, vibrating beams,

```python
# Example 2: Pendulum Released from Horizontal Position

print("=" * 60)
print("Example 2: Energy Conservation - Pendulum")
print("=" * 60)

# Given
m = 2.0  # kg
L = 1.5  # m (length of pendulum)
theta_0 = np.radians(90)  # Released from horizontal (90 degrees)
g = 9.81  # m/s^2

print(f"\nGiven:")
print(f"Mass: m = {m} kg")
print(f"Length: L = {L} m")
print(
    f" Initial angle: theta_0 = {
        np.degrees(theta_0):.0f} degrees (horizontal)")
print(f"Released from rest")

# Define reference: lowest point (theta = 0) as zero potential energy
# Position 1: theta = 90 degrees (horizontal)
h1 = L * (1 - np.cos(theta_0))  # Height above lowest point
v1 = 0  # Released from rest
T1 = 0.5 * m * v1**2
V1 = m * g * h1
E1 = T1 + V1

print(f"\nInitial State (horizontal position):")
print(f"Height: h1 = {h1:.3f} m")
print(f"Velocity: v1 = {v1:.2f} m/s")
print(f"Kinetic energy: T1 = {T1:.2f} J")
print(f"Potential energy: V1 = {V1:.2f} J")
print(f"Total energy: E1 = {E1:.2f} J")

# Position 2: theta = 0 (lowest point)
h2 = 0  # Reference level
V2 = m * g * h2

# Conservation of energy (no friction)
T2 = E1 - V2
v2 = np.sqrt(2 * T2 / m)

print(f"\nFinal State (lowest point):")
print(f"Height: h2 = {h2:.3f} m")
print(f"Potential energy: V2 = {V2:.2f} J")
print(f"Kinetic energy: T2 = {T2:.2f} J")
print(f"Velocity: v2 = {v2:.2f} m/s")
print(f"Total energy: E2 = {T2 + V2:.2f} J")

# Verify conservation
print(f"\nEnergy Conservation Check:")
print(f"E1 = {E1:.6f} J")
print(f"E2 = {T2 + V2:.6f} J")
print(f"Difference: {abs(E1 - (T2 + V2)):.9f} J (machine precision)")

# Theoretical maximum speed
v_max_theory = np.sqrt(2 * g * L)  # For release from horizontal
print(f"\nTheoretical maximum: v_max = sqrt(2gL) = {v_max_theory:.2f} m/s")
print(f"Our calculation: v2 = {v2:.2f} m/s")
print(f"Match: {abs(v2 - v_max_theory) < 1e-10}")

# Simulate full swing
theta_range = np.linspace(0, np.pi/2, 100)  # 0 to 90 degrees
h_range = L * (1 - np.cos(theta_range))
V_range = m * g * h_range
T_range = E1 - V_range
v_range = np.sqrt(2 * T_range / m)

# Plotting
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Energy vs angle
ax1 = axes[0, 0]
ax1.plot(
    np.degrees(theta_range),
    T_range,
    'b-',
    linewidth=2,
    label='Kinetic T')
ax1.plot(
    np.degrees(theta_range),
    V_range,
    'r-',
    linewidth=2,
    label='Potential V')
ax1.plot(
    np.degrees(theta_range),
    T_range + V_range,
    'g--',
    linewidth=2,
    label='Total E')
ax1.axvline(x=0, color='blue', linestyle=':', alpha=0.5, label='Lowest point')
ax1.axvline(x=90, color='red', linestyle=':', alpha=0.5, label='Release point')
ax1.set_xlabel('Angle (degrees)', fontsize=11)
ax1.set_ylabel('Energy (J)', fontsize=11)
ax1.set_title('Energy vs Angle', fontsize=12, fontweight='bold')
ax1.legend(fontsize=9)
ax1.grid(True, alpha=0.3)

# Velocity vs angle
ax2 = axes[0, 1]
ax2.plot(np.degrees(theta_range), v_range, 'b-', linewidth=2)
ax2.axhline(y=v2, color='red', linestyle='--', label=f'v_max = {v2:.2f} m/s')
ax2.set_xlabel('Angle (degrees)', fontsize=11)
ax2.set_ylabel('Velocity (m/s)', fontsize=11)
ax2.set_title('Velocity vs Angle', fontsize=12, fontweight='bold')
ax2.legend(fontsize=9)
ax2.grid(True, alpha=0.3)

# Pendulum diagram
ax3 = axes[1, 0]
angles_plot = [np.pi/2, np.pi/4, 0]  # 90, 45, 0 degrees
colors_plot = ['red', 'orange', 'blue']
labels_plot = ['Release (v=0)', 'Mid-swing', 'Bottom (v_max)']

for angle, color, label in zip(angles_plot, colors_plot, labels_plot):
    x = L * np.sin(angle)
    y = -L * np.cos(angle)
    ax3.plot([0, x], [0, y], 'o-', color=color,
             linewidth=2, markersize=8, label=label)

# Draw arc
arc_angles = np.linspace(0, np.pi/2, 50)
arc_x = L * np.sin(arc_angles)
arc_y = -L * np.cos(arc_angles)
ax3.plot(arc_x, arc_y, 'k--', alpha=0.3, linewidth=1)

ax3.axhline(y=0, color='brown', linewidth=3, label='Reference (V=0)')
ax3.set_xlim(-0.3, L+0.3)
ax3.set_ylim(-L-0.3, 0.5)
ax3.set_aspect('equal')
ax3.set_xlabel('x (m)', fontsize=11)
ax3.set_ylabel('y (m)', fontsize=11)
ax3.set_title('Pendulum Positions', fontsize=12, fontweight='bold')
ax3.legend(fontsize=9, loc='upper right')
ax3.grid(True, alpha=0.3)

# Energy bar chart
ax4 = axes[1, 1]
positions = ['Start\n(horizontal)', 'Bottom']
T_vals = [T1, T2]
V_vals = [V1, V2]
x_pos = np.arange(len(positions))
width = 0.35

ax4.bar(
    x_pos - width/2,
    T_vals,
    width,
    label='Kinetic T',
    color='blue',
    alpha=0.7)
ax4.bar(
    x_pos + width/2,
    V_vals,
    width,
    label='Potential V',
    color='red',
    alpha=0.7)
ax4.axhline(
    y=E1,
    color='green',
    linestyle='--',
    linewidth=2,
    label=f'Total E = {
        E1:.2f} J')
ax4.set_ylabel('Energy (J)', fontsize=11)
ax4.set_title('Energy Transformation', fontsize=12, fontweight='bold')
ax4.set_xticks(x_pos)
ax4.set_xticklabels(positions)
ax4.legend(fontsize=9)
ax4.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

print(f"\nKey observations:")
print(f"1. Total mechanical energy E = T + V is constant (no friction)")
print(f"2. Maximum speed occurs at lowest point (all PE converted to KE)")
print(f"3. Much simpler than solving F=ma in polar coordinates!")
print(f"4. Energy method gives speed directly, no trajectory needed")

# TODO: Use these parameters in your solutions below
```

### What This Graph Tells You

**Energy vs. Angle (top-left):** PE and KE are **mirror images** — when one increases, the other decreases by exactly the same amount. Total energy (dashed line) is perfectly constant.

**Velocity vs. Angle (top-right):** Maximum velocity occurs at the **lowest point** ($\theta = 0$) where all PE has converted to KE. The pendulum momentarily stops at the extremes.

> **Student Takeaway:** For conservative systems, the energy method bypasses force analysis entirely. You can find velocity at any position just from energy conservation. This is why Unit 6 exists — energy methods are often the fastest path to the answer.

## Part C: Power and Efficiency

### Power Definition

**Power = Rate of doing work**

$$P = \frac{dW}{dt}$$

Units: Watts (W) = J/s

### Power from Force and Velocity

$$P = \frac{dW}{dt} = \frac{d}{dt}\left(\int \mathbf{F} \cdot d\mathbf{s}\right) = \mathbf{F} \cdot \frac{d\mathbf{s}}{dt} = \mathbf{F} \cdot \mathbf{v}$$

**Power = Force · Velocity**

For 1D motion:
$$P = F \cdot v = Fv\cos\theta$$

where $\theta$ is angle between force and velocity

### Power from Torque and Angular Velocity

For rotational motion:
$$P = \boldsymbol{\tau} \cdot \boldsymbol{\omega}$$

For 1D rotation:
$$P = \tau \omega$$

### Average vs Instantaneous Power

**Average power**:
$$P_{avg} = \frac{W}{\Delta t}$$

**Instantaneous power**:
$$P = \\lim_{\Delta t \\to 0} \frac{\Delta W}{\Delta t} = \frac{dW}{dt}$$

### Efficiency

**Mechanical efficiency**:
$$\eta = \frac{P_{out}}{P_{in}} = \frac{\text{Useful work output}}{\text{Total energy input}}$$

Always: $0 < \eta < 1$ (or 0% to 100%)

Energy losses due to:
- Friction
- Heat dissipation
- Air resistance
- Mechanical imperfections

### Practical Units

- **Watts (W)**: SI unit
- **Kilowatts (kW)**: 1000 W
- **Horsepower (hp)**: 1 hp = 746 W
- **Energy**: 1 kWh = 3.6 MJ

### Applications

1. **Motor sizing**: Required power = Force × Velocity
2. **Vehicle performance**: P = F_drag × v (max speed)
3. **Machine design**: Torque-speed curves
4. **Energy consumption**: P × time = energy used

---


The diagram below shows a car climbing a hill at constant speed. Since velocity is constant,
all the engine's power goes into overcoming gravity and drag — none goes into acceleration.

![Car Climbing Hill — Power Analysis](figs/U6_L1_car_hill.png)

*Car Climbing Hill — Power Analysis*

> **Engineering Application:** The power equation $P = F_{drive} \cdot v$ directly sizes the
> engine or motor. For an EV climbing a 10% grade at 100 km/h, the required power is
> about 30 kW just to overcome gravity — this is why hill climbing drains EV batteries so fast!


```python
# Example 3: Car Climbing Hill - Power Analysis

print("=" * 60)
print("Example 3: Power Requirements for Car Climbing Hill")
print("=" * 60)

# Given
m = 1500.0  # kg (car mass)
v = 25.0  # m/s (constant speed = 90 km/h)
theta = np.radians(5)  # 5 degree incline
C_d = 0.3  # Drag coefficient
A = 2.5  # m^2 (frontal area)
rho = 1.2  # kg/m^3 (air density)
mu_r = 0.02  # Rolling resistance coefficient
eta = 0.85  # Powertrain efficiency (85%)

g = 9.81  # m/s^2

print(f"\nGiven:")
print(f"Mass: m = {m:.0f} kg")
print(f"Speed: v = {v:.1f} m/s ({v*3.6:.0f} km/h)")
print(f"Incline: theta = {np.degrees(theta):.1f} degrees")
print(f"Drag coefficient: C_d = {C_d}")
print(f"Frontal area: A = {A:.1f} m^2")
print(f"Rolling resistance: mu_r = {mu_r}")
print(f"Powertrain efficiency: eta = {eta*100:.0f}%")

# Calculate forces opposing motion
# 1. Gravity component (up slope)
F_gravity = m * g * np.sin(theta)

# 2. Air drag
F_drag = 0.5 * rho * C_d * A * v**2

# 3. Rolling resistance
N = m * g * np.cos(theta)  # Normal force
F_rolling = mu_r * N

# Total resistance
F_total = F_gravity + F_drag + F_rolling

print(f"\nForce Analysis:")
print(f"Gravity component: F_g = {F_gravity:.2f} N")
print(f"Air drag: F_drag = {F_drag:.2f} N")
print(f"Rolling resistance: F_r = {F_rolling:.2f} N")
print(f"Total resistance: F_total = {F_total:.2f} N")

# Power at wheels (to overcome resistance)
P_wheels = F_total * v
P_wheels_kW = P_wheels / 1000
P_wheels_hp = P_wheels / 746

print(f"\nPower at Wheels:")
print(f"P = F × v = {P_wheels:.2f} W")
print(f"P = {P_wheels_kW:.2f} kW")
print(f"P = {P_wheels_hp:.2f} hp")

# Engine power (accounting for efficiency)
P_engine = P_wheels / eta
P_engine_kW = P_engine / 1000
P_engine_hp = P_engine / 746

print(f"\nEngine Power Required:")
print(f"P_engine = P_wheels / eta = {P_engine:.2f} W")
print(f"P_engine = {P_engine_kW:.2f} kW")
print(f"P_engine = {P_engine_hp:.2f} hp")

# Power losses
P_loss = P_engine - P_wheels
print(f"\nPower Losses (transmission, friction):")
print(f"P_loss = {P_loss:.2f} W ({P_loss/P_engine*100:.1f}%)")

# Breakdown by component
P_gravity = F_gravity * v
P_drag = F_drag * v
P_rolling = F_rolling * v

print(f"\nPower Breakdown:")
print(
    f" Climbing (gravity): {
        P_gravity:.2f} W ({
            P_gravity /
            P_wheels *
            100:.1f}%)")
print(f"Air drag: {P_drag:.2f} W ({P_drag/P_wheels*100:.1f}%)")
print(f"Rolling: {P_rolling:.2f} W ({P_rolling/P_wheels*100:.1f}%)")

# Analyze power vs speed
v_range = np.linspace(5, 40, 50)  # 5 to 40 m/s
F_gravity_range = np.full_like(v_range, F_gravity)
F_drag_range = 0.5 * rho * C_d * A * v_range**2
F_rolling_range = np.full_like(v_range, F_rolling)
F_total_range = F_gravity_range + F_drag_range + F_rolling_range

P_wheels_range = F_total_range * v_range / 1000  # kW
P_engine_range = P_wheels_range / eta

# Component powers
P_gravity_range = F_gravity_range * v_range / 1000
P_drag_range = F_drag_range * v_range / 1000
P_rolling_range = F_rolling_range * v_range / 1000

# Plotting
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Power vs speed (total)
ax1 = axes[0, 0]
ax1.plot(v_range*3.6, P_engine_range, 'r-', linewidth=2, label='Engine power')
ax1.plot(v_range*3.6, P_wheels_range, 'b-', linewidth=2, label='Wheel power')
ax1.axvline(x=v*3.6, color='green', linestyle='--', alpha=0.7,
            label=f'Operating point ({v*3.6:.0f} km/h)')
ax1.axhline(y=P_engine_kW, color='red', linestyle=':', alpha=0.5)
ax1.set_xlabel('Speed (km/h)', fontsize=11)
ax1.set_ylabel('Power (kW)', fontsize=11)
ax1.set_title('Power vs Speed', fontsize=12, fontweight='bold')
ax1.legend(fontsize=9)
ax1.grid(True, alpha=0.3)

# Power breakdown by component
ax2 = axes[0, 1]
ax2.plot(
    v_range*3.6,
    P_gravity_range,
    'g-',
    linewidth=2,
    label='Climbing (gravity)')
ax2.plot(v_range*3.6, P_drag_range, 'r-', linewidth=2, label='Air drag')
ax2.plot(
    v_range*3.6,
    P_rolling_range,
    'b-',
    linewidth=2,
    label='Rolling resistance')
ax2.axvline(x=v*3.6, color='black', linestyle='--', alpha=0.5)
ax2.set_xlabel('Speed (km/h)', fontsize=11)
ax2.set_ylabel('Power (kW)', fontsize=11)
ax2.set_title('Power Breakdown by Component', fontsize=12, fontweight='bold')
ax2.legend(fontsize=9)
ax2.grid(True, alpha=0.3)

# Force breakdown
ax3 = axes[1, 0]
forces = [F_gravity, F_drag, F_rolling]
labels = ['Gravity\n(climbing)', 'Air\nDrag', 'Rolling\nResistance']
colors = ['green', 'red', 'blue']
x_pos = np.arange(len(labels))

bars = ax3.bar(x_pos, forces, color=colors, alpha=0.7)
ax3.set_ylabel('Force (N)', fontsize=11)
ax3.set_title(
    f'Force Components at {
        v*3.6:.0f} km/h',
    fontsize=12,
    fontweight='bold')
ax3.set_xticks(x_pos)
ax3.set_xticklabels(labels)
ax3.grid(True, alpha=0.3, axis='y')

# Add values on bars
for bar, force in zip(bars, forces):
    height = bar.get_height()
    ax3.text(bar.get_x() + bar.get_width()/2., height,
             f'{force:.0f} N', ha='center', va='bottom',
             fontsize=9, fontweight='bold')

# Energy flow diagram
ax4 = axes[1, 1]
energy_labels = ['Engine\nOutput', 'Transmission\nLoss', 'Wheel\nPower']
energy_values = [P_engine_kW, (P_engine-P_wheels)/1000, P_wheels_kW]
energy_colors = ['red', 'gray', 'blue']
x_pos_e = np.arange(len(energy_labels))

ax4.bar(x_pos_e, energy_values, color=energy_colors, alpha=0.7)
ax4.set_ylabel('Power (kW)', fontsize=11)
ax4.set_title('Power Flow', fontsize=12, fontweight='bold')
ax4.set_xticks(x_pos_e)
ax4.set_xticklabels(energy_labels)
ax4.grid(True, alpha=0.3, axis='y')

# Add efficiency annotation
ax4.text(
    1,
    max(energy_values)*0.5,
    f'eta = {
        eta*100:.0f}%',
    ha='center',
    fontsize=11,
    bbox=dict(
        boxstyle='round',
        facecolor='yellow',
        alpha=0.5))

plt.tight_layout()
plt.show()

print(f"\nKey observations:")
print(f"1. Power increases with speed (especially drag: P ~ v^3)")
print(f"2. At low speeds: gravity dominates")
print(f"3. At high speeds: air drag dominates")
print(f"4. Efficiency losses significant: {(1-eta)*100:.0f}% wasted")
print(f"5. Engine must produce {P_engine_hp:.0f} hp for this scenario")

# TODO: Use these parameters in your solutions below
```

### What This Graph Tells You

**Power vs. Speed (top-left):** Three power demands grow at different rates:
- **Hill climbing** ($P_{hill} = mgv\sin\theta$): Linear in $v$ — constant grade, constant weight
- **Air drag** ($P_{drag} = \frac{1}{2}C_D A \rho v^3$): Cubic!! This dominates at high speed
- **Rolling resistance** ($P_{roll} = C_r mg v$): Linear, small but constant

**Component Breakdown (top-right):** At highway speed, air drag consumes **>50%** of engine power. At low speed, hill climbing dominates.

**Force Balance (bottom-left):** Shows the force each effect imposes. Drag force grows as $v^2$ — doubling speed quadruples the drag force.

**Power Flow (bottom-right):** Only a fraction of engine power reaches the wheels (drivetrain efficiency ~85%). Of that, the split between hill/drag/rolling shows where your fuel goes.

**Key Observations:**
- Power required grows roughly as $v^3$ at high speed (drag-dominated) — this is why fuel economy drops drastically above 100 km/h
- Engine must be sized for PEAK power demand (uphill at max speed), not average demand
- Efficiency: only 25–30% of fuel energy reaches the wheels (engine + drivetrain losses)
- Hybrid/electric vehicles recover braking energy — the power flow reverses downhill

## Summary

### Key Concepts Covered

1. **Work-Energy Theorem**:
 - Work = integral of force along path
 - $W_{1\\to 2} = T_2 - T_1 = \Delta T$
 - Alternative to force-based analysis

2. **Conservative Forces**:
 - Work is path-independent
 - Can define potential energy: $W = -\Delta V$
 - Examples: gravity, springs

3. **Conservation of Energy**:
 - For conservative systems: $T + V = $ constant
 - With non-conservative forces: $T_1 + V_1 + W_{nc} = T_2 + V_2$

4. **Power**:
 - $P = \mathbf{F} \cdot \mathbf{v} = Fv\cos\theta$
 - Rotational: $P = \tau \omega$
 - Efficiency: $\eta = P_{out} / P_{in}$

### Problem-Solving Procedures

**Method 1: Work-Energy Theorem**
1. Identify initial and final states
2. Calculate work by each force: $W = \int F \\, ds$
3. Apply: $W_{total} = \Delta T$
4. Solve for unknowns

**Method 2: Conservation of Energy**
1. Check if all forces are conservative
2. Define potential energy reference (V = 0)
3. Calculate energies at states: $T_1 + V_1$, $T_2 + V_2$
4. Apply: $E_1 = E_2$ (or include $W_{nc}$ if present)
5. Solve for unknowns

**Method 3: Power Analysis**
1. Identify forces and velocities
2. Calculate: $P = \mathbf{F} \cdot \mathbf{v}$
3. Account for efficiency losses
4. Find required input power

### When to Use Energy Methods

**Advantages**:
- Scalar equations (simpler than vector F=ma)
- No need for detailed trajectory
- Works well with constraints
- Natural for conservative systems

**Limitations**:
- Don't give accelerations directly
- Can't find forces (only speeds, positions)
- Require knowledge of work done

**Use energy methods when**:
- Finding speeds at specific positions
- System has conservative forces
- Path is complex or unknown
- Interested in power/efficiency

**Use force methods (F=ma) when**:
- Need accelerations or forces
- Want complete time history
- System heavily damped/dissipative
