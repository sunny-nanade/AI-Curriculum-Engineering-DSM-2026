## 🎓 Why Rotating Reference Frames?

### Scenario: Person Walking on Rotating Platform

Imagine:
- **Circular platform** rotating with angular velocity **Ω**
- **Person** walking radially outward on the platform

**Two observers**:
1. **Observer A** (on ground, fixed frame): Sees person moving in a curved path
2. **Observer B** (on platform, rotating frame): Sees person walking in a straight line radially

**Question**: Who is right?

**Answer**: Both! They're describing the same motion from different perspectives.

### The Challenge

How do we relate what Observer A sees to what Observer B sees?

**Key insight**: Time derivatives (velocity, acceleration) depend on your reference frame!

$$\left(\frac{d\vec{r}}{dt}\right)_{\text{fixed}} \neq \left(\frac{d\vec{r}}{dt}\right)_{\text{rotating}}$$

We need a mathematical tool to connect these: **The Transport Theorem**

---

## 📝 Notation and Setup

### Two Reference Frames

**Fixed Frame (Inertial)**: Capital letters **XYZ**
- Axes: **Î**, **Ĵ**, **K̂** (constant direction)
- Observer on ground
- Newton's laws apply directly

**Rotating Frame**: Lowercase letters **xyz**
- Axes: **î**, **ĵ**, **k̂** (rotating with angular velocity **Ω**)
- Observer on rotating platform
- Rotates with body

### Vector Representation

Any vector **r** can be expressed in either frame:

**Fixed frame**:
$$\vec{r} = r_X \hat{I} + r_Y \hat{J} + r_Z \hat{K}$$

**Rotating frame**:
$$\vec{r} = r_x \hat{i} + r_y \hat{j} + r_z \hat{k}$$

**Important**: Components (r_X, r_Y, r_Z) vs (r_x, r_y, r_z) are different!

### Angular Velocity of Rotating Frame

$$\vec{\Omega} = \Omega_X \hat{I} + \Omega_Y \hat{J} + \Omega_Z \hat{K}$$

This is the angular velocity of the **rotating frame** relative to the fixed frame.

---

## 🧮 Deriving the Transport Theorem

### Setup: General Vector in Rotating Frame

Consider any vector **Q** expressed in rotating frame:
$$\vec{Q} = Q_x \hat{i} + Q_y \hat{j} + Q_z \hat{k}$$

### Goal: Find Time Derivative in Fixed Frame

We want:
$$\left(\frac{d\vec{Q}}{dt}\right)_{\text{fixed}}$$

But **Q** is expressed using rotating basis vectors **î**, **ĵ**, **k̂**.

### Step 1: Apply Product Rule

$$\frac{d\vec{Q}}{dt} = \frac{dQ_x}{dt}\hat{i} + Q_x\frac{d\hat{i}}{dt} + \frac{dQ_y}{dt}\hat{j} + Q_y\frac{d\hat{j}}{dt} + \frac{dQ_z}{dt}\hat{k} + Q_z\frac{d\hat{k}}{dt}$$

Rearrange:
$$\frac{d\vec{Q}}{dt} = \left(\frac{dQ_x}{dt}\hat{i} + \frac{dQ_y}{dt}\hat{j} + \frac{dQ_z}{dt}\hat{k}\right) + \left(Q_x\frac{d\hat{i}}{dt} + Q_y\frac{d\hat{j}}{dt} + Q_z\frac{d\hat{k}}{dt}\right)$$

**First term**: Rate of change of components in rotating frame
**Second term**: Rate of change of basis vectors

---

### Step 2: Time Derivatives of Rotating Basis Vectors

**Key fact**: Rotating basis vectors **î**, **ĵ**, **k̂** are constant in magnitude but change direction.

From rigid body kinematics (Lecture 1):
$$\frac{d\hat{i}}{dt} = \vec{\Omega} \times \hat{i}$$
$$\frac{d\hat{j}}{dt} = \vec{\Omega} \times \hat{j}$$
$$\frac{d\hat{k}}{dt} = \vec{\Omega} \times \hat{k}$$

**Why?** Each basis vector is a point on a rigid body (the rotating frame) rotating with **Ω**.

### Substitute Back

$$\frac{d\vec{Q}}{dt} = \left(\frac{dQ_x}{dt}\hat{i} + \frac{dQ_y}{dt}\hat{j} + \frac{dQ_z}{dt}\hat{k}\right) + Q_x(\vec{\Omega} \times \hat{i}) + Q_y(\vec{\Omega} \times \hat{j}) + Q_z(\vec{\Omega} \times \hat{k})$$

Factor out cross product:
$$\frac{d\vec{Q}}{dt} = \left(\frac{dQ_x}{dt}\hat{i} + \frac{dQ_y}{dt}\hat{j} + \frac{dQ_z}{dt}\hat{k}\right) + \vec{\Omega} \times (Q_x\hat{i} + Q_y\hat{j} + Q_z\hat{k})$$

$$\frac{d\vec{Q}}{dt} = \left(\frac{dQ_x}{dt}\hat{i} + \frac{dQ_y}{dt}\hat{j} + \frac{dQ_z}{dt}\hat{k}\right) + \vec{\Omega} \times \vec{Q}$$

---

### Transport Theorem (Final Form)

$$\boxed{\left(\frac{d\vec{Q}}{dt}\right)_{\text{fixed}} = \left(\frac{d\vec{Q}}{dt}\right)_{\text{rotating}} + \vec{\Omega} \times \vec{Q}}$$

**In words**:
> The time derivative in the fixed frame equals the time derivative in the rotating frame plus a correction term Ω × Q.

### Understanding the Terms

| Term | Symbol | Physical Meaning |
|------|--------|------------------|
| **Fixed frame derivative** | $\left(\frac{d\vec{Q}}{dt}\right)_{\text{fixed}}$ | Rate of change seen by fixed observer |
| **Rotating frame derivative** | $\left(\frac{d\vec{Q}}{dt}\right)_{\text{rotating}}$ | Rate of change seen by rotating observer |
| **Transport term** | $\vec{\Omega} \times \vec{Q}$ | Correction due to frame rotation |

### Special Notation

Often written as:
$$\frac{^F d\vec{Q}}{dt} = \frac{^R d\vec{Q}}{dt} + \vec{\Omega} \times \vec{Q}$$

Where:
- **^F** = "as seen from Fixed frame"
- **^R** = "as seen from Rotating frame"

---

## 🎯 Application: Velocity

Let **Q = r** (position vector), apply transport theorem:

$$\left(\frac{d\vec{r}}{dt}\right)_{\text{fixed}} = \left(\frac{d\vec{r}}{dt}\right)_{\text{rotating}} + \vec{\Omega} \times \vec{r}$$

**Left side**: Velocity as seen from fixed frame = **v_fixed**

**Right side, first term**: Velocity as seen from rotating frame = **v_rel** (relative velocity)

**Right side, second term**: Transport velocity = **v_trans**

### Velocity Relation

$$\boxed{\vec{v}_{\text{fixed}} = \vec{v}_{\text{rel}} + \vec{\Omega} \times \vec{r}}$$

**In words**:
> Velocity in fixed frame = Velocity in rotating frame + Transport velocity

### Physical Interpretation

**v_rel**: How fast the point moves **relative to the rotating frame**
- What observer on platform sees
- "Relative motion"

**Ω × r**: Velocity due to frame rotation itself
- Even if point is stationary in rotating frame (v_rel = 0)
- Frame rotation carries the point
- "Transport velocity"

**v_fixed**: Total velocity as seen by fixed observer
- Combination of both effects

---

## 📐 Simple Example: Point Fixed in Rotating Frame

**Scenario**: Point P is painted on a rotating disk at distance r = 0.5 m from center.
- Disk rotates: Ω = 3.0 rad/s
- Point is **fixed** in rotating frame: v_rel = 0

**Question**: What is the velocity in the fixed frame?

**Solution**:

Using velocity relation:
$$\vec{v}_{\text{fixed}} = \vec{v}_{\text{rel}} + \vec{\Omega} \times \vec{r}$$

Since point is fixed in rotating frame:
$$\vec{v}_{\text{rel}} = 0$$

Therefore:
$$\vec{v}_{\text{fixed}} = \vec{\Omega} \times \vec{r}$$

**Magnitude**:
$$|\vec{v}_{\text{fixed}}| = \Omega r = 3.0 \times 0.5 = 1.5 \text{ m/s}$$

**This is exactly what we expect** - same as rigid body velocity formula from Lecture 1!

**Key insight**: Transport theorem **generalizes** rigid body formulas to allow motion within the rotating frame.

## 📝 Example 1: Object on Rotating Platform

### Problem Statement

A circular platform rotates with constant angular velocity **Ω = 2.0 rad/s** counterclockwise.
A small object moves on the platform with velocity (measured in the rotating frame):

$$\vec{v}_{\text{rel}} = 1.5\hat{i} + 0.8\hat{j} \text{ m/s}$$

At the instant shown, the object is at position:
$$\vec{r} = 2.0\hat{i} \text{ m}$$

(measured from platform center, in rotating frame coordinates)

**Find**:
1. Transport velocity (Ω × r)
2. Velocity in fixed frame
3. Speed in fixed frame

**Given**:
| Quantity | Value | Units |
|----------|-------|-------|
| Ω | 2.0 | rad/s (CCW) |
| v_rel | (1.5, 0.8) | m/s (in rotating frame) |
| r | (2.0, 0) | m (in rotating frame) |

---

### Step 1: Reference Frames Setup

**Rotating Frame** (on platform):
- **î**: Radial direction (outward)
- **ĵ**: Tangential direction (CCW)
- **k̂**: Upward (out of page)
- Angular velocity: **Ω = 2.0k̂** rad/s

**Fixed Frame** (on ground):
- **Î**, **Ĵ**, **K̂** (stationary)

**At the instant shown**: Rotating frame aligned with fixed frame for simplicity
- **î** aligned with **Î**
- **ĵ** aligned with **Ĵ**
- **k̂** aligned with **K̂**

(This is just for the instant of calculation; frames are rotating relative to each other)

---

### Step 2: Calculate Transport Velocity

Transport velocity is the velocity due to frame rotation:
$$\vec{v}_{\text{trans}} = \vec{\Omega} \times \vec{r}$$

**Given**:
- $\vec{\Omega} = 2.0\hat{k}$ rad/s
- $\vec{r} = 2.0\hat{i}$ m

**Calculate**:
$$\vec{\Omega} \times \vec{r} = (2.0\hat{k}) \times (2.0\hat{i})$$

Using **k̂ × î = ĵ**:
$$\vec{v}_{\text{trans}} = 2.0 \times 2.0 (\hat{k} \times \hat{i}) = 4.0\hat{j} \text{ m/s}$$

**Physical meaning**: 
- Point at r = 2.0 m rotating at Ω = 2.0 rad/s
- Tangential velocity = Ωr = 2.0 × 2.0 = 4.0 m/s
- Direction: Tangential (perpendicular to radius, CCW)

---

### Step 3: Find Velocity in Fixed Frame

Use transport theorem for velocity:
$$\vec{v}_{\text{fixed}} = \vec{v}_{\text{rel}} + \vec{\Omega} \times \vec{r}$$

**Substitute values**:
$$\vec{v}_{\text{fixed}} = (1.5\hat{i} + 0.8\hat{j}) + 4.0\hat{j}$$

$$\boxed{\vec{v}_{\text{fixed}} = 1.5\hat{i} + 4.8\hat{j} \text{ m/s}}$$

**Components**:
- **X-component**: 1.5 m/s (all from relative motion)
- **Y-component**: 4.8 m/s (0.8 from relative + 4.0 from transport)

---

### Step 4: Calculate Speed

Speed = magnitude of velocity:
$$|\vec{v}_{\text{fixed}}| = \sqrt{v_x^2 + v_y^2}$$

$$|\vec{v}_{\text{fixed}}| = \sqrt{(1.5)^2 + (4.8)^2}$$

$$|\vec{v}_{\text{fixed}}| = \sqrt{2.25 + 23.04} = \sqrt{25.29}$$

$$\boxed{|\vec{v}_{\text{fixed}}| = 5.03 \text{ m/s}}$$

**Direction** (angle from X-axis):
$$\theta = \arctan\left(\frac{v_y}{v_x}\right) = \arctan\left(\frac{4.8}{1.5}\right) = 72.6°$$

Object moving mostly in Y-direction (tangential to platform rotation).

---

```
# Example 1: Numerical Calculation

# Given
Omega = 2.0  # rad/s
v_rel = np.array([1.5, 0.8, 0])  # m/s (in rotating frame)
r = np.array([2.0, 0, 0])  # m (in rotating frame)
Omega_vec = np.array([0, 0, Omega])  # rad/s

print("Example 1: Object on Rotating Platform")
print("="*60)
print(f"Given:")
print(f"  Ω = {Omega} rad/s (platform rotation)")
print(f"  v_rel = ({v_rel[0]}, {v_rel[1]}) m/s (in rotating frame)")
print(f"  r = ({r[0]}, {r[1]}) m (in rotating frame)")
print()

# Calculate transport velocity
v_trans = np.cross(Omega_vec, r)
print(f"Step 1: Transport velocity")
print(f"  v_trans = Ω × r = ({v_trans[0]:.2f}, {v_trans[1]:.2f}) m/s")
print()

# Calculate fixed frame velocity
v_fixed = v_rel + v_trans
print(f"Step 2: Fixed frame velocity")
print(f"  v_fixed = v_rel + v_trans")
print(f"  v_fixed = ({v_fixed[0]:.2f}, {v_fixed[1]:.2f}) m/s")
print()

# Calculate magnitude
v_magnitude = np.linalg.norm(v_fixed[:2])
angle_deg = np.degrees(np.arctan2(v_fixed[1], v_fixed[0]))
print(f"Step 3: Speed")
print(f"  |v_fixed| = {v_magnitude:.2f} m/s")
print(f"  Direction: {angle_deg:.1f}° from X-axis")
print("="*60)
print()

# Visualization
fig, ax = plt.subplots(figsize=(10, 10))
ax.set_xlim(-1, 6)
ax.set_ylim(-1, 6)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.set_title('Velocity Components', fontsize=14, fontweight='bold')

# Platform (circle)
circle = plt.Circle((0, 0), 3, fill=False, edgecolor='gray', linewidth=2, linestyle='--', alpha=0.5)
ax.add_patch(circle)

# Object position
ax.plot(r[0], r[1], 'ro', markersize=15, label='Object', zorder=5)

# Position vector
ax.arrow(0, 0, r[0], r[1], head_width=0.15, head_length=0.1,
         fc='gray', ec='gray', linestyle='--', alpha=0.5)
ax.text(r[0]/2, r[1]/2 - 0.3, 'r', fontsize=14)

# Velocity vectors (scaled for visualization)
scale = 0.4
ax.arrow(r[0], r[1], v_rel[0]*scale, v_rel[1]*scale,
         head_width=0.2, head_length=0.15,
         fc='blue', ec='blue', linewidth=3, label=f'v_rel = {np.linalg.norm(v_rel[:2]):.2f} m/s')

ax.arrow(r[0], r[1], v_trans[0]*scale, v_trans[1]*scale,
         head_width=0.2, head_length=0.15,
         fc='green', ec='green', linewidth=3, label=f'v_trans = {np.linalg.norm(v_trans[:2]):.2f} m/s')

ax.arrow(r[0], r[1], v_fixed[0]*scale, v_fixed[1]*scale,
         head_width=0.2, head_length=0.15,
         fc='red', ec='red', linewidth=3, label=f'v_fixed = {v_magnitude:.2f} m/s')

# Angular velocity indicator
ax.annotate('Ω', xy=(0, 0), xytext=(-0.5, 0.5),
           fontsize=18, color='purple', fontweight='bold')

ax.set_xlabel('X (m)', fontsize=12)
ax.set_ylabel('Y (m)', fontsize=12)
ax.legend(loc='upper left', fontsize=10)

plt.tight_layout()
plt.show()

print("✓ Visualization complete!")
```

### Step 5: Verification

**Units Check**:
- All velocities in m/s ✓

- Cross product dimensionally correct ✓

**Physical Reasoning**:

## 📚 Lecture Summary

### The Transport Theorem (Fundamental)

$$\boxed{\left(\frac{d\vec{Q}}{dt}\right)_{\text{fixed}} = \left(\frac{d\vec{Q}}{dt}\right)_{\text{rotating}} + \vec{\Omega} \times \vec{Q}}$$

**Applies to any vector Q**!

### Velocity Relation

$$\boxed{\vec{v}_{\text{fixed}} = \vec{v}_{\text{rel}} + \vec{\Omega} \times \vec{r}}$$

| Term | Meaning |
|------|---------|
| **v_fixed** | Velocity in fixed (inertial) frame |
| **v_rel** | Velocity in rotating frame (relative velocity) |
| **Ω × r** | Transport velocity (due to frame rotation) |

### Key Concepts

1. **Reference frame matters**:
   - Same motion, different descriptions
   - Fixed vs rotating observers see different velocities

2. **Transport term**:
   - Always present when frame rotates
   - v_trans = Ω × r
   - Perpendicular to position vector

3. **Vector addition**:
   - Must add vectorially
   - Components in same coordinate system

4. **Path shapes differ**:
   - Straight in rotating frame → curved in fixed frame
   - Radial motion → spiral path

### Special Cases

- **Object fixed in rotating frame**: v_rel = 0 → v_fixed = Ω × r (rigid body)
- **Fixed frame motion**: Ω = 0 → v_fixed = v_rel (no rotation)

### Applications

- Weather systems (hurricanes)
- Rotating machinery
- Earth's rotation effects
- Spacecraft dynamics
- Robot arms on moving bases

## 🔜 Next Lecture Preview: Acceleration and Coriolis!

### What's Coming in Lecture 4

We derived velocity relation today:
$$\vec{v}_{\text{fixed}} = \vec{v}_{\text{rel}} + \vec{\Omega} \times \vec{r}$$

**Next**: Apply transport theorem **again** to get acceleration!

$$\vec{a}_{\text{fixed}} = \vec{a}_{\text{rel}} + \vec{\alpha} \times \vec{r} + 2\vec{\Omega} \times \vec{v}_{\text{rel}} + \vec{\Omega} \times (\vec{\Omega} \times \vec{r})$$

**The star term**: $2\vec{\Omega} \times \vec{v}_{\text{rel}}$ is the **Coriolis acceleration**! 🌀

This is THE most important concept in rotating frames:
- Causes hurricanes to spin
- Deflects projectiles
- Makes Foucault pendulum precess
- Essential for navigation
