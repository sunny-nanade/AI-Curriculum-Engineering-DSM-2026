## Part A: Rigid Body Equations of Motion 

### What is a Rigid Body?
**Definition**: An extended object where distances between all points remain constant
- **Idealization**: No deformation (perfectly stiff)
- **Reality**: Steel beam, wheel, robot link → nearly rigid
- **Not rigid**: Spring, rope, fluid

### Two Types of Motion

#### 1. Translation (ΣF = m·a_cm)
Every point has the **same velocity**:
$$\mathbf{v}_P = \mathbf{v}_{cm} \quad \text{(for all points P)}$$

**Newton's 2nd Law for CM**:
$$\sum \mathbf{F}_{\text{external}} = m \mathbf{a}_{cm}$$

**Key insight**: Center of mass moves like a particle with total mass!

#### 2. Rotation (Στ = I·α)
Object spins about an axis. Each point has different velocity!

**Velocity of point P on rigid body**:
$$\mathbf{v}_P = \mathbf{v}_{cm} + \boldsymbol{\omega} \times \mathbf{r}_{P/cm}$$

where:
- $\boldsymbol{\omega}$ = angular velocity vector (rad/s)
- $\mathbf{r}_{P/cm}$ = position of P relative to CM

**Angular Equation of Motion** (about CM or fixed axis):
$$\sum \tau = I \alpha$$

where:
- $\tau$ = torque (N·m)
- $I$ = moment of inertia (kg·m²) — **resistance to rotation**
- $\alpha = \dot{\omega}$ = angular acceleration (rad/s²)

### Moment of Inertia
**Definition**: How mass is distributed about rotation axis
$$I = \int r^2 \, dm = \sum m_i r_i^2$$

**Common Shapes** (about center):
- **Point mass**: $I = m r^2$
- **Thin rod** (length L, mass m, about center): $I = \frac{1}{12} m L^2$
- **Solid disk/cylinder** (radius R): $I = \frac{1}{2} m R^2$
- **Hollow cylinder** (radius R): $I = m R^2$
- **Solid sphere**: $I = \frac{2}{5} m R^2$

**Parallel Axis Theorem**: Shift axis from CM to distance $d$ away
$$I_{\text{new}} = I_{cm} + m d^2$$

### Combined Motion: Rolling
**Rolling without slipping**: Point at contact has zero velocity!
$$v_{cm} = \omega R \quad \text{(constraint)}$$

**Consequences**:
- Acceleration: $a_{cm} = \alpha R$
- Total kinetic energy: $KE = \frac{1}{2} m v_{cm}^2 + \frac{1}{2} I_{cm} \omega^2$

- $I$ measures how hard it is to change an object's angular velocity — it's the rotational equivalent of mass
- Mass distribution matters MORE than total mass: a hollow pipe is harder to spin than a solid rod of equal mass
- The **parallel axis theorem** $I_{new} = I_{cm} + md^2$ shows that $I$ always INCREASES when the axis moves away from the center of mass
- Physical analogy: $I \cdot \alpha = \tau$ (rotation) is exactly like $m \cdot a = F$ (translation)

> **Student Takeaway:** Moment of inertia depends on both mass AND how that mass is distributed relative to the rotation axis. This is why flywheels have mass at the rim (high $I$ for energy storage) and why figure skaters pull their arms in to spin faster (reducing $I$).

## Part B: Rolling Motion Analysis 

### Classic Problem: Rolling Down an Incline

**Setup**: Object (mass m, radius R, moment of inertia I) rolls down incline (angle θ)

**Forces**:
- Weight: $mg$ (down)
- Normal: $N$ (perpendicular to incline)
- Friction: $f$ (up the incline) — **provides torque for rotation!**

**Equations**:

**Translation** (down incline, +x direction):
$$m a_{cm} = mg \sin\theta - f$$

**Rotation** (about center of mass):
$$I \alpha = f R$$

**Rolling constraint** (no slip):
$$a_{cm} = \alpha R$$

### Solution Strategy
1. Use constraint to eliminate $\alpha$: $\alpha = a_{cm}/R$
2. From rotation equation: $f = I \alpha / R = I a_{cm} / R^2$
3. Substitute into translation equation:
 $$m a_{cm} = mg \sin\theta - \frac{I a_{cm}}{R^2}$$
4. Solve for $a_{cm}$:
 $$\boxed{a_{cm} = \frac{g \sin\theta}{1 + I/(mR^2)}}$$

### Key Results

**Define**: $\beta = I/(mR^2)$ (dimensionless shape factor)
- Solid sphere: $\beta = 2/5$
- Solid disk: $\beta = 1/2$
- Hollow cylinder: $\beta = 1$

**Acceleration**:
$$a_{cm} = \frac{g \sin\theta}{1 + \beta}$$

**Who wins the race down the incline?**
- **Smallest $\beta$ → fastest** (less rotational inertia)
- Order: Solid sphere > Solid disk > Hollow cylinder
- **Independent of mass and radius!** (shape only)

**Friction force**:
$$f = \frac{\beta}{1 + \beta} mg \sin\theta$$

**No friction → no rotation** (would just slide with $a = g\sin\theta$, faster but no roll)

![Rolling Disk on Incline \u2014 Translation + Rotation = Rolling Without Slip](figs/U5_L1_rolling_disk.png)

```python
# Example 2: Rolling Down Incline - Compare Different Shapes

print("=" * 60)
print("Example 2: Rolling Race Down Incline")
print("=" * 60)

# Given
g = 9.81  # m/s^2
theta = np.radians(30)  # 30 degree incline
m = 1.0  # kg (same for all)
R = 0.1  # m (same radius)
distance = 5.0  # m (length of incline)

# Shape factors (beta = I/(mR^2))
shapes = ['Solid Sphere', 'Solid Disk', 'Hollow Cylinder']
betas = [2/5, 1/2, 1]  # beta values
colors = ['purple', 'green', 'red']

# Calculate accelerations
accels = [g * np.sin(theta) / (1 + beta) for beta in betas]

# Calculate time to reach bottom
# Using s = (1/2) a t^2 → t = sqrt(2s/a)
times = [np.sqrt(2 * distance / a) for a in accels]

# Calculate final velocities
# v = a*t or v = sqrt(2*a*s)
velocities = [np.sqrt(2 * a * distance) for a in accels]

# Calculate friction forces
frictions = [beta/(1+beta) * m * g * np.sin(theta) for beta in betas]

print(f"\nIncline: {np.degrees(theta):.1f}°, Length: {distance} m")
print(f"Mass: {m} kg, Radius: {R} m (same for all)\n")

for i, (shape, beta, a, t, v, f) in enumerate(
        zip(shapes, betas, accels, times, velocities, frictions)):
    print(f"{i+1}. {shape}:")
    print(f"β = I/(mR²) = {beta:.3f}")
    print(f"Acceleration: {a:.3f} m/s²")
    print(f"Time to bottom: {t:.3f} s")
    print(f"Final velocity: {v:.3f} m/s")
    print(f"Friction force: {f:.3f} N")
    print()

# Simulate motion
t_max = max(times) * 1.1
t = np.linspace(0, t_max, 500)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Position vs time
ax1 = axes[0, 0]
for i, (shape, a, t_finish, color) in enumerate(
        zip(shapes, accels, times, colors)):
    # Position until it reaches bottom, then constant
    pos = np.where(t <= t_finish, 0.5 * a * t**2, distance)
    ax1.plot(t, pos, color=color, linewidth=2,
             label=f'{shape} (t={t_finish:.2f}s)')

ax1.axhline(
    y=distance,
    color='k',
    linestyle='--',
    alpha=0.3,
    label='Finish line')
ax1.set_xlabel('Time (s)', fontsize=11)
ax1.set_ylabel('Distance (m)', fontsize=11)
ax1.set_title('Position vs Time', fontsize=12, fontweight='bold')
ax1.legend(fontsize=9)
ax1.grid(True, alpha=0.3)

# Velocity vs time
ax2 = axes[0, 1]
for i, (shape, a, t_finish, color) in enumerate(
        zip(shapes, accels, times, colors)):
    vel = np.where(t <= t_finish, a * t, a * t_finish)
    ax2.plot(t, vel, color=color, linewidth=2, label=shape)

ax2.set_xlabel('Time (s)', fontsize=11)
ax2.set_ylabel('Velocity (m/s)', fontsize=11)
ax2.set_title('Velocity vs Time', fontsize=12, fontweight='bold')
ax2.legend(fontsize=9)
ax2.grid(True, alpha=0.3)

# Comparison bars - Time to finish
ax3 = axes[1, 0]
bars = ax3.bar(
    shapes,
    times,
    color=colors,
    alpha=0.7,
    edgecolor='black',
    linewidth=2)
ax3.set_ylabel('Time to bottom (s)', fontsize=11)
ax3.set_title('Race Results (Who Wins?)', fontsize=12, fontweight='bold')
ax3.grid(axis='y', alpha=0.3)
for bar, t_val in zip(bars, times):
    height = bar.get_height()
    ax3.text(bar.get_x() + bar.get_width()/2,
             height + 0.02,
             f'{t_val:.3f}s',
             ha='center',
             va='bottom',
             fontsize=10,
             fontweight='bold')

# Energy comparison
ax4 = axes[1, 1]
KE_trans = [0.5 * m * v**2 for v in velocities]
KE_rot = [0.5 * beta * m * R**2 *
          (v/R)**2 for beta, v in zip(betas, velocities)]
KE_total = [kt + kr for kt, kr in zip(KE_trans, KE_rot)]
PE_lost = m * g * distance * np.sin(theta)

x_pos = np.arange(len(shapes))
width = 0.25
ax4.bar(
    x_pos - width,
    KE_trans,
    width,
    label='KE translational',
    color='blue',
    alpha=0.7)
ax4.bar(x_pos, KE_rot, width, label='KE rotational', color='red', alpha=0.7)
ax4.bar(
    x_pos + width,
    KE_total,
    width,
    label='KE total',
    color='purple',
    alpha=0.7)
ax4.axhline(
    y=PE_lost,
    color='green',
    linestyle='--',
    linewidth=2,
    label=f'PE lost = {
        PE_lost:.2f} J')
ax4.set_ylabel('Energy (J)', fontsize=11)
ax4.set_title(
    'Energy Distribution (Conservation Check)',
    fontsize=12,
    fontweight='bold')
ax4.set_xticks(x_pos)
ax4.set_xticklabels(shapes, fontsize=9)
ax4.legend(fontsize=9)
ax4.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print("\n Winner: Solid Sphere (smallest β → fastest!)")
print("Key insight: Energy splits between translation and rotation.")
print("Objects with smaller β convert more PE → translational KE → faster!")

# TODO: Use these parameters in your solutions below
```

### What This Graph Tells You

**Position and Velocity (left panels):** The solid sphere reaches the bottom FIRST, followed by the solid disk, with the hollow cylinder arriving LAST. The race results may seem counterintuitive!

**Race Results (bar chart):** Finish times: Sphere (1.52s) < Disk (1.56s) < Hollow Cylinder (1.76s). The sphere wins despite having the same mass and starting height.

- Sphere: 71% translational KE, 29% rotational KE — **most energy goes to moving forward**

- A frictionless sliding block ($\beta = 0$) would beat ALL rolling objects

> **Student Takeaway:** Rolling motion splits gravitational PE into translational AND rotational KE. Objects with lower $\beta$ (mass closer to center) convert more energy to forward motion and win the race. This explains why solid wheels roll faster than hollow ones.

## Part C: 1-DOF Rigid Body Systems 

### Example: Physical Pendulum

**Setup**: Extended object (not point mass) swinging about pivot

**Difference from simple pendulum**:
- Simple: Point mass on massless string → $\theta'' + \frac{g}{L}\theta = 0$
- **Physical**: Extended body → inertia matters!

**Equation of Motion** (small angles):
$$I_O \alpha = -mg d \sin\theta \approx -mg d \theta$$

where:
- $I_O$ = moment of inertia about pivot O
- $d$ = distance from pivot to center of mass
- $\alpha = \ddot{\theta}$ = angular acceleration

**Result**:
$$\ddot{\theta} + \frac{mgd}{I_O} \theta = 0$$

**Natural frequency**:
$$\omega_n = \sqrt{\frac{mgd}{I_O}}$$

**Period**:
$$T = 2\pi \sqrt{\frac{I_O}{mgd}}$$

### Example: Uniform Rod (length L, mass m)
Pivoting about one end:
- $I_O = \frac{1}{3}mL^2$ (about end)
- $d = L/2$ (CM at center)
- Period: $T = 2\pi \sqrt{\frac{2L}{3g}}$

Compare to simple pendulum of length L:
- Simple: $T = 2\pi \sqrt{\frac{L}{g}}$
- Rod: $T = 2\pi \sqrt{\frac{2L}{3g}} = \sqrt{\frac{2}{3}} T_{simple} \approx 0.816 T_{simple}$

**Physical pendulum is faster!** (less equivalent length)

---


```python
# Example 3: Physical Pendulum (Uniform Rod)

print("=" * 60)
print("Example 3: Physical Pendulum - Swinging Rod")
print("=" * 60)

# Given
m = 2.0  # kg
L = 1.0  # m (rod length)
g = 9.81  # m/s^2

# Calculate properties
I_O = (1/3) * m * L**2  # Inertia about end
d = L / 2  # Distance to CM
omega_n = np.sqrt(m * g * d / I_O)
T_rod = 2 * np.pi / omega_n

# Compare to simple pendulum
T_simple = 2 * np.pi * np.sqrt(L / g)

print(f"\nRod Properties:")
print(f"Mass: {m} kg")
print(f"Length: {L} m")
print(f"I_O (about end): {I_O:.4f} kg·m²")
print(f"Distance to CM: {d} m")
print(f"\nDynamics:")
print(f"Natural frequency: {omega_n:.4f} rad/s")
print(f"Period (rod): {T_rod:.4f} s")
print(f"Period (simple, L={L}m): {T_simple:.4f} s")
print(f"Ratio: T_rod/T_simple = {T_rod/T_simple:.4f} = √(2/3) ")

# Equation of motion: θ'' + ω_n^2 θ = 0


def pendulum_ode(y, t):
    """
    State: y = [theta, omega]
    """
    theta, omega = y

    # Small angle approximation
    alpha = -(m * g * d / I_O) * theta

    return [omega, alpha]


# Initial conditions
theta_0 = np.radians(20)  # 20 degree initial angle
omega_0 = 0.0  # Released from rest
y0 = [theta_0, omega_0]

# Simulate
t = np.linspace(0, 3*T_rod, 1000)
solution = odeint(pendulum_ode, y0, t)
theta = solution[:, 0]
omega = solution[:, 1]

# Analytical solution for comparison
theta_analytical = theta_0 * np.cos(omega_n * t)

# Calculate energy
PE = m * g * d * (1 - np.cos(theta))  # Relative to lowest point
KE = 0.5 * I_O * omega**2
E_total = PE + KE

# Plotting
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Angle vs time
ax1 = axes[0, 0]
ax1.plot(t, np.degrees(theta), 'b-', linewidth=2, label='Numerical')
ax1.plot(
    t,
    np.degrees(theta_analytical),
    'r--',
    linewidth=2,
    alpha=0.7,
    label='Analytical')
ax1.set_xlabel('Time (s)', fontsize=11)
ax1.set_ylabel('Angle (degrees)', fontsize=11)
ax1.set_title('Angular Position', fontsize=12, fontweight='bold')
ax1.legend(fontsize=10)
ax1.grid(True, alpha=0.3)

# Angular velocity vs time
ax2 = axes[0, 1]
ax2.plot(t, omega, 'g-', linewidth=2)
ax2.set_xlabel('Time (s)', fontsize=11)
ax2.set_ylabel('Angular Velocity (rad/s)', fontsize=11)
ax2.set_title('Angular Velocity', fontsize=12, fontweight='bold')
ax2.grid(True, alpha=0.3)

# Phase portrait
ax3 = axes[1, 0]
ax3.plot(np.degrees(theta), omega, 'purple', linewidth=2)
ax3.plot(np.degrees(theta[0]), omega[0], 'go', markersize=10, label='Start')
ax3.set_xlabel('Angle (degrees)', fontsize=11)
ax3.set_ylabel('Angular Velocity (rad/s)', fontsize=11)
ax3.set_title('Phase Portrait', fontsize=12, fontweight='bold')
ax3.legend(fontsize=10)
ax3.grid(True, alpha=0.3)

# Energy vs time
ax4 = axes[1, 1]
ax4.plot(t, PE, 'r-', linewidth=2, label='PE')
ax4.plot(t, KE, 'b-', linewidth=2, label='KE')
ax4.plot(t, E_total, 'k--', linewidth=2, label='Total')
ax4.set_xlabel('Time (s)', fontsize=11)
ax4.set_ylabel('Energy (J)', fontsize=11)
ax4.set_title('Energy (Conservation Check)', fontsize=12, fontweight='bold')
ax4.legend(fontsize=10)
ax4.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\n Period from simulation: {t[np.argmax(theta[50:])+50] * 2:.4f} s")
print(f"(matches theoretical: {T_rod:.4f} s)")
print(
    f"\n Energy conservation: {E_total[0]:.6f} → {E_total[-1]:.6f} J (excellent!)")

# TODO: Use these parameters in your solutions below
```

### What This Graph Tells You

**Angle vs. Time (top-left):** The rod oscillates with period $T_{rod} = 1.62$ s, which is **shorter** than a simple pendulum of the same length ($T_{simple} = 2.01$ s). The physical pendulum swings faster!

**Phase Portrait (top-right):** The closed ellipse confirms energy-conserving, periodic motion. The orbit never spirals — no energy is lost.

**Energy (bottom panels):** KE and PE exchange continuously, with total energy perfectly constant. Maximum KE occurs at the bottom (fastest swing), maximum PE at the extremes (momentarily stationary).

**Key Observations:**
- Physical pendulum period: $T = 2\pi\sqrt{\frac{I_O}{mgd}}$ where $d$ is the distance from pivot to center of mass
- For a uniform rod: $I_O = mL^2/3$ and $d = L/2$, giving effective length $L_{eff} = 2L/3 < L$
- The "equivalent simple pendulum length" $L_{eq} = I_O/(md) = 2L/3$ is always LESS than the physical length for most shapes
- This is why a swinging baseball bat has a different "sweet spot" period than a simple bob

> **Student Takeaway:** Real swinging objects (legs, baseball bats, pendulum clocks) are physical pendulums, not simple pendulums. Their period depends on the mass distribution through the moment of inertia $I_O$, not just the length.

## Part D: Pure Rotation - Spinning Disk 

### Problem: Disk with Applied Torque

**Setup**: Solid disk (mass m, radius R) spins about center. Torque τ applied.

**Equation of Motion**:
$$\tau = I \alpha$$

where:
- $I = \frac{1}{2}mR^2$ (solid disk)
- $\alpha = \dot{\omega}$ (angular acceleration)

### Case 1: Constant Torque
$$\alpha = \frac{\tau}{I} = \text{constant}$$
$$\omega(t) = \omega_0 + \alpha t$$
$$\theta(t) = \theta_0 + \omega_0 t + \frac{1}{2}\alpha t^2$$

**Analogy**: Just like $F = ma$ for linear motion!
- $\tau \leftrightarrow F$
- $I \leftrightarrow m$
- $\alpha \leftrightarrow a$
- $\omega \leftrightarrow v$
- $\theta \leftrightarrow x$

### Case 2: Friction (Damping)
**Damping torque**: $\tau_{damp} = -c \omega$ (proportional to angular velocity)

**Equation**:
$$I \alpha = -c \omega$$
$$\alpha + \frac{c}{I}\omega = 0$$

**Solution**: Exponential decay
$$\omega(t) = \omega_0 e^{-t/\tau_d}$$

where $\tau_d = I/c$ = time constant

---


```python
# Example 4: Spinning Disk with Torque and Friction

print("=" * 60)
print("Example 4: Spinning Disk Dynamics")
print("=" * 60)

# Given
m = 5.0  # kg
R = 0.3  # m
I = 0.5 * m * R**2  # Solid disk

# Case 1: Constant torque
tau_applied = 2.0  # N·m
omega_0_case1 = 0.0  # rad/s (starts from rest)
alpha_case1 = tau_applied / I

# Case 2: Free spin with friction
c = 0.5  # N·m·s/rad (damping coefficient)
omega_0_case2 = 10.0  # rad/s (initial spin)
tau_d = I / c  # Time constant

print(f"\nDisk Properties:")
print(f"Mass: {m} kg")
print(f"Radius: {R} m")
print(f"Moment of inertia: {I:.4f} kg·m²")

print(f"\nCase 1: Constant Torque τ = {tau_applied} N·m")
print(f"Angular acceleration: {alpha_case1:.4f} rad/s²")
print(f"Time to reach 10 rad/s: {10/alpha_case1:.4f} s")

print(f"\nCase 2: Free Spin with Friction (c = {c} N·m·s/rad)")
print(f"Initial ω: {omega_0_case2} rad/s")
print(f"Time constant τ_d: {tau_d:.4f} s")
print(f"Time to decay to 37%: {tau_d:.4f} s (1 time constant)")
print(f"Time to decay to 5%: {3*tau_d:.4f} s (3 time constants)")

# Simulate
t = np.linspace(0, 10, 500)

# Case 1: Constant torque
omega_case1 = omega_0_case1 + alpha_case1 * t
theta_case1 = 0.5 * alpha_case1 * t**2

# Case 2: Exponential decay
omega_case2 = omega_0_case2 * np.exp(-t / tau_d)
theta_case2 = omega_0_case2 * tau_d * (1 - np.exp(-t / tau_d))

# Plotting
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Angular velocity - Case 1
ax1 = axes[0, 0]
ax1.plot(t, omega_case1, 'b-', linewidth=2.5, label='Case 1: Constant τ')
ax1.set_xlabel('Time (s)', fontsize=11)
ax1.set_ylabel('Angular Velocity (rad/s)', fontsize=11)
ax1.set_title(
    'Case 1: Acceleration (Constant Torque)',
    fontsize=12,
    fontweight='bold')
ax1.legend(fontsize=10)
ax1.grid(True, alpha=0.3)

# Angular position - Case 1
ax2 = axes[0, 1]
ax2.plot(t, theta_case1, 'g-', linewidth=2.5, label='Case 1')
ax2.set_xlabel('Time (s)', fontsize=11)
ax2.set_ylabel('Angular Position (rad)', fontsize=11)
ax2.set_title('Case 1: Position (Parabolic)', fontsize=12, fontweight='bold')
ax2.legend(fontsize=10)
ax2.grid(True, alpha=0.3)

# Angular velocity - Case 2
ax3 = axes[1, 0]
ax3.plot(t, omega_case2, 'r-', linewidth=2.5, label='Case 2: Friction')
ax3.axhline(y=omega_0_case2 * np.exp(-1), color='k', linestyle='--', alpha=0.5,
            label=f'37% at t={tau_d:.2f}s')
ax3.set_xlabel('Time (s)', fontsize=11)
ax3.set_ylabel('Angular Velocity (rad/s)', fontsize=11)
ax3.set_title(
    'Case 2: Deceleration (Exponential Decay)',
    fontsize=12,
    fontweight='bold')
ax3.legend(fontsize=10)
ax3.grid(True, alpha=0.3)

# Angular position - Case 2
ax4 = axes[1, 1]
ax4.plot(t, theta_case2, 'purple', linewidth=2.5, label='Case 2')
ax4.axhline(y=omega_0_case2 * tau_d, color='k', linestyle='--', alpha=0.5,
            label=f'Asymptote: {omega_0_case2 * tau_d:.2f} rad')
ax4.set_xlabel('Time (s)', fontsize=11)
ax4.set_ylabel('Angular Position (rad)', fontsize=11)
ax4.set_title(
    'Case 2: Position (Approaches Limit)',
    fontsize=12,
    fontweight='bold')
ax4.legend(fontsize=10)
ax4.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\n Insight:")
print("Case 1: Linear acceleration → just like F=ma for rotation!")
print("Case 2: Exponential decay → energy dissipated by friction")

# TODO: Use these parameters in your solutions below
```

### What This Graph Tells You

**Case 1 — Constant Torque (left panels):**
- Angular velocity increases **linearly**: $\omega = \alpha t = (\tau/I) t$ — just like $v = at$ for translation
- Angular position grows **quadratically**: $\theta = \frac{1}{2}\alpha t^2$ — just like $x = \frac{1}{2}at^2$
- This is pure rotational kinematics with constant angular acceleration

**Case 2 — Friction Damping (right panels):**
- Angular velocity decays **exponentially**: $\omega(t) = \omega_0 e^{-t/\tau_c}$ where $\tau_c = I/c$
- After 1 time constant ($\tau_c = 0.45$ s): $\omega$ drops to 37% of initial
- After 3 time constants: $\omega < 5%$ (practically stopped)
- Position approaches a finite limit (the disk doesn't spin forever)

**Key Observations:**
- Rotational kinematics perfectly mirrors translational kinematics (just swap $F \to \tau$, $m \to I$, $a \to \alpha$)
- Time constant $\tau_c = I/c$: higher $I$ \u2192 slower decay (flywheel effect), higher $c$ \u2192 faster decay (brake effect)
- Real systems: bearings provide viscous friction ($\tau_{friction} = c\omega$), brakes provide dry friction ($\tau_{friction} =$ constant)

> **Student Takeaway:** The time constant $\tau_c = I/c$ is the single most important parameter for spinning systems. It tells you how quickly a motor spins up, how long a flywheel stores energy, and how fast a brake stops rotation.

## Summary

### Key Equations

**Rigid Body Dynamics**:
$$\boxed{\sum \mathbf{F} = m \mathbf{a}_{cm}} \quad \boxed{\sum \tau = I \alpha}$$

**Moment of Inertia** (resistance to rotation):
$$I = \int r^2 \, dm = \sum m_i r_i^2$$

**Parallel Axis Theorem**:
$$I_{\text{new}} = I_{cm} + m d^2$$

**Rolling Constraint** (no slip):
$$v_{cm} = \omega R \quad \text{and} \quad a_{cm} = \alpha R$$

**Rolling Down Incline**:
$$a_{cm} = \frac{g \sin\theta}{1 + I/(mR^2)}$$

**Physical Pendulum**:
$$\omega_n = \sqrt{\frac{mgd}{I_O}} \quad \text{where } d = \text{distance from pivot to CM}$$

### Problem-Solving Strategy

1. **Identify motion type**: Translation only? Rotation only? Combined?
2. **Draw FBD**: Show all forces and where they act
3. **Choose reference**: CM or fixed pivot
4. **Apply equations**:
 - Translation: ΣF = m·a_cm
 - Rotation: Στ = I·α
5. **Constraints**: Rolling? v = ωR, a = αR
6. **Solve**: Eliminate unknowns, find accelerations

### Physical Insights

- **Moment of inertia**: Shape matters! Hollow > Solid for same m, R
- **Rolling race**: Smaller I/(mR²) → faster (solid sphere wins)
- **Friction enables rolling**: Without it, objects just slide
- **Physical pendulum**: Faster than simple pendulum of same length
- **Rotational damping**: Exponential decay (just like linear damping)
