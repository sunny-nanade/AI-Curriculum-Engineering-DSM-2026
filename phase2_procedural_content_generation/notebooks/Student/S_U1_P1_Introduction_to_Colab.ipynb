# PART 4: Basic Computations with NumPy

---

## Creating Arrays

In dynamics, we work with **vectors** and **arrays** of numbers.

NumPy makes this easy:

---

````python
# Create arrays (like mathematical vectors)
position = np.array([3, 4, 0]) # Position vector in 3D
time = np.array([0, 1, 2, 3, 4]) # Time array

print("Position vector:", position)
print("Time array:", time)

# Calculate magnitude of position vector
magnitude = np.sqrt(np.sum(position**2))
print(f"\nMagnitude of position: {magnitude:.2f} meters")
````

## Question: What Did This Code Do?

Look at the code above. Can you identify:

1. How we created the position vector?
2. How we calculated its magnitude?
3. What does `**2` mean?
4. What does `np.sum()` do?

**Discuss with your neighbor...**

*(Scroll for explanation)*

---

## Explanation

```python
position = np.array([3, 4, 0]) # Creates array with 3 elements
```
- This creates a NumPy array (like a vector) with components [3, 4, 0]

```python
magnitude = np.sqrt(np.sum(position**2))
```
- `position**2` → squares each element: [9, 16, 0]
- `np.sum(...)` → adds them up: 9 + 16 + 0 = 25
- `np.sqrt(...)` → takes square root: √25 = 5

This is the **Euclidean distance formula**: $|r| = \sqrt{x^2 + y^2 + z^2}$

# PART 5: Creating Your First Plot

---

## Why Plotting Matters

In Dynamic Systems Modeling, we **visualize**:
- Position vs Time
- Velocity vs Time 
- Trajectories in space
- System behavior

**Seeing is understanding!**

---

## Simple Plot Example

Let's plot a sine wave (represents oscillation):

---

````python
# Create data for plotting
t = np.linspace(0, 2*np.pi, 100) # Time from 0 to 2π
y = np.sin(t) # Sine wave (oscillation)

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(t, y, 'b-', linewidth=2, label='sin(t)')
plt.xlabel('Time (radians)', fontsize=12)
plt.ylabel('Amplitude', fontsize=12)
plt.title('Simple Harmonic Motion', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

print("Plot created successfully!")
print("This represents oscillation - common in mechanical systems!")
````

## Understanding the Plot Code

Let's break down what happened:

```python
t = np.linspace(0, 2*np.pi, 100)
```
- Creates 100 evenly spaced points from 0 to 2π

```python
y = np.sin(t)
```
- Calculates sine value for each t

```python
plt.plot(t, y, 'b-', linewidth=2)
```
- `t, y` → x and y data
- `'b-'` → blue solid line
- `linewidth=2` → thickness

```python
plt.show()
```
- Displays the plot!

# PART 8: Troubleshooting Common Issues

---

## Problem 1: "NameError: name 'np' is not defined"

**Cause**: You forgot to import NumPy!

**Solution**: Run the imports cell first:
```python
import numpy as np
import matplotlib.pyplot as plt
```

**Remember**: Always run imports at the beginning!

---

## Problem 2: Code Doesn't Work / Weird Results

**Possible causes**:
1. Ran cells out of order
2. Modified a variable without re-running dependent cells

**Solution**: 
- Click **Runtime → Restart and run all**
- This runs everything from the beginning in order

---

## Problem 3: Plot Doesn't Show

**Cause**: Forgot `plt.show()` or didn't import matplotlib

**Solution**:
```python
import matplotlib.pyplot as plt # At the top
# ... your plotting code ...
plt.show() # At the end
```

---

## Problem 4: Lost Internet Connection

**What happens**: Code stops running

**Solution**:
1. Reconnect to internet
2. Click **Reconnect** button at top right
3. Re-run your cells (previous results might be lost)

**Tip**: Save frequently to avoid losing work!

---

## Problem 5: Can't Access Notebook Later

**Cause**: Notebook not saved to Drive

**Solution**:
1. Make sure you're logged in with NMIMS account
2. Click **File → Locate in Drive** to see where it's saved
3. Check your Drive's "Colab Notebooks" folder
