### Key Concepts

**Standard SDOF equation**:
$$m\ddot{x} + c\dot{x} + kx = 0$$

**Parameters**:
- $\omega_n = \sqrt{k/m}$ → **Natural frequency** (rad/s)
- $f_n = \omega_n / (2\pi)$ → Natural frequency (Hz)
- $T_n = 1/f_n = 2\pi/\omega_n$ → Period (s)
- $\zeta = c/(2\sqrt{km}) = c/c_c$ → **Damping ratio** (dimensionless)
- $c_c = 2\sqrt{km} = 2m\omega_n$ → Critical damping

**Damping regimes**:
1. **Underdamped** ($\zeta < 1$): Oscillates with decay
2. **Critically damped** ($\zeta = 1$): Fastest return without overshoot
3. **Overdamped** ($\zeta > 1$): Slow exponential decay, no oscillation

## Part A: Undamped Free Vibrations ($\zeta = 0$)

### System: Mass-Spring

**Equation of motion**:
$$m\ddot{x} + kx = 0$$

Divide by $m$:
$$\ddot{x} + \omega_n^2 x = 0, \quad \omega_n = \sqrt{\frac{k}{m}}$$

### Solution

**General solution** (assume $x = e^{\lambda t}$):
$$\lambda^2 + \omega_n^2 = 0 \quad \Rightarrow \quad \lambda = \pm i\omega_n$$

Real form:
$$x(t) = A\cos(\omega_n t) + B\sin(\omega_n t)$$

Or in amplitude-phase form:
$$x(t) = X_0\cos(\omega_n t - \phi)$$

where:
- $X_0 = \sqrt{A^2 + B^2}$ → amplitude
- $\phi = \arctan(B/A)$ → phase angle

### Initial Conditions

Given $x(0) = x_0$ and $\dot{x}(0) = v_0$:
$$A = x_0, \quad B = \frac{v_0}{\omega_n}$$

Therefore:
$$x(t) = x_0\cos(\omega_n t) + \frac{v_0}{\omega_n}\sin(\omega_n t)$$

### Key Properties

**Frequency**:
- Independent of amplitude (linear system)
- Depends only on $k$ and $m$
- $\omega_n \uparrow$ when $k \uparrow$ (stiffer) or $m \downarrow$ (lighter)

**Period**: $T_n = \frac{2\pi}{\omega_n} = 2\pi\sqrt{\frac{m}{k}}$

**Energy**: Conserved!
$$E = \frac{1}{2}kX_0^2 = KE + PE$$

**Phase portrait**: Perfect circle (energy level curve)

![Single DOF System \u2014 Mass, Spring, and Damper Model](figs/U5_L4_sdof_system.png)

```python
# Example 1: Undamped SDOF Vibration

print("=" * 60)
print("Example 1: Undamped Free Vibration")
print("=" * 60)

# Given
m = 2.0 # kg
k = 200.0 # N/m
c = 0.0 # N·s/m (no damping)

# Calculate natural frequency
omega_n = np.sqrt(k / m)
f_n = omega_n / (2 * np.pi)
T_n = 1 / f_n

print(f"\nSystem Properties:")
print(f" Mass: {m} kg")
print(f" Stiffness: {k} N/m")
print(f" Damping: {c} N·s/m (undamped)")

print(f"\nNatural Frequency:")
print(f" ωₙ = {omega_n:.3f} rad/s")
print(f" fₙ = {f_n:.3f} Hz")
print(f" Tₙ = {T_n:.3f} s (period)")

# Initial conditions
x0 = 0.1 # m (10 cm initial displacement)
v0 = 0.0 # m/s (released from rest)

print(f"\nInitial Conditions:")
print(f" x(0) = {x0*100} cm")
print(f" v(0) = {v0} m/s (released from rest)")

# ODE system
def sdof_ode(y, t, m, c, k):
 x, v = y
 a = (-c * v - k * x) / m
 return [v, a]

# Time array
t = np.linspace(0, 3*T_n, 1000)

# Solve
y0 = [x0, v0]
solution = odeint(sdof_ode, y0, t, args=(m, c, k))
x = solution[:, 0]
v = solution[:, 1]

# Energy
KE = 0.5 * m * v**2
PE = 0.5 * k * x**2
E_total = KE + PE

# Analytical solution for comparison
A = x0
B = v0 / omega_n
X0 = np.sqrt(A**2 + B**2)
x_analytical = x0 * np.cos(omega_n * t) + (v0 / omega_n) * np.sin(omega_n * t)

# Plotting
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Displacement vs time
ax1 = axes[0, 0]
ax1.plot(t, x*100, 'b-', linewidth=2, label='Numerical')
ax1.plot(t, x_analytical*100, 'r--', linewidth=1.5, alpha=0.7, label='Analytical')
ax1.axhline(y=0, color='k', linestyle='-', alpha=0.3)
ax1.set_xlabel('Time (s)', fontsize=11)
ax1.set_ylabel('Displacement (cm)', fontsize=11)
ax1.set_title(f'Displacement vs Time (ωₙ = {omega_n:.2f} rad/s)', fontsize=12, fontweight='bold')
ax1.legend(fontsize=10)
ax1.grid(True, alpha=0.3)

# Phase portrait
ax2 = axes[0, 1]
ax2.plot(x*100, v*100, 'b-', linewidth=2)
ax2.plot(x[0]*100, v[0]*100, 'go', markersize=10, label='Start')
ax2.set_xlabel('Displacement (cm)', fontsize=11)
ax2.set_ylabel('Velocity (cm/s)', fontsize=11)
ax2.set_title('Phase Portrait (Perfect Circle!)', fontsize=12, fontweight='bold')
ax2.legend(fontsize=10)
ax2.grid(True, alpha=0.3)
ax2.set_aspect('equal', adjustable='box')

# Energy
ax3 = axes[1, 0]
ax3.plot(t, KE, 'b-', linewidth=2, label='Kinetic')
ax3.plot(t, PE, 'r-', linewidth=2, label='Potential')
ax3.plot(t, E_total, 'k--', linewidth=2, label='Total')
ax3.set_xlabel('Time (s)', fontsize=11)
ax3.set_ylabel('Energy (J)', fontsize=11)
ax3.set_title('Energy (Conserved!)', fontsize=12, fontweight='bold')
ax3.legend(fontsize=10)
ax3.grid(True, alpha=0.3)

# Velocity vs time
ax4 = axes[1, 1]
ax4.plot(t, v*100, 'b-', linewidth=2)
ax4.axhline(y=0, color='k', linestyle='-', alpha=0.3)
ax4.set_xlabel('Time (s)', fontsize=11)
ax4.set_ylabel('Velocity (cm/s)', fontsize=11)
ax4.set_title('Velocity vs Time', fontsize=12, fontweight='bold')
ax4.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\n Analysis complete!")
print(f"\n Key observations:")
print(f" 1. Oscillates forever (no energy loss)")
print(f" 2. Period Tₙ = {T_n:.3f}s is constant")
print(f" 3. Energy conserved: E = {E_total[0]:.6f} J")
print(f" 4. Phase portrait is perfect circle (no spiraling)")
print(f" 5. Amplitude X₀ = {X0*100:.2f} cm (constant)")

# TODO: Use these parameters in your solutions below

# TODO: Use these parameters in your solutions below
```

**Displacement vs. Time (top-left):** A perfect sinusoid that oscillates FOREVER with constant amplitude. Natural frequency $f_n = 1.59$ Hz, period $T_n = 0.628$ s. This is the baseline behavior with zero energy loss.

## Part B: Damped Free Vibrations ($\zeta > 0$)

### Equation with Damping

$$m\ddot{x} + c\dot{x} + kx = 0$$

Standard form:
$$\ddot{x} + 2\zeta\omega_n\dot{x} + \omega_n^2 x = 0$$

where:
- $\omega_n = \sqrt{k/m}$ → undamped natural frequency
- $\zeta = \frac{c}{2\sqrt{km}} = \frac{c}{c_c}$ → damping ratio
- $c_c = 2\sqrt{km} = 2m\omega_n$ → critical damping coefficient

### Solution (assume $x = e^{\lambda t}$)

Characteristic equation:
$$\lambda^2 + 2\zeta\omega_n\lambda + \omega_n^2 = 0$$

Roots:
$$\lambda = -\zeta\omega_n \pm \omega_n\sqrt{\zeta^2 - 1}$$

### Three Cases

#### 1. Underdamped ($\zeta < 1$)

**Roots**: Complex conjugates $\lambda = -\zeta\omega_n \pm i\omega_d$

where $\omega_d = \omega_n\sqrt{1 - \zeta^2}$ → **damped frequency**

**Solution**:
$$x(t) = X_0 e^{-\zeta\omega_n t}\cos(\omega_d t - \phi)$$

**Characteristics**:
- Oscillates with **exponentially decaying envelope**
- Frequency $\omega_d < \omega_n$ (damping slows oscillation)
- **Logarithmic decrement**: $\delta = \ln\frac{x_n}{x_{n+1}} = \frac{2\pi\zeta}{\sqrt{1-\zeta^2}}$

#### 2. Critically Damped ($\zeta = 1$)

**Roots**: Real, repeated $\lambda = -\omega_n$

**Solution**:
$$x(t) = (A + Bt)e^{-\omega_n t}$$

**Characteristics**:
- **Fastest return** to equilibrium without overshoot
- No oscillation
- Used in: door closers, automotive dampers

#### 3. Overdamped ($\zeta > 1$)

**Roots**: Real, distinct $\lambda_1, \lambda_2 < 0$

**Solution**:
$$x(t) = A e^{\lambda_1 t} + B e^{\lambda_2 t}$$

**Characteristics**:
- **Slow exponential decay**
- No oscillation (over-damped)
- Slower than critical damping

### Logarithmic Decrement

For underdamped systems, measure decay:

$$\delta = \ln\frac{x_n}{x_{n+1}} = \frac{2\pi\zeta}{\sqrt{1-\zeta^2}} \approx 2\pi\zeta \quad (\text{for } \zeta \ll 1)$$

Invert to find damping:
$$\zeta = \frac{\delta}{\sqrt{4\pi^2 + \delta^2}}$$

**Experimental use**: Measure successive peak amplitudes to extract $\zeta$!

```python
# Example 3: Experimental Modal Parameter Extraction

print("=" * 60)
print("Example 3: Extract Modal Parameters from Measurement")
print("=" * 60)

print("\n Scenario: Unknown system, we measured free response")
print(" Goal: Find ωₙ and ζ from time-series data")

# "Unknown" system (we'll pretend we measured this)
m_actual = 1.5 # kg (unknown to us)
k_actual = 150.0 # N/m (unknown to us)
zeta_actual = 0.15 # (unknown to us)
c_actual = zeta_actual * 2 * np.sqrt(k_actual * m_actual)

omega_n_actual = np.sqrt(k_actual / m_actual)

print(f"\n (Hidden) Actual System:")
print(f" m = {m_actual} kg, k = {k_actual} N/m")
print(f" ζ_actual = {zeta_actual}")
print(f" ωₙ_actual = {omega_n_actual:.3f} rad/s")

# Generate "measurement" data
x0_meas = 0.05 # m
v0_meas = 0.0
t_meas = np.linspace(0, 5, 500)
sol_meas = odeint(sdof_ode, [x0_meas, v0_meas], t_meas, args=(m_actual, c_actual, k_actual))
x_meas = sol_meas[:, 0]

# Add some noise to simulate real measurement
np.random.seed(42)
noise = 0.001 * np.random.randn(len(x_meas)) # 1mm noise
x_meas_noisy = x_meas + noise

print(f"\n We measured vibration for {t_meas[-1]} seconds")
print(f" Initial displacement: {x0_meas*100:.1f} cm")
print(f" Measurement noise: ~1 mm RMS")

# Method 1: Extract natural frequency from zero-crossings or peaks
from scipy.signal import find_peaks
peaks, _ = find_peaks(x_meas_noisy)
peak_times = t_meas[peaks]
peak_values = x_meas_noisy[peaks]

if len(peaks) >= 3:
 # Average period from peaks
 periods = np.diff(peak_times)
 T_d_measured = np.mean(periods)
 omega_d_measured = 2 * np.pi / T_d_measured
 
 print(f"\n Step 1: Measure damped frequency")
 print(f" Found {len(peaks)} peaks")
 print(f" Average period: Td = {T_d_measured:.4f} s")
 print(f" Damped frequency: ωd = {omega_d_measured:.3f} rad/s")

# Method 2: Logarithmic decrement from peak decay
if len(peaks) >= 2:
 # Calculate log dec from multiple consecutive peaks
 deltas = []
 for i in range(len(peaks)-1):
     delta_i = np.log(peak_values[i] / peak_values[i+1])
 deltas.append(delta_i)
 
 delta_measured = np.mean(deltas)
 
 # Extract damping ratio from log dec
 zeta_measured = delta_measured / np.sqrt(4*np.pi**2 + delta_measured**2)
 
 print(f"\n Step 2: Measure logarithmic decrement")
 print(f" Average δ = {delta_measured:.4f}")
 print(f" Extracted ζ = {zeta_measured:.4f}")

# Method 3: Calculate undamped natural frequency
omega_n_measured = omega_d_measured / np.sqrt(1 - zeta_measured**2)
f_n_measured = omega_n_measured / (2 * np.pi)

print(f"\n Step 3: Calculate undamped natural frequency")
print(f" ωₙ = ωd / √(1-ζ²) = {omega_n_measured:.3f} rad/s")
print(f" fₙ = {f_n_measured:.3f} Hz")

# Plotting
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Measured data with identified peaks
ax1 = axes[0]
ax1.plot(t_meas, x_meas_noisy*100, 'b-', linewidth=1, alpha=0.7, label='Measured data')
ax1.plot(peak_times, peak_values*100, 'ro', markersize=8, label=f'Peaks (n={len(peaks)})')

# Fitted envelope
envelope_fit = x0_meas * 100 * np.exp(-zeta_measured * omega_n_measured * t_meas)
ax1.plot(t_meas, envelope_fit, 'k--', linewidth=2, alpha=0.6, label='Fitted envelope')
ax1.plot(t_meas, -envelope_fit, 'k--', linewidth=2, alpha=0.6)

ax1.axhline(y=0, color='k', linestyle='-', alpha=0.3)
ax1.set_xlabel('Time (s)', fontsize=11)
ax1.set_ylabel('Displacement (cm)', fontsize=11)
ax1.set_title('Measured Response + Identified Parameters', fontsize=12, fontweight='bold')
ax1.legend(fontsize=10)
ax1.grid(True, alpha=0.3)

# Verification: compare with actual
ax2 = axes[1]
params = ['ωₙ (rad/s)', 'ζ', 'fₙ (Hz)']
actual_vals = [omega_n_actual, zeta_actual, omega_n_actual/(2*np.pi)]
measured_vals = [omega_n_measured, zeta_measured, f_n_measured]
errors = [abs(m-a)/a*100 for m, a in zip(measured_vals, actual_vals)]

x_pos = np.arange(len(params))
width = 0.35

ax2.bar(x_pos - width/2, actual_vals, width, label='Actual', color='green', alpha=0.7)
ax2.bar(x_pos + width/2, measured_vals, width, label='Measured', color='blue', alpha=0.7)

ax2.set_ylabel('Value', fontsize=11)
ax2.set_title('Parameter Extraction Accuracy', fontsize=12, fontweight='bold')
ax2.set_xticks(x_pos)
ax2.set_xticklabels(params, fontsize=10)
ax2.legend(fontsize=10)
ax2.grid(True, alpha=0.3, axis='y')

# Add error text
for i, (a, m, e) in enumerate(zip(actual_vals, measured_vals, errors)):
 ax2.text(i, max(a, m)*1.05, f'Error: {e:.2f}%', ha='center', fontsize=9)

plt.tight_layout()
plt.show()

print(f"\n Verification:")
print(f" Parameter Actual Measured Error")
print(f" ωₙ (rad/s) {omega_n_actual:.3f} {omega_n_measured:.3f} {errors[0]:.2f}%")
print(f" ζ {zeta_actual:.4f} {zeta_measured:.4f} {errors[1]:.2f}%")
print(f" fₙ (Hz) {actual_vals[2]:.3f} {measured_vals[2]:.3f} {errors[2]:.2f}%")

print(f"\n This is how modal testing works in practice!")
print(f" 1. Impact structure (hammer test)")
print(f" 2. Measure free response (accelerometer)")
print(f" 3. Extract ωₙ and ζ from decay")
print(f" 4. Build mathematical model")

# TODO: Use these parameters in your solutions below

# TODO: Use these parameters in your solutions below
```

## Summary

### Key Equations

**Equation of motion**:
$$m\ddot{x} + c\dot{x} + kx = 0 \quad \Rightarrow \quad \ddot{x} + 2\zeta\omega_n\dot{x} + \omega_n^2 x = 0$$

**Natural frequency**:
$$\omega_n = \sqrt{\frac{k}{m}}, \quad f_n = \frac{\omega_n}{2\pi}, \quad T_n = \frac{1}{f_n}$$

**Damping ratio**:
$$\zeta = \frac{c}{c_c} = \frac{c}{2\sqrt{km}} = \frac{c}{2m\omega_n}$$

**Damped frequency** (underdamped):
$$\omega_d = \omega_n\sqrt{1 - \zeta^2}$$

**Logarithmic decrement**:
$$\delta = \ln\frac{x_n}{x_{n+1}} = \frac{2\pi\zeta}{\sqrt{1-\zeta^2}} \quad \Rightarrow \quad \zeta = \frac{\delta}{\sqrt{4\pi^2 + \delta^2}}$$

### Three Damping Regimes

| Regime | Condition | Behavior | Application |
|--------|-----------|----------|-------------|
| **Underdamped** | $\zeta < 1$ | Oscillates with decay | Musical instruments, sensors |
| **Critically Damped** | $\zeta = 1$ | Fastest return, no overshoot | Door closers, dampers |
| **Overdamped** | $\zeta > 1$ | Slow decay, no oscillation | Heavy damping systems |

### Experimental Modal Analysis

**From free response measurements**:
1. **Period measurement**: Count peaks → $T_d$ → $\omega_d = 2\pi/T_d$
2. **Amplitude decay**: Measure peak amplitudes → $\delta = \ln(x_n/x_{n+1})$ → $\zeta$
3. **Natural frequency**: $\omega_n = \omega_d / \sqrt{1-\zeta^2}$

**Tools**:
- Accelerometer for vibration measurement
- Impact hammer for excitation
- FFT for frequency domain analysis

### Design Guidelines

**Natural frequency**:
- Increase $k$ (stiffer) → $\omega_n \uparrow$
- Decrease $m$ (lighter) → $\omega_n \uparrow$
- Want $\omega_n$ far from forcing frequencies (avoid resonance!)

**Damping**:
- $\zeta < 0.05$: Very lightly damped (many oscillations)
- $0.05 < \zeta < 0.3$: Light damping (typical for structures)
- $\zeta \approx 0.7$: Good compromise (overshoot ~5%)
- $\zeta = 1.0$: Critical damping (optimal for settling time)
- $\zeta > 1.0$: Overdamped (sluggish response)

### Connection to Multi-DOF

**Each mode of multi-DOF system behaves as SDOF**!

For mode $i$:
- Natural frequency: $\omega_i$
- Mode shape: $\phi_i$
- Damping ratio: $\zeta_i$

Total response = sum of modal responses (next lecture!)
