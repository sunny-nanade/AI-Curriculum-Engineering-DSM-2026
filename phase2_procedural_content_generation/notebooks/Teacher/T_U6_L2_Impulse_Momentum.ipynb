### Context

- **This lecture**: Impulse-momentum (vector, involves time)
- Both derive from Newton's 2nd law

**When to use momentum methods**:
- Collisions and impacts (very short time)
- Force varies with time (not position)
- Want to find velocities after impact
- Conservation of momentum applies

### Key Concepts

**Linear Momentum** (particle):
$$\mathbf{L} = m\mathbf{v}$$

**Impulse** (force over time):
$$\mathbf{J} = \int_{t_1}^{t_2} \mathbf{F} \\, dt$$

**Impulse-Momentum Theorem**:
$$\mathbf{J}_{1\\to 2} = \mathbf{L}_2 - \mathbf{L}_1 = \Delta \mathbf{L}$$

Impulse equals change in momentum!

**Angular Momentum**:
$$\mathbf{H}_O = \mathbf{r} \times m\mathbf{v} \quad \text{(particle about point O)}$$
$$\mathbf{H}_G = I\boldsymbol{\omega} \quad \text{(rigid body about center of mass)}$$

**Angular Impulse-Momentum**:
$$\int_{t_1}^{t_2} \mathbf{M}_O \\, dt = \mathbf{H}_{O,2} - \mathbf{H}_{O,1}$$

## Part A: Linear Impulse-Momentum

### Derivation from Newton's 2nd Law

Start with: $\mathbf{F} = m\mathbf{a} = m\frac{d\mathbf{v}}{dt}$

Rearrange: $\mathbf{F} \\, dt = m \\, d\mathbf{v}$

Integrate over time:
$$\int_{t_1}^{t_2} \mathbf{F} \\, dt = \int_{\mathbf{v}_1}^{\mathbf{v}_2} m \\, d\mathbf{v}$$

$$\mathbf{J}_{1\\to 2} = m\mathbf{v}_2 - m\mathbf{v}_1 = \Delta \mathbf{L}$$

### Components

**In 2D**:
$$J_x = \int F_x \\, dt = m(v_{2x} - v_{1x})$$
$$J_y = \int F_y \\, dt = m(v_{2y} - v_{1y})$$

Apply component-wise for each direction!

### Special Cases

**1. Constant force**: $\mathbf{J} = \mathbf{F} \Delta t$

**2. Average force**: $\mathbf{F}_{avg} = \frac{\mathbf{J}}{\Delta t} = \frac{\Delta \mathbf{L}}{\Delta t}$

**3. Impulsive forces** (very large, very short duration):
- Collision forces dominate
- Can neglect gravity, friction during impact
- $\Delta t \\to 0$ but $\mathbf{J}$ finite

### Conservation of Momentum

**If total external force = 0**:
$$\mathbf{J}_{ext} = 0 \quad \Rightarrow \quad \mathbf{L}_1 = \mathbf{L}_2$$

Total momentum conserved!

**For a system of particles**:
$$\sum m_i \mathbf{v}_{i,1} = \sum m_i \mathbf{v}_{i,2}$$

Internal forces cancel (Newton's 3rd law)

### Key Differences: Energy vs Momentum

| Property | Work-Energy | Impulse-Momentum |
|----------|-------------|------------------|
| Type | Scalar | Vector |
| Integrated over | Displacement | Time |
| Form | $W = \int F \\, ds$ | $J = \int F \\, dt$ |
| Result | Speed | Velocity (direction!) |
| Best for | Position-dependent F | Time-dependent F |
| Collisions | Energy lost (inelastic) | Always conserved (if no ext F) |

```python
# Example 1: Baseball Hit by Bat

print("=" * 60)
print("Example 1: Baseball Impact - Impulse-Momentum")
print("=" * 60)

# Given
m = 0.145  # kg (baseball mass)
v1 = -40.0  # m/s (pitched ball, approaching)
v2 = 50.0  # m/s (hit ball, leaving)
dt = 0.001  # s (contact time = 1 ms)

print(f"\nGiven:")
print(f"Ball mass: m = {m} kg")
print(f"Initial velocity: v1 = {v1} m/s (approaching)")
print(f"Final velocity: v2 = {v2} m/s (leaving)")
print(f"Contact time: dt = {dt*1000} ms")

# Calculate momentum change
L1 = m * v1
L2 = m * v2
delta_L = L2 - L1

print(f"\nMomentum Analysis:")
print(f"Initial momentum: L1 = {L1:.3f} kg·m/s")
print(f"Final momentum: L2 = {L2:.3f} kg·m/s")
print(f"Change in momentum: ΔL = {delta_L:.3f} kg·m/s")

# Impulse-momentum theorem
J = delta_L
F_avg = J / dt

print(f"\nImpulse-Momentum Theorem:")
print(f"Impulse: J = ΔL = {J:.3f} N·s")
print(f"Average force: F_avg = J/Δt = {F_avg:.1f} N")
print(f"Force in multiples of weight: {F_avg/(m*9.81):.0f} × mg")

# Energy analysis (for comparison)
T1 = 0.5 * m * v1**2
T2 = 0.5 * m * v2**2
delta_T = T2 - T1
W = delta_T

print(f"\nEnergy Analysis (for comparison):")
print(f"Initial KE: T1 = {T1:.2f} J")
print(f"Final KE: T2 = {T2:.2f} J")
print(f"Change in KE: ΔT = {delta_T:.2f} J")
print(f"Work done: W = {W:.2f} J")

# Model force as half-sine wave (realistic)
t_array = np.linspace(0, dt, 1000)
F_array = F_avg * np.pi / 2 * np.sin(np.pi * t_array / dt)

# Verify impulse
J_numerical = np.trapz(F_array, t_array)
print(f"\nForce Model Verification:")
print(f"Modeled as half-sine wave: F(t) = F_max sin(πt/Δt)")
print(f"Peak force: F_max = {np.max(F_array):.1f} N")
print(f"Impulse (numerical): J = {J_numerical:.3f} N·s")
print(f"Impulse (analytical): J = {J:.3f} N·s")
print(f"Error: {abs(J_numerical - J)/J * 100:.3f}%")

# Plotting
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Force vs time
ax1 = axes[0, 0]
ax1.plot(t_array*1000, F_array, 'b-', linewidth=2)
ax1.axhline(
    y=F_avg,
    color='r',
    linestyle='--',
    linewidth=2,
    label=f'Average: {
        F_avg:.0f} N')
ax1.fill_between(t_array*1000, F_array, alpha=0.3)
ax1.set_xlabel('Time (ms)', fontsize=11)
ax1.set_ylabel('Force (N)', fontsize=11)
ax1.set_title('Contact Force vs Time', fontsize=12, fontweight='bold')
ax1.legend(fontsize=9)
ax1.grid(True, alpha=0.3)
ax1.text(
    dt*1000/2,
    F_avg*0.5,
    f'Area = Impulse\n= {
        J:.2f} N·s',
    ha='center',
    fontsize=10,
    bbox=dict(
        boxstyle='round',
        facecolor='yellow',
        alpha=0.5))

# Velocity change
ax2 = axes[0, 1]
velocities = [v1, v2]
labels = ['Before\nImpact', 'After\nImpact']
colors = ['red', 'green']
x_pos = np.arange(len(labels))

bars = ax2.bar(x_pos, velocities, color=colors, alpha=0.7, width=0.6)
ax2.axhline(y=0, color='k', linestyle='-', linewidth=1)
ax2.set_ylabel('Velocity (m/s)', fontsize=11)
ax2.set_title('Velocity Change', fontsize=12, fontweight='bold')
ax2.set_xticks(x_pos)
ax2.set_xticklabels(labels)
ax2.grid(True, alpha=0.3, axis='y')

# Add velocity values
for i, (bar, v) in enumerate(zip(bars, velocities)):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height,
             f'{v:.1f} m/s', ha='center',
             va='bottom' if height > 0 else 'top',
             fontsize=10, fontweight='bold')

# Add arrow showing change
ax2.annotate('', xy=(1, v2), xytext=(0, v1),
             arrowprops=dict(arrowstyle='->', lw=2, color='blue'))
ax2.text(
    0.5,
    (v1+v2)/2,
    f'Δv = {
        v2-v1:.0f} m/s',
    ha='center',
    fontsize=10,
    bbox=dict(
        boxstyle='round',
        facecolor='wheat',
        alpha=0.7))

# Momentum diagram
ax3 = axes[1, 0]
momenta = [L1, L2]
m_labels = ['Initial\nMomentum', 'Final\nMomentum']
m_colors = ['red', 'green']
x_pos_m = np.arange(len(m_labels))

ax3.bar(x_pos_m, momenta, color=m_colors, alpha=0.7, width=0.6)
ax3.axhline(y=0, color='k', linestyle='-', linewidth=1)
ax3.set_ylabel('Momentum (kg·m/s)', fontsize=11)
ax3.set_title('Momentum Change', fontsize=12, fontweight='bold')
ax3.set_xticks(x_pos_m)
ax3.set_xticklabels(m_labels)
ax3.grid(True, alpha=0.3, axis='y')

# Add impulse arrow
ax3.annotate(f'J = {J:.2f} N·s', xy=(0.5, (L1+L2)/2), xytext=(1.5, (L1+L2)/2),
             arrowprops=dict(arrowstyle='->', lw=2, color='blue'),
             fontsize=11, fontweight='bold',
             bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.7))

# Energy comparison
ax4 = axes[1, 1]
energies = [T1, T2]
e_labels = ['Before', 'After']
ax4.bar(e_labels, energies, color=['blue', 'orange'], alpha=0.7, width=0.5)
ax4.set_ylabel('Kinetic Energy (J)', fontsize=11)
ax4.set_title(
    'Energy Change (Momentum conserved, Energy added)',
    fontsize=12,
    fontweight='bold')
ax4.grid(True, alpha=0.3, axis='y')

# Add energy gain annotation
ax4.text(0.5,
         max(energies)*0.9,
         f'ΔT = +{delta_T:.1f} J\n(Bat adds energy)',
         ha='center',
         fontsize=10,
         bbox=dict(boxstyle='round',
                   facecolor='lightgreen',
                   alpha=0.7))

plt.tight_layout()
plt.show()

print(f"\nKey observations:")
print(f"1. Impulse changes momentum: J = ΔL")
print(f"2. Average force is enormous: {F_avg/(m*9.81):.0f} times weight!")
print(f"3. Direction reverses: momentum sign changes")
print(f"4. Energy increases (bat does work on ball)")
print(f"5. Impulse-momentum works perfectly for impacts")
```

### What This Graph Tells You

**Force vs. Time (top-left):** The contact force is ENORMOUS (~8000+ N) but lasts only milliseconds. The force profile is roughly half-sinusoidal — zero at contact, peaks mid-impact, zero at separation.

**Velocity Change (top-right):** The ball's velocity **reverses direction** — from negative (approaching) to positive (departing). The velocity change is ~60 m/s in just 1 ms!

**Momentum Change (bottom-left):** The impulse (area under force-time curve) equals the momentum change: $J = \int F\,dt = \Delta p$. This is the impulse-momentum theorem in action.

**Energy (bottom-right):** The ball **gains** kinetic energy — the bat does work on the ball. The coefficient of restitution ($e > 0$) means the collision is partially elastic.

**Key Observations:**
- Peak force > 8000 N from a bat — equivalent to ~800 kg resting on the ball!
- Even though the force is huge, the TIME is tiny, so the displacement during impact is negligible
- This is why impulse (not force) is the useful quantity for impacts: $J = F_{avg} \cdot \Delta t = m \cdot \Delta v$
- The ball's speed INCREASES because the bat is much heavier and moving fast

> **Student Takeaway:** Impulse-momentum is the right tool for impacts and collisions where forces are large but brief. You don't need to know the exact force history — just the total impulse gives you the velocity change.

## Part B: Conservation of Momentum & Collisions

### Conservation Law

**If no external forces** (or external impulse negligible):
$$\sum \mathbf{L}_1 = \sum \mathbf{L}_2$$

Total momentum before = Total momentum after

**Important**: Applies even in **inelastic** collisions!
- Momentum: Always conserved (no external force)
- Energy: Lost in inelastic collisions (heat, deformation)

### Collision Types

**1. Perfectly Elastic** (e = 1):
- Both momentum and kinetic energy conserved
- $v_{rel,after} = -v_{rel,before}$
- Example: Billiard balls, atomic particles

**2. Perfectly Inelastic** (e = 0):
- Objects stick together after collision
- Maximum energy loss
- $v_A = v_B$ after collision
- Example: Car crash, clay balls

**3. Partially Elastic** (0 < e < 1):
- Most real collisions
- Some energy lost

### Coefficient of Restitution

**Definition**:
$$e = \frac{|v_{B,2} - v_{A,2}|}{|v_{A,1} - v_{B,1}|} = \frac{\text{Relative velocity of separation}}{\text{Relative velocity of approach}}$$

- $e = 1$: Perfectly elastic
- $e = 0$: Perfectly inelastic
- $0 < e < 1$: Partially elastic

### Direct Central Impact

**Two objects colliding head-on**:

Conservation of momentum:
$$m_A v_{A,1} + m_B v_{B,1} = m_A v_{A,2} + m_B v_{B,2}$$

Coefficient of restitution:
$$e = \frac{v_{B,2} - v_{A,2}}{v_{A,1} - v_{B,1}}$$

**Two equations, two unknowns** ($v_{A,2}$, $v_{B,2}$)

### Oblique Impact

**2D collision**: Apply conservation in components

**Tangential direction** (along contact surface):
- No impulse (frictionless)
- Velocities unchanged: $v_{t,1} = v_{t,2}$

**Normal direction** (perpendicular to contact):
- Apply momentum conservation
- Apply coefficient of restitution

### Energy Loss in Collisions

**Energy dissipated**:
$$\Delta T_{loss} = T_1 - T_2 = \frac{1}{2}\frac{m_A m_B}{m_A + m_B}(1-e^2)(v_{A,1} - v_{B,1})^2$$

- Proportional to $(1-e^2)$
- Maximum when $e=0$ (perfectly inelastic)
- Zero when $e=1$ (perfectly elastic)

```python
# Example 2: Two-Car Collision

print("=" * 60)
print("Example 2: Two-Car Collision - Conservation Analysis")
print("=" * 60)

# Given
m_A = 1200.0  # kg (car A)
m_B = 1500.0  # kg (car B)
v_A1 = 20.0  # m/s (car A initial)
v_B1 = 0.0  # m/s (car B at rest)
e = 0.5  # coefficient of restitution (partially elastic)

print(f"\nGiven:")
print(f"Car A: m_A = {m_A} kg, v_A1 = {v_A1} m/s")
print(f"Car B: m_B = {m_B} kg, v_B1 = {v_B1} m/s (at rest)")
print(f"Coefficient of restitution: e = {e}")

# Conservation of momentum
# m_A * v_A1 + m_B * v_B1 = m_A * v_A2 + m_B * v_B2

# Coefficient of restitution
# e = (v_B2 - v_A2) / (v_A1 - v_B1)

# Solve system of equations
# From momentum: m_A*v_A1 + m_B*v_B1 = m_A*v_A2 + m_B*v_B2
# From restitution: v_B2 - v_A2 = e*(v_A1 - v_B1)

# Solving:
total_mass = m_A + m_B
v_A2 = ((m_A - e*m_B)*v_A1 + (1+e)*m_B*v_B1) / total_mass
v_B2 = ((1+e)*m_A*v_A1 + (m_B - e*m_A)*v_B1) / total_mass

print(f"\nSolution:")
print(f"Car A final velocity: v_A2 = {v_A2:.3f} m/s")
print(f"Car B final velocity: v_B2 = {v_B2:.3f} m/s")

# Verify momentum conservation
L_before = m_A * v_A1 + m_B * v_B1
L_after = m_A * v_A2 + m_B * v_B2
print(f"\nMomentum Conservation Check:")
print(f"Before: L1 = {L_before:.3f} kg·m/s")
print(f"After: L2 = {L_after:.3f} kg·m/s")
print(f"Error: {abs(L_before - L_after):.9f} kg·m/s")

# Verify coefficient of restitution
v_rel_before = v_A1 - v_B1
v_rel_after = v_B2 - v_A2
e_check = v_rel_after / v_rel_before
print(f"\nCoefficient of Restitution Check:")
print(f"Relative velocity before: {v_rel_before:.3f} m/s")
print(f"Relative velocity after: {v_rel_after:.3f} m/s")
print(f"e (calculated): {e_check:.6f}")
print(f"e (given): {e:.6f}")
print(f"Error: {abs(e - e_check):.9f}")

# Energy analysis
T1 = 0.5 * m_A * v_A1**2 + 0.5 * m_B * v_B1**2
T2 = 0.5 * m_A * v_A2**2 + 0.5 * m_B * v_B2**2
T_loss = T1 - T2
energy_retained = T2 / T1 * 100

print(f"\nEnergy Analysis:")
print(f"Initial KE: T1 = {T1:.2f} J")
print(f"Final KE: T2 = {T2:.2f} J")
print(f"Energy lost: ΔT = {T_loss:.2f} J ({T_loss/T1*100:.1f}%)")
print(f"Energy retained: {energy_retained:.1f}%")

# Compare with perfectly inelastic (e=0)
v_stuck = (m_A * v_A1 + m_B * v_B1) / (m_A + m_B)
T_stuck = 0.5 * (m_A + m_B) * v_stuck**2
T_loss_max = T1 - T_stuck

print(f"\nComparison with Perfectly Inelastic (e=0):")
print(f"If stuck together: v = {v_stuck:.3f} m/s")
print(f"Energy lost (e=0): {T_loss_max:.2f} J ({T_loss_max/T1*100:.1f}%)")
print(f"Our case (e={e}): {T_loss:.2f} J ({T_loss/T1*100:.1f}%)")

# Plotting
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Velocity comparison
ax1 = axes[0, 0]
cars = ['Car A', 'Car B']
v_before = [v_A1, v_B1]
v_after = [v_A2, v_B2]
x = np.arange(len(cars))
width = 0.35

ax1.bar(x - width/2, v_before, width, label='Before', color='red', alpha=0.7)
ax1.bar(x + width/2, v_after, width, label='After', color='green', alpha=0.7)
ax1.set_ylabel('Velocity (m/s)', fontsize=11)
ax1.set_title(
    'Velocities Before and After Collision',
    fontsize=12,
    fontweight='bold')
ax1.set_xticks(x)
ax1.set_xticklabels(cars)
ax1.legend(fontsize=10)
ax1.grid(True, alpha=0.3, axis='y')

# Add values on bars
for i in range(len(cars)):
    ax1.text(i - width/2, v_before[i] + 0.5, f'{v_before[i]:.1f}',
             ha='center', fontsize=9, fontweight='bold')
    ax1.text(i + width/2, v_after[i] + 0.5, f'{v_after[i]:.1f}',
             ha='center', fontsize=9, fontweight='bold')

# Momentum bar chart
ax2 = axes[0, 1]
L_A1 = m_A * v_A1
L_B1 = m_B * v_B1
L_A2 = m_A * v_A2
L_B2 = m_B * v_B2

states = ['Before', 'After']
L_A_vals = [L_A1, L_A2]
L_B_vals = [L_B1, L_B2]
x_m = np.arange(len(states))

ax2.bar(x_m, L_A_vals, width, label='Car A', color='blue', alpha=0.7)
ax2.bar(
    x_m,
    L_B_vals,
    width,
    bottom=L_A_vals,
    label='Car B',
    color='orange',
    alpha=0.7)
ax2.axhline(
    y=L_before,
    color='red',
    linestyle='--',
    linewidth=2,
    label=f'Total: {
        L_before:.0f}')
ax2.set_ylabel('Momentum (kg·m/s)', fontsize=11)
ax2.set_title('Momentum Conservation', fontsize=12, fontweight='bold')
ax2.set_xticks(x_m)
ax2.set_xticklabels(states)
ax2.legend(fontsize=9)
ax2.grid(True, alpha=0.3, axis='y')

# Energy breakdown
ax3 = axes[1, 0]
energy_labels = ['Initial\nKE', 'Final\nKE', 'Energy\nLost']
energy_vals = [T1/1000, T2/1000, T_loss/1000]  # Convert to kJ
energy_colors = ['blue', 'green', 'red']

bars_e = ax3.bar(energy_labels, energy_vals, color=energy_colors, alpha=0.7)
ax3.set_ylabel('Energy (kJ)', fontsize=11)
ax3.set_title('Energy in Collision', fontsize=12, fontweight='bold')
ax3.grid(True, alpha=0.3, axis='y')

# Add percentages
for bar, val in zip(bars_e, energy_vals):
    height = bar.get_height()
    ax3.text(bar.get_x() + bar.get_width()/2., height,
             f'{val:.1f} kJ', ha='center', va='bottom',
             fontsize=9, fontweight='bold')

# Effect of coefficient of restitution
ax4 = axes[1, 1]
e_range = np.linspace(0, 1, 50)
energy_loss_range = []

for e_val in e_range:
    v_A2_temp = ((m_A - e_val*m_B)*v_A1 + (1+e_val)*m_B*v_B1) / total_mass
    v_B2_temp = ((1+e_val)*m_A*v_A1 + (m_B - e_val*m_A)*v_B1) / total_mass
    T2_temp = 0.5 * m_A * v_A2_temp**2 + 0.5 * m_B * v_B2_temp**2
    energy_loss_range.append((T1 - T2_temp) / T1 * 100)

ax4.plot(e_range, energy_loss_range, 'b-', linewidth=2)
ax4.axvline(
    x=e,
    color='red',
    linestyle='--',
    linewidth=2,
    label=f'Our case: e={e}')
ax4.axhline(y=T_loss/T1*100, color='red', linestyle=':', alpha=0.5)
ax4.scatter([e], [T_loss/T1*100], color='red', s=100, zorder=5)
ax4.set_xlabel('Coefficient of Restitution (e)', fontsize=11)
ax4.set_ylabel('Energy Loss (%)', fontsize=11)
ax4.set_title('Energy Loss vs Elasticity', fontsize=12, fontweight='bold')
ax4.legend(fontsize=9)
ax4.grid(True, alpha=0.3)
ax4.set_xlim(0, 1)

# Annotate extremes
ax4.text(0.05, 95, 'Perfectly\nInelastic', ha='left', fontsize=9,
         bbox=dict(boxstyle='round', facecolor='red', alpha=0.3))
ax4.text(0.95, 5, 'Perfectly\nElastic', ha='right', fontsize=9,
         bbox=dict(boxstyle='round', facecolor='green', alpha=0.3))

plt.tight_layout()
plt.show()

print(f"\nKey observations:")
print(f"1. Momentum conserved exactly (no external forces)")
print(f"2. Energy lost due to inelastic collision (e < 1)")
print(f"3. Car A slows down, Car B speeds up")
print(f"4. Lower e → more energy lost")
print(f"5. Both momentum and restitution equations needed")
```

### What This Graph Tells You

**Velocity Comparison (top-left):** Before vs. after velocities for both cars. Car 1 slows down, Car 2 speeds up. The split depends on masses and the coefficient of restitution $e$.

**Momentum Stack (top-right):** Total momentum (top of stacked bar) is **IDENTICAL** before and after — this is conservation of momentum visualized. Individual momenta redistribute, but the sum is constant.

**Energy Loss (bottom-left):** KE curve shows energy is LOST during collision. For $e = 0.5$: roughly 25% of the collision KE is converted to heat, sound, and deformation.

**Energy Breakdown (bottom-right):** Shows where the energy goes: remaining KE (both cars) + energy lost. The lost fraction increases as $e$ decreases (more inelastic).

**Key Observations:**
- **Momentum is ALWAYS conserved** in collisions (no external horizontal forces during brief impact)
- **Energy is NOT conserved** (except when $e = 1$, perfectly elastic)
- $e = 0$: perfectly inelastic (cars stick together, maximum energy loss)
- $e = 1$: perfectly elastic (KE conserved, objects bounce cleanly)
- Real car crashes: $e \approx 0.1–0.3$ (most energy absorbed by crumpling — this is BY DESIGN for safety)

> **Student Takeaway:** Momentum conservation + coefficient of restitution = complete solution for any 1D collision. Momentum tells you the final velocities; energy analysis tells you how much was dissipated. This is used in crash analysis, sports engineering, and particle physics.

## Part C: Angular Impulse-Momentum

### Angular Momentum

**Particle about point O**:
$$\mathbf{H}_O = \mathbf{r} \times m\mathbf{v}$$

where $\mathbf{r}$ is position vector from O to particle

**Magnitude** (2D): $H_O = r \cdot mv \sin\phi$

where $\phi$ is angle between $\mathbf{r}$ and $\mathbf{v}$

**Rigid body about center of mass**:
$$\mathbf{H}_G = I_G \boldsymbol{\omega}$$

**For 2D rotation**: $H_G = I_G \omega$

### Angular Impulse-Momentum Theorem

Start with moment equation: $\mathbf{M}_O = \frac{d\mathbf{H}_O}{dt}$

Integrate over time:
$$\int_{t_1}^{t_2} \mathbf{M}_O \\, dt = \mathbf{H}_{O,2} - \mathbf{H}_{O,1}$$

**Angular impulse = Change in angular momentum**

### Conservation of Angular Momentum

**If total external moment = 0**:
$$\mathbf{H}_{O,1} = \mathbf{H}_{O,2}$$

Angular momentum conserved!

**Examples**:
- Figure skater pulling arms in (increases $\omega$)
- Diver tucking for somersault
- Satellite rotation control

### Combined Linear and Angular

**For rigid bodies with both translation and rotation**:

**Linear momentum**:
$$\int \mathbf{F} \\, dt = m\mathbf{v}_{G,2} - m\mathbf{v}_{G,1}$$

**Angular momentum** (about G):
$$\int \mathbf{M}_G \\, dt = I_G \omega_2 - I_G \omega_1$$

Apply both equations independently!

### Impact with Rotation

**Example**: Baseball bat hitting ball

**For bat**:
- Linear impulse changes velocity of center of mass
- Angular impulse (about G) changes rotation rate

**Sweet spot**: Point where no jarring force at handle
- Located at center of percussion
- Optimal impact location

### System Angular Momentum

**Total angular momentum about O**:
$$\mathbf{H}_O = \mathbf{r}_G \times m\mathbf{v}_G + I_G \boldsymbol{\omega}$$

- First term: Angular momentum due to translation of G about O
- Second term: Angular momentum due to rotation about G

**Special case** (O = G): $\mathbf{H}_G = I_G \boldsymbol{\omega}$

```python
# Example 3: Ball Striking Door Edge

print("=" * 60)
print("Example 3: Ball Striking Door - Angular Impulse-Momentum")
print("=" * 60)

# Given
m_ball = 0.5  # kg
v_ball = 10.0  # m/s (ball velocity)
m_door = 20.0  # kg (door mass)
L = 1.0  # m (door width)
d = 0.8  # m (impact distance from hinge)
e_door = 0.6  # coefficient of restitution

# Door moment of inertia (about hinge, treat as thin rod)
I_door = (1/3) * m_door * L**2

print(f"\nGiven:")
print(f"Ball: m = {m_ball} kg, v = {v_ball} m/s")
print(f"Door: m = {m_door} kg, L = {L} m")
print(f"Impact distance from hinge: d = {d} m")
print(f"Door moment of inertia: I = {I_door:.4f} kg·m²")
print(f"Coefficient of restitution: e = {e_door}")

# Initial conditions
omega_door_1 = 0  # Door initially at rest
v_ball_1 = v_ball

# Let v_ball_2 = velocity of ball after impact (unknown)
# Let omega_door_2 = angular velocity of door after impact (unknown)

# Conservation of angular momentum about hinge:
# L_ball_1 + L_door_1 = L_ball_2 + L_door_2
# m_ball * v_ball_1 * d + 0 = m_ball * v_ball_2 * d + I_door * omega_door_2

# Coefficient of restitution:
# e = -(v_ball_2 - v_door_at_d_2) / (v_ball_1 - 0)
# where v_door_at_d_2 = omega_door_2 * d
# e = -(v_ball_2 - omega_door_2 * d) / v_ball_1

# From restitution:
# v_ball_2 = omega_door_2 * d - e * v_ball_1

# Substitute into momentum:
# m_ball * v_ball_1 * d = m_ball * (omega_door_2 * d - e * v_ball_1) * d + I_door * omega_door_2
# m_ball * v_ball_1 * d = m_ball * omega_door_2 * d^2 - m_ball * e * v_ball_1 * d + I_door * omega_door_2
# m_ball * v_ball_1 * d * (1 + e) = omega_door_2 * (m_ball * d^2 + I_door)

omega_door_2 = (m_ball * v_ball_1 * d * (1 + e_door)) / \
    (m_ball * d**2 + I_door)
v_ball_2 = omega_door_2 * d - e_door * v_ball_1

print(f"\nSolution:")
print(f"Door angular velocity after impact: ω = {omega_door_2:.4f} rad/s")
print(f"Ball velocity after impact: v = {v_ball_2:.4f} m/s")
print(f"Ball velocity change: Δv = {v_ball_2 - v_ball_1:.4f} m/s")

# Verify angular momentum conservation
H_before = m_ball * v_ball_1 * d
H_after = m_ball * v_ball_2 * d + I_door * omega_door_2

print(f"\nAngular Momentum Conservation (about hinge):")
print(f"Before: H1 = {H_before:.6f} kg·m²/s")
print(f"After: H2 = {H_after:.6f} kg·m²/s")
print(f"Error: {abs(H_before - H_after):.9f}")

# Verify coefficient of restitution
v_door_at_d = omega_door_2 * d
v_rel_before = v_ball_1 - 0
v_rel_after = -(v_ball_2 - v_door_at_d)
e_check = v_rel_after / v_rel_before

print(f"\nCoefficient of Restitution Check:")
print(f"Door velocity at impact point: {v_door_at_d:.4f} m/s")
print(f"Relative approach velocity: {v_rel_before:.4f} m/s")
print(f"Relative separation velocity: {v_rel_after:.4f} m/s")
print(f"e (calculated): {e_check:.6f}")
print(f"e (given): {e_door:.6f}")

# Energy analysis
T1 = 0.5 * m_ball * v_ball_1**2
T2 = 0.5 * m_ball * v_ball_2**2 + 0.5 * I_door * omega_door_2**2
T_loss = T1 - T2

print(f"\nEnergy Analysis:")
print(f"Initial KE: T1 = {T1:.3f} J")
print(f"Final KE: T2 = {T2:.3f} J")
print(f"- Ball: {0.5 * m_ball * v_ball_2**2:.3f} J")
print(f"- Door: {0.5 * I_door * omega_door_2**2:.3f} J")
print(f"Energy lost: {T_loss:.3f} J ({T_loss/T1*100:.1f}%)")

# Impulse on ball
J_ball = m_ball * (v_ball_2 - v_ball_1)
print(f"\nImpulse on Ball:")
print(f"J = {J_ball:.3f} N·s (negative = rebounds)")

# Plotting
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Before/after diagram
ax1 = axes[0, 0]
ax1.text(
    0.5,
    0.9,
    'BEFORE IMPACT',
    ha='center',
    fontsize=12,
    fontweight='bold',
    transform=ax1.transAxes)
# Door (vertical line at x=0, hinge at origin)
ax1.plot([0, 0], [0, L], 'brown', linewidth=8, label='Door')
ax1.plot(0, 0, 'ko', markersize=12, label='Hinge')
# Ball approaching
ax1.arrow(-0.3, d, 0.25, 0, head_width=0.05, head_length=0.03,
          fc='red', ec='red', linewidth=2)
ax1.plot(-0.3, d, 'ro', markersize=15, label='Ball')
ax1.text(-0.3, d-0.15, f'v={v_ball} m/s', ha='center', fontsize=9)
ax1.text(-0.05, d+0.1, f'd={d}m', ha='left', fontsize=9)
ax1.set_xlim(-0.5, 0.5)
ax1.set_ylim(-0.2, 1.3)
ax1.set_aspect('equal')
ax1.axis('off')

# After diagram (on same subplot, right side)
ax1.text(0.5, 0.4, 'AFTER IMPACT', ha='center', fontsize=12, fontweight='bold',
         transform=ax1.transAxes)
# Door rotated
theta_viz = 0.3  # rad (for visualization)
door_x = L * np.sin(theta_viz)
door_y = L * np.cos(theta_viz)
ax1.plot([0, door_x], [0, door_y], 'brown', linewidth=8, alpha=0.5)
ax1.plot(0, 0, 'ko', markersize=12)
# Ball rebounding
ball_x_after = door_x + 0.3
ball_y_after = d
if v_ball_2 < 0:
    ax1.arrow(ball_x_after, ball_y_after, -0.2, 0, head_width=0.05,
              head_length=0.03, fc='blue', ec='blue', linewidth=2)
ax1.plot(ball_x_after, ball_y_after, 'bo', markersize=15)
ax1.text(
    ball_x_after,
    ball_y_after-0.15,
    f'v={
        v_ball_2:.1f} m/s',
    ha='center',
    fontsize=9)
# Angular velocity arrow
ax1.annotate(
    '', xy=(
        0.1, 0.3), xytext=(
            0.15, 0.2), arrowprops=dict(
                arrowstyle='->', lw=2, color='green', connectionstyle="arc3,rad=.5"))
ax1.text(0.2, 0.25, f'ω={omega_door_2:.2f}\nrad/s',
         fontsize=9, color='green', fontweight='bold')

# Angular momentum
ax2 = axes[0, 1]
H_ball_before = m_ball * v_ball_1 * d
H_door_before = 0
H_ball_after = m_ball * v_ball_2 * d
H_door_after = I_door * omega_door_2

labels_H = [
    'Ball\n(before)',
    'Door\n(before)',
    'Ball\n(after)',
    'Door\n(after)']
H_values = [H_ball_before, H_door_before, H_ball_after, H_door_after]
colors_H = ['red', 'brown', 'blue', 'green']
x_H = np.arange(len(labels_H))

bars_H = ax2.bar(x_H, H_values, color=colors_H, alpha=0.7)
ax2.axhline(y=0, color='k', linewidth=1)
ax2.set_ylabel('Angular Momentum (kg·m²/s)', fontsize=11)
ax2.set_title('Angular Momentum About Hinge', fontsize=12, fontweight='bold')
ax2.set_xticks(x_H)
ax2.set_xticklabels(labels_H, fontsize=9)
ax2.grid(True, alpha=0.3, axis='y')

# Draw conservation
ax2.plot([0.5, 2.5], [H_before, H_after],
         'k--', linewidth=2, label='Conserved')
ax2.legend(fontsize=9)

# Velocity comparison
ax3 = axes[1, 0]
vel_labels = ['Ball\nBefore', 'Ball\nAfter', 'Door edge\nAfter']
vel_values = [v_ball_1, v_ball_2, v_door_at_d]
vel_colors = ['red', 'blue', 'green']

bars_v = ax3.bar(vel_labels, vel_values, color=vel_colors, alpha=0.7)
ax3.axhline(y=0, color='k', linewidth=1)
ax3.set_ylabel('Velocity (m/s)', fontsize=11)
ax3.set_title('Velocities', fontsize=12, fontweight='bold')
ax3.grid(True, alpha=0.3, axis='y')

for bar, val in zip(bars_v, vel_values):
    height = bar.get_height()
    ax3.text(bar.get_x() + bar.get_width()/2., height,
             f'{val:.2f}', ha='center',
             va='bottom' if height > 0 else 'top',
             fontsize=9, fontweight='bold')

# Energy distribution
ax4 = axes[1, 1]
energy_labels = ['Initial\nKE', 'Ball\nFinal', 'Door\nFinal', 'Lost']
energy_values = [
    T1,
    0.5 *
    m_ball *
    v_ball_2**2,
    0.5 *
    I_door *
    omega_door_2**2,
    T_loss]
energy_colors = ['blue', 'lightblue', 'lightgreen', 'red']

bars_e = ax4.bar(energy_labels, energy_values, color=energy_colors, alpha=0.7)
ax4.set_ylabel('Energy (J)', fontsize=11)
ax4.set_title('Energy Distribution', fontsize=12, fontweight='bold')
ax4.grid(True, alpha=0.3, axis='y')

for bar, val in zip(bars_e, energy_values):
    height = bar.get_height()
    ax4.text(bar.get_x() + bar.get_width()/2., height,
             f'{val:.2f} J', ha='center', va='bottom',
             fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()

print(f"\nKey observations:")
print(f"1. Angular momentum conserved about hinge")
print(f"2. Ball rebounds backward (v < 0)")
print(f"3. Door rotates (gains angular momentum)")
print(f"4. Energy lost in collision (e < 1)")
print(f"5. Must use angular momentum (not linear) for rotation")
```

**Before/After Geometry (top-left):** The ball strikes the door at distance $d$ from the hinge. The ball rebounds; the door starts rotating. The hinge provides an external force but does NO work (no displacement at hinge point).

**Angular Momentum (top-right):** Angular momentum about the HINGE is conserved (hinge force has zero moment arm). This is the key insight — choose the right reference point to eliminate unknown forces!

**Velocity Comparison (bottom-left):** The ball rebounds slower than it arrived (it transferred momentum to the door). The door's edge velocity depends on where the ball struck.

**Key Observations:**
- Choosing the **hinge as the moment center** eliminates the unknown hinge reaction force from the equation
- Angular impulse-momentum: $L_{before} = L_{after}$ about the hinge
- Striking farther from the hinge transfers MORE angular momentum to the door (larger moment arm)
- This is the principle behind door handles being far from hinges — maximum leverage!

> **Student Takeaway:** Angular impulse-momentum extends the collision framework to rotating bodies. The KEY skill: choose your reference point wisely to eliminate unknown constraint forces. About the hinge, the angular impulse of the hinge reaction is zero.

## Summary

### Key Concepts Covered

1. **Impulse-Momentum Theorem**:
 - $\mathbf{J} = \int \mathbf{F} \\, dt = \Delta \mathbf{L}$
 - Vector equation (apply component-wise)
 - Ideal for time-dependent forces

2. **Conservation of Momentum**:
 - If $\mathbf{F}_{ext} = 0$: $\mathbf{L}_1 = \mathbf{L}_2$
 - Always conserved in collisions (no external force during impact)
 - Applies even when energy is lost

3. **Collisions**:
 - Elastic: Both momentum and energy conserved
 - Inelastic: Momentum conserved, energy lost
 - Coefficient of restitution: $e = v_{rel,after} / v_{rel,before}$

4. **Angular Impulse-Momentum**:
 - $\int \mathbf{M} \\, dt = \Delta \mathbf{H}$
 - For rigid bodies: $\mathbf{H}_G = I_G \boldsymbol{\omega}$
 - Conservation: If $\mathbf{M}_{ext} = 0$: $\mathbf{H}_1 = \mathbf{H}_2$

### Problem-Solving Procedures

**Method 1: Impulse-Momentum (Single Object)**
1. Identify initial and final velocities
2. Calculate impulse: $\mathbf{J} = \int \mathbf{F} \\, dt$
3. Apply: $\mathbf{J} = m(\mathbf{v}_2 - \mathbf{v}_1)$
4. Solve component-wise if needed

**Method 2: Conservation (Collisions)**
1. Draw diagram with velocities before/after
2. Apply momentum conservation: $\sum m_i \mathbf{v}_{i,1} = \sum m_i \mathbf{v}_{i,2}$
3. Apply coefficient of restitution: $e = (v_{B,2} - v_{A,2}) / (v_{A,1} - v_{B,1})$
4. Solve system of equations
5. Check: Energy should decrease unless e=1

**Method 3: Angular Impulse-Momentum**
1. Choose reference point (hinge, pivot, or G)
2. Calculate angular momenta: $\mathbf{H} = \mathbf{r} \times m\mathbf{v}$ or $I\omega$
3. Apply: $\int \mathbf{M} \\, dt = \mathbf{H}_2 - \mathbf{H}_1$
4. Use restitution if collision involved

### When to Use: Energy vs Momentum

**Use Energy Methods When**:
- Want to find speeds (not directions)
- Forces are position-dependent
- Conservative system (no friction)
- Work is easy to calculate

**Use Momentum Methods When**:
- Want to find velocities (with directions)
- Forces are time-dependent
- Collisions/impacts
- External force = 0 (conservation)
- Very short time intervals

**Often Use Both**:
- Momentum: Find velocities after collision
- Energy: Check energy loss, find final positions

### Applications

- **Vehicle safety**: Crash analysis, airbag timing
- **Sports**: Ball impacts, bat/racket design
- **Aerospace**: Rocket staging, satellite maneuvering
- **Manufacturing**: Hammering, forging, impact testing
- **Robotics**: Collision detection and response
