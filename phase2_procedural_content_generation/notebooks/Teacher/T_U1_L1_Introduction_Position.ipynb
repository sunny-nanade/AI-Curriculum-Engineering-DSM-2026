## What is Dynamics?

**Dynamics** is the branch of mechanics that deals with the motion of objects.

It answers three fundamental questions:

| **Question** | **Field** | **Example** |
|--------------|-----------|-------------|
| **HOW does it move?** | **Kinematics** | Describing a car's trajectory |
| **WHY does it move?** | **Kinetics** | Analyzing forces on the car |
| **What will it do next?** | **Prediction** | Forecasting future motion |

---

## Kinematics vs Kinetics

### Kinematics

> Study of motion **WITHOUT** considering forces
>
> "HOW does it move?"

**Questions answered:**
- What is the position at time t?
- How fast is it moving? (velocity)
- Is it speeding up or slowing down? (acceleration)

**Example**: The car moved from point A to B in 5 seconds, reaching 20 m/s.

---

### Kinetics

> Study of motion **INCLUDING** forces
>
> "WHY does it move?"

**Questions answered:**
- What forces cause the motion?
- How much force is needed?
- What is the effect of mass?

**Example**: The car accelerated because the engine applied 500 N forward force.

## Newton's Laws of Motion

These are the foundation of classical dynamics. Let's review each law.

---

### Newton's First Law: Law of Inertia

> An object at rest stays at rest, and an object in motion stays in motion with constant velocity, unless acted upon by a net external force.

**Mathematical Form**:

$$\sum \vec{F} = 0 \quad \Rightarrow \quad \vec{v} = \text{constant}$$

**Example**: A hockey puck sliding on frictionless ice continues moving forever.

---

### Newton's Second Law: F = ma

> The acceleration of an object is directly proportional to the net force acting on it and inversely proportional to its mass.

**Mathematical Form**:

$$\sum \vec{F} = m\vec{a}$$

Where:
- $\vec{F}$ = Net force (Newtons, N)
- $m$ = Mass (kilograms, kg)
- $\vec{a}$ = Acceleration (m/s²)

**Example**: Pushing a shopping cart - heavier cart (larger m) requires more force for the same acceleration.

---

### Newton's Third Law: Action-Reaction

> For every action, there is an equal and opposite reaction.

**Mathematical Form**:

$$\vec{F}_{AB} = -\vec{F}_{BA}$$

**Example**: When you push on a wall, the wall pushes back on you with equal force.

---

### Answer

The **second law** (F = ma) connects forces to motion.

However, in kinematics we focus on the motion itself (position, velocity, acceleration) **without worrying about forces yet**.

## Reference Frames

To describe motion, we need a **reference frame** - a coordinate system from which we measure position, velocity, and acceleration.

---

### Inertial Reference Frame

> A reference frame where Newton's laws hold true.
> 
> An object with no net force has zero acceleration.

**Examples**:
- The ground (for everyday problems)
- The stars (for spacecraft)
- Any frame moving at constant velocity

---

### Non-Inertial Reference Frame

> A reference frame that is accelerating.
> 
> Newton's laws don't work directly - need "fictitious forces".

**Examples**:
- Inside an accelerating car (you feel pushed back)
- On a rotating merry-go-round (centrifugal force)
- Inside an elevator accelerating upward

---

## What is a Position Vector?

The **position vector** $\vec{r}(t)$ describes the **location** of a particle at time $t$ relative to a chosen origin.

---

### Definition

In **Cartesian coordinates** (3D):

$$\vec{r}(t) = x(t)\hat{i} + y(t)\hat{j} + z(t)\hat{k}$$

Where:
- $x(t)$ = position along x-axis (meters)
- $y(t)$ = position along y-axis (meters)
- $z(t)$ = position along z-axis (meters)
- $\hat{i}, \hat{j}, \hat{k}$ = unit vectors (fixed directions)

---

In **2D** (most common for now):

$$\vec{r}(t) = x(t)\hat{i} + y(t)\hat{j}$$

---

### Key Points

1. Position vector **starts at the origin** and **points to the particle**
2. It's a **function of time** $t$ (position changes as object moves)
3. **Units**: meters (m) in SI system
4. **Direction matters** - it's a vector!

---

## Visualizing Position Vectors

Let's visualize this concept using Python. The code below creates a simple plot showing a position vector.

**What the code does**: 
- Draws coordinate axes
- Plots a position vector as an arrow from origin to point (3, 4)
- Shows the x and y components as dashed lines

---

````python
# Example: Position vector at a single instant
# Particle at position (3, 4) meters at time t = 2 seconds

x = 3 # x-coordinate in meters
y = 4 # y-coordinate in meters

# Create the plot
fig, ax = plt.subplots(figsize=(8, 8))

# Draw coordinate axes
ax.axhline(0, color='black', linewidth=0.5)
ax.axvline(0, color='black', linewidth=0.5)
ax.grid(True, alpha=0.3)

# Draw the position vector (red arrow from origin to point)
ax.quiver(0, 0, x, y, angles='xy', scale_units='xy', scale=1, 
 color='red', width=0.01, headwidth=5, label='Position vector')

# Draw component lines (dashed)
ax.plot([0, x], [0, 0], 'b--', linewidth=2, alpha=0.5, label=f'x = {x} m')
ax.plot([x, x], [0, y], 'g--', linewidth=2, alpha=0.5, label=f'y = {y} m')

# Mark the particle position
ax.plot(x, y, 'ro', markersize=10)
ax.text(x+0.2, y+0.2, f'Particle at ({x}, {y})', fontsize=12, fontweight='bold')

# Mark origin
ax.plot(0, 0, 'ko', markersize=8)
ax.text(0.2, -0.3, 'Origin', fontsize=12, fontweight='bold')

# Labels and formatting
ax.set_xlabel('x (meters)', fontsize=14, fontweight='bold')
ax.set_ylabel('y (meters)', fontsize=14, fontweight='bold')
ax.set_title('Position Vector Example\n$\\vec{r} = 3\\hat{i} + 4\\hat{j}$ m', fontsize=16, fontweight='bold')
ax.set_xlim(-1, 6)
ax.set_ylim(-1, 6)
ax.set_aspect('equal')
ax.legend(loc='upper left', fontsize=12)

plt.tight_layout()
plt.show()

# Calculate magnitude
magnitude = np.sqrt(x**2 + y**2)
print(f"\nPosition vector: r = {x}i + {y}j meters")
print(f"Magnitude: |r| = sqrt({x}² + {y}²) = {magnitude:.2f} meters")
````

### Understanding the Plot

1. The **red arrow** is the position vector - it tells us where the particle is
2. The **dashed lines** show the x and y components
3. The **magnitude** (5 m) is the straight-line distance from origin to particle
4. **Direction matters** - this vector is different from $-3\hat{i} - 4\hat{j}$

## Example 1: Particle at Rest

A particle is located at position (2, 3) meters and doesn't move.

**Question**: What is the position vector $\vec{r}(t)$?

**Think about it...**

*(Scroll down for solution)*

---

### Solution

Since the particle doesn't move:

$$\vec{r}(t) = 2\hat{i} + 3\hat{j} \quad \text{(constant for all time } t\text{)}$$

At $t = 0$: $\vec{r}(0) = 2\hat{i} + 3\hat{j}$ m

At $t = 5$: $\vec{r}(5) = 2\hat{i} + 3\hat{j}$ m

At any time $t$: $\vec{r}(t) = 2\hat{i} + 3\hat{j}$ m

The position vector is **constant** because there's no motion.

---

## Example 2: Particle Moving in Straight Line

A particle moves along the x-axis with position:

$$x(t) = 2 + 3t \quad \text{meters}$$

Where $t$ is time in seconds.

**Questions**: 
1. What is the position vector $\vec{r}(t)$?
2. What is the position at $t = 0, 1, 2$ seconds?

**Try to solve first...**

*(Scroll for solution)*

---

### Solution

Since motion is only along x-axis, $y(t) = 0$ and $z(t) = 0$:

$$\vec{r}(t) = (2 + 3t)\hat{i} + 0\hat{j} = (2 + 3t)\hat{i} \quad \text{m}$$

At $t = 0$: $\vec{r}(0) = 2\hat{i}$ m (starts at 2 m from origin)

At $t = 1$: $\vec{r}(1) = 5\hat{i}$ m

At $t = 2$: $\vec{r}(2) = 8\hat{i}$ m

The particle is moving in the positive x-direction.

---

### Visualization

Let's plot this motion to see how position changes with time.

**What the code does**: 
- Calculates position at different times (t = 0, 1, 2, 3, 4 seconds)
- Draws position vectors at each time with different colors
- Shows the trajectory along the x-axis

---

````python
# Time array (seconds)
t = np.array([0, 1, 2, 3, 4])

# Position along x-axis: x(t) = 2 + 3t
x = 2 + 3*t
y = np.zeros_like(t) # y is always 0 (moving along x-axis only)

# Create the plot
fig, ax = plt.subplots(figsize=(12, 6))

# Draw axes
ax.axhline(0, color='black', linewidth=1)
ax.axvline(0, color='black', linewidth=1)
ax.grid(True, alpha=0.3)

# Plot trajectory line
ax.plot(x, y, 'b-', linewidth=2, alpha=0.3, label='Trajectory')

# Plot position vectors at each time
colors = plt.cm.viridis(np.linspace(0, 1, len(t)))
for i in range(len(t)):
 # Draw position vector
 ax.quiver(0, 0, x[i], y[i], angles='xy', scale_units='xy', scale=1,
 color=colors[i], width=0.008, headwidth=5, alpha=0.6)
 # Mark position
 ax.plot(x[i], y[i], 'o', markersize=10, color=colors[i])
 # Label with time and position
 ax.text(x[i], 0.3, f't={t[i]}s\nx={x[i]:.0f}m', 
 ha='center', fontsize=10, fontweight='bold')

# Labels and formatting
ax.set_xlabel('x (meters)', fontsize=14, fontweight='bold')
ax.set_ylabel('y (meters)', fontsize=14, fontweight='bold')
ax.set_title('Particle Motion Along X-Axis\n$\\vec{r}(t) = (2+3t)\\hat{i}$ m', fontsize=16, fontweight='bold')
ax.set_xlim(-1, 16)
ax.set_ylim(-2, 2)

plt.tight_layout()
plt.show()

print("\nPosition at different times:")
for i in range(len(t)):
 print(f"t = {t[i]} s → x = {x[i]:.1f} m → r = {x[i]:.1f}i m")
````

## Example 3: Projectile Motion (2D)

**Problem Statement:** 
A cricket ball is thrown from ground level (origin) with the following initial velocity:
- Horizontal component: $v_{x0} = 10$ m/s (constant, no air resistance)
- Vertical component: $v_{y0} = 15$ m/s (upward)

The ball experiences gravitational acceleration $g = 9.8$ m/s² downward.

The position components follow:
$$x(t) = 10t \quad \text{m}$$
$$y(t) = 15t - 4.9t^2 \quad \text{m}$$

**Questions:** 
A. What is the maximum height reached by the ball? 
B. At what time does the ball reach maximum height? 
C. What is the total time the ball is in the air? 
D. What is the horizontal range (how far does it land)? 
E. What is the position vector at $t = 1$ second?

**Try solving manually first...**

*(Scroll for solution)*

---


### Step 1⃣: Given & Find (Organize the Information)

** GIVEN:**
- Initial position: $(x_0, y_0) = (0, 0)$ (ground level, origin)
- Initial velocity components:
 - Horizontal: $v_{x0} = 10$ m/s (constant)
 - Vertical: $v_{y0} = 15$ m/s (upward)
- Acceleration: 
 - Horizontal: $a_x = 0$ (no forces in x-direction)
 - Vertical: $a_y = -g = -9.8$ m/s² (gravity, downward)

** FIND:**
- Part A: Maximum height $h_{\text{max}}$
- Part B: Time to reach max height $t_{\text{max}}$
- Part C: Total time in air $T$
- Part D: Horizontal range $R$
- Part E: Position vector at $t = 1$ s

---


### Step 2⃣: Concept Review (Key Formulas for Projectile Motion)

**Position in 2D:**
$$\vec{r}(t) = \vec{r}_0 + \vec{v}_0 t + \frac{1}{2}\vec{a}t^2$$

**Component form:**
$$x(t) = x_0 + v_{x0}t + \frac{1}{2}a_x t^2 = 0 + v_{x0}t + 0 = v_{x0}t$$
$$y(t) = y_0 + v_{y0}t + \frac{1}{2}a_y t^2 = 0 + v_{y0}t - \frac{1}{2}g t^2$$

**Key relationships:**
- At maximum height: $v_y = 0$ → $t_{\text{max}} = \frac{v_{y0}}{g}$
- Maximum height: $h_{\text{max}} = y(t_{\text{max}}) = \frac{v_{y0}^2}{2g}$
- Total time: $T = 2t_{\text{max}}$ (symmetry for projectile from ground)
- Range: $R = v_{x0} \cdot T$

---


### Step 3⃣: Manual Solution (Step-by-Step Algebra)

Let's solve each part systematically:

---

**Part A: Maximum Height**

Method 1 - Using the formula:
$$h_{\text{max}} = \frac{v_{y0}^2}{2g} = \frac{(15)^2}{2(9.8)} = \frac{225}{19.6} = 11.48 \text{ m}$$

Method 2 - Substituting into $y(t)$ at $t_{\text{max}}$:
$$t_{\text{max}} = \frac{v_{y0}}{g} = \frac{15}{9.8} = 1.53 \text{ s}$$
$$h_{\text{max}} = 15(1.53) - 4.9(1.53)^2 = 22.96 - 11.47 = 11.49 \text{ m}$$

$$\boxed{h_{\text{max}} = 11.48 \text{ m}}$$

---

**Part B: Time to Reach Maximum Height**

At max height, vertical velocity = 0:
$$v_y = v_{y0} - gt = 0$$
$$t_{\text{max}} = \frac{v_{y0}}{g} = \frac{15}{9.8} = 1.53 \text{ s}$$

$$\boxed{t_{\text{max}} = 1.53 \text{ s}}$$

---

**Part C: Total Time in Air**

Ball returns to ground when $y(T) = 0$:
$$15T - 4.9T^2 = 0$$
$$T(15 - 4.9T) = 0$$
$$T = 0 \text{ (launch)} \quad \text{or} \quad T = \frac{15}{4.9} = 3.06 \text{ s (landing)}$$

Alternatively: $T = 2t_{\text{max}} = 2(1.53) = 3.06$ s

$$\boxed{T = 3.06 \text{ s}}$$

---

**Part D: Horizontal Range**

Horizontal distance when ball lands:
$$R = v_{x0} \cdot T = 10 \times 3.06 = 30.6 \text{ m}$$

$$\boxed{R = 30.6 \text{ m}}$$

---

**Part E: Position at t = 1 s**

Substitute $t = 1$ into position equations:
$$x(1) = 10(1) = 10 \text{ m}$$
$$y(1) = 15(1) - 4.9(1)^2 = 15 - 4.9 = 10.1 \text{ m}$$

Position vector:
$$\vec{r}(1) = 10\hat{i} + 10.1\hat{j} \text{ m}$$

Magnitude:
$$|\vec{r}(1)| = \sqrt{10^2 + 10.1^2} = \sqrt{100 + 102.01} = \sqrt{202.01} = 14.21 \text{ m}$$

$$\boxed{\vec{r}(1) = 10\hat{i} + 10.1\hat{j} \text{ m}, \quad |\vec{r}| = 14.21 \text{ m}}$$

---


### Step 4⃣: Summary of Results

| Part | Quantity | Manual Calculation | Result |
|------|----------|-------------------|--------|
| A | Max Height | $\frac{15^2}{2(9.8)}$ | **11.48 m** |
| B | Time to Max Height | $\frac{15}{9.8}$ | **1.53 s** |
| C | Total Time in Air | $2 \times 1.53$ | **3.06 s** |
| D | Range | $10 \times 3.06$ | **30.6 m** |
| E | Position at t=1s | $(10, 15-4.9)$ | **(10, 10.1) m** |

Now let's verify with Python code...

---


### Step 5⃣: Verification (Compare Manual vs Python)

Now we'll implement the same calculations in Python to verify our manual work:


````python
# Given values
vx0 = 10 # m/s (horizontal velocity)
vy0 = 15 # m/s (vertical velocity)
g = 9.8 # m/s² (gravitational acceleration)

# Manual calculations (for verification)
import math

# Part A: Maximum height
t_max = vy0 / g
h_max = (vy0**2) / (2 * g)

# Part B: Time to max height (already calculated)
# t_max = t_max

# Part C: Total time in air
T = 2 * t_max

# Part D: Range
R = vx0 * T

# Part E: Position at t=1s
x_1 = vx0 * 1
y_1 = vy0 * 1 - 0.5 * g * 1**2
r_1_mag = math.sqrt(x_1**2 + y_1**2)

print("=" * 60)
print("VERIFICATION: Manual Calculations vs Python Code")
print("=" * 60)
print(f"Part A - Maximum height: h_max = {h_max:.2f} m")
print(f"Part B - Time to max height: t_max = {t_max:.2f} s")
print(f"Part C - Total time in air: T = {T:.2f} s")
print(f"Part D - Range: R = {R:.2f} m")
print(f"Part E - Position at t=1s: r(1) = ({x_1:.1f}, {y_1:.1f}) m")
print(f"Magnitude: |r(1)| = {r_1_mag:.2f} m")
print("=" * 60)
print("All manual calculations verified by Python code!")
print("=" * 60)

# Time array for smooth trajectory
t_full = np.linspace(0, T, 100)

# Position components
x_full = vx0 * t_full
y_full = vy0 * t_full - 4.9 * t_full**2

# Specific times to show vectors
t_points = np.array([0, 0.5, 1, 1.5, 2, 2.5, 3])
x_points = vx0 * t_points
y_points = vy0 * t_points - 4.9 * t_points**2

# Create figure with two subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

# LEFT PLOT: Trajectory with position vectors
ax1.plot(x_full, y_full, 'b-', linewidth=3, label='Trajectory', alpha=0.5)
ax1.axhline(0, color='black', linewidth=1, linestyle='--', alpha=0.3)
ax1.grid(True, alpha=0.3)

# Draw position vectors at specific times
colors = plt.cm.plasma(np.linspace(0, 1, len(t_points)))
for i in range(len(t_points)):
 if y_points[i] >= 0: # Only show while above ground
  ax1.quiver(0, 0, x_points[i], y_points[i], angles='xy', scale_units='xy', scale=1,
 color=colors[i], width=0.005, headwidth=5, alpha=0.6)
 ax1.plot(x_points[i], y_points[i], 'o', markersize=8, color=colors[i])
 ax1.text(x_points[i]+1, y_points[i]+1, f't={t_points[i]:.1f}s', 
 fontsize=9, fontweight='bold')

ax1.set_xlabel('x (meters)', fontsize=14, fontweight='bold')
ax1.set_ylabel('y (meters)', fontsize=14, fontweight='bold')
ax1.set_title('Projectile Trajectory with Position Vectors', fontsize=14, fontweight='bold')
ax1.set_xlim(-2, 35)
ax1.set_ylim(-2, 14)
ax1.legend(loc='upper right', fontsize=12)

# RIGHT PLOT: Position components vs time
ax2.plot(t_full, x_full, 'b-', linewidth=2, label=f'x(t) = {vx0}t (linear)')
ax2.plot(t_full, y_full, 'r-', linewidth=2, label=f'y(t) = {vy0}t - 4.9t² (parabolic)')
ax2.axhline(0, color='black', linewidth=1, linestyle='--', alpha=0.3)
ax2.grid(True, alpha=0.3)
ax2.set_xlabel('Time (seconds)', fontsize=14, fontweight='bold')
ax2.set_ylabel('Position (meters)', fontsize=14, fontweight='bold')
ax2.set_title('Position Components vs Time', fontsize=14, fontweight='bold')
ax2.legend(loc='upper left', fontsize=12)

plt.tight_layout()
plt.show()

# Print position at specific times
print("\nPosition vector at different times:")
for i in range(len(t_points)):
 if y_points[i] >= 0:
     mag = np.sqrt(x_points[i]**2 + y_points[i]**2)
 print(f"t = {t_points[i]:.1f} s → r = {x_points[i]:.1f}i + {y_points[i]:.1f}j m (|r| = {mag:.1f} m)")
````

### Step 7⃣: Discussion & Physical Insights

** What the numbers tell us:**

1. **Trajectory Shape**: Classic parabola - symmetric rise and fall
 - Takes 1.53 s to go up, another 1.53 s to come down
 - Maximum height (11.48 m) occurs at midpoint of flight

2. **Horizontal vs Vertical Motion**:
 - Horizontal: Constant speed (10 m/s) → linear x(t) = 10t
 - Vertical: Deceleration by gravity → parabolic y(t) = 15t - 4.9t²
 - These are **independent** - key projectile motion principle!

3. **Energy Perspective**:
 - At launch: Maximum kinetic energy (both vx and vy)
 - At peak: Only horizontal KE remains (vy = 0), max potential energy
 - At landing: Symmetric to launch (same speed, opposite vy direction)

4. **Range Formula Derivation**:
 - $R = v_{x0} \cdot T = v_{x0} \cdot \frac{2v_{y0}}{g} = \frac{2v_{x0}v_{y0}}{g}$
 - In our case: $R = \frac{2(10)(15)}{9.8} = \frac{300}{9.8} = 30.6$ m 

---

** Exam Strategy:**

When solving projectile problems manually:

1. **Always separate x and y components** - they're independent!
2. **Know your key formulas**:
 - $t_{\text{max}} = \frac{v_{y0}}{g}$
 - $h_{\text{max}} = \frac{v_{y0}^2}{2g}$
 - $T = 2t_{\text{max}}$ (if launch and land at same height)
 - $R = v_{x0} \cdot T$

3. **Check your answers**:
 - Does the ball go higher if thrown faster upward? (yes)
 - Does horizontal speed affect maximum height? (no!)
 - Are your units consistent? (m, m/s, m/s²)

4. **Common mistakes to avoid**:
 - Forgetting the factor of 1/2 in $y = v_{y0}t - \frac{1}{2}gt^2$
 - Using total time instead of time to max height
 - Mixing up x and y components

---


````python
# Time array for smooth trajectory
t_full = np.linspace(0, 3.06, 100) # Until ball hits ground

# Position components
x_full = 10 * t_full
y_full = 15 * t_full - 4.9 * t_full**2

# Specific times to show vectors
t_points = np.array([0, 0.5, 1, 1.5, 2, 2.5, 3])
x_points = 10 * t_points
y_points = 15 * t_points - 4.9 * t_points**2

# Create figure with two subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

# LEFT PLOT: Trajectory with position vectors
ax1.plot(x_full, y_full, 'b-', linewidth=3, label='Trajectory', alpha=0.5)
ax1.axhline(0, color='black', linewidth=1, linestyle='--', alpha=0.3)
ax1.grid(True, alpha=0.3)

# Draw position vectors at specific times
colors = plt.cm.plasma(np.linspace(0, 1, len(t_points)))
for i in range(len(t_points)):
 if y_points[i] >= 0: # Only show while above ground
  ax1.quiver(0, 0, x_points[i], y_points[i], angles='xy', scale_units='xy', scale=1,
 color=colors[i], width=0.005, headwidth=5, alpha=0.6)
 ax1.plot(x_points[i], y_points[i], 'o', markersize=8, color=colors[i])
 ax1.text(x_points[i]+1, y_points[i]+1, f't={t_points[i]:.1f}s', 
 fontsize=9, fontweight='bold')

ax1.set_xlabel('x (meters)', fontsize=14, fontweight='bold')
ax1.set_ylabel('y (meters)', fontsize=14, fontweight='bold')
ax1.set_title('Projectile Trajectory with Position Vectors', fontsize=14, fontweight='bold')
ax1.set_xlim(-2, 35)
ax1.set_ylim(-2, 14)
ax1.legend(loc='upper right', fontsize=12)

# RIGHT PLOT: Position components vs time
ax2.plot(t_full, x_full, 'b-', linewidth=2, label='x(t) = 10t (linear)')
ax2.plot(t_full, y_full, 'r-', linewidth=2, label='y(t) = 15t - 4.9t² (parabolic)')
ax2.axhline(0, color='black', linewidth=1, linestyle='--', alpha=0.3)
ax2.grid(True, alpha=0.3)
ax2.set_xlabel('Time (seconds)', fontsize=14, fontweight='bold')
ax2.set_ylabel('Position (meters)', fontsize=14, fontweight='bold')
ax2.set_title('Position Components vs Time', fontsize=14, fontweight='bold')
ax2.legend(loc='upper left', fontsize=12)

plt.tight_layout()
plt.show()

# Print position at specific times
print("\nPosition vector at different times:")
for i in range(len(t_points)):
 if y_points[i] >= 0:
     mag = np.sqrt(x_points[i]**2 + y_points[i]**2)
 print(f"t = {t_points[i]:.1f} s → r = {x_points[i]:.1f}i + {y_points[i]:.1f}j m (|r| = {mag:.1f} m)")
````

### Key Observations

1. **Trajectory shape**: Parabola (characteristic of projectile motion)
2. **Horizontal motion**: Linear (constant speed in x-direction)
3. **Vertical motion**: Parabolic (gravity pulls down with 9.8 m/s²)
4. **Position vectors**: Change in both magnitude AND direction as particle moves
5. **At peak**: Vertical component reaches maximum, then decreases

## Practice: Circular Motion

A robot arm moves such that the position of its end effector is given by:

$$x(t) = 0.5 \cos(t) \quad \text{m}$$
$$y(t) = 0.5 \sin(t) \quad \text{m}$$

Where $t$ is time in seconds.

**Tasks**:
1. Write the position vector $\vec{r}(t)$
2. Find the position at $t = 0, \pi/2, \pi$ seconds
3. Calculate the magnitude of the position vector
4. Describe the motion

**Work on this now...**

*(Scroll for solution)*

---

### Solution - Part 1

**Position vector**:

$$\vec{r}(t) = 0.5\cos(t)\hat{i} + 0.5\sin(t)\hat{j} \quad \text{m}$$

---

### Solution - Part 2

**Position at specific times**:

At $t = 0$:
$$\vec{r}(0) = 0.5\cos(0)\hat{i} + 0.5\sin(0)\hat{j} = 0.5\hat{i} + 0\hat{j} \text{ m}$$

At $t = \pi/2$:
$$\vec{r}(\pi/2) = 0.5\cos(\pi/2)\hat{i} + 0.5\sin(\pi/2)\hat{j} = 0\hat{i} + 0.5\hat{j} \text{ m}$$

At $t = \pi$:
$$\vec{r}(\pi) = 0.5\cos(\pi)\hat{i} + 0.5\sin(\pi)\hat{j} = -0.5\hat{i} + 0\hat{j} \text{ m}$$

---

### Solution - Part 3

**Magnitude**:

$$|\vec{r}(t)| = \sqrt{(0.5\cos t)^2 + (0.5\sin t)^2}$$

$$= \sqrt{0.25(\cos^2 t + \sin^2 t)}$$

$$= \sqrt{0.25 \times 1} = 0.5 \text{ m}$$

**The magnitude is constant!** The distance from origin never changes.

---

### Solution - Part 4

**Description of motion**:

The robot arm end effector moves in a **circular path** with:
- Radius: 0.5 m
- Center: Origin
- Period: $2\pi$ seconds (one complete circle)

This is called **uniform circular motion**.

---

### Visualization

**What the code does**: 
- Plots circular trajectory
- Shows position vectors at 8 different times around the circle
- Demonstrates that all vectors have the same length (0.5 m)

---

````python
# Time array (one complete circle)
t = np.linspace(0, 2*np.pi, 200)

# Position components for circular motion
x = 0.5 * np.cos(t)
y = 0.5 * np.sin(t)

# Specific times to show vectors
t_points = np.array([0, np.pi/4, np.pi/2, 3*np.pi/4, np.pi, 5*np.pi/4, 3*np.pi/2, 7*np.pi/4])
x_points = 0.5 * np.cos(t_points)
y_points = 0.5 * np.sin(t_points)

# Create the plot
fig, ax = plt.subplots(figsize=(10, 10))

# Plot circle trajectory
circle = plt.Circle((0, 0), 0.5, color='blue', fill=False, linewidth=3, label='Circular path')
ax.add_patch(circle)

# Draw axes
ax.axhline(0, color='black', linewidth=1)
ax.axvline(0, color='black', linewidth=1)
ax.grid(True, alpha=0.3)

# Draw position vectors at specific times
colors = plt.cm.rainbow(np.linspace(0, 1, len(t_points)))
for i in range(len(t_points)):
 ax.quiver(0, 0, x_points[i], y_points[i], angles='xy', scale_units='xy', scale=1,
 color=colors[i], width=0.01, headwidth=5, alpha=0.7)
 ax.plot(x_points[i], y_points[i], 'o', markersize=10, color=colors[i])
 
 # Add time labels
 label_x = x_points[i] * 1.4
 label_y = y_points[i] * 1.4
 ax.text(label_x, label_y, f't={t_points[i]:.2f}', 
 ha='center', va='center', fontsize=9, fontweight='bold')

# Mark origin
ax.plot(0, 0, 'ko', markersize=8)
ax.text(0.05, -0.08, 'Origin', fontsize=12, fontweight='bold')

# Labels and formatting
ax.set_xlabel('x (meters)', fontsize=14, fontweight='bold')
ax.set_ylabel('y (meters)', fontsize=14, fontweight='bold')
ax.set_title('Circular Motion of Robot Arm\n$\\vec{r}(t) = 0.5\\cos(t)\\hat{i} + 0.5\\sin(t)\\hat{j}$ m\nRadius = 0.5 m (constant)', 
 fontsize=14, fontweight='bold')
ax.set_xlim(-0.8, 0.8)
ax.set_ylim(-0.8, 0.8)
ax.set_aspect('equal')

plt.tight_layout()
plt.show()

print("\nKey Observations:")
print("- Trajectory: Perfect circle with radius 0.5 m")
print("- All position vectors have the SAME magnitude (0.5 m)")
print("- Direction changes continuously as arm rotates")
print("- Period: 2π seconds (approx 6.28 seconds) for one complete rotation")
print("- This is UNIFORM CIRCULAR MOTION")
````

### Important Points

1. **Circular motion** is common in robotics and machines
2. Position vector **magnitude stays constant** (always 0.5 m from origin)
3. But the **direction** changes continuously
4. In **Lecture 3**, we'll see that **polar coordinates** make circular motion much simpler
5. This is a preview of why we need different coordinate systems
