# PART 1: From Position to Velocity

---


## The Kinematic Chain

**Building on Lecture 1:** 
In Lecture 1, we studied **position** $\vec{r}(t)$ - where an object is located in space. We analyzed projectile motion and saw how position changes with time.

**Today's Goal:** 
But objects don't just exist at one location - they **move**! How fast? In what direction? Is the speed changing? To answer these questions, we need three related concepts:

```
POSITION (r) → VELOCITY (v) → ACCELERATION (a)
 ↓ ↓ ↓
 WHERE? HOW FAST? HOW FAST IS
 SPEED CHANGING?
```

Each concept is related by **differentiation**:

$$\vec{r}(t) \xrightarrow{\frac{d}{dt}} \vec{v}(t) \xrightarrow{\frac{d}{dt}} \vec{a}(t)$$

---


## What is Velocity?

**Velocity** is the **rate of change of position** with respect to time.

### Mathematical Definition:

$$\vec{v}(t) = \frac{d\vec{r}}{dt} = \lim_{\Delta t \to 0} \frac{\vec{r}(t + \Delta t) - \vec{r}(t)}{\Delta t}$$

### Component Form:

If $\vec{r}(t) = x(t)\hat{i} + y(t)\hat{j} + z(t)\hat{k}$, then:

$$\vec{v}(t) = \frac{dx}{dt}\hat{i} + \frac{dy}{dt}\hat{j} + \frac{dz}{dt}\hat{k} = v_x\hat{i} + v_y\hat{j} + v_z\hat{k}$$

### Physical Interpretation:

- **Magnitude**: $|\vec{v}| = $ speed (how fast)
- **Direction**: Direction of motion (where heading)
- **Units**: m/s, km/h, ft/s, etc.

---


## Example 1: Velocity from Position (1D Motion)

**Problem Statement:** 
A particle moves along the x-axis with position given by:

$$x(t) = 2 + 3t + 5t^2 \quad \text{meters}$$

where $t$ is time in seconds.

**Questions:** 
A. Find the velocity function $v_x(t)$ 
B. What is the velocity at $t = 0, 1, 2$ seconds? 
C. When (if ever) is the particle at rest?

---


### Step 1⃣: Given & Find

** GIVEN:**
- Position function: $x(t) = 2 + 3t + 5t^2$ m
- Time range of interest: $t = 0, 1, 2$ s

** FIND:**
- Part A: Velocity function $v_x(t)$
- Part B: Velocity at specific times
- Part C: Time when particle is at rest ($v = 0$)

---


### Step 2⃣: Concept Review (Differentiation Rules)

**Power Rule:**
$$\frac{d}{dt}[t^n] = nt^{n-1}$$

**Constant Rule:**
$$\frac{d}{dt}[c] = 0 \quad \text{(constant disappears)}$$

**Constant Multiple Rule:**
$$\frac{d}{dt}[ct^n] = cnt^{n-1}$$

**Sum Rule:**
$$\frac{d}{dt}[f + g] = \frac{df}{dt} + \frac{dg}{dt}$$

---


### Step 3⃣: Manual Solution (Step-by-Step Differentiation)

**Part A: Find Velocity Function**

Given: $x(t) = 2 + 3t + 5t^2$

Velocity is the derivative:
$$v_x(t) = \frac{dx}{dt}$$

Apply differentiation term-by-term:

$$v_x(t) = \frac{d}{dt}[2] + \frac{d}{dt}[3t] + \frac{d}{dt}[5t^2]$$

$$v_x(t) = 0 + 3(1) + 5(2t^{2-1})$$

$$v_x(t) = 3 + 10t \quad \text{m/s}$$

$$\boxed{v_x(t) = 3 + 10t \text{ m/s}}$$

---

**Part B: Velocity at Specific Times**

At $t = 0$ s:
$$v_x(0) = 3 + 10(0) = 3 \text{ m/s}$$

At $t = 1$ s:
$$v_x(1) = 3 + 10(1) = 13 \text{ m/s}$$

At $t = 2$ s:
$$v_x(2) = 3 + 10(2) = 23 \text{ m/s}$$

$$\boxed{\begin{array}{ll}
v_x(0) = 3 \text{ m/s} \\
v_x(1) = 13 \text{ m/s} \\
v_x(2) = 23 \text{ m/s}
\end{array}}$$

---

**Part C: When is Particle at Rest?**

Particle is at rest when $v_x = 0$:
$$3 + 10t = 0$$
$$10t = -3$$
$$t = -0.3 \text{ s}$$

Since $t = -0.3$ is **negative** (before our observation started), the particle is **never at rest** during $t \geq 0$.

$$\boxed{\text{Particle never at rest for } t \geq 0}$$

---


### Step 4⃣: Summary Table

| Time (s) | Position x(t) (m) | Velocity vₓ(t) (m/s) | Interpretation |
|----------|-------------------|---------------------|----------------|
| 0 | 2 | 3 | Starting 2m from origin, moving right at 3 m/s |
| 1 | 10 | 13 | At 10m, moving right at 13 m/s (faster!) |
| 2 | 26 | 23 | At 26m, moving right at 23 m/s (even faster!) |

**Observation**: Velocity is **increasing** (particle is accelerating) 

---


### Step 5⃣: Python Verification

Let's verify our manual calculations with Python:


````python
# Define position function
def x(t):
 return 2 + 3*t + 5*t**2

# Define velocity function (our manual result)
def vx(t):
 return 3 + 10*t

# Test at specific times
times = np.array([0, 1, 2])

print("=" * 60)
print("VERIFICATION: Position and Velocity")
print("=" * 60)
print(f"{'Time (s)':<10} {'Position (m)':<15} {'Velocity (m/s)':<15}")
print("-" * 60)
for t in times:
 print(f"{t:<10.1f} {x(t):<15.1f} {vx(t):<15.1f}")
print("=" * 60)

# Numerical derivative (to double-check our calculus)
dt = 0.0001 # Very small time step
print("\nNumerical verification (using derivative definition):")
for t in times:
 vx_numerical = (x(t + dt) - x(t)) / dt
 vx_analytical = vx(t)
 print(f" t={t}s: Analytical={vx_analytical:.2f} m/s, Numerical={vx_numerical:.2f} m/s ")
````

### Step 6⃣: Visualization

Let's visualize how position and velocity relate:


````python
# Time array
t = np.linspace(0, 2.5, 100)
x_vals = x(t)
vx_vals = vx(t)

# Create figure with two subplots
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 10))

# TOP: Position vs Time
ax1.plot(t, x_vals, 'b-', linewidth=2, label='x(t) = 2 + 3t + 5t²')
ax1.scatter([0, 1, 2], [x(0), x(1), x(2)], color='red', s=100, zorder=5, label='Sample points')
ax1.grid(True, alpha=0.3)
ax1.set_xlabel('Time (seconds)', fontsize=14, fontweight='bold')
ax1.set_ylabel('Position (meters)', fontsize=14, fontweight='bold')
ax1.set_title('Position vs Time (Parabolic)', fontsize=16, fontweight='bold')
ax1.legend(fontsize=12)

# BOTTOM: Velocity vs Time
ax2.plot(t, vx_vals, 'r-', linewidth=2, label='vₓ(t) = 3 + 10t')
ax2.scatter([0, 1, 2], [vx(0), vx(1), vx(2)], color='darkred', s=100, zorder=5, label='Sample points')
ax2.axhline(0, color='black', linewidth=1, linestyle='--', alpha=0.5)
ax2.grid(True, alpha=0.3)
ax2.set_xlabel('Time (seconds)', fontsize=14, fontweight='bold')
ax2.set_ylabel('Velocity (m/s)', fontsize=14, fontweight='bold')
ax2.set_title('Velocity vs Time (Linear)', fontsize=16, fontweight='bold')
ax2.legend(fontsize=12)

plt.tight_layout()
plt.show()

print("\n Key Observation:")
print(" Position is PARABOLIC (curved) → x(t) has t² term")
print(" Velocity is LINEAR (straight line) → v(t) has only t term")
print(" Differentiating reduces the power by 1: t² → t → constant")
````

### Step 7⃣: Physical Insights

**What do the numbers mean?**

1. **Initial velocity = 3 m/s**
 - Particle starts with speed 3 m/s (positive = moving right)
 - This comes from the "3t" term in position

2. **Velocity increases linearly**
 - Velocity grows by 10 m/s every second
 - This is because there's a "5t²" term in position
 - When you differentiate t², you get 2t, multiplied by 5 gives 10t

3. **Never at rest**
 - Since velocity starts at 3 m/s and only increases, it never hits zero
 - Particle always moves in positive direction (to the right)

**The calculus connection:**
- **Slope of position curve = velocity** at that instant
- Notice how position curve gets **steeper** over time
- This matches velocity **increasing** over time!

# PART 2: From Velocity to Acceleration

---


## What is Acceleration?

**Acceleration** is the **rate of change of velocity** with respect to time.

### Mathematical Definition:

$$\vec{a}(t) = \frac{d\vec{v}}{dt} = \frac{d^2\vec{r}}{dt^2}$$

(Acceleration is the **second derivative** of position!)

### Component Form:

$$\vec{a}(t) = \frac{dv_x}{dt}\hat{i} + \frac{dv_y}{dt}\hat{j} + \frac{dv_z}{dt}\hat{k} = a_x\hat{i} + a_y\hat{j} + a_z\hat{k}$$

### Physical Interpretation:

- **Magnitude**: How quickly speed is changing
- **Direction**: Direction of velocity change (not necessarily direction of motion!)
- **Units**: m/s², ft/s², etc.
- **Special cases**:
 - $a = 0$: Constant velocity (no speed change)
 - $a > 0$ (same direction as $v$): Speeding up
 - $a < 0$ (opposite to $v$): Slowing down

---


## Example 2: Complete Kinematic Analysis (Projectile Motion)

**Connection to L1:** 
Remember in Lecture 1 Example 3, we analyzed projectile position $x(t) = 10t$, $y(t) = 15t - 4.9t^2$. We STARTED with position equations.

**Now in L2:** We'll see HOW to DERIVE velocity and acceleration FROM position using calculus!

---

**Problem Statement:** 
A ball is thrown from ground level with initial velocity components:
- Horizontal: $v_{x0} = 12$ m/s (constant)
- Vertical: $v_{y0} = 16$ m/s (upward)

Gravity acts downward with $g = 9.8$ m/s².

The position components are:
$$x(t) = 12t \quad \text{m}$$
$$y(t) = 16t - 4.9t^2 \quad \text{m}$$

**Questions:** 
A. Find velocity functions $v_x(t)$ and $v_y(t)$ 
B. Find acceleration functions $a_x(t)$ and $a_y(t)$ 
C. What is the velocity vector at $t = 1$ s? 
D. What is the speed at $t = 1$ s? 
E. Verify that acceleration equals gravity

---


### Step 1⃣: Given & Find

** GIVEN:**
- Position: $x(t) = 12t$ m, $y(t) = 16t - 4.9t^2$ m
- Initial velocities: $v_{x0} = 12$ m/s, $v_{y0} = 16$ m/s
- Gravity: $g = 9.8$ m/s²
- Time of interest: $t = 1$ s

** FIND:**
- Part A: $v_x(t)$, $v_y(t)$
- Part B: $a_x(t)$, $a_y(t)$
- Part C: $\vec{v}(1)$
- Part D: $|\vec{v}(1)|$ (speed)
- Part E: Verify $\vec{a} = -g\hat{j}$

---


### Step 2⃣: Concept Review (Derivative Rules)

**The kinematic chain:**
$$\vec{r}(t) \xrightarrow{\frac{d}{dt}} \vec{v}(t) \xrightarrow{\frac{d}{dt}} \vec{a}(t)$$

Or equivalently:
$$\vec{r}(t) \xrightarrow{\frac{d^2}{dt^2}} \vec{a}(t)$$

**Differentiation rules:**
- $\frac{d}{dt}[ct] = c$ (linear term → constant)
- $\frac{d}{dt}[ct^2] = 2ct$ (quadratic → linear)
- $\frac{d}{dt}[c] = 0$ (constant → zero)

---


### Step 3⃣: Manual Solution (Complete Differentiation)

**Part A: Find Velocity Functions**

Horizontal velocity:
$$v_x(t) = \frac{dx}{dt} = \frac{d}{dt}[12t] = 12 \text{ m/s}$$

Vertical velocity:
$$v_y(t) = \frac{dy}{dt} = \frac{d}{dt}[16t - 4.9t^2] = 16 - 9.8t \text{ m/s}$$

$$\boxed{\begin{array}{l}
v_x(t) = 12 \text{ m/s (constant)} \\
v_y(t) = 16 - 9.8t \text{ m/s}
\end{array}}$$

---

**Part B: Find Acceleration Functions**

Horizontal acceleration:
$$a_x(t) = \frac{dv_x}{dt} = \frac{d}{dt}[12] = 0 \text{ m/s}^2$$

Vertical acceleration:
$$a_y(t) = \frac{dv_y}{dt} = \frac{d}{dt}[16 - 9.8t] = -9.8 \text{ m/s}^2$$

$$\boxed{\begin{array}{l}
a_x(t) = 0 \text{ m/s}^2 \\
a_y(t) = -9.8 \text{ m/s}^2 = -g
\end{array}}$$

---

**Part C: Velocity at t = 1 s**

$$v_x(1) = 12 \text{ m/s}$$
$$v_y(1) = 16 - 9.8(1) = 6.2 \text{ m/s}$$

$$\vec{v}(1) = 12\hat{i} + 6.2\hat{j} \text{ m/s}$$

$$\boxed{\vec{v}(1) = 12\hat{i} + 6.2\hat{j} \text{ m/s}}$$

---

**Part D: Speed at t = 1 s**

Speed is the magnitude of velocity:
$$|\vec{v}(1)| = \sqrt{v_x^2 + v_y^2} = \sqrt{12^2 + 6.2^2}$$
$$= \sqrt{144 + 38.44} = \sqrt{182.44} = 13.51 \text{ m/s}$$

$$\boxed{\text{Speed} = 13.51 \text{ m/s}}$$

---

**Part E: Verify Acceleration = Gravity**

Acceleration vector:
$$\vec{a}(t) = 0\hat{i} + (-9.8)\hat{j} = -9.8\hat{j} \text{ m/s}^2$$

Gravity vector:
$$\vec{g} = -9.8\hat{j} \text{ m/s}^2$$

$$\vec{a} = \vec{g} \quad $$

**Verified!** The only force acting is gravity (no air resistance).

---


### Step 4⃣: Summary Table

| Time (s) | Position | Velocity | Speed | Acceleration |
|----------|----------|----------|-------|-------------|
| 0 | (0, 0) | (12, 16) m/s | 20 m/s | (0, -9.8) m/s² |
| 1 | (12, 11.1) m | (12, 6.2) m/s | 13.5 m/s | (0, -9.8) m/s² |
| 1.63 | (19.6, 13.1) m | (12, 0) m/s | 12 m/s | (0, -9.8) m/s² |
| 2 | (24, 12.4) m | (12, -3.6) m/s | 12.5 m/s | (0, -9.8) m/s² |

**Key observations:**
- Horizontal velocity **constant** (no horizontal forces)
- Vertical velocity **decreases** (gravity pulls down)
- At peak (t≈1.63s), vertical velocity = 0
- Acceleration **constant throughout** (gravity never changes)

---


### Step 5⃣: Python Implementation


````python
# Position functions
def pos_x(t):
 return 12 * t

def pos_y(t):
 return 16 * t - 4.9 * t**2

# Velocity functions
def vel_x(t):
 return np.full_like(np.asarray(t, dtype=float), 12.0) # Constant

def vel_y(t):
 return 16 - 9.8 * t

# Acceleration functions
def acc_x(t):
 return np.zeros_like(np.asarray(t, dtype=float)) # No horizontal acceleration

def acc_y(t):
 return np.full_like(np.asarray(t, dtype=float), -9.8) # Gravity

# Speed (magnitude of velocity)
def speed(t):
 vx = vel_x(t)
 vy = vel_y(t)
 return np.sqrt(vx**2 + vy**2)

# Test at t = 1 s
t = 1.0

print("=" * 70)
print("KINEMATIC ANALYSIS AT t = 1.0 s")
print("=" * 70)
print(f"Position: r({t}) = ({pos_x(t):.1f}, {pos_y(t):.1f}) m")
print(f"Velocity: v({t}) = ({vel_x(t):.1f}, {vel_y(t):.1f}) m/s")
print(f"Speed: |v({t})| = {speed(t):.2f} m/s")
print(f"Acceleration: a({t}) = ({acc_x(t):.1f}, {acc_y(t):.1f}) m/s²")
print("=" * 70)

# Verify using numerical derivatives
dt = 0.0001
print("\nNumerical verification:")
vx_num = (pos_x(t + dt) - pos_x(t)) / dt
vy_num = (pos_y(t + dt) - pos_y(t)) / dt
print(f" vx: Analytical = {vel_x(t):.2f}, Numerical = {vx_num:.2f} ")
print(f" vy: Analytical = {vel_y(t):.2f}, Numerical = {vy_num:.2f} ")

ax_num = (vel_x(t + dt) - vel_x(t)) / dt
ay_num = (vel_y(t + dt) - vel_y(t)) / dt
print(f" ax: Analytical = {acc_x(t):.2f}, Numerical = {ax_num:.2f} ")
print(f" ay: Analytical = {acc_y(t):.2f}, Numerical = {ay_num:.2f} ")
````

### Step 6⃣: Comprehensive Visualization


````python
# Time array (until ball hits ground)
t_flight = 2 * 16 / 9.8 # Total time in air
t_array = np.linspace(0, t_flight, 200)

# Calculate all quantities
x_vals = pos_x(t_array)
y_vals = pos_y(t_array)
vx_vals = vel_x(t_array)
vy_vals = vel_y(t_array)
speed_vals = speed(t_array)
ax_vals = acc_x(t_array)
ay_vals = acc_y(t_array)

# Create comprehensive plot
fig = plt.figure(figsize=(16, 12))

# 1. Trajectory (position)
ax1 = plt.subplot(3, 2, 1)
ax1.plot(x_vals, y_vals, 'b-', linewidth=2)
ax1.scatter([pos_x(1)], [pos_y(1)], color='red', s=100, zorder=5)
ax1.set_xlabel('x (m)', fontweight='bold')
ax1.set_ylabel('y (m)', fontweight='bold')
ax1.set_title('Trajectory (Position)', fontsize=14, fontweight='bold')
ax1.grid(True, alpha=0.3)
ax1.axhline(0, color='black', linewidth=1)

# 2. Position components vs time
ax2 = plt.subplot(3, 2, 2)
ax2.plot(t_array, x_vals, 'b-', linewidth=2, label='x(t) = 12t')
ax2.plot(t_array, y_vals, 'r-', linewidth=2, label='y(t) = 16t - 4.9t²')
ax2.scatter([1, 1], [pos_x(1), pos_y(1)], color='darkred', s=60, zorder=5)
ax2.set_xlabel('Time (s)', fontweight='bold')
ax2.set_ylabel('Position (m)', fontweight='bold')
ax2.set_title('Position vs Time', fontsize=14, fontweight='bold')
ax2.legend()
ax2.grid(True, alpha=0.3)

# 3. Velocity components vs time
ax3 = plt.subplot(3, 2, 3)
ax3.plot(t_array, vx_vals, 'b-', linewidth=2, label='vx(t) = 12')
ax3.plot(t_array, vy_vals, 'r-', linewidth=2, label='vy(t) = 16 - 9.8t')
ax3.scatter([1, 1], [vel_x(1), vel_y(1)], color='darkred', s=60, zorder=5)
ax3.axhline(0, color='black', linewidth=1, linestyle='--')
ax3.set_xlabel('Time (s)', fontweight='bold')
ax3.set_ylabel('Velocity (m/s)', fontweight='bold')
ax3.set_title('Velocity vs Time', fontsize=14, fontweight='bold')
ax3.legend()
ax3.grid(True, alpha=0.3)

# 4. Speed vs time
ax4 = plt.subplot(3, 2, 4)
ax4.plot(t_array, speed_vals, 'purple', linewidth=2, label='|v(t)|')
ax4.scatter([1], [speed(1)], color='darkred', s=100, zorder=5)
ax4.set_xlabel('Time (s)', fontweight='bold')
ax4.set_ylabel('Speed (m/s)', fontweight='bold')
ax4.set_title('Speed vs Time', fontsize=14, fontweight='bold')
ax4.legend()
ax4.grid(True, alpha=0.3)

# 5. Acceleration components vs time
ax5 = plt.subplot(3, 2, 5)
ax5.plot(t_array, ax_vals, 'b-', linewidth=2, label='ax(t) = 0')
ax5.plot(t_array, ay_vals, 'r-', linewidth=2, label='ay(t) = -9.8')
ax5.set_xlabel('Time (s)', fontweight='bold')
ax5.set_ylabel('Acceleration (m/s²)', fontweight='bold')
ax5.set_title('Acceleration vs Time (Constant!)', fontsize=14, fontweight='bold')
ax5.legend()
ax5.grid(True, alpha=0.3)
ax5.set_ylim(-12, 2)

# 6. Summary text
ax6 = plt.subplot(3, 2, 6)
ax6.axis('off')
summary_text = f"""
KINEMATIC SUMMARY AT t = 1.0 s

Position: r = ({pos_x(1):.1f}, {pos_y(1):.1f}) m
Velocity: v = ({vel_x(1):.1f}, {vel_y(1):.1f}) m/s
Speed: |v| = {speed(1):.2f} m/s
Acceleration: a = (0, -9.8) m/s²

KEY RELATIONSHIPS:

    v = dr/dt (velocity = slope of position)
a = dv/dt (acceleration = slope of velocity)
a = d²r/dt² (acceleration = curvature of position)

PHYSICAL INSIGHTS:

 Horizontal: No forces → constant velocity
 Vertical: Gravity → constant acceleration
 Speed minimum at peak (vy = 0)
 Acceleration ALWAYS points down (-g)
"""
ax6.text(0.1, 0.5, summary_text, fontsize=11, family='monospace',
 verticalalignment='center')

plt.tight_layout()
plt.show()
````

### Step 7⃣: Discussion & Exam Tips

**Physical Insights:**

1. **Derivative relationships are fundamental:**
 - Position → Velocity: Take first derivative
 - Velocity → Acceleration: Take first derivative again
 - Position → Acceleration: Take second derivative

2. **Independence of horizontal and vertical motion:**
 - Horizontal: $x = v_{x0}t$ (linear), $v_x = $ constant, $a_x = 0$
 - Vertical: $y = v_{y0}t - \frac{1}{2}gt^2$ (parabolic), $v_y = v_{y0} - gt$, $a_y = -g$

3. **Acceleration tells you about forces:**
 - $\vec{a} = -g\hat{j}$ means only gravity acts (Newton's 2nd law)
 - Zero horizontal acceleration → no horizontal forces
 - Constant acceleration → constant force

4. **Speed vs velocity:**
 - Velocity is a vector (has direction)
 - Speed is magnitude of velocity (always positive)
 - Speed can be minimum even when velocity is not zero!

---

** Exam Strategy:**

**For differentiation problems:**
1. Write given position function clearly
2. Identify which derivative you need (velocity or acceleration)
3. Apply power rule term-by-term
4. Simplify your answer
5. Check units: position (m) → velocity (m/s) → acceleration (m/s²)

**Common mistakes to avoid:**
- Forgetting to reduce the power: $\frac{d}{dt}[t^2] = 2t$, not $t^2$!
- Constants disappear when differentiated: $\frac{d}{dt}[5] = 0$
- Second derivative: Apply chain twice, don't skip steps
- Confusing speed with velocity (speed = $|\vec{v}|$, always positive)

**Quick verification checks:**
- Does velocity have one power less than position? 
- Does acceleration have one power less than velocity? 
- Are units consistent throughout? 
- Does acceleration equal -g for projectiles?

# Summary: The Kinematic Hierarchy

---


## Quick Reference

| Quantity | Symbol | Definition | Units | Physical Meaning |
|----------|--------|------------|-------|------------------|
| **Position** | $\vec{r}(t)$ | Location in space | m | Where? |
| **Velocity** | $\vec{v}(t) = \frac{d\vec{r}}{dt}$ | Rate of position change | m/s | How fast & which way? |
| **Speed** | $|\vec{v}(t)|$ | Magnitude of velocity | m/s | How fast? (no direction) |
| **Acceleration** | $\vec{a}(t) = \frac{d\vec{v}}{dt} = \frac{d^2\vec{r}}{dt^2}$ | Rate of velocity change | m/s² | How is speed changing? |

**Key Relationships:**
$$\boxed{\vec{r}(t) \xrightarrow{\text{differentiate}} \vec{v}(t) \xrightarrow{\text{differentiate}} \vec{a}(t)}$$

$$\boxed{\vec{a}(t) \xrightarrow{\text{integrate}} \vec{v}(t) \xrightarrow{\text{integrate}} \vec{r}(t)}$$
