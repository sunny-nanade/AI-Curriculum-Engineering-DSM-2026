## What is a Rigid Body?

### Definition

**Rigid Body**: An object where the distance between any two points remains **constant** during motion.

**Key Properties**:
1. Has **size** and **shape** (not a point mass)
2. Does **not deform** (no bending, stretching, compression)
3. Internal structure remains **fixed**

**Examples**:
- Steel gear, aluminum wheel, robot link
- Wooden rod, metal disk, ceramic plate
- Rubber ball (compresses), spring (stretches)
- Cloth, rope, liquid (deforms easily)

### Real vs Ideal

**Real Objects**: Always deform slightly under force
- Steel beam bends (very little)
- Car tire compresses at contact patch

**Rigid Body Approximation**: Valid when deformation is **negligible** compared to overall motion
- Good for: most metal parts, structural analysis, mechanism design
- Not good for: soft materials, crash analysis, vibrations

## Degrees of Freedom (DOF)

### Definition

**Degrees of Freedom (DOF)**: The minimum number of **independent coordinates** needed to completely describe the configuration of a system.

### Particle vs Rigid Body

#### Particle (Point Mass)
- **Position**: 3 coordinates (x, y, z)
- **DOF = 3** (in 3D space)
- Example: GPS coordinates of a car (treats car as point)

#### Rigid Body (Extended Object)
- **Position**: 3 coordinates (x, y, z) for center of mass
- **Orientation**: 3 angles (rotations about 3 axes)
- **DOF = 6** (in 3D space)
- Example: Airplane position + roll, pitch, yaw

### DOF in 2D (Planar Motion)

For motion in a plane:
- **Particle**: DOF = 2 (x, y)
- **Rigid Body**: DOF = 3 (x, y, θ)
  - (x, y): Position of center
  - θ: Rotation angle

**Example**: Wheel rolling on ground
- Position: x, y of wheel center
- Orientation: θ (rotation angle)
- Total: 3 DOF

### Constraints Reduce DOF

**Unconstrained**: Maximum DOF
- 3D rigid body: 6 DOF
- 2D rigid body: 3 DOF

**Constrained**: DOF reduced
- Wheel rolling without slipping: 2 DOF (v = ωR relates x and θ)
- Door on hinge: 1 DOF (only rotation angle)
- Fixed object: 0 DOF (cannot move)

---

## Rigid Body Motion = Translation + Rotation

### The Big Idea

Any rigid body motion can be decomposed into:
1. **Translation**: Motion of the center of mass
2. **Rotation**: Rotation about the center of mass

**Mathematically**:
$$\vec{v}_{\text{point}} = \vec{v}_{\text{center}} + \vec{\omega} \times \vec{r}$$

Where:
- $\vec{v}_{\text{point}}$: Velocity of any point on the rigid body
- $\vec{v}_{\text{center}}$: Velocity of center of mass (translation)
- $\vec{\omega}$: Angular velocity vector (rotation)
- $\vec{r}$: Position vector from center to point

### Pure Translation

**All points move with same velocity**

Example: Car sliding on ice (no rotation)
- Every point on car has same velocity
- No rotation: ω = 0
- v_point = v_center (constant for all points)

### Pure Rotation

**Object rotates about fixed point, no translation**

Example: Door opening (hinge is fixed)
- Hinge point: v = 0 (not moving)
- Other points: v = ω × r (velocity depends on distance from hinge)
- Center moves: v_center ≠ 0 (unless center is at hinge)

### General Motion = Both

**Translation + Rotation combined**

Example: Wheel rolling on ground
- Center translates: v_center (moves forward)
- Wheel rotates: ω (spins)
- Any point: v = v_center + ω × r

**Key Insight**: We can analyze translation and rotation **separately**, then combine!

## Angular Velocity Vector **ω**

### Definition

**Angular Velocity (ω)**: Vector describing the **rate** and **axis** of rotation.

**Properties**:
- **Magnitude**: |ω| = rate of rotation (rad/s)
- **Direction**: Axis of rotation (right-hand rule)
- **Units**: rad/s (radians per second)

### Right-Hand Rule

**How to find ω direction**:
1. Curl fingers of right hand in direction of rotation
2. Thumb points in direction of ω

**Example**: Wheel spinning counterclockwise (viewed from right)
- Rotation: CCW
- ω direction: Points to the right (away from you)

### 2D vs 3D

#### 2D (Planar Motion)
- Rotation always about axis perpendicular to plane
- ω = ω **k̂** (always in z-direction)
- Only magnitude ω matters (scalar)

#### 3D (General Motion)
- Rotation can be about any axis
- ω = ωₓ **î** + ωᵧ **ĵ** + ω_z **k̂** (vector)
- Both magnitude and direction matter

### Relation to Angular Displacement

**Angular displacement**: θ (rad)

**Angular velocity**: 
$$\omega = \frac{d\theta}{dt}$$

**Example**: 
- Wheel rotates at 5 revolutions per second
- ω = 5 rev/s × 2π rad/rev = 10π rad/s ≈ 31.4 rad/s

## Velocity of Points on a Rigid Body

### The Fundamental Formula

For any point P on a rigid body:

$$\boxed{\vec{v}_P = \vec{v}_{\text{center}} + \vec{\omega} \times \vec{r}}$$

Where:
- $\vec{v}_P$: Velocity of point P (what we want to find)
- $\vec{v}_{\text{center}}$: Velocity of center of mass (translation)
- $\vec{\omega}$: Angular velocity vector (rotation rate and axis)
- $\vec{r}$: Position vector **from center to P**

### Physical Interpretation

**Term 1: $\vec{v}_{\text{center}}$**
- Translation component
- Same for all points on the body
- "Carrying velocity" - body moving as a whole

**Term 2: $\vec{\omega} \times \vec{r}$**
- Rotation component
- Different for each point (depends on $\vec{r}$)
- "Spinning velocity" - due to rotation about center

**Total**: Every point has translation velocity PLUS rotation velocity

---

## Cross Product Review: **ω × r**

### Definition

The **cross product** (vector product) of two vectors:

$$\vec{a} \times \vec{b} = |\vec{a}| |\vec{b}| \sin\theta \, \hat{n}$$

Where:
- |a|, |b|: Magnitudes of vectors
- θ: Angle between vectors
- $\hat{n}$: Unit vector perpendicular to both (right-hand rule)

### Properties

1. **Perpendicular**: $\vec{a} \times \vec{b}$ is perpendicular to both $\vec{a}$ and $\vec{b}$
2. **Magnitude**: $|\vec{a} \times \vec{b}| = |\vec{a}| |\vec{b}| \sin\theta$
3. **Anti-commutative**: $\vec{a} \times \vec{b} = -(\vec{b} \times \vec{a})$
4. **Zero if parallel**: $\vec{a} \times \vec{a} = \vec{0}$

### Component Form (3D)

$$\vec{a} \times \vec{b} = \begin{vmatrix} \hat{i} & \hat{j} & \hat{k} \\ a_x & a_y & a_z \\ b_x & b_y & b_z \end{vmatrix}$$

$$= (a_y b_z - a_z b_y)\hat{i} + (a_z b_x - a_x b_z)\hat{j} + (a_x b_y - a_y b_x)\hat{k}$$

### For Rigid Body: **ω × r**

In 2D (planar motion):
- $\vec{\omega} = \omega \hat{k}$ (rotation about z-axis)
- $\vec{r} = x\hat{i} + y\hat{j}$ (position in xy-plane)
- $\vec{\omega} \times \vec{r} = \omega \hat{k} \times (x\hat{i} + y\hat{j})$
- $= \omega x (\hat{k} \times \hat{i}) + \omega y (\hat{k} \times \hat{j})$
- $= \omega x \hat{j} - \omega y \hat{i}$
- $= -\omega y \hat{i} + \omega x \hat{j}$

**Result**: $\vec{\omega} \times \vec{r} = (-\omega y, \omega x, 0)$

**Magnitude**: $|\vec{\omega} \times \vec{r}| = \omega \sqrt{x^2 + y^2} = \omega r$

**Direction**: Perpendicular to $\vec{r}$, in direction of rotation

---

## Derivation: Why v = v_center + ω × r?

### Starting Point

Consider two points on a rigid body:
- Point O: Center of mass (or any reference point)
- Point P: Any other point

**Position vectors** (in fixed frame):
- $\vec{r}_O$: Position of O
- $\vec{r}_P$: Position of P
- $\vec{r} = \vec{r}_P - \vec{r}_O$: **Relative position** (O to P)

### Key Constraint: Rigid Body

**Rigid body constraint**: Distance between O and P is **constant**

$$|\vec{r}| = \text{constant}$$

This means $\vec{r}$ can **rotate** but its magnitude cannot change.

### Taking Time Derivative

**Velocity of P** (in fixed frame):
$$\vec{v}_P = \frac{d\vec{r}_P}{dt}$$

**Using** $\vec{r}_P = \vec{r}_O + \vec{r}$:
$$\vec{v}_P = \frac{d\vec{r}_O}{dt} + \frac{d\vec{r}}{dt}$$

$$\vec{v}_P = \vec{v}_O + \frac{d\vec{r}}{dt}$$

### What is d**r**/dt?

Since $|\vec{r}|$ is constant, $\vec{r}$ can only **rotate** (not stretch).

**For a rotating vector**:
$$\frac{d\vec{r}}{dt} = \vec{\omega} \times \vec{r}$$

This is a fundamental result from kinematics of rotation!

**Proof** (in components): If $\vec{r} = r\cos\theta \hat{i} + r\sin\theta \hat{j}$ and $\theta = \omega t$:
$$\frac{d\vec{r}}{dt} = -r\omega\sin\theta \hat{i} + r\omega\cos\theta \hat{j} = \vec{\omega} \times \vec{r}$$

### Final Result

$$\boxed{\vec{v}_P = \vec{v}_O + \vec{\omega} \times \vec{r}}$$

Or, using center of mass as reference:

$$\boxed{\vec{v}_P = \vec{v}_{\text{center}} + \vec{\omega} \times \vec{r}}$$

**This is the fundamental velocity formula for rigid bodies!**

---

## Special Cases

### Case 1: Pure Translation (ω = 0)

**No rotation**:
$$\vec{v}_P = \vec{v}_{\text{center}} + \vec{0} \times \vec{r} = \vec{v}_{\text{center}}$$

**All points have same velocity** (sliding block, car on straight road without rotation)

### Case 2: Pure Rotation (v_center = 0)

**Center is fixed**:
$$\vec{v}_P = \vec{0} + \vec{\omega} \times \vec{r} = \vec{\omega} \times \vec{r}$$

**Velocity magnitude**: $|\vec{v}_P| = \omega r$ (farther from center → faster)

**Examples**: Door on hinge, spinning top (if center doesn't move)

### Case 3: Rolling Without Slipping

**Special constraint**: $v_{\text{center}} = \omega R$ (R = radius)

**At contact point** (bottom of wheel): $\vec{v}_{\text{contact}} = 0$ (stationary!)

This is the **instantaneous center of rotation** - the point that's momentarily at rest.

### Case 4: General Motion

**Both translation and rotation**:
$$\vec{v}_P = \vec{v}_{\text{center}} + \vec{\omega} \times \vec{r}$$

**Example**: Wheel rolling and slipping, airplane in flight

## Example 1: Rolling Wheel Without Slipping

### Problem Statement

A wheel of radius **R = 0.5 m** rolls without slipping on a horizontal surface.
The wheel's center moves with constant velocity **v_center = 2.0 m/s** to the right.

**Find**:
1. Angular velocity ω of the wheel
2. Velocity of the **top** point of the wheel
3. Velocity of the **bottom** point (contact point)
4. Velocity of a point on the **side** of the wheel

**Given Information**:
| Quantity | Symbol | Value | Units |
|----------|--------|-------|-------|
| Wheel radius | R | 0.5 | m |
| Center velocity | v_center | 2.0 | m/s (→) |
| Constraint | Rolling without slipping | v_center = ωR | - |

---

### STEP 1: Understand the Physics

**Physical Situation**:
- Wheel rolling on ground (like a car tire)
- **Translation**: Center moves forward at v_center
- **Rotation**: Wheel spins with angular velocity ω

**Key Constraint: Rolling Without Slipping**
- Contact point is **instantaneously at rest** (v_contact = 0)
- This gives us: **v_center = ω × R**

**Forces** (not needed for kinematics, but good to know):
- Weight (mg) downward
- Normal force (N) upward
- Friction (f) provides torque for rolling (static friction, no slipping!)

**What Makes This Interesting**:
- Different points on wheel have **different velocities**
- Top point moves faster than center
- Bottom point is momentarily at rest!

---

### STEP 2: Choose Coordinate System

**Fixed Frame** (ground frame):
- Origin: Contact point when t=0
- x-axis: Horizontal, direction of motion (→)
- y-axis: Vertical, upward (↑)
- z-axis: Out of page (⊙) [right-hand rule]

**Center of Wheel**:
- Position: (x_center, R, 0)
- Velocity: v_center **î** = 2.0 **î** m/s

**Angular Velocity**:
- Direction: About z-axis (perpendicular to plane of wheel)
- For rolling to the right: ω points **out of page** (+z direction)
- $\vec{\omega} = \omega \hat{k}$

**Points of Interest**:
- **Top**: Position relative to center: $\vec{r}_{top} = R\hat{j}$ (directly above)
- **Bottom**: Position relative to center: $\vec{r}_{bottom} = -R\hat{j}$ (directly below)
- **Side** (right): Position relative to center: $\vec{r}_{side} = R\hat{i}$ (to the right)

---

### STEP 3: Write the Equations

**Velocity Formula**:
$$\vec{v}_P = \vec{v}_{\text{center}} + \vec{\omega} \times \vec{r}$$

**Given**:
- $\vec{v}_{\text{center}} = v_c \hat{i} = 2.0 \hat{i}$ m/s
- $\vec{\omega} = \omega \hat{k}$ (to be found)
- $\vec{r}$: depends on which point

**For Rolling Without Slipping**:

At the **bottom** point (contact):
$$\vec{v}_{\text{bottom}} = \vec{0}$$

Using velocity formula:
$$\vec{0} = \vec{v}_{\text{center}} + \vec{\omega} \times \vec{r}_{\text{bottom}}$$

$$\vec{0} = v_c \hat{i} + (\omega \hat{k}) \times (-R\hat{j})$$

**Calculating cross product**:
$$\hat{k} \times \hat{j} = -\hat{i}$$

$$\vec{0} = v_c \hat{i} + \omega(-R)(-\hat{i}) = v_c \hat{i} + \omega R \hat{i}$$

**Setting equal to zero**:
$$v_c + \omega R = 0$$

**Solving for ω**:
$$\omega = \frac{v_c}{R}$$

---

### STEP 4: Solve Symbolically (using SymPy)

```
# Define symbolic variables
R_sym, vc_sym, omega_sym = sp.symbols('R v_c omega', real=True, positive=True)

# Velocity vectors (as matrices for cross product)
i, j, k = sp.symbols('i j k')

print("Given:")
print(f"v_center = v_c î")
print(f"ω = ω k̂")
print(f"R = radius")
print()

# Rolling without slipping constraint
print("Rolling without slipping constraint:")
print("At bottom: v_bottom = 0")
print("Using v = v_center + ω × r:")
print()

# Bottom point: r = -R ĵ
print("Bottom point: r_bottom = -R ĵ")
print("ω × r_bottom = (ω k̂) × (-R ĵ) = ω R (k̂ × ĵ) = -ω R (-î) = ω R î")
print()
print("v_bottom = v_c î + ω R î = 0")
print("Therefore: ω = v_c / R")
print()

# Now find velocities at other points
omega_formula = vc_sym / R_sym

print("Now we can find velocities at all points:")
print()

# Top point: r = R ĵ
print("Top point: r_top = R ĵ")
print("ω × r_top = (ω k̂) × (R ĵ) = ω R (k̂ × ĵ) = -ω R î")
print("v_top = v_c î + (-ω R î) = v_c î - ω R î")
print("Substituting ω = v_c/R:")
print("v_top = v_c î - (v_c/R)·R î = v_c î - v_c î... NO WAIT!")
print()
print("Let me recalculate the cross products correctly:")
print()

# Correct cross products
print("Cross product identities:")
print("î × ĵ = k̂")
print("ĵ × k̂ = î")
print("k̂ × î = ĵ")
print("k̂ × ĵ = -î (important!)")
print()

print("Bottom point: r = -R ĵ")
print("ω × r = (ω k̂) × (-R ĵ) = -ω R (k̂ × ĵ) = -ω R (-î) = ω R î")
print("v_bottom = v_c î + ω R î = (v_c + ω R) î = 0")
print("So: ω R = -v_c  →  Wait, this gives negative ω?")
print()

print("ISSUE: Sign convention!")
print("For wheel rolling RIGHT, it must spin CLOCKWISE (from side view)")
print("CLOCKWISE rotation → ω points INTO page (-k̂ direction)")
print()
print("Correcting: ω = -ω_mag k̂ (negative because into page)")
print("Then: ω_mag = v_c / R")
print()

# Store result
omega_result = vc_sym / R_sym
print(f"Angular speed: |ω| = {omega_result}")
```

### STEP 5: Calculate Numerical Values

## Example 2: Rotating Rod

### Problem Statement

A uniform rigid rod of length **L = 1.0 m** rotates about one end (point O) in a horizontal plane.
The rod rotates with constant angular velocity **ω = 3.0 rad/s** counterclockwise.

**Find**:
1. Velocity of the **free end** (point B, opposite to pivot)
2. Velocity of the **midpoint** (point M)
3. Velocity of a point **1/4 of the way** from the pivot (point Q)

**Given Information**:
| Quantity | Symbol | Value | Units |
|----------|--------|-------|-------|
| Rod length | L | 1.0 | m |
| Angular velocity | ω | 3.0 | rad/s (CCW) |
| Pivot at | O | Fixed | - |

---

### Solution

**Setup**:
- Pivot point O is **fixed**: v_O = 0
- This is **pure rotation** (no translation)
- Formula simplifies: **v = ω × r** (since v_center = 0)

**Coordinate System**:
- Origin at O (pivot point)
- Rod lies along x-axis initially
- z-axis out of page
- ω = 3.0 **k̂** rad/s (CCW → out of page)

### Point B (Free End)

**Position**: $\vec{r}_B = L\hat{i}$ (assuming rod along x-axis at t=0)

**But wait!** Rod is rotating, so direction changes. Let's use angle θ = ωt:
$$\vec{r}_B = L\cos\theta \, \hat{i} + L\sin\theta \, \hat{j}$$

**Velocity**:
$$\vec{v}_B = \vec{\omega} \times \vec{r}_B$$

**Magnitude**: 
$$|\vec{v}_B| = |\vec{\omega}| \cdot |\vec{r}_B| \cdot \sin(90°) = \omega L = 3.0 \times 1.0 = 3.0 \text{ m/s}$$

**Direction**: Perpendicular to rod, in direction of rotation (CCW)

### Point M (Midpoint)

**Position**: $\vec{r}_M = \frac{L}{2}(\cos\theta \, \hat{i} + \sin\theta \, \hat{j})$

**Magnitude**:
$$|\vec{v}_M| = \omega \cdot \frac{L}{2} = 3.0 \times 0.5 = 1.5 \text{ m/s}$$

**Direction**: Perpendicular to rod, CCW

### Point Q (1/4 from pivot)

**Position**: $\vec{r}_Q = \frac{L}{4}(\cos\theta \, \hat{i} + \sin\theta \, \hat{j})$

**Magnitude**:
$$|\vec{v}_Q| = \omega \cdot \frac{L}{4} = 3.0 \times 0.25 = 0.75 \text{ m/s}$$

**Direction**: Perpendicular to rod, CCW

---

```
# Example 2: Rotating Rod Visualization

# Given
L = 1.0  # m
omega = 3.0  # rad/s (CCW)

# Calculate velocities
v_B = omega * L  # Free end
v_M = omega * L/2  # Midpoint
v_Q = omega * L/4  # Quarter point

print("Example 2: Rotating Rod")
print("="*50)
print(f"Rod length: L = {L} m")
print(f"Angular velocity: ω = {omega} rad/s (CCW)")
print()
print("Velocities:")
print(f"Point B (free end): v_B = ω·L = {omega}×{L} = {v_B:.2f} m/s")
print(f"Point M (midpoint): v_M = ω·(L/2) = {omega}×{L/2} = {v_M:.2f} m/s")
print(f"Point Q (L/4): v_Q = ω·(L/4) = {omega}×{L/4} = {v_Q:.2f} m/s")
print("="*50)
print()

# Create visualization
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Left plot: Rod at one instant with velocity vectors
theta = np.pi/4  # 45 degrees for visualization
ax1.set_xlim(-0.3, 1.3)
ax1.set_ylim(-0.3, 1.3)
ax1.set_aspect('equal')
ax1.grid(True, alpha=0.3)
ax1.set_title('Rotating Rod: Velocity Vectors', fontsize=14, fontweight='bold')

# Rod
x_B = L * np.cos(theta)
y_B = L * np.sin(theta)
ax1.plot([0, x_B], [0, y_B], 'k-', linewidth=6, label='Rod')

# Points
ax1.plot(0, 0, 'ko', markersize=15, label='O (pivot)')
ax1.plot(x_B, y_B, 'ro', markersize=12, label='B (free end)')
ax1.plot(x_B/2, y_B/2, 'go', markersize=12, label='M (midpoint)')
ax1.plot(x_B/4, y_B/4, 'bo', markersize=12, label='Q (L/4)')

# Velocity vectors (perpendicular to rod)
# Direction perpendicular to rod (CCW rotation)
perp_x = -np.sin(theta)
perp_y = np.cos(theta)

scale = 0.2  # Scale for visualization
ax1.arrow(x_B, y_B, v_B*scale*perp_x, v_B*scale*perp_y,
         head_width=0.05, head_length=0.03,
         fc='red', ec='red', linewidth=2)
ax1.arrow(x_B/2, y_B/2, v_M*scale*perp_x, v_M*scale*perp_y,
         head_width=0.05, head_length=0.03,
         fc='green', ec='green', linewidth=2)
ax1.arrow(x_B/4, y_B/4, v_Q*scale*perp_x, v_Q*scale*perp_y,
         head_width=0.05, head_length=0.03,
         fc='blue', ec='blue', linewidth=2)

# Angular velocity indicator
ax1.plot(0, 0, 'mo', markersize=20, alpha=0.3)
ax1.text(0.05, 0.05, 'ω', fontsize=18, color='magenta', fontweight='bold')

ax1.set_xlabel('x (m)', fontsize=12)
ax1.set_ylabel('y (m)', fontsize=12)
ax1.legend(loc='upper right', fontsize=10)

# Right plot: Velocity vs distance from pivot
r_points = np.linspace(0, L, 100)
v_points = omega * r_points

ax2.plot(r_points, v_points, 'b-', linewidth=3)
ax2.plot([0, L/4, L/2, L], [0, v_Q, v_M, v_B], 'ro', markersize=10)

ax2.grid(True, alpha=0.3)
ax2.set_xlabel('Distance from Pivot (m)', fontsize=12)
ax2.set_ylabel('Velocity (m/s)', fontsize=12)
ax2.set_title('Velocity vs Distance: v = ω·r', fontsize=14, fontweight='bold')

# Annotate points
ax2.text(0, 0, 'O\n(pivot)', ha='center', va='bottom', fontsize=10)
ax2.text(L/4, v_Q, f'  Q\n  v={v_Q:.2f}', ha='left', va='bottom', fontsize=10)
ax2.text(L/2, v_M, f'  M\n  v={v_M:.2f}', ha='left', va='bottom', fontsize=10)
ax2.text(L, v_B, f'  B\n  v={v_B:.2f}', ha='left', va='bottom', fontsize=10)

plt.tight_layout()
plt.show()

print("Visualization complete!")
print()
print("KEY OBSERVATIONS:")
print("1. Velocity is PROPORTIONAL to distance from pivot: v = ω·r")
print("2. Direction is PERPENDICULAR to rod")
print("3. All points have same angular velocity ω")
print("4. Farther from pivot → higher linear velocity")
```

### Key Takeaways from Example 2

1. **Pure rotation** (pivot fixed):
   - v_pivot = 0 (fixed point)
   - Formula simplifies: v = ω × r

2. **Velocity proportional to distance**:
   - v = ω · r (linear relationship)
   - Farther from pivot → faster motion

3. **Direction always perpendicular**:
   - Cross product ensures v ⊥ r
   - Direction determined by rotation direction (right-hand rule)

4. **Comparison with rolling wheel**:
   - Rolling: v = v_center + ω × r (both translation and rotation)
   - Rotating rod: v = ω × r (only rotation, no translation)

## Lecture Summary

### Key Concepts Learned

1. **Rigid Body Definition**
   - Object with constant internal distances
   - Has size and shape (not a point mass)
   - DOF: 3 (in 2D), 6 (in 3D)

2. **Rigid Body Motion = Translation + Rotation**
   - Translation: All points move together (v_center)
   - Rotation: Points move around center (ω × r)
   - General motion combines both

3. **Angular Velocity ω**
   - Vector describing rotation rate and axis
   - Magnitude: |ω| = rotation rate (rad/s)
   - Direction: Right-hand rule (axis of rotation)

4. **Fundamental Velocity Formula**
   $$\boxed{\vec{v}_P = \vec{v}_{\text{center}} + \vec{\omega} \times \vec{r}}$$
   - Works for ANY point on rigid body
   - r is position FROM center TO point P
   - Cross product gives rotation contribution

5. **Special Cases**
   - Pure translation: ω = 0 → v_P = v_center
   - Pure rotation: v_center = 0 → v_P = ω × r
   - Rolling without slipping: v_center = ωR

### Examples Covered

- **Example 1**: Rolling wheel (translation + rotation)
  - Top moves at 2×v_center
  - Bottom at rest (instantaneous center)
  
- **Example 2**: Rotating rod (pure rotation)
  - v = ω·r (linear relationship)
  - Velocity ⊥ to rod
