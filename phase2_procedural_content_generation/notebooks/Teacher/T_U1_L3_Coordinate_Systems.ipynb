## 3. Polar Coordinates (2D)

### Definition

Position in 2D plane using distance and angle:

$$\vec{r} = r\hat{e}_r$$

where:
- $r$ = distance from origin (radial coordinate)
- $\theta$ = angle from x-axis (angular coordinate, in radians)
- $\hat{e}_r$ = unit vector pointing radially outward
- $\hat{e}_\theta$ = unit vector perpendicular to $\hat{e}_r$ (tangential direction)

### Conversion Formulas

**Cartesian → Polar:**
$$r = \sqrt{x^2 + y^2}$$
$$\theta = \tan^{-1}\left(\frac{y}{x}\right) \quad \text{(use atan2 for all quadrants)}$$

**Polar → Cartesian:**
$$x = r\cos(\theta)$$
$$y = r\sin(\theta)$$

### Unit Vectors

$$\hat{e}_r = \cos(\theta)\hat{i} + \sin(\theta)\hat{j}$$
$$\hat{e}_\theta = -\sin(\theta)\hat{i} + \cos(\theta)\hat{j}$$

**Important**: $\hat{e}_r$ and $\hat{e}_\theta$ **change direction** as $\theta$ changes!

### When to Use Polar

- Circular motion (constant $r$)
- Central force problems
- Rotating disks, turntables
- Any 2D motion with obvious center

---

### Example 1: Polar Coordinates

A particle moves in a circle of radius 5 m at angle $\theta = 60°$ from the x-axis.

---

**GIVEN:**
- Polar coordinates: $r = 5$ m, $\theta = 60° = \frac{\pi}{3}$ rad

**FIND:**
- Cartesian coordinates $(x, y)$
- Verify magnitude

---

### Concept Review

Polar → Cartesian conversion:
$$x = r\cos(\theta), \quad y = r\sin(\theta)$$

---

### Manual Solution

**Step 1**: Convert angle to radians
$$\theta = 60° = 60 \times \frac{\pi}{180} = \frac{\pi}{3} \text{ rad}$$

**Step 2**: Calculate x-coordinate
$$x = r\cos(\theta) = 5\cos\left(\frac{\pi}{3}\right)$$

$$\cos\left(\frac{\pi}{3}\right) = \cos(60°) = 0.5$$

$$x = 5 \times 0.5 = 2.5 \text{ m}$$

**Step 3**: Calculate y-coordinate
$$y = r\sin(\theta) = 5\sin\left(\frac{\pi}{3}\right)$$

$$\sin\left(\frac{\pi}{3}\right) = \sin(60°) = \frac{\sqrt{3}}{2} \approx 0.866$$

$$y = 5 \times 0.866 = 4.33 \text{ m}$$

**Step 4**: Verify magnitude
$$|\vec{r}| = \sqrt{x^2 + y^2} = \sqrt{(2.5)^2 + (4.33)^2}$$
$$= \sqrt{6.25 + 18.75} = \sqrt{25} = 5 \text{ m} \quad $$

---

### Answer

$$\boxed{\begin{array}{l}
x = 2.5 \text{ m} \\
y = 4.33 \text{ m} \\
|\vec{r}| = 5 \text{ m (verified)}
\end{array}}$$

---

```python
# Example 1 - Python Verification
import numpy as np
import matplotlib.pyplot as plt

# Given
r = 5  # m
theta_deg = 60  # degrees
theta_rad = np.deg2rad(theta_deg)

# Convert to Cartesian
x = r * np.cos(theta_rad)
y = r * np.sin(theta_rad)

# Verify magnitude
magnitude = np.sqrt(x**2 + y**2)

print("Example 1: Polar → Cartesian Conversion")
print("="*50)
print(f"Given: r = {r} m, θ = {theta_deg}° = {theta_rad:.4f} rad")
print(f"\nCartesian coordinates:")
print(f" x = r cos(θ) = {r} × {np.cos(theta_rad):.3f} = {x:.2f} m")
print(f" y = r sin(θ) = {r} × {np.sin(theta_rad):.3f} = {y:.2f} m")
print(f"\nVerification:")
print(f" |r| = √(x² + y²) = √({x:.2f}² + {y:.2f}²) = {magnitude:.2f} m ")

# Visualize
fig, ax = plt.subplots(figsize=(8, 8))

# Draw circle
circle_theta = np.linspace(0, 2*np.pi, 100)
circle_x = r * np.cos(circle_theta)
circle_y = r * np.sin(circle_theta)
ax.plot(circle_x, circle_y, 'b--', alpha=0.3, label=f'Circle (r={r}m)')

# Draw axes
ax.axhline(0, color='k', linewidth=0.5)
ax.axvline(0, color='k', linewidth=0.5)

# Draw position vector
ax.quiver(0, 0, x, y, angles='xy', scale_units='xy', scale=1,
          color='red', width=0.01, label=f'r={r}m, θ={theta_deg}°')

# Draw angle arc
arc_theta = np.linspace(0, theta_rad, 30)
arc_r = 1.5
ax.plot(
    arc_r *
    np.cos(arc_theta),
    arc_r *
    np.sin(arc_theta),
    'g-',
    linewidth=2)
ax.text(2, 1, f'θ = {theta_deg}°', fontsize=12, color='green')

# Mark point
ax.scatter([x], [y], color='red', s=100, zorder=5)
ax.text(x+0.3, y+0.3, f'({x:.2f}, {y:.2f})', fontsize=11)

ax.set_xlabel('x (m)', fontsize=12)
ax.set_ylabel('y (m)', fontsize=12)
ax.set_title('Polar to Cartesian Conversion', fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend()
ax.set_aspect('equal')
ax.set_xlim([-6, 6])
ax.set_ylim([-6, 6])
plt.tight_layout()
plt.show()

print("\n Conversion verified!")
```

## 4. Cylindrical Coordinates (3D)

### Definition

Extension of polar coordinates to 3D by adding vertical component:

$$\vec{r} = \rho\hat{e}_\rho + z\hat{k}$$

where:
- $\rho$ = radial distance from z-axis (like $r$ in polar)
- $\phi$ = angle in xy-plane from x-axis
- $z$ = height (same as Cartesian z)

### Conversion Formulas

**Cartesian → Cylindrical:**
$$\rho = \sqrt{x^2 + y^2}$$
$$\phi = \tan^{-1}\left(\frac{y}{x}\right)$$
$$z = z$$

**Cylindrical → Cartesian:**
$$x = \rho\cos(\phi)$$
$$y = \rho\sin(\phi)$$
$$z = z$$

### When to Use Cylindrical

- Helical motion (spiral staircase)
- Rotating machinery with vertical motion
- Cylindrical objects (pipes, columns)
- Any problem with cylindrical symmetry

---

### Example 2: Cylindrical Coordinates

A particle moves on a helix with radius $\rho = 3$ m at angle $\phi = 45°$ and height $z = 4$ m.

---

**GIVEN:**
- Cylindrical coordinates: $\rho = 3$ m, $\phi = 45°$, $z = 4$ m

**FIND:**
- Cartesian coordinates $(x, y, z)$
- Distance from origin

---

### Manual Solution

**Step 1**: Convert to Cartesian
$$x = \rho\cos(\phi) = 3\cos(45°) = 3 \times 0.707 = 2.12 \text{ m}$$
$$y = \rho\sin(\phi) = 3\sin(45°) = 3 \times 0.707 = 2.12 \text{ m}$$
$$z = 4 \text{ m}$$

**Step 2**: Distance from origin
$$|\vec{r}| = \sqrt{\rho^2 + z^2} = \sqrt{3^2 + 4^2} = \sqrt{25} = 5 \text{ m}$$

---

### Answer

$$\boxed{(x, y, z) = (2.12, 2.12, 4) \text{ m}, \quad |\vec{r}| = 5 \text{ m}}$$

---

```python
# Example 2 - Cylindrical Coordinates
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

rho, phi_deg, z = 3, 45, 4
phi_rad = np.deg2rad(phi_deg)

x = rho * np.cos(phi_rad)
y = rho * np.sin(phi_rad)
magnitude = np.sqrt(rho**2 + z**2)

print(f"Example 2: Cylindrical (ρ={rho}, φ={phi_deg}°, z={z})")
print(f"Cartesian: ({x:.2f}, {y:.2f}, {z}) m")
print(f"Distance: {magnitude:.2f} m")

# 3D Plot
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Helix
t = np.linspace(0, 2*np.pi, 100)
helix_x = rho * np.cos(t)
helix_y = rho * np.sin(t)
helix_z = np.linspace(0, z, 100)
ax.plot(helix_x, helix_y, helix_z, 'b--', alpha=0.3)

# Position vector
ax.quiver(0, 0, 0, x, y, z, arrow_length_ratio=0.1, color='red', linewidth=2)
ax.scatter([x], [y], [z], color='red', s=100)

ax.set_xlabel('x (m)')
ax.set_ylabel('y (m)')
ax.set_zlabel('z (m)')
ax.set_title('Cylindrical Coordinates')
plt.show()
```

## 5. Spherical Coordinates (3D)

### Definition

Position using distance and two angles:

$$\vec{r} = R\hat{e}_R$$

where:
- $R$ = distance from origin
- $\theta$ = polar angle from z-axis ($0 \leq \theta \leq \pi$)
- $\phi$ = azimuthal angle in xy-plane ($0 \leq \phi < 2\pi$)

### Conversion Formulas

**Spherical → Cartesian:**
$$x = R\sin(\theta)\cos(\phi)$$
$$y = R\sin(\theta)\sin(\phi)$$
$$z = R\cos(\theta)$$

### When to Use Spherical

- Satellite orbits
- GPS systems
- Planetary motion
- Spherically symmetric problems

---

### Example 3: Spherical Coordinates

Satellite at $R = 7000$ km, $\theta = 30°$, $\phi = 60°$.

---

**MANUAL SOLUTION:**

$$x = 7000 \times \sin(30°) \times \cos(60°) = 7000 \times 0.5 \times 0.5 = 1750 \text{ km}$$
$$y = 7000 \times \sin(30°) \times \sin(60°) = 7000 \times 0.5 \times 0.866 = 3031 \text{ km}$$
$$z = 7000 \times \cos(30°) = 7000 \times 0.866 = 6062 \text{ km}$$

$$\boxed{(x, y, z) = (1750, 3031, 6062) \text{ km}}$$

---

```python
# Example 3 - Spherical
R, theta_deg, phi_deg = 7000, 30, 60
theta_rad = np.deg2rad(theta_deg)
phi_rad = np.deg2rad(phi_deg)

x = R * np.sin(theta_rad) * np.cos(phi_rad)
y = R * np.sin(theta_rad) * np.sin(phi_rad)
z = R * np.cos(theta_rad)

print(f"Spherical: R={R}, θ={theta_deg}°, φ={phi_deg}°")
print(f"Cartesian: ({x:.0f}, {y:.0f}, {z:.0f}) km")
print(
    f"Verify: √({x:.0f}² + {y:.0f}² + {z:.0f}²) = {np.sqrt(x**2+y**2+z**2):.0f} km ")
```

## 6. Summary & Choosing Coordinates

### Decision Framework

| Motion Type | Best System | Reason |
|-------------|-------------|--------|
| Projectile | Cartesian | Linear, independent x, y |
| Circular (2D) | Polar | Constant r |
| Helix/Spiral | Cylindrical | Rotation + vertical |
| Satellite | Spherical | Constant R from center |

### Key Takeaways

1. Match coordinates to problem geometry
2. Polar/Cylindrical/Spherical unit vectors change direction
3. Verify conversions by checking magnitude
4. Right choice simplifies equations dramatically

### Problem 1: Polar Coordinate Conversion

A point has Cartesian coordinates $(x, y) = (6, 8)$ m.

**Tasks**:
1. Convert to polar coordinates $(r, \theta)$
2. Express $\theta$ in both radians and degrees
3. Verify by converting back to Cartesian
4. Plot both representations

---

---

**SOLUTION TO PROBLEM 1:**

**Given**: $(x, y) = (6, 8)$ m

**Find**: Polar coordinates $(r, \theta)$

**Conversion formulas**:
$$r = \sqrt{x^2 + y^2}, \quad \theta = \tan^{-1}\left(\frac{y}{x}\right)$$

**Task 1 & 2**: Convert to polar
$$r = \sqrt{x^2 + y^2} = \sqrt{6^2 + 8^2} = \sqrt{36 + 64} = \sqrt{100} = 10 \text{ m}$$
$$\theta = \tan^{-1}\left(\frac{y}{x}\right) = \tan^{-1}\left(\frac{8}{6}\right) = \tan^{-1}(1.333) = 0.9273 \text{ rad}$$
$$\theta = 0.9273 \times \frac{180°}{\pi} = 53.13°$$

**Task 3**: Verify by converting back
$$x = r\cos(\theta) = 10\cos(53.13°) = 10 \times 0.6 = 6.0 \text{ m} \quad \checkmark$$
$$y = r\sin(\theta) = 10\sin(53.13°) = 10 \times 0.8 = 8.0 \text{ m} \quad \checkmark$$

$$\boxed{r = 10.0 \text{ m}, \quad \theta = 0.9273 \text{ rad} = 53.13°}$$

---

```python
# Problem 1 - Complete Solution Code
import numpy as np
import matplotlib.pyplot as plt

# Given
x = 6  # m
y = 8  # m

# Convert to polar
r = np.sqrt(x**2 + y**2)
theta_rad = np.arctan2(y, x)
theta_deg = np.degrees(theta_rad)

print("Polar Coordinates:")
print(f" r = {r:.2f} m")
print(f" θ = {theta_rad:.4f} rad = {theta_deg:.2f}°")

# Verify by converting back
x_check = r * np.cos(theta_rad)
y_check = r * np.sin(theta_rad)
print(f"\nVerification (back to Cartesian):")
print(f" x = {x_check:.2f} m ")
print(f" y = {y_check:.2f} m ")

# Plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Cartesian
ax1.plot([0, x], [0, y], 'b-', linewidth=2, label='Position vector')
ax1.plot(x, y, 'ro', markersize=10, label=f'Point ({x}, {y})')
ax1.grid(True, alpha=0.3)
ax1.set_xlabel('x (m)', fontsize=12)
ax1.set_ylabel('y (m)', fontsize=12)
ax1.set_title('Cartesian Representation', fontsize=14, fontweight='bold')
ax1.axis('equal')
ax1.legend()
ax1.set_xlim(-1, 10)
ax1.set_ylim(-1, 10)

# Polar
ax2 = plt.subplot(122, projection='polar')
ax2.plot([0, theta_rad], [0, r], 'b-', linewidth=2)
ax2.plot(theta_rad, r, 'ro', markersize=10)
ax2.set_title(
    f'Polar: r={
        r:.1f}m, θ={
            theta_deg:.1f}°',
    fontsize=14,
    fontweight='bold',
    pad=20)
ax2.grid(True)

plt.tight_layout()
plt.show()

print("\n Conversion verified!")
```

### Problem 3: Spherical Satellite Tracking

A satellite moves from:
- **Initial position**: $(x_0, y_0, z_0) = (5000, 3000, 4000)$ km
- **Final position**: $(x_1, y_1, z_1) = (4000, 5000, 4000)$ km

**Tasks**:
1. Convert both positions to spherical coordinates $(R, \theta, \phi)$
2. Calculate the change in radius $\Delta R$
3. Calculate the change in polar angle $\Delta\theta$
4. Calculate the change in azimuthal angle $\Delta\phi$
5. Which coordinate system better describes this motion?

---

---

**SOLUTION TO PROBLEM 3:**

**Given**: Initial $(5000, 3000, 4000)$ km, Final $(4000, 5000, 4000)$ km

**Conversion formulas**:
$$R = \sqrt{x^2 + y^2 + z^2}, \quad \theta = \cos^{-1}\left(\frac{z}{R}\right), \quad \phi = \tan^{-1}\left(\frac{y}{x}\right)$$

**Initial position**:
$$R_0 = \sqrt{5000^2 + 3000^2 + 4000^2} = \sqrt{50,000,000} = 7071 \text{ km}$$
$$\theta_0 = \cos^{-1}\left(\frac{4000}{7071}\right) = 55.55°$$
$$\phi_0 = \tan^{-1}\left(\frac{3000}{5000}\right) = 30.96°$$

**Final position**:
$$R_1 = \sqrt{4000^2 + 5000^2 + 4000^2} = \sqrt{57,000,000} = 7550 \text{ km}$$
$$\theta_1 = \cos^{-1}\left(\frac{4000}{7550}\right) = 58.01°$$
$$\phi_1 = \tan^{-1}\left(\frac{5000}{4000}\right) = 51.34°$$

**Changes**:
$$\Delta R = 7550 - 7071 = 479 \text{ km}$$
$$\Delta\theta = 58.01° - 55.55° = 2.46°$$
$$\Delta\phi = 51.34° - 30.96° = 20.38°$$

**Best system**: Spherical coordinates clearly show the satellite moved to higher orbit (ΔR = 479 km) while changing angular position.

$$\boxed{\Delta R = 479 \text{ km}, \quad \Delta\theta = 2.46°, \quad \Delta\phi = 20.38°}$$

```python
# Problem 3 - Complete Solution Code
import numpy as np

# Given positions (km)
x0, y0, z0 = 5000, 3000, 4000
x1, y1, z1 = 4000, 5000, 4000

# Convert to spherical


def cart_to_spherical(x, y, z):
    R = np.sqrt(x**2 + y**2 + z**2)
    theta = np.degrees(np.arccos(z / R))
    phi = np.degrees(np.arctan2(y, x))
    return R, theta, phi


R0, theta0, phi0 = cart_to_spherical(x0, y0, z0)
R1, theta1, phi1 = cart_to_spherical(x1, y1, z1)

print("Spherical Coordinates:")
print("="*70)
print(f"Initial position:")
print(f" R₀ = {R0:.0f} km")
print(f" θ₀ = {theta0:.2f}°")
print(f" φ₀ = {phi0:.2f}°")
print()
print(f"Final position:")
print(f" R₁ = {R1:.0f} km")
print(f" θ₁ = {theta1:.2f}°")
print(f" φ₁ = {phi1:.2f}°")
print("="*70)

# Calculate changes
dR = R1 - R0
dtheta = theta1 - theta0
dphi = phi1 - phi0

print(f"\nChanges:")
print(f" ΔR = {dR:.0f} km (satellite moved to higher orbit)")
print(f" Δθ = {dtheta:.2f}° (polar angle change)")
print(f" Δφ = {dphi:.2f}° (azimuthal angle change)")
print("\n Spherical coordinates reveal orbital changes clearly!")
```
