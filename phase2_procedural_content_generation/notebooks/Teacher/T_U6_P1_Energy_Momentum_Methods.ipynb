## Phase 1: Package Sliding Down Ramp

### Problem Setup
Package starts from rest at height h, slides down ramp with friction.

### Method 1: Work-Energy (Recommended)

**Energy equation**:
$$T_1 + V_1 + W_{nc} = T_2 + V_2$$

Where:
- $T_1 = 0$ (starts from rest)
- $V_1 = mgh$ (potential energy at top)
- $V_2 = 0$ (reference at bottom)
- $W_{nc} = -\\mu_k N s$ (friction work, negative)
- $T_2 = \\frac{1}{2}mv_2^2$ (kinetic energy at bottom)

**Normal force**: $N = mg\\cos\\theta$

**Distance along ramp**: $s = \\frac{h}{\\sin\\theta}$

**Friction work**: $W_f = -\\mu_k mg\\cos\\theta \\cdot \\frac{h}{\\sin\\theta} = -\\mu_k mgh\\cot\\theta$

**Final velocity**:
$$0 + mgh - \\mu_k mgh\\cot\\theta = \\frac{1}{2}mv_2^2$$
$$v_2 = \\sqrt{2gh(1 - \\mu_k\\cot\\theta)}$$

### Method 2: Force-Based (Alternative)

**Net force down ramp**: $F = mg\\sin\\theta - \\mu_k mg\\cos\\theta$

**Acceleration**: $a = g(\\sin\\theta - \\mu_k\\cos\\theta)$

**Kinematics**: $v_2^2 = v_1^2 + 2as$

Should give same result!

---


```python
# Phase 1: Package Sliding Down Ramp

print("=" * 70)
print("PHASE 1: Package Sliding Down Ramp")
print("=" * 70)

# Given parameters
m_pkg = 5.0  # kg (package mass)
h = 2.0  # m (ramp height)
theta = np.radians(30)  # 30 degrees
mu_ramp = 0.15  # friction coefficient on ramp
g = 9.81  # m/s^2

print(f"\nGiven:")
print(f"Package mass: m = {m_pkg} kg")
print(f"Ramp height: h = {h} m")
print(f"Ramp angle: theta = {np.degrees(theta):.0f} degrees")
print(f"Friction coefficient: mu = {mu_ramp}")

# Method 1: Work-Energy
print(f"\n" + "-" * 70)
print("METHOD 1: Work-Energy Approach")
print("-" * 70)

# Calculate distance along ramp
s_ramp = h / np.sin(theta)
print(f"\nRamp length: s = h/sin(theta) = {s_ramp:.3f} m")

# Energy at top
T1 = 0  # Starts from rest
V1 = m_pkg * g * h
E1 = T1 + V1

print(f"\nEnergy at top:")
print(f"Kinetic: T1 = {T1:.2f} J")
print(f"Potential: V1 = {V1:.2f} J")
print(f"Total: E1 = {E1:.2f} J")

# Work by friction
N_ramp = m_pkg * g * np.cos(theta)
W_friction = -mu_ramp * N_ramp * s_ramp

print(f"\nFriction work:")
print(f"Normal force: N = {N_ramp:.2f} N")
print(f"Friction work: W_f = {W_friction:.2f} J (negative: dissipates energy)")

# Energy at bottom
V2 = 0  # Reference at bottom
T2 = E1 + W_friction - V2
v_bottom_energy = np.sqrt(2 * T2 / m_pkg)

print(f"\nEnergy at bottom:")
print(f"Potential: V2 = {V2:.2f} J")
print(f"Kinetic: T2 = {T2:.2f} J")
print(f"Velocity: v = {v_bottom_energy:.3f} m/s")

# Method 2: Force-Based (verification)
print(f"\n" + "-" * 70)
print("METHOD 2: Force-Based Approach (Verification)")
print("-" * 70)

F_gravity = m_pkg * g * np.sin(theta)
F_friction_ramp = mu_ramp * N_ramp
F_net = F_gravity - F_friction_ramp
a_ramp = F_net / m_pkg

print(f"\nForce analysis:")
print(f"Gravity component: F_g = {F_gravity:.2f} N (down ramp)")
print(f"Friction force: F_f = {F_friction_ramp:.2f} N (up ramp)")
print(f"Net force: F_net = {F_net:.2f} N")
print(f"Acceleration: a = {a_ramp:.3f} m/s^2")

# Kinematics: v^2 = u^2 + 2as
v_bottom_kinematics = np.sqrt(2 * a_ramp * s_ramp)

print(f"\nKinematics:")
print(f"v^2 = 0 + 2*a*s")
print(f"v = {v_bottom_kinematics:.3f} m/s")

# Comparison
print(f"\n" + "=" * 70)
print("COMPARISON OF METHODS")
print("=" * 70)
print(f"Energy method: v = {v_bottom_energy:.6f} m/s")
print(f"Force method:  v = {v_bottom_kinematics:.6f} m/s")
print(f"Difference: {abs(v_bottom_energy - v_bottom_kinematics):.10f} m/s")
print(f"Match: {abs(v_bottom_energy - v_bottom_kinematics) < 1e-10}")

# Energy dissipated
energy_dissipated_phase1 = abs(W_friction)
print(f"\nEnergy dissipated in Phase 1: {energy_dissipated_phase1:.2f} J ({energy_dissipated_phase1/E1*100:.1f}%)")

# Store result for next phase
v_pkg_before_collision = v_bottom_energy
print(f"\nPackage velocity entering Phase 2: {v_pkg_before_collision:.3f} m/s")
```

## Phase 2: Collision Between Package and Cart

### Problem Setup
Package (moving) collides with cart (at rest). Partially inelastic collision.

### Method: Impulse-Momentum (Required for Collisions)

**Conservation of momentum**:
$$m_{pkg} v_{pkg,1} + m_{cart} v_{cart,1} = m_{pkg} v_{pkg,2} + m_{cart} v_{cart,2}$$

**Coefficient of restitution**:
$$e = \\frac{v_{cart,2} - v_{pkg,2}}{v_{pkg,1} - v_{cart,1}}$$

**Two equations, two unknowns**: Solve for $v_{pkg,2}$ and $v_{cart,2}$

### Solution

From conservation:
$$m_{pkg} v_{pkg,1} = m_{pkg} v_{pkg,2} + m_{cart} v_{cart,2}$$

From restitution:
$$v_{cart,2} - v_{pkg,2} = e \\cdot v_{pkg,1}$$

Solving:
$$v_{pkg,2} = \\frac{(m_{pkg} - e \\cdot m_{cart})v_{pkg,1}}{m_{pkg} + m_{cart}}$$
$$v_{cart,2} = \\frac{(1+e) m_{pkg} v_{pkg,1}}{m_{pkg} + m_{cart}}$$

### Energy Loss

**Before**: $T_1 = \\frac{1}{2}m_{pkg}v_{pkg,1}^2$

**After**: $T_2 = \\frac{1}{2}m_{pkg}v_{pkg,2}^2 + \\frac{1}{2}m_{cart}v_{cart,2}^2$

**Loss**: $\\Delta T = T_1 - T_2$ (dissipated as heat, deformation, sound)

---


```python
# Phase 2: Collision Between Package and Cart

print("\n" + "=" * 70)
print("PHASE 2: Collision Between Package and Cart")
print("=" * 70)

# Given parameters
m_cart = 10.0  # kg
v_cart_before = 0.0  # m/s (at rest)
e = 0.3  # coefficient of restitution

print(f"\nGiven:")
print(f"Package: m = {m_pkg} kg, v = {v_pkg_before_collision:.3f} m/s")
print(f"Cart: m = {m_cart} kg, v = {v_cart_before} m/s (at rest)")
print(f"Coefficient of restitution: e = {e}")

# Apply conservation of momentum
total_mass = m_pkg + m_cart
v_pkg_after = ((m_pkg - e*m_cart)*v_pkg_before_collision + (1+e)*m_cart*v_cart_before) / total_mass
v_cart_after = ((1+e)*m_pkg*v_pkg_before_collision + (m_cart - e*m_pkg)*v_cart_before) / total_mass

print(f"\n" + "-" * 70)
print("SOLUTION: Momentum + Restitution")
print("-" * 70)
print(f"\nAfter collision:")
print(f"Package velocity: v_pkg = {v_pkg_after:.3f} m/s")
print(f"Cart velocity: v_cart = {v_cart_after:.3f} m/s")

# Verify momentum conservation
L_before = m_pkg * v_pkg_before_collision + m_cart * v_cart_before
L_after = m_pkg * v_pkg_after + m_cart * v_cart_after

print(f"\nMomentum conservation check:")
print(f"Before: L = {L_before:.6f} kg·m/s")
print(f"After: L = {L_after:.6f} kg·m/s")
print(f"Error: {abs(L_before - L_after):.10f} (should be ~0)")

# Verify coefficient of restitution
v_rel_before = v_pkg_before_collision - v_cart_before
v_rel_after = v_cart_after - v_pkg_after
e_check = v_rel_after / v_rel_before

print(f"\nCoefficient of restitution check:")
print(f"Relative velocity before: {v_rel_before:.6f} m/s")
print(f"Relative velocity after: {v_rel_after:.6f} m/s")
print(f"e (calculated): {e_check:.6f}")
print(f"e (given): {e:.6f}")
print(f"Error: {abs(e - e_check):.10f}")

# Energy analysis
T_before_collision = 0.5 * m_pkg * v_pkg_before_collision**2
T_after_collision = 0.5 * m_pkg * v_pkg_after**2 + 0.5 * m_cart * v_cart_after**2
energy_dissipated_phase2 = T_before_collision - T_after_collision

print(f"\nEnergy analysis:")
print(f"KE before collision: T1 = {T_before_collision:.2f} J")
print(f"KE after collision: T2 = {T_after_collision:.2f} J")
print(f"Energy dissipated: {energy_dissipated_phase2:.2f} J ({energy_dissipated_phase2/T_before_collision*100:.1f}%)")
print(f"Energy retained: {T_after_collision/T_before_collision*100:.1f}%")

# Compare with perfectly inelastic (e=0)
v_stuck = L_before / total_mass
T_stuck = 0.5 * total_mass * v_stuck**2
loss_max = T_before_collision - T_stuck

print(f"\nComparison with perfectly inelastic (e=0):")
print(f"If stuck together: v = {v_stuck:.3f} m/s")
print(f"Maximum energy loss: {loss_max:.2f} J ({loss_max/T_before_collision*100:.1f}%)")
print(f"Our case (e={e}): {energy_dissipated_phase2:.2f} J ({energy_dissipated_phase2/T_before_collision*100:.1f}%)")

# Store results for next phase
print(f"\n" + "=" * 70)
print(f"Velocities entering Phase 3:")
print(f"Package: {v_pkg_after:.3f} m/s")
print(f"Cart: {v_cart_after:.3f} m/s")
print("=" * 70)
```

## Phase 3: Combined System Slides to Stop

### Problem Setup
After collision, both package and cart slide on floor with friction until they stop.

**Note**: Objects moving at different velocities after collision!

### Method 1: Work-Energy (Simplest)

For each object separately:
$$T_{initial} + W_{friction} = 0$$
$$\\frac{1}{2}mv^2 - \\mu_k mg \\cdot d = 0$$
$$d = \\frac{v^2}{2\\mu_k g}$$

### Method 2: Impulse-Momentum

For each object:
$$J = \\Delta L$$
$$-\\mu_k mg \\cdot t = 0 - mv$$
$$t = \\frac{v}{\\mu_k g}$$

Then: $d = \\bar{v} \\cdot t = \\frac{v}{2} \\cdot \\frac{v}{\\mu_k g} = \\frac{v^2}{2\\mu_k g}$

Same result!

### Method 3: Force-Based

Deceleration: $a = -\\mu_k g$

Kinematics: $v_f^2 = v_i^2 + 2ad$

With $v_f = 0$: $d = \\frac{v_i^2}{2\\mu_k g}$

All methods give same answer!

---


```python
# Phase 3: Combined System Slides to Stop

print("\n" + "=" * 70)
print("PHASE 3: Objects Slide to Stop on Floor")
print("=" * 70)

# Given parameters
mu_floor = 0.20  # friction coefficient on floor

print(f"\nGiven:")
print(f"Floor friction coefficient: mu = {mu_floor}")
print(f"Package initial velocity: v = {v_pkg_after:.3f} m/s")
print(f"Cart initial velocity: v = {v_cart_after:.3f} m/s")

# Method 1: Work-Energy (Package)
print(f"\n" + "-" * 70)
print("METHOD 1: Work-Energy Approach")
print("-" * 70)

# Package
T_pkg_initial = 0.5 * m_pkg * v_pkg_after**2
W_friction_pkg = -mu_floor * m_pkg * g  # Force × distance (but we solve for distance)
# T_initial + W_friction * d = 0
# Solving for d:
d_pkg_energy = v_pkg_after**2 / (2 * mu_floor * g)

print(f"\nPackage:")
print(f"Initial KE: T = {T_pkg_initial:.2f} J")
print(f"Friction force: F = {mu_floor * m_pkg * g:.2f} N")
print(f"Stopping distance: d = v²/(2μg) = {d_pkg_energy:.3f} m")

# Cart
T_cart_initial = 0.5 * m_cart * v_cart_after**2
d_cart_energy = v_cart_after**2 / (2 * mu_floor * g)

print(f"\nCart:")
print(f"Initial KE: T = {T_cart_initial:.2f} J")
print(f"Friction force: F = {mu_floor * m_cart * g:.2f} N")
print(f"Stopping distance: d = v²/(2μg) = {d_cart_energy:.3f} m")

# Method 2: Force-Based (Verification)
print(f"\n" + "-" * 70)
print("METHOD 2: Force-Based Approach (Verification)")
print("-" * 70)

a_floor = -mu_floor * g
print(f"\nDeceleration: a = -μg = {a_floor:.3f} m/s²")

# Package (using v² = u² + 2as with v=0)
d_pkg_kinematics = -v_pkg_after**2 / (2 * a_floor)
t_pkg = -v_pkg_after / a_floor

print(f"\nPackage:")
print(f"Stopping distance: d = {d_pkg_kinematics:.3f} m")
print(f"Stopping time: t = {t_pkg:.3f} s")

# Cart
d_cart_kinematics = -v_cart_after**2 / (2 * a_floor)
t_cart = -v_cart_after / a_floor

print(f"\nCart:")
print(f"Stopping distance: d = {d_cart_kinematics:.3f} m")
print(f"Stopping time: t = {t_cart:.3f} s")

# Comparison
print(f"\n" + "=" * 70)
print("COMPARISON: Energy vs Force Methods")
print("=" * 70)
print(f"\nPackage stopping distance:")
print(f"Energy method: {d_pkg_energy:.6f} m")
print(f"Force method:  {d_pkg_kinematics:.6f} m")
print(f"Match: {abs(d_pkg_energy - d_pkg_kinematics) < 1e-10}")

print(f"\nCart stopping distance:")
print(f"Energy method: {d_cart_energy:.6f} m")
print(f"Force method:  {d_cart_kinematics:.6f} m")
print(f"Match: {abs(d_cart_energy - d_cart_kinematics) < 1e-10}")

# Energy dissipated in Phase 3
energy_dissipated_phase3 = T_pkg_initial + T_cart_initial
print(f"\nEnergy dissipated in Phase 3: {energy_dissipated_phase3:.2f} J (all KE → heat)")
```

```python
# Overall System Analysis

print("\n" + "#" * 70)
print("#" + " " * 68 + "#")
print("#" + "  COMPLETE SYSTEM ANALYSIS".center(68) + "#")
print("#" + " " * 68 + "#")
print("#" * 70)

# Total energy budget
E_initial = m_pkg * g * h  # Potential energy at start
E_dissipated_total = energy_dissipated_phase1 + energy_dissipated_phase2 + energy_dissipated_phase3

print(f"\nENERGY BUDGET:")
print(f"Initial potential energy: {E_initial:.2f} J")
print(f"" + "-" * 50)
print(f"Phase 1 (ramp friction): {energy_dissipated_phase1:.2f} J ({energy_dissipated_phase1/E_initial*100:.1f}%)")
print(f"Phase 2 (collision): {energy_dissipated_phase2:.2f} J ({energy_dissipated_phase2/E_initial*100:.1f}%)")
print(f"Phase 3 (floor friction): {energy_dissipated_phase3:.2f} J ({energy_dissipated_phase3/E_initial*100:.1f}%)")
print(f"" + "-" * 50)
print(f"Total dissipated: {E_dissipated_total:.2f} J ({E_dissipated_total/E_initial*100:.1f}%)")
print(f"Energy conservation check: {abs(E_initial - E_dissipated_total):.6f} J")

# Summary of key results
print(f"\nKEY RESULTS:")
print(f"Package velocity at bottom of ramp: {v_pkg_before_collision:.3f} m/s")
print(f"Package velocity after collision: {v_pkg_after:.3f} m/s")
print(f"Cart velocity after collision: {v_cart_after:.3f} m/s")
print(f"Package stopping distance: {d_pkg_energy:.3f} m")
print(f"Cart stopping distance: {d_cart_energy:.3f} m")
print(f"Total horizontal travel: {max(d_pkg_energy, d_cart_energy):.3f} m")

# Method effectiveness summary
print(f"\nMETHOD EFFECTIVENESS:")
print(f"Phase 1 (sliding): Energy method BEST (friction over distance)")
print(f"Phase 2 (collision): Momentum method REQUIRED (impact)")
print(f"Phase 3 (braking): Energy method SIMPLEST (direct formula)")

print(f"\n" + "#" * 70)
```

```python
# Comprehensive Visualization

fig = plt.figure(figsize=(16, 12))
gs = fig.add_gridspec(3, 3, hspace=0.3, wspace=0.3)

# === Phase Diagrams ===
ax1 = fig.add_subplot(gs[0, :])
ax1.set_xlim(-1, 10)
ax1.set_ylim(-0.5, 2.5)
ax1.set_aspect('equal')
ax1.axis('off')
ax1.set_title('System Overview: Three Phases', fontsize=14, fontweight='bold', pad=20)

# Phase 1: Ramp
ramp_length = s_ramp
ramp_x = np.array([0, ramp_length*np.cos(theta)])
ramp_y = np.array([h, 0])
ax1.plot(ramp_x, ramp_y, 'brown', linewidth=8, label='Ramp')
ax1.plot(0, h, 'ro', markersize=12)
ax1.text(-0.3, h, 'Start', ha='right', fontsize=10, fontweight='bold')
ax1.text(ramp_x[1]/2, ramp_y[1]/2 + 0.3, f'Phase 1\nμ={mu_ramp}', ha='center',
        fontsize=9, bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.7))

# Phase 2: Collision point
collision_x = ramp_x[1] + 0.5
ax1.plot([ramp_x[1], collision_x], [0, 0], 'k-', linewidth=2)
ax1.plot(collision_x, 0, 'go', markersize=15)
ax1.text(collision_x, -0.3, 'Collision\n(Phase 2)', ha='center', fontsize=9,
        fontweight='bold', bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.7))

# Phase 3: Sliding to stop
stop_x = collision_x + max(d_pkg_energy, d_cart_energy) + 0.5
ax1.plot([collision_x, stop_x], [0, 0], 'gray', linewidth=4, linestyle='--')
ax1.text((collision_x + stop_x)/2, 0.3, f'Phase 3\nμ={mu_floor}', ha='center',
        fontsize=9, bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.7))
ax1.plot(collision_x + d_pkg_energy, 0, 'rs', markersize=10, label='Package stops')
ax1.plot(collision_x + d_cart_energy, 0, 'bs', markersize=10, label='Cart stops')
ax1.legend(loc='upper right', fontsize=9)

# === Energy Flow Diagram ===
ax2 = fig.add_subplot(gs[1, 0])
energy_stages = ['Initial\nPE', 'After\nRamp', 'After\nCollision', 'Final']
E_stages = [E_initial,
           E_initial - energy_dissipated_phase1,
           E_initial - energy_dissipated_phase1 - energy_dissipated_phase2,
           0]
colors_e = ['green', 'blue', 'orange', 'red']

ax2.bar(energy_stages, E_stages, color=colors_e, alpha=0.7)
ax2.set_ylabel('Energy (J)', fontsize=11)
ax2.set_title('Energy Cascade', fontsize=12, fontweight='bold')
ax2.grid(True, alpha=0.3, axis='y')

for i, (stage, E) in enumerate(zip(energy_stages, E_stages)):
    ax2.text(i, E + 2, f'{E:.1f} J', ha='center', fontsize=9, fontweight='bold')

# === Energy Dissipation Pie Chart ===
ax3 = fig.add_subplot(gs[1, 1])
dissipation_labels = ['Ramp\nFriction', 'Collision', 'Floor\nFriction']
dissipation_values = [energy_dissipated_phase1, energy_dissipated_phase2, energy_dissipated_phase3]
colors_pie = ['#ff9999', '#66b3ff', '#99ff99']

wedges, texts, autotexts = ax3.pie(dissipation_values, labels=dissipation_labels,
                                     colors=colors_pie, autopct='%1.1f%%',
                                     startangle=90)
for autotext in autotexts:
    autotext.set_color('black')
    autotext.set_fontweight('bold')
ax3.set_title('Energy Dissipation Breakdown', fontsize=12, fontweight='bold')

# === Velocity Timeline ===
ax4 = fig.add_subplot(gs[1, 2])
phases_v = ['Start', 'Bottom\nRamp', 'After\nCollision\n(pkg)', 'After\nCollision\n(cart)', 'Final']
velocities = [0, v_pkg_before_collision, v_pkg_after, v_cart_after, 0]
colors_v = ['red', 'blue', 'orange', 'green', 'red']

ax4.bar(range(len(phases_v)), velocities, color=colors_v, alpha=0.7)
ax4.set_xticks(range(len(phases_v)))
ax4.set_xticklabels(phases_v, fontsize=8)
ax4.set_ylabel('Velocity (m/s)', fontsize=11)
ax4.set_title('Velocity Through System', fontsize=12, fontweight='bold')
ax4.grid(True, alpha=0.3, axis='y')

for i, v in enumerate(velocities):
    ax4.text(i, v + 0.2, f'{v:.2f}', ha='center', fontsize=9, fontweight='bold')

# === Momentum Analysis ===
ax5 = fig.add_subplot(gs[2, 0])
momentum_labels = ['Before\nCollision', 'After\nCollision']
L_pkg_before = m_pkg * v_pkg_before_collision
L_pkg_after = m_pkg * v_pkg_after
L_cart_after = m_cart * v_cart_after

x_m = np.arange(len(momentum_labels))
width_m = 0.4
ax5.bar(x_m[0], L_pkg_before, width_m, label='Package', color='blue', alpha=0.7)
ax5.bar(x_m[1] - width_m/2, L_pkg_after, width_m/2, label='Package', color='blue', alpha=0.7)
ax5.bar(x_m[1], L_cart_after, width_m/2, bottom=L_pkg_after, label='Cart', color='orange', alpha=0.7)
ax5.axhline(y=L_before, color='red', linestyle='--', linewidth=2, label=f'Total={L_before:.1f}')
ax5.set_ylabel('Momentum (kg·m/s)', fontsize=11)
ax5.set_title('Momentum Conservation (Phase 2)', fontsize=12, fontweight='bold')
ax5.set_xticks(x_m)
ax5.set_xticklabels(momentum_labels)
ax5.legend(fontsize=9)
ax5.grid(True, alpha=0.3, axis='y')

# === Stopping Distances ===
ax6 = fig.add_subplot(gs[2, 1])
objects_stop = ['Package', 'Cart']
distances = [d_pkg_energy, d_cart_energy]
colors_stop = ['blue', 'orange']

bars_stop = ax6.barh(objects_stop, distances, color=colors_stop, alpha=0.7)
ax6.set_xlabel('Stopping Distance (m)', fontsize=11)
ax6.set_title('Phase 3: Stopping Distances', fontsize=12, fontweight='bold')
ax6.grid(True, alpha=0.3, axis='x')

for bar, d in zip(bars_stop, distances):
    width = bar.get_width()
    ax6.text(width + 0.05, bar.get_y() + bar.get_height()/2,
            f'{d:.3f} m', va='center', fontsize=10, fontweight='bold')

# === Method Comparison Table ===
ax7 = fig.add_subplot(gs[2, 2])
ax7.axis('off')
ax7.set_title('Method Selection Guide', fontsize=12, fontweight='bold', pad=20)

table_data = [
    ['Phase', 'Best Method', 'Why?'],
    ['1: Ramp', 'Energy', 'Friction over distance'],
    ['2: Collision', 'Momentum', 'Impact (required)'],
    ['3: Braking', 'Energy', 'Direct formula']
]

table = ax7.table(cellText=table_data, cellLoc='left', loc='center',
                 colWidths=[0.2, 0.3, 0.5])
table.auto_set_font_size(False)
table.set_fontsize(9)
table.scale(1, 2)

# Style header row
for i in range(3):
    table[(0, i)].set_facecolor('#4CAF50')
    table[(0, i)].set_text_props(weight='bold', color='white')

# Color other rows
colors_table = ['#FFE082', '#90CAF9', '#A5D6A7']
for i in range(1, 4):
    for j in range(3):
        table[(i, j)].set_facecolor(colors_table[i-1])

plt.suptitle('Complete System Analysis: Package Delivery Practical',
            fontsize=16, fontweight='bold', y=0.98)

plt.show()

print("\nVisualization complete!")
```

## Summary and Key Takeaways

### What We Learned

1. **Energy Methods (Work-Energy)**:
   - Best for friction over distance
   - Direct formulas available
   - Scalar equations (simpler)
   - Used in Phases 1 and 3

2. **Momentum Methods (Impulse-Momentum)**:
   - Required for collisions
   - Handles impact forces
   - Vector equations (includes direction)
   - Used in Phase 2

3. **Multiple Approaches Often Possible**:
   - Different methods give same answer
   - Choose simplest for the problem
   - Use both for verification

### Decision Framework

**Use Energy Methods When**:
- Forces depend on position
- Want to find speeds
- Friction over known distance
- Conservative forces present

**Use Momentum Methods When**:
- Collisions or impacts
- Forces depend on time
- Want velocities (with direction)
- Very short time intervals

**Use Force Methods When**:
- Need accelerations
- Want complete trajectory
- Time history important

### Results Summary

**Energy Budget**:
- Started with potential energy at height h
- Dissipated in three phases:
  1. Ramp friction (~20-30%)
  2. Inelastic collision (~40-50%)
  3. Floor friction (~20-30%)
- Total: 100% → All energy becomes heat!

**Key Insights**:
- Momentum always conserved in collisions (no external force)
- Energy lost in inelastic collisions (e < 1)
- Different objects stop at different distances (different velocities after collision)
- Total system energy always conserved (including dissipated energy)

### Practical Applications

**Engineering**:
- Conveyor belt systems
- Package handling
- Material transport
- Safety analysis

**Design Considerations**:
- Minimize energy loss (use low friction, elastic collisions)
- Control speeds (ramp angle, friction)
- Predict stopping distances (safety zones)
- Size braking systems appropriately
