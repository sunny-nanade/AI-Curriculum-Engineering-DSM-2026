## Acceleration of a Point on a Rigid Body

### Starting Point: Velocity Formula

From Lecture 1, we know:
$$\vec{v}_P = \vec{v}_{\text{center}} + \vec{\omega} \times \vec{r}$$

### Goal: Find Acceleration

Acceleration is the time derivative of velocity:
$$\vec{a}_P = \frac{d\vec{v}_P}{dt}$$

Let's carefully differentiate the velocity formula...

---

### Step 1: Take Time Derivative

$$\vec{a}_P = \frac{d}{dt}\left(\vec{v}_{\text{center}} + \vec{\omega} \times \vec{r}\right)$$

Apply derivative to each term:
$$\vec{a}_P = \frac{d\vec{v}_{\text{center}}}{dt} + \frac{d}{dt}(\vec{\omega} \times \vec{r})$$

$$\boxed{\vec{a}_P = \vec{a}_{\text{center}} + \frac{d}{dt}(\vec{\omega} \times \vec{r})}$$

**First term**: $\vec{a}_{\text{center}}$ is the acceleration of the center (given or known)

**Second term**: Requires **product rule** because both $\vec{\omega}$ and $\vec{r}$ can change with time!

---

### Step 2: Apply Product Rule

For the rotation term:
$$\frac{d}{dt}(\vec{\omega} \times \vec{r}) = \frac{d\vec{\omega}}{dt} \times \vec{r} + \vec{\omega} \times \frac{d\vec{r}}{dt}$$

**Define angular acceleration**:
$$\vec{\alpha} = \frac{d\vec{\omega}}{dt}$$
- Rate of change of angular velocity
- Units: rad/s²
- Same direction as $\vec{\omega}$ if speeding up, opposite if slowing down

**What is** $\frac{d\vec{r}}{dt}$**?**

Remember: $\vec{r}$ is position vector FROM center TO point P.
- In fixed frame, $\vec{r}$ rotates with the rigid body
- From Lecture 1: $\frac{d\vec{r}}{dt} = \vec{\omega} \times \vec{r}$

---

### Step 3: Substitute

$$\frac{d}{dt}(\vec{\omega} \times \vec{r}) = \vec{\alpha} \times \vec{r} + \vec{\omega} \times (\vec{\omega} \times \vec{r})$$

Combine with center acceleration:
$$\boxed{\vec{a}_P = \vec{a}_{\text{center}} + \vec{\alpha} \times \vec{r} + \vec{\omega} \times (\vec{\omega} \times \vec{r})}$$

This is the **fundamental acceleration formula** for rigid body kinematics!

---

## Understanding the Components

$$\vec{a}_P = \vec{a}_{\text{center}} + \vec{\alpha} \times \vec{r} + \vec{\omega} \times (\vec{\omega} \times \vec{r})$$

### Three Terms:

| Term | Name | Physical Meaning | Direction |
|------|------|------------------|-----------|
| $\vec{a}_{\text{center}}$ | **Translation** | Acceleration of center point | Same as center |
| $\vec{\alpha} \times \vec{r}$ | **Tangential** | Due to angular acceleration | ⊥ to r, in plane |
| $\vec{\omega} \times (\vec{\omega} \times \vec{r})$ | **Normal (Centripetal)** | Due to change in direction | Toward center |

### Tangential Acceleration

$$\vec{a}_{\text{tangential}} = \vec{\alpha} \times \vec{r}$$

- **Magnitude**: $|\vec{a}_t| = |\vec{\alpha}| \cdot |\vec{r}|$
- **Direction**: Perpendicular to $\vec{r}$, tangent to circular path
- **Causes**: Change in speed along the path
- **Zero when**: Angular velocity is constant (α = 0)

### Normal (Centripetal) Acceleration

$$\vec{a}_{\text{normal}} = \vec{\omega} \times (\vec{\omega} \times \vec{r})$$

- **Magnitude**: $|\vec{a}_n| = |\vec{\omega}|^2 \cdot |\vec{r}| = \omega^2 r$
- **Direction**: Toward center (perpendicular to velocity)
- **Causes**: Change in direction of velocity
- **Zero when**: No rotation (ω = 0)

### Double Cross Product Identity

Using vector identity: $\vec{A} \times (\vec{B} \times \vec{C}) = \vec{B}(\vec{A} \cdot \vec{C}) - \vec{C}(\vec{A} \cdot \vec{B})$

For $\vec{\omega} \times (\vec{\omega} \times \vec{r})$:
- If ω ⊥ r (planar motion): $\vec{\omega} \times (\vec{\omega} \times \vec{r}) = -\omega^2 \vec{r}$
- Points toward center with magnitude $\omega^2 r$

---

## Special Cases

### Case 1: Constant Angular Velocity (α = 0)

If $\vec{\alpha} = 0$ (steady rotation):
$$\vec{a}_P = \vec{a}_{\text{center}} + \vec{\omega} \times (\vec{\omega} \times \vec{r})$$

Only normal acceleration from rotation (no tangential component).

**Example**: Wheel rolling at constant speed.

### Case 2: Pure Rotation (a_center = 0, fixed point)

If center is fixed:
$$\vec{a}_P = \vec{\alpha} \times \vec{r} + \vec{\omega} \times (\vec{\omega} \times \vec{r})$$

Both tangential and normal components present.

**Example**: Door closing (hinged at one end).

### Case 3: Pure Translation (ω = 0, α = 0)

If no rotation:
$$\vec{a}_P = \vec{a}_{\text{center}}$$

All points have same acceleration (rigid body moves as particle).

**Example**: Elevator (no rotation, just vertical acceleration).

### Case 4: Rolling Without Slipping

Constraint: $v_{\text{center}} = \omega R$ (differentiate)
$$a_{\text{center}} = \alpha R$$

This relates linear and angular acceleration.

## Example 1: Accelerating Wheel

### Problem Statement

A wheel of radius **R = 0.4 m** is rolling without slipping on a horizontal surface.
At a particular instant:
- Center velocity: **v_center = 3.0 m/s** to the right
- Center acceleration: **a_center = 1.5 m/s²** to the right
- Angular acceleration: **α = ?** (to be determined from rolling constraint)

**Find**:
1. Angular velocity ω
2. Angular acceleration α
3. Acceleration of the **top** point
4. Acceleration of the **bottom** point (contact with ground)
5. Acceleration of the **center-right** point

**Given**:
| Quantity | Value | Units |
|----------|-------|-------|
| R | 0.4 | m |
| v_center | 3.0 | m/s → |
| a_center | 1.5 | m/s² → |
| Constraint | Rolling without slipping | - |

---

### Step 1: Physical Situation

**Motion Type**: Rolling with acceleration (translation + rotation, both changing)

**Rolling Without Slipping Constraint**:
- Velocity: $v_{\text{center}} = \omega R$
- Acceleration: $a_{\text{center}} = \alpha R$ (differentiate velocity constraint)

**Why this constraint?**
- Contact point has zero velocity (no slipping)
- This relates linear and angular motion

**Implications**:
- α and a_center are NOT independent
- If center accelerates, angular velocity MUST change

---

### Step 2: Coordinate System & Kinematics

**Fixed Frame**:
- Origin: Ground at initial center position
- x-axis: → (direction of motion)
- y-axis: ↑ (upward)
- z-axis: ⊙ (out of page)

**Given in vector form**:
- $\vec{v}_{\text{center}} = 3.0\hat{i}$ m/s
- $\vec{a}_{\text{center}} = 1.5\hat{i}$ m/s²
- $\vec{\omega} = \omega\hat{k}$ (CCW, out of page)
- $\vec{\alpha} = \alpha\hat{k}$ (angular acceleration)

**From rolling constraint**:
$$\omega = \frac{v_{\text{center}}}{R} = \frac{3.0}{0.4} = 7.5 \text{ rad/s}$$

$$\alpha = \frac{a_{\text{center}}}{R} = \frac{1.5}{0.4} = 3.75 \text{ rad/s}^2$$

---

### Step 3: Position Vectors

**From center to each point** (at instant when wheel is in standard position):

**Top point (T)**:
$$\vec{r}_T = R\hat{j} = 0.4\hat{j} \text{ m}$$

**Bottom point (B)** (contact):
$$\vec{r}_B = -R\hat{j} = -0.4\hat{j} \text{ m}$$

**Center-right point (CR)**:
$$\vec{r}_{CR} = R\hat{i} = 0.4\hat{i} \text{ m}$$

---

### Step 4: Apply Acceleration Formula

$$\vec{a}_P = \vec{a}_{\text{center}} + \vec{\alpha} \times \vec{r} + \vec{\omega} \times (\vec{\omega} \times \vec{r})$$

Where:
- $\vec{a}_{\text{center}} = 1.5\hat{i}$ m/s²
- $\vec{\alpha} = 3.75\hat{k}$ rad/s²
- $\vec{\omega} = 7.5\hat{k}$ rad/s

**Calculate each term**:

#### Tangential acceleration: α × r
- Uses right-hand rule
- $\hat{k} \times \hat{i} = \hat{j}$
- $\hat{k} \times \hat{j} = -\hat{i}$

#### Normal acceleration: ω × (ω × r)
- Double cross product
- Results in: $-\omega^2 \vec{r}$ (toward center)
- Magnitude: $\omega^2 r = (7.5)^2 \times r$

---

### Step 5a: Top Point Acceleration

**Position**: $\vec{r}_T = 0.4\hat{j}$ m

**Tangential**:
$$\vec{\alpha} \times \vec{r}_T = (3.75\hat{k}) \times (0.4\hat{j}) = 3.75 \times 0.4 (\hat{k} \times \hat{j}) = -1.5\hat{i} \text{ m/s}^2$$

**Normal**:
$$\vec{\omega} \times (\vec{\omega} \times \vec{r}_T) = -\omega^2 \vec{r}_T = -(7.5)^2 \times 0.4\hat{j} = -22.5\hat{j} \text{ m/s}^2$$

**Total**:
$$\vec{a}_T = 1.5\hat{i} + (-1.5\hat{i}) + (-22.5\hat{j})$$
$$\boxed{\vec{a}_T = 0\hat{i} - 22.5\hat{j} \text{ m/s}^2}$$

**Magnitude**: $|\vec{a}_T| = 22.5$ m/s² **downward**

---

### Step 5b: Bottom Point Acceleration

**Position**: $\vec{r}_B = -0.4\hat{j}$ m

**Tangential**:
$$\vec{\alpha} \times \vec{r}_B = (3.75\hat{k}) \times (-0.4\hat{j}) = 3.75 \times (-0.4) (\hat{k} \times \hat{j}) = 1.5\hat{i} \text{ m/s}^2$$

**Normal**:
$$\vec{\omega} \times (\vec{\omega} \times \vec{r}_B) = -\omega^2 \vec{r}_B = -(7.5)^2 \times (-0.4\hat{j}) = 22.5\hat{j} \text{ m/s}^2$$

**Total**:
$$\vec{a}_B = 1.5\hat{i} + 1.5\hat{i} + 22.5\hat{j}$$
$$\boxed{\vec{a}_B = 3.0\hat{i} + 22.5\hat{j} \text{ m/s}^2}$$

**Magnitude**: $|\vec{a}_B| = \sqrt{3.0^2 + 22.5^2} = 22.7$ m/s²

**Direction**: Mostly upward, slightly forward

**Key insight**: Even though bottom point has **zero velocity** (instantaneous center), it has **non-zero acceleration**!

---

### Step 5c: Center-Right Point Acceleration

**Position**: $\vec{r}_{CR} = 0.4\hat{i}$ m

**Tangential**:
$$\vec{\alpha} \times \vec{r}_{CR} = (3.75\hat{k}) \times (0.4\hat{i}) = 3.75 \times 0.4 (\hat{k} \times \hat{i}) = 1.5\hat{j} \text{ m/s}^2$$

**Normal**:
$$\vec{\omega} \times (\vec{\omega} \times \vec{r}_{CR}) = -\omega^2 \vec{r}_{CR} = -(7.5)^2 \times 0.4\hat{i} = -22.5\hat{i} \text{ m/s}^2$$

**Total**:
$$\vec{a}_{CR} = 1.5\hat{i} + 1.5\hat{j} + (-22.5\hat{i})$$
$$\boxed{\vec{a}_{CR} = -21.0\hat{i} + 1.5\hat{j} \text{ m/s}^2}$$

**Magnitude**: $|\vec{a}_{CR}| = \sqrt{21.0^2 + 1.5^2} = 21.1$ m/s²

**Direction**: Mostly backward (toward center)

---

```
# Example 1: Numerical Calculation using SymPy

# Given
R = 0.4  # m
v_center = 3.0  # m/s
a_center = 1.5  # m/s²

# From rolling constraint
omega = v_center / R
alpha = a_center / R

print("Example 1: Accelerating Wheel")
print("="*60)
print(f"Given:")
print(f"R = {R} m")
print(f"v_center = {v_center} m/s")
print(f"a_center = {a_center} m/s²")
print()
print(f"From rolling constraint:")
print(f"ω = v/R = {v_center}/{R} = {omega} rad/s")
print(f"α = a/R = {a_center}/{R} = {alpha} rad/s²")
print()

# Position vectors (at standard configuration)
r_T = np.array([0, R, 0])  # Top
r_B = np.array([0, -R, 0])  # Bottom
r_CR = np.array([R, 0, 0])  # Center-right

# Vectors
a_center_vec = np.array([a_center, 0, 0])
omega_vec = np.array([0, 0, omega])
alpha_vec = np.array([0, 0, alpha])

# Helper function to calculate acceleration
def calc_accel(r, a_c, om_vec, al_vec):
    # Tangential: alpha cross r
    a_tan = np.cross(al_vec, r)
    # Normal: omega cross (omega cross r)
    a_norm = np.cross(om_vec, np.cross(om_vec, r))
    # Total
    a_total = a_c + a_tan + a_norm
    return a_tan, a_norm, a_total

# Calculate for each point
a_tan_T, a_norm_T, a_total_T = calc_accel(r_T, a_center_vec, omega_vec, alpha_vec)
a_tan_B, a_norm_B, a_total_B = calc_accel(r_B, a_center_vec, omega_vec, alpha_vec)
a_tan_CR, a_norm_CR, a_total_CR = calc_accel(r_CR, a_center_vec, omega_vec, alpha_vec)

print("RESULTS:")
print()
print("Top Point:")
print(f"Tangential: ({a_tan_T[0]:.2f}, {a_tan_T[1]:.2f}) m/s²")
print(f"Normal: ({a_norm_T[0]:.2f}, {a_norm_T[1]:.2f}) m/s²")
print(f"TOTAL: ({a_total_T[0]:.2f}, {a_total_T[1]:.2f}) m/s²")
print(f"Magnitude: {np.linalg.norm(a_total_T[:2]):.2f} m/s²")
print()
print("Bottom Point (contact):")
print(f"Tangential: ({a_tan_B[0]:.2f}, {a_tan_B[1]:.2f}) m/s²")
print(f"Normal: ({a_norm_B[0]:.2f}, {a_norm_B[1]:.2f}) m/s²")
print(f"TOTAL: ({a_total_B[0]:.2f}, {a_total_B[1]:.2f}) m/s²")
print(f"Magnitude: {np.linalg.norm(a_total_B[:2]):.2f} m/s²")
print()
print("Center-Right Point:")
print(f"Tangential: ({a_tan_CR[0]:.2f}, {a_tan_CR[1]:.2f}) m/s²")
print(f"Normal: ({a_norm_CR[0]:.2f}, {a_norm_CR[1]:.2f}) m/s²")
print(f"TOTAL: ({a_total_CR[0]:.2f}, {a_total_CR[1]:.2f}) m/s²")
print(f"Magnitude: {np.linalg.norm(a_total_CR[:2]):.2f} m/s²")
print("="*60)
```

### Step 6: Verification

**Units Check**:
- All accelerations in m/s² 
- ω in rad/s, α in rad/s² 
- Cross products dimensionally correct 

**Physical Reasoning**:

1. **Top point**: a = 22.5 m/s² downward
   - Makes sense! Moving in circular path, needs centripetal acceleration toward center
   - Tangential components cancel (translation vs rotation)

2. **Bottom point**: a ≠ 0 even though v = 0!
   - **Critical insight**: Zero velocity does NOT mean zero acceleration

3. **Normal acceleration dominates**:

   - This is typical for high-speed rotation

**Rolling Constraint Verification**:
- α = a_center/R = 1.5/0.4 = 3.75 rad/s² 
- Consistent throughout problem

## Example 2: Rotating Disk

### Problem Statement

A disk of radius **R = 0.6 m** rotates about its center (fixed point).
At t = 0:
- Angular velocity: **ω₀ = 4.0 rad/s** (CCW)
- Angular acceleration: **α = 2.0 rad/s²** (CCW, speeding up)

**Find** acceleration of a point on the **edge** of the disk at t = 0.

---

### Solution

**Given**:
- R = 0.6 m
- ω₀ = 4.0 rad/s (CCW)
- α = 2.0 rad/s² (CCW)
- Center is fixed

**Acceleration Formula** (center fixed, so a_center = 0):
$$\vec{a}_P = \vec{\alpha} \times \vec{r} + \vec{\omega} \times (\vec{\omega} \times \vec{r})$$

**Tangential Component**:
$$|\vec{a}_t| = |\vec{\alpha}| \cdot R = 2.0 \times 0.6 = 1.2 \text{ m/s}^2$$

Direction: Tangent to circle (perpendicular to radius, in direction of α)

**Normal Component**:
$$|\vec{a}_n| = \omega^2 R = (4.0)^2 \times 0.6 = 9.6 \text{ m/s}^2$$

Direction: Toward center (radially inward)

**Total Magnitude**:
$$|\vec{a}| = \sqrt{a_t^2 + a_n^2} = \sqrt{1.2^2 + 9.6^2} = \sqrt{1.44 + 92.16} = 9.67 \text{ m/s}^2$$

**Angle from radial direction**:
$$\theta = \arctan\left(\frac{a_t}{a_n}\right) = \arctan\left(\frac{1.2}{9.6}\right) = 7.1°$$

Total acceleration points **mostly radially inward** with slight tangential component.

---

```
# Example 2: Rotating Disk

# Given
R = 0.6  # m
omega_0 = 4.0  # rad/s
alpha = 2.0  # rad/s²

# Calculate components
a_tangential = alpha * R
a_normal = omega_0**2 * R
a_total = np.sqrt(a_tangential**2 + a_normal**2)
angle_deg = np.degrees(np.arctan(a_tangential / a_normal))

print("Example 2: Rotating Disk")
print("="*60)
print(f"Given:")
print(f"R = {R} m")
print(f"ω₀ = {omega_0} rad/s")
print(f"α = {alpha} rad/s²")
print()
print(f"Results at t = 0:")
print(f"Tangential acceleration: a_t = αR = {alpha}×{R} = {a_tangential:.2f} m/s²")
print(f"Normal acceleration: a_n = ω²R = {omega_0}²×{R} = {a_normal:.2f} m/s²")
print(f"Total magnitude: |a| = √(a_t² + a_n²) = {a_total:.2f} m/s²")
print(f"Angle from radial: θ = {angle_deg:.1f}°")
print("="*60)
print()

# Visualization
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Left: Vector diagram
ax1.set_xlim(-1.5, 1.5)
ax1.set_ylim(-1.5, 1.5)
ax1.set_aspect('equal')
ax1.grid(True, alpha=0.3)
ax1.set_title('Acceleration Components at Edge Point', fontsize=14, fontweight='bold')

# Disk
circle = plt.Circle((0, 0), R, fill=False, edgecolor='black', linewidth=2)
ax1.add_patch(circle)
ax1.plot(0, 0, 'ko', markersize=10, label='Center (fixed)')

# Point on right edge (for visualization)
px, py = R, 0
ax1.plot(px, py, 'ro', markersize=12, label='Point P')

# Radius vector
ax1.arrow(0, 0, px, py, head_width=0.05, head_length=0.03,
         fc='gray', ec='gray', linestyle='--', alpha=0.5)
ax1.text(px/2, py/2 - 0.1, 'r', fontsize=12)

# Tangential acceleration (upward for CCW)
scale = 0.1
ax1.arrow(px, py, 0, a_tangential*scale,
         head_width=0.08, head_length=0.05,
         fc='blue', ec='blue', linewidth=3, label=f'a_t = {a_tangential:.2f} m/s²')

# Normal acceleration (toward center, leftward)
ax1.arrow(px, py, -a_normal*scale, 0,
         head_width=0.08, head_length=0.05,
         fc='red', ec='red', linewidth=3, label=f'a_n = {a_normal:.2f} m/s²')

# Total acceleration
ax1.arrow(px, py, -a_normal*scale, a_tangential*scale,
         head_width=0.08, head_length=0.05,
         fc='black', ec='black', linewidth=3, label=f'a_total = {a_total:.2f} m/s²')

# Angular velocity indicator
ax1.annotate('ω', xy=(0, 0), xytext=(-0.3, 0.3),
            fontsize=18, color='green', fontweight='bold')
ax1.annotate('α', xy=(0, 0), xytext=(-0.3, 0.5),
            fontsize=18, color='purple', fontweight='bold')

ax1.set_xlabel('x (m)', fontsize=12)
ax1.set_ylabel('y (m)', fontsize=12)
ax1.legend(loc='upper left', fontsize=9)

# Right: Magnitude comparison
components = ['Tangential\n(α×r)', 'Normal\n(ω²r)', 'Total']
magnitudes = [a_tangential, a_normal, a_total]
colors = ['blue', 'red', 'black']

bars = ax2.bar(components, magnitudes, color=colors, alpha=0.7, edgecolor='black', linewidth=2)
ax2.set_ylabel('Acceleration (m/s²)', fontsize=12)
ax2.set_title('Acceleration Component Magnitudes', fontsize=14, fontweight='bold')
ax2.grid(True, axis='y', alpha=0.3)

# Add value labels on bars
for bar, mag in zip(bars, magnitudes):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height,
            f'{mag:.2f}', ha='center', va='bottom', fontsize=11, fontweight='bold')

plt.tight_layout()
plt.show()

print("Visualization complete!")
print()
print("KEY OBSERVATIONS:")
print("1. Normal acceleration (9.6 m/s²) >> tangential (1.2 m/s²)")
print("2. At higher speeds, normal component dominates")
print("3. Total acceleration points mostly toward center")
print("4. As ω increases, a_n grows quadratically (ω²)")
```

### Key Takeaways from Example 2

1. **Two components for rotating disk**:
   - Tangential: a_t = αR (causes speed change)
   - Normal: a_n = ω²R (causes direction change)

2. **Normal often larger**:
   - a_n ∝ ω² (quadratic growth)
   - a_t ∝ α (linear)
   - At high ω, normal dominates

3. **Perpendicular components**:
   - Tangential ⊥ normal
   - Use Pythagorean theorem for total

4. **Simpler than rolling**:
   - No translation term (center fixed)
   - Only rotation terms remain

## Lecture Summary

### Fundamental Formula

$$\boxed{\vec{a}_P = \vec{a}_{\text{center}} + \vec{\alpha} \times \vec{r} + \vec{\omega} \times (\vec{\omega} \times \vec{r})}$$

### Three Components

| Component | Formula | Magnitude | Direction | Cause |
|-----------|---------|-----------|-----------|-------|
| **Translation** | $\vec{a}_{\text{center}}$ | Given | Same as center | Center acceleration |
| **Tangential** | $\vec{\alpha} \times \vec{r}$ | αr | ⊥ to r, tangent | Angular acceleration |
| **Normal** | $-\omega^2\vec{r}$ | ω²r | Toward center | Centripetal |

### Key Insights

1. **Zero velocity ≠ zero acceleration**
   - Bottom of rolling wheel: v = 0 but a ≠ 0
   - Instantaneous center has acceleration!

2. **Normal acceleration often dominates**
   - a_n = ω²r (quadratic in ω)
   - a_t = αr (linear in α)
   - At high speeds, a_n >> a_t

3. **Rolling constraint**
   - Velocity: v = ωR
   - Acceleration: a = αR (differentiate)
   - Links linear and angular

4. **Components are perpendicular**
   - Can use Pythagorean theorem
   - Or vector addition

### Special Cases

- **Constant ω** (α = 0): Only normal acceleration
- **Fixed center**: Only rotation terms
- **Pure translation**: Only a_center term

### Examples Covered

1. **Accelerating wheel**: Rolling with a_center ≠ 0, α ≠ 0
2. **Rotating disk**: Pure rotation, fixed center
