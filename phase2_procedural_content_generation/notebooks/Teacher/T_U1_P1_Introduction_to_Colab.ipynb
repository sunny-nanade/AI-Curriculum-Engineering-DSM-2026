# PART 4: Basic Computations with NumPy

---

## Creating Arrays

In dynamics, we work with **vectors** and **arrays** of numbers.

NumPy makes this easy:

---

````python
# Create arrays (like mathematical vectors)
position = np.array([3, 4, 0])  # Position vector in 3D
time = np.array([0, 1, 2, 3, 4])  # Time array

print("Position vector:", position)
print("Time array:", time)

# Calculate magnitude of position vector
magnitude = np.sqrt(np.sum(position**2))
print(f"\nMagnitude of position: {magnitude:.2f} meters")
````

## Question: What Did This Code Do?

Look at the code above. Can you identify:

1. How we created the position vector?
2. How we calculated its magnitude?
3. What does `**2` mean?
4. What does `np.sum()` do?

**Discuss with your neighbor...**

*(Scroll for explanation)*

---

## Explanation

```python
position = np.array([3, 4, 0]) # Creates array with 3 elements
```
- This creates a NumPy array (like a vector) with components [3, 4, 0]

```python
magnitude = np.sqrt(np.sum(position**2))
```
- `position**2` → squares each element: [9, 16, 0]
- `np.sum(...)` → adds them up: 9 + 16 + 0 = 25
- `np.sqrt(...)` → takes square root: √25 = 5

This is the **Euclidean distance formula**: $|r| = \sqrt{x^2 + y^2 + z^2}$

# PART 5: Creating Your First Plot

---

## Why Plotting Matters

In Dynamic Systems Modeling, we **visualize**:
- Position vs Time
- Velocity vs Time 
- Trajectories in space
- System behavior

**Seeing is understanding!**

---

## Simple Plot Example

Let's plot a sine wave (represents oscillation):

---

````python
# Create data for plotting
t = np.linspace(0, 2*np.pi, 100)  # Time from 0 to 2π
y = np.sin(t)  # Sine wave (oscillation)

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(t, y, 'b-', linewidth=2, label='sin(t)')
plt.xlabel('Time (radians)', fontsize=12)
plt.ylabel('Amplitude', fontsize=12)
plt.title('Simple Harmonic Motion', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

print(" Plot created successfully!")
print("This represents oscillation - common in mechanical systems!")
````

## Understanding the Plot Code

Let's break down what happened:

```python
t = np.linspace(0, 2*np.pi, 100)
```
- Creates 100 evenly spaced points from 0 to 2π

```python
y = np.sin(t)
```
- Calculates sine value for each t

```python
plt.plot(t, y, 'b-', linewidth=2)
```
- `t, y` → x and y data
- `'b-'` → blue solid line
- `linewidth=2` → thickness

```python
plt.show()
```
- Displays the plot!

## Exercise 3: Create a Position Plot (25 points)

**Task**: Plot position vs time for a moving particle

Given:
- Time: t = 0 to 5 seconds
- Position: x(t) = 2t + 3 (linear motion)

Write code to:
1. Create a time array with 50 points from 0 to 5
2. Calculate position array using the formula
3. Create a plot with:
 - Blue line
 - Title: "Particle Motion"
 - X-label: "Time (s)"
 - Y-label: "Position (m)"

---

### TEACHER SOLUTION (Hidden in Student Version)

---

````python
# Exercise 3: COMPLETE SOLUTION

# Step 1: Create time array
t = np.linspace(0, 5, 50)  # 50 points from 0 to 5 seconds

# Step 2: Calculate position using x(t) = 2t + 3
x = 2*t + 3

# Step 3: Create plot
plt.figure(figsize=(10, 6))
plt.plot(t, x, 'b-', linewidth=2)
plt.xlabel('Time (s)', fontsize=12)
plt.ylabel('Position (m)', fontsize=12)
plt.title('Particle Motion', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.show()

print(" Plot created successfully!")
print(f" Initial position (t=0): {x[0]:.1f} m")
print(f" Final position (t=5): {x[-1]:.1f} m")
print(f" This is uniform motion with constant velocity = 2 m/s")
````

## Exercise 4: Challenge Problem (25 points)

**Task**: Plot a 2D trajectory

A ball is thrown with:
- Horizontal position: x(t) = 10t
- Vertical position: y(t) = 15t - 5t²

Create a single plot showing:
1. The trajectory (y vs x) - NOT vs time!
2. Red line with markers
3. Title: "Projectile Trajectory"
4. Axis labels with units
5. Mark the starting point (green dot) and landing point (red dot)

**Hint**: The ball lands when y = 0

---

### TEACHER SOLUTION (Hidden in Student Version)

---

````python
# Exercise 4: COMPLETE SOLUTION (CHALLENGE!)

# Step 1: Create time array (0 to 3 seconds, ball lands at t=3)
t = np.linspace(0, 3, 100)

# Step 2: Calculate x(t) and y(t)
x = 10 * t  # Horizontal position
y = 15*t - 5*t**2  # Vertical position (parabola)

# Step 3: Create trajectory plot (y vs x, NOT vs time!)
plt.figure(figsize=(10, 6))
plt.plot(x, y, 'r-', linewidth=2, marker='o', markevery=10, label='Trajectory')

# Step 4: Mark special points
# Starting point (t=0): x=0, y=0
plt.plot(0, 0, 'go', markersize=12, label='Start', zorder=5)

# Landing point (t=3): x=30, y=0
plt.plot(30, 0, 'ro', markersize=12, label='Landing', zorder=5)

# Maximum height point (t=1.5): x=15, y=11.25
t_max = 1.5
x_max = 10 * t_max
y_max = 15*t_max - 5*t_max**2
plt.plot(x_max, y_max, 'bs', markersize=10, label='Peak', zorder=5)

plt.xlabel('Horizontal Position (m)', fontsize=12)
plt.ylabel('Vertical Position (m)', fontsize=12)
plt.title('Projectile Trajectory', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.legend()
plt.axis('equal')
plt.show()

print(" Projectile trajectory plotted!")
print(f" Maximum height: {y_max:.2f} m at x = {x_max:.1f} m")
print(f" Range (landing distance): {x[-1]:.1f} m")
print(f" Time of flight: {t[-1]:.1f} s")
print("\n This is classic projectile motion - a parabolic path!")
````

1. Ran cells out of order

## Problem 3: Plot Doesn't Show
