## Question A1: Basketball Trajectory (6 marks)

### Problem Statement

A basketball is thrown from position (0, 2) meters (origin at ground level, y = 2m is release height). The ball's position is given by:

$$x(t) = 8t \quad \text{meters}$$
$$y(t) = 2 + 12t - 5t^2 \quad \text{meters}$$

where $t$ is time in seconds.

### Tasks (Show all working!)

**Part (a)** [2 marks]: Write the position vector $\vec{r}(t)$ in component form using unit vectors $\hat{i}$ and $\hat{j}$.

**Part (b)** [2 marks]: Calculate the position vector at $t = 1$ second. Find the magnitude $|\vec{r}(1)|$.

**Part (c)** [2 marks]: At what time $t$ does the ball hit the ground (y = 0)? What is the horizontal distance traveled (range)?

### Your Solution Space

**SOLVE MANUALLY ON PAPER FIRST!** Show:
- Given data
- Formula for position vector
- Substitution at t = 1s
- Magnitude calculation: $|\vec{r}| = \sqrt{x^2 + y^2}$
- For ground: Set y(t) = 0, solve quadratic
- Use quadratic formula: $t = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}$
- Find x at that time

**Hints**:
- Reference: Lecture 1 Example 3 (Projectile motion)
- For quadratic $y = 2 + 12t - 5t^2 = 0$, rearrange: $-5t^2 + 12t + 2 = 0$
- Or multiply by -1: $5t^2 - 12t - 2 = 0$
- Take positive root (t > 0)

---

### VERIFICATION CODE - Question A1

**After solving manually, run this cell to verify your answers:**

```python
print("="*70)
print("QUESTION A1 - VERIFICATION")
print("="*70)

# Given functions


def x_A1(t):
    return 8*t


def y_A1(t):
    return 2 + 12*t - 5*t**2


# Part (a): Position vector
print("\nPart (a): Position vector r(t)")
print(" r(t) = x(t)î + y(t)ĵ")
print(" r(t) = (8t)î + (2 + 12t - 5t²)ĵ meters")
print(" Write this in your solution")

# Part (b): Position at t = 1s
t1 = 1.0
x1 = x_A1(t1)
y1 = y_A1(t1)
magnitude = np.sqrt(x1**2 + y1**2)

print("\nPart (b): Position at t = 1 second")
print(f" x(1) = 8(1) = {x1:.1f} m")
print(f" y(1) = 2 + 12(1) - 5(1)² = 2 + 12 - 5 = {y1:.1f} m")
print(f" r(1) = {x1:.1f}î + {y1:.1f}ĵ meters")
print(f" |r(1)| = √({x1}² + {y1}²) = √{x1**2 + y1**2:.1f} = {magnitude:.2f} m")

# Part (c): Time when ball hits ground
# Solve 2 + 12t - 5t² = 0 → 5t² - 12t - 2 = 0
a_coef = 5
b_coef = -12
c_coef = -2

discriminant = b_coef**2 - 4*a_coef*c_coef
t_ground_1 = (-b_coef + np.sqrt(discriminant)) / (2*a_coef)
t_ground_2 = (-b_coef - np.sqrt(discriminant)) / (2*a_coef)

# Take positive root
t_ground = t_ground_1 if t_ground_1 > 0 else t_ground_2
x_range = x_A1(t_ground)

print("\nPart (c): When ball hits ground (y = 0)")
print(" Solve: 2 + 12t - 5t² = 0 → 5t² - 12t - 2 = 0")
print(f" Quadratic formula: a={a_coef}, b={b_coef}, c={c_coef}")
print(
    f" Discriminant = b² - 4ac = {b_coef}² - 4({a_coef})({c_coef}) = {discriminant}")
print(f" t = [12 ± √{discriminant}] / (2×5)")
print(f" t = [12 ± {np.sqrt(discriminant):.3f}] / 10")
print(f" t₁ = {t_ground_1:.3f} s (positive, physical)")
print(f" t₂ = {t_ground_2:.3f} s (negative, before throw)")
print(f"\n Time to hit ground: t = {t_ground:.3f} seconds")
print(
    f" Horizontal range: x({
        t_ground:.3f}) = 8 × {
            t_ground:.3f} = {
                x_range:.2f} meters")

# Visualization
t_plot = np.linspace(0, t_ground, 100)
x_plot = x_A1(t_plot)
y_plot = y_A1(t_plot)

plt.figure()
plt.plot(x_plot, y_plot, 'b-', linewidth=2, label='Basketball trajectory')
plt.plot(x1, y1, 'go', markersize=10, label=f't=1s: ({x1:.1f}, {y1:.1f})m')
plt.plot(x_range, 0, 'ro', markersize=10, label=f'Lands: ({x_range:.2f}, 0)m')
plt.plot(0, 2, 'ko', markersize=8, label='Release: (0, 2)m')
plt.axhline(
    0,
    color='brown',
    linewidth=2,
    linestyle='--',
    alpha=0.5,
    label='Ground')
plt.xlabel('x (meters)', fontsize=12, fontweight='bold')
plt.ylabel('y (meters)', fontsize=12, fontweight='bold')
plt.title('Question A1: Basketball Trajectory', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.legend()
plt.axis('equal')
plt.tight_layout()
plt.show()

print("\n" + "="*70)
print("EXPECTED ANSWERS (verify your manual solution):")
print("="*70)
print(f"Part (a): r(t) = (8t)î + (2 + 12t - 5t²)ĵ m")
print(f"Part (b): r(1) = {x1:.1f}î + {y1:.1f}ĵ m, |r(1)| = {magnitude:.2f} m")
print(f"Part (c): t = {t_ground:.3f} s, Range = {x_range:.2f} m")
print("="*70)
```

## Question A2: 3D Drone Motion (6 marks)

### Problem Statement

A drone's position in 3D space is tracked as:

$$x(t) = 5 + 3t \quad \text{meters}$$
$$y(t) = 2t \quad \text{meters}$$
$$z(t) = 10 + 4t - t^2 \quad \text{meters}$$

where $t$ is time in seconds. The drone starts at $t = 0$ and flies until $t = 3$ seconds.

### Tasks (Show all working!)

**Part (a)** [2 marks]: Write the 3D position vector $\vec{r}(t)$ and evaluate it at $t = 0$ (starting position).

**Part (b)** [2 marks]: Find the position vector at $t = 2$ seconds. Calculate its magnitude $|\vec{r}(2)|$ (distance from origin).

**Part (c)** [2 marks]: Calculate the displacement vector $\Delta\vec{r}$ from $t = 0$ to $t = 3$ seconds. Find the magnitude of displacement.

### Your Solution Space

**SOLVE MANUALLY ON PAPER!** Show:
- Position vector: $\vec{r}(t) = x(t)\hat{i} + y(t)\hat{j} + z(t)\hat{k}$
- Evaluate at specific times by substitution
- 3D magnitude: $|\vec{r}| = \sqrt{x^2 + y^2 + z^2}$
- Displacement: $\Delta\vec{r} = \vec{r}(t_2) - \vec{r}(t_1)$
- Component-wise subtraction

**Hints**:
- Reference: Lecture 1 Section on 3D vectors
- Be careful with signs when subtracting vectors
- Displacement is NOT the same as distance traveled
- Draw a 3D diagram if it helps!

---

### VERIFICATION CODE - Question A2

**After solving manually, run this cell:**

```python
print("="*70)
print("QUESTION A2 - VERIFICATION")
print("="*70)

# Given functions


def x_A2(t):
    return 5 + 3*t


def y_A2(t):
    return 2*t


def z_A2(t):
    return 10 + 4*t - t**2


def r_A2(t):
    """Position vector as numpy array"""
    return np.array([x_A2(t), y_A2(t), z_A2(t)])


# Part (a): Position vector and starting position
r0 = r_A2(0)
print("\nPart (a): Position vector and starting position")
print(" r(t) = x(t)î + y(t)ĵ + z(t)k̂")
print(" r(t) = (5 + 3t)î + (2t)ĵ + (10 + 4t - t²)k̂ meters")
print("\n At t = 0:")
print(f" x(0) = 5 + 3(0) = {r0[0]:.1f} m")
print(f" y(0) = 2(0) = {r0[1]:.1f} m")
print(f" z(0) = 10 + 4(0) - (0)² = {r0[2]:.1f} m")
print(f" r(0) = {r0[0]:.1f}î + {r0[1]:.1f}ĵ + {r0[2]:.1f}k̂ meters")

# Part (b): Position at t = 2s
r2 = r_A2(2)
mag_r2 = np.linalg.norm(r2)

print("\nPart (b): Position at t = 2 seconds")
print(f" x(2) = 5 + 3(2) = {r2[0]:.1f} m")
print(f" y(2) = 2(2) = {r2[1]:.1f} m")
print(f" z(2) = 10 + 4(2) - (2)² = 10 + 8 - 4 = {r2[2]:.1f} m")
print(f" r(2) = {r2[0]:.1f}î + {r2[1]:.1f}ĵ + {r2[2]:.1f}k̂ meters")
print(f"\n Magnitude:")
print(f" |r(2)| = √(x² + y² + z²)")
print(f" |r(2)| = √({r2[0]}² + {r2[1]}² + {r2[2]}²)")
print(f" |r(2)| = √({r2[0]**2} + {r2[1]**2} + {r2[2]**2})")
print(f" |r(2)| = √{np.sum(r2**2):.1f} = {mag_r2:.2f} m")

# Part (c): Displacement from t=0 to t=3
r3 = r_A2(3)
displacement = r3 - r0
mag_disp = np.linalg.norm(displacement)

print("\nPart (c): Displacement from t=0 to t=3 seconds")
print(f" r(3) = (5 + 3×3)î + (2×3)ĵ + (10 + 4×3 - 3²)k̂")
print(f" r(3) = {r3[0]:.1f}î + {r3[1]:.1f}ĵ + {r3[2]:.1f}k̂ meters")
print(f"\n Displacement: Δr = r(3) - r(0)")
print(f" Δr = ({r3[0]:.1f}î + {r3[1]:.1f}ĵ + {r3[2]:.1f}k̂) - ({r0[0]:.1f}î + {r0[1]:.1f}ĵ + {r0[2]:.1f}k̂)")
print(f" Δr = ({r3[0] -
                r0[0]:.1f})î + ({r3[1] -
                                 r0[1]:.1f})ĵ + ({r3[2] -
                                                  r0[2]:.1f})k̂ meters")
print(f"\n Magnitude of displacement:")
print(
    f" |Δr| = √({displacement[0]}² + {displacement[1]}² + {displacement[2]}²)")
print(f" |Δr| = √{np.sum(displacement**2):.1f} = {mag_disp:.2f} m")

# Visualization
t_vals = np.linspace(0, 3, 50)
x_vals = x_A2(t_vals)
y_vals = y_A2(t_vals)
z_vals = z_A2(t_vals)

fig = plt.figure(figsize=(12, 5))

# 3D plot
ax1 = fig.add_subplot(121, projection='3d')
ax1.plot(x_vals, y_vals, z_vals, 'b-', linewidth=2, label='Drone path')
ax1.scatter(
    *r0,
    color='green',
    s=100,
    label=f't=0: ({
        r0[0]:.0f}, {
            r0[1]:.0f}, {
                r0[2]:.0f})')
ax1.scatter(
    *r2,
    color='orange',
    s=100,
    label=f't=2: ({
        r2[0]:.0f}, {
            r2[1]:.0f}, {
                r2[2]:.0f})')
ax1.scatter(
    *r3,
    color='red',
    s=100,
    label=f't=3: ({
        r3[0]:.0f}, {
            r3[1]:.0f}, {
                r3[2]:.0f})')
ax1.plot([r0[0], r3[0]], [r0[1], r3[1]], [r0[2], r3[2]],
         'r--', linewidth=2, alpha=0.5, label='Displacement')
ax1.set_xlabel('x (m)', fontweight='bold')
ax1.set_ylabel('y (m)', fontweight='bold')
ax1.set_zlabel('z (m)', fontweight='bold')
ax1.set_title('3D Drone Path', fontweight='bold')
ax1.legend()

# Component plots
ax2 = fig.add_subplot(122)
ax2.plot(t_vals, x_vals, 'r-', linewidth=2, label='x(t)')
ax2.plot(t_vals, y_vals, 'g-', linewidth=2, label='y(t)')
ax2.plot(t_vals, z_vals, 'b-', linewidth=2, label='z(t)')
ax2.scatter([0, 2, 3], [r0[0], r2[0], r3[0]], color='red', s=50, zorder=5)
ax2.scatter([0, 2, 3], [r0[1], r2[1], r3[1]], color='green', s=50, zorder=5)
ax2.scatter([0, 2, 3], [r0[2], r2[2], r3[2]], color='blue', s=50, zorder=5)
ax2.set_xlabel('Time (seconds)', fontweight='bold')
ax2.set_ylabel('Position (meters)', fontweight='bold')
ax2.set_title('Position Components vs Time', fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.legend()

plt.tight_layout()
plt.show()

print("\n" + "="*70)
print("EXPECTED ANSWERS:")
print("="*70)
print(f"Part (a): r(t) = (5+3t)î + (2t)ĵ + (10+4t-t²)k̂ m")
print(f" r(0) = {r0[0]:.0f}î + {r0[1]:.0f}ĵ + {r0[2]:.0f}k̂ m")
print(
    f"Part (b): r(2) = {r2[0]:.0f}î + {r2[1]:.0f}ĵ + {r2[2]:.0f}k̂ m, |r(2)| = {mag_r2:.2f} m")
print(
    f"Part (c): Δr = {
        displacement[0]:.0f}î + {
            displacement[1]:.0f}ĵ + {
                displacement[2]:.0f}k̂ m, |Δr| = {
                    mag_disp:.2f} m")
print("="*70)
```

## Question B1: Rocket Launch Analysis (7 marks)

### Problem Statement

A model rocket is launched vertically upward from ground level. Its height (vertical position) is:

$$y(t) = 60t - 10t^2 \quad \text{meters}$$

where $t$ is time in seconds after launch.

### Tasks (Show all derivatives step-by-step!)

**Part (a)** [2 marks]: Find the velocity function $v_y(t) = \frac{dy}{dt}$. What is the initial velocity at $t = 0$?

**Part (b)** [2 marks]: Find the acceleration function $a_y(t) = \frac{dv_y}{dt}$. Is it constant? What does this represent physically?

**Part (c)** [2 marks]: At what time does the rocket reach maximum height? (Hint: At max height, velocity = 0)

**Part (d)** [1 mark]: What is the maximum height reached?

### Your Solution Space

**SOLVE MANUALLY!** Show:
- Differentiation: $\frac{d}{dt}[ct] = c$, $\frac{d}{dt}[ct^2] = 2ct$
- Apply to each term of $y(t)$
- For velocity: $v_y(t) = \frac{dy}{dt} = \frac{d}{dt}[60t - 10t^2]$
- For acceleration: $a_y(t) = \frac{dv_y}{dt}$
- Max height condition: Set $v_y(t) = 0$, solve for $t$
- Substitute $t$ back into $y(t)$

**Hints**:
- Reference: Lecture 2 Example 1 (velocity from position)
- Constant negative acceleration = gravity pulling down
- At max height, rocket momentarily stops (v = 0) before falling

---

### VERIFICATION CODE - Question B1

```python
print("="*70)
print("QUESTION B1 - VERIFICATION")
print("="*70)

# Symbolic calculation
t_sym = sp.Symbol('t', real=True, positive=True)
y_sym = 60*t_sym - 10*t_sym**2
v_sym = sp.diff(y_sym, t_sym)
a_sym = sp.diff(v_sym, t_sym)

print("\nPart (a): Velocity function (derivative of position)")
print(" Given: y(t) = 60t - 10t²")
print(" Velocity: v_y(t) = dy/dt")
print(" ")
print(" v_y(t) = d/dt[60t - 10t²]")
print(" v_y(t) = d/dt[60t] - d/dt[10t²]")
print(" v_y(t) = 60 - 10(2t) [using power rule]")
print(f" v_y(t) = {v_sym} m/s")
print(" ")
print(f" Initial velocity (t=0): v_y(0) = 60 - 20(0) = 60 m/s ")

print("\nPart (b): Acceleration function (derivative of velocity)")
print(f" Given: v_y(t) = {v_sym}")
print(" Acceleration: a_y(t) = dv_y/dt")
print(" ")
print(" a_y(t) = d/dt[60 - 20t]")
print(" a_y(t) = d/dt[60] - d/dt[20t]")
print(" a_y(t) = 0 - 20")
print(f" a_y(t) = {a_sym} m/s² (CONSTANT) ")
print(" ")
print(" Physical meaning: Constant downward acceleration = 20 m/s²")
print(" This is twice Earth's gravity (simplified model rocket)")

# Solve for max height time
t_max = sp.solve(v_sym, t_sym)[0]
y_max = y_sym.subs(t_sym, t_max)

print("\nPart (c): Time to reach maximum height")
print(" At max height: velocity = 0")
print(" Set v_y(t) = 0:")
print(" 60 - 20t = 0")
print(" 20t = 60")
print(f" t = 60/20 = {float(t_max)} seconds ")

print("\nPart (d): Maximum height")
print(f" Substitute t = {float(t_max)} into y(t):")
print(f" y({float(t_max)}) = 60({float(t_max)}) - 10({float(t_max)})²")
print(f" y({float(t_max)}) = {60*float(t_max)} - 10({float(t_max)**2})")
print(f" y({float(t_max)}) = {60*float(t_max)} - {10*float(t_max)**2}")
print(f" y_max = {float(y_max)} meters ")

# Visualization
t_vals = np.linspace(0, 6, 100)
y_vals = 60*t_vals - 10*t_vals**2
v_vals = 60 - 20*t_vals
a_vals = -20*np.ones_like(t_vals)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Position
axes[0].plot(t_vals, y_vals, 'b-', linewidth=2)
axes[0].scatter([float(t_max)], [float(y_max)], color='red', s=100,
                zorder=5, label=f'Max: ({float(t_max)}s, {float(y_max)}m)')
axes[0].axhline(
    0,
    color='brown',
    linewidth=2,
    linestyle='--',
    alpha=0.5,
    label='Ground')
axes[0].set_xlabel('Time (s)', fontweight='bold')
axes[0].set_ylabel('Height (m)', fontweight='bold')
axes[0].set_title('Position: y(t) = 60t - 10t²', fontweight='bold')
axes[0].grid(True, alpha=0.3)
axes[0].legend()

# Velocity
axes[1].plot(t_vals, v_vals, 'r-', linewidth=2)
axes[1].axhline(0, color='black', linewidth=1, linestyle='--', alpha=0.5)
axes[1].scatter([0, float(t_max)], [60, 0], color='darkred', s=100, zorder=5)
axes[1].set_xlabel('Time (s)', fontweight='bold')
axes[1].set_ylabel('Velocity (m/s)', fontweight='bold')
axes[1].set_title('Velocity: v(t) = 60 - 20t', fontweight='bold')
axes[1].grid(True, alpha=0.3)
axes[1].annotate(
    'v=0 at max height', xy=(
        float(t_max), 0), xytext=(
            float(t_max)+0.5, 10), arrowprops=dict(
                arrowstyle='->', color='red'), fontsize=10, fontweight='bold')

# Acceleration
axes[2].plot(t_vals, a_vals, 'g-', linewidth=3)
axes[2].set_xlabel('Time (s)', fontweight='bold')
axes[2].set_ylabel('Acceleration (m/s²)', fontweight='bold')
axes[2].set_title('Acceleration: a(t) = -20 (constant)', fontweight='bold')
axes[2].grid(True, alpha=0.3)
axes[2].set_ylim(-25, -15)

plt.tight_layout()
plt.show()

print("\n" + "="*70)
print("EXPECTED ANSWERS:")
print("="*70)
print(f"Part (a): v_y(t) = 60 - 20t m/s, v_y(0) = 60 m/s")
print(f"Part (b): a_y(t) = -20 m/s² (constant, downward)")
print(f"Part (c): t_max = {float(t_max)} seconds")
print(f"Part (d): y_max = {float(y_max)} meters")
print("="*70)
```

## Question D1: Elliptical Race Track Analysis (10 marks)

### Problem Statement

A race car moves on an elliptical track. Its position is described by the **parametric equations**:

$$x(t) = 50\cos(0.2t) \quad \text{meters}$$
$$y(t) = 30\sin(0.2t) \quad \text{meters}$$

where $t$ is time in seconds, and $t \geq 0$.

This describes an ellipse with semi-major axis $a = 50$ m (along x) and semi-minor axis $b = 30$ m (along y). The car completes one full lap when $0.2t = 2\pi$, i.e., at $t = 10\pi \approx 31.4$ seconds.

### Tasks (Show complete working with calculus!)

**Part (a)** [3 marks]: Find the **velocity vector** $\vec{v}(t)$ by differentiating the position components.

Calculate:
- $v_x(t) = \frac{dx}{dt}$
- $v_y(t) = \frac{dy}{dt}$
- Write $\vec{v}(t) = v_x(t)\hat{i} + v_y(t)\hat{j}$

**Part (b)** [3 marks]: Calculate the velocity vector at $t = 0$ seconds (starting point). Find the **speed** $|\vec{v}(0)|$.

**Part (c)** [2 marks]: Find the **acceleration vector** $\vec{a}(t)$ by differentiating velocity.

Calculate:
- $a_x(t) = \frac{dv_x}{dt}$
- $a_y(t) = \frac{dv_y}{dt}$

**Part (d)** [2 marks]: Evaluate the acceleration vector at $t = 0$ seconds. Calculate the magnitude $|\vec{a}(0)|$.

### Your Solution Space

**SOLVE MANUALLY!** Show:

**For Part (a) - Velocity:**
- Use chain rule: $\frac{d}{dt}[\cos(\omega t)] = -\omega\sin(\omega t)$
- Use chain rule: $\frac{d}{dt}[\sin(\omega t)] = \omega\cos(\omega t)$
- Here $\omega = 0.2$ rad/s
- $v_x(t) = \frac{d}{dt}[50\cos(0.2t)] = 50 \times (-0.2\sin(0.2t)) = -10\sin(0.2t)$
- Similarly for $v_y(t)$

**For Part (b) - Evaluate at t=0:**
- Substitute $t = 0$ into velocity components
- Remember: $\sin(0) = 0$, $\cos(0) = 1$
- Speed: $|\vec{v}| = \sqrt{v_x^2 + v_y^2}$

**For Part (c) - Acceleration:**
- Differentiate velocity components
- $a_x(t) = \frac{d}{dt}[v_x(t)]$
- Use chain rule again

**For Part (d) - Evaluate at t=0:**
- Substitute $t = 0$ into acceleration components
- Calculate magnitude

**Hints**:
- Reference: Practical 2 Circular Motion (this is similar but ellipse instead of circle)
- Chain rule is essential for trigonometric functions
- At $t = 0$: car is at position $(50, 0)$ m (rightmost point of ellipse)

### VERIFICATION CODE - Question D1

```python
print("="*70)
print("QUESTION D1 - VERIFICATION")
print("="*70)

# Define symbolic variable
t_sym = sp.Symbol('t', real=True, positive=True)

# Given parametric equations
a_ellipse = 50  # semi-major axis (m)
b_ellipse = 30  # semi-minor axis (m)
omega = 0.2  # angular frequency (rad/s)

x_sym = a_ellipse * sp.cos(omega * t_sym)
y_sym = b_ellipse * sp.sin(omega * t_sym)

print("\nGiven: Elliptical motion (parametric equations)")
print(f" x(t) = {a_ellipse}cos({omega}t) meters")
print(f" y(t) = {b_ellipse}sin({omega}t) meters")
print(f" ω = {omega} rad/s")
print()
print(f" Ellipse: Semi-major axis a = {a_ellipse} m")
print(f" Semi-minor axis b = {b_ellipse} m")
print(f" Period: T = 2π/ω = 2π/{omega} = {2*np.pi/omega:.2f} seconds")

print("\n" + "="*70)
print("Part (a): Velocity vector (derivatives of position)")
print("="*70)

# Calculate velocity (first derivative)
vx_sym = sp.diff(x_sym, t_sym)
vy_sym = sp.diff(y_sym, t_sym)

print("\n Using chain rule:")
print(" d/dt[cos(ωt)] = -ω sin(ωt)")
print(" d/dt[sin(ωt)] = ω cos(ωt)")
print()
print(f" v_x(t) = d/dt[{a_ellipse}cos({omega}t)]")
print(f" v_x(t) = {a_ellipse} × d/dt[cos({omega}t)]")
print(f" v_x(t) = {a_ellipse} × (-{omega}sin({omega}t))")
print(f" v_x(t) = {vx_sym} m/s ")
print()
print(f" v_y(t) = d/dt[{b_ellipse}sin({omega}t)]")
print(f" v_y(t) = {b_ellipse} × d/dt[sin({omega}t)]")
print(f" v_y(t) = {b_ellipse} × ({omega}cos({omega}t))")
print(f" v_y(t) = {vy_sym} m/s ")
print()
print(f" Velocity vector:")
print(f" v(t) = v_x(t)î + v_y(t)ĵ")
print(f" v(t) = [{vx_sym}]î + [{vy_sym}]ĵ m/s ")

print("\n" + "="*70)
print("Part (b): Velocity at t = 0 seconds")
print("="*70)

t_eval = 0
x_0 = float(x_sym.subs(t_sym, t_eval))
y_0 = float(y_sym.subs(t_sym, t_eval))
vx_0 = float(vx_sym.subs(t_sym, t_eval))
vy_0 = float(vy_sym.subs(t_sym, t_eval))
speed_0 = np.sqrt(vx_0**2 + vy_0**2)

print(f"\n Position at t = 0:")
print(f" x(0) = {a_ellipse}cos(0) = {a_ellipse} × 1 = {x_0:.1f} m")
print(f" y(0) = {b_ellipse}sin(0) = {b_ellipse} × 0 = {y_0:.1f} m")
print(f" r(0) = {x_0:.1f}î + {y_0:.1f}ĵ m (rightmost point)")
print()
print(f" Velocity at t = 0:")
print(f" Remember: sin(0) = 0, cos(0) = 1")
print()
print(f" v_x(0) = -{a_ellipse * omega}sin(0)")
print(f" v_x(0) = -{a_ellipse * omega} × 0 = {vx_0:.1f} m/s ")
print()
print(f" v_y(0) = {b_ellipse * omega}cos(0)")
print(f" v_y(0) = {b_ellipse * omega} × 1 = {vy_0:.1f} m/s ")
print()
print(f" v(0) = {vx_0:.1f}î + {vy_0:.1f}ĵ m/s")
print(f" (moving upward at rightmost point)")
print()
print(f" Speed at t = 0:")
print(f" |v(0)| = √(v_x² + v_y²)")
print(f" |v(0)| = √({vx_0:.1f}² + {vy_0:.1f}²)")
print(f" |v(0)| = √({vx_0**2:.1f} + {vy_0**2:.1f})")
print(f" |v(0)| = √{vx_0**2 + vy_0**2:.1f}")
print(f" |v(0)| = {speed_0:.1f} m/s ")

print("\n" + "="*70)
print("Part (c): Acceleration vector (derivatives of velocity)")
print("="*70)

# Calculate acceleration (second derivative)
ax_sym = sp.diff(vx_sym, t_sym)
ay_sym = sp.diff(vy_sym, t_sym)

print(f"\n a_x(t) = dv_x/dt")
print(f" a_x(t) = d/dt[{vx_sym}]")
print(f" a_x(t) = -{a_ellipse * omega} × d/dt[sin({omega}t)]")
print(f" a_x(t) = -{a_ellipse * omega} × {omega}cos({omega}t)")
print(f" a_x(t) = {ax_sym} m/s² ")
print()
print(f" a_y(t) = dv_y/dt")
print(f" a_y(t) = d/dt[{vy_sym}]")
print(f" a_y(t) = {b_ellipse * omega} × d/dt[cos({omega}t)]")
print(f" a_y(t) = {b_ellipse * omega} × (-{omega}sin({omega}t))")
print(f" a_y(t) = {ay_sym} m/s² ")
print()
print(f" Acceleration vector:")
print(f" a(t) = a_x(t)î + a_y(t)ĵ")
print(f" a(t) = [{ax_sym}]î + [{ay_sym}]ĵ m/s² ")
print()
print(" Observation: a(t) = -ω²r(t)")
print(f" Acceleration points toward center (like centripetal acceleration)")

print("\n" + "="*70)
print("Part (d): Acceleration at t = 0 seconds")
print("="*70)

ax_0 = float(ax_sym.subs(t_sym, t_eval))
ay_0 = float(ay_sym.subs(t_sym, t_eval))
accel_mag_0 = np.sqrt(ax_0**2 + ay_0**2)

print(f"\n Acceleration at t = 0:")
print(f" a_x(0) = -{a_ellipse * omega**2}cos(0)")
print(f" a_x(0) = -{a_ellipse * omega**2} × 1 = {ax_0:.2f} m/s² ")
print()
print(f" a_y(0) = -{b_ellipse * omega**2}sin(0)")
print(f" a_y(0) = -{b_ellipse * omega**2} × 0 = {ay_0:.1f} m/s² ")
print()
print(f" a(0) = {ax_0:.2f}î + {ay_0:.1f}ĵ m/s²")
print(f" (pointing toward center from rightmost point)")
print()
print(f" Magnitude of acceleration at t = 0:")
print(f" |a(0)| = √(a_x² + a_y²)")
print(f" |a(0)| = √({ax_0:.2f}² + {ay_0:.1f}²)")
print(f" |a(0)| = √{ax_0**2:.2f}")
print(f" |a(0)| = {accel_mag_0:.2f} m/s² ")
print()
print(
    f" Alternative: |a(0)| = ω²|r(0)| = {
        omega**2} × {x_0} = {
            omega**2 *
        x_0:.2f} m/s² ")

# Visualization
t_vals = np.linspace(0, 2*np.pi/omega, 200)
x_vals = a_ellipse * np.cos(omega * t_vals)
y_vals = b_ellipse * np.sin(omega * t_vals)

# Calculate velocity and acceleration arrays
vx_vals = -a_ellipse * omega * np.sin(omega * t_vals)
vy_vals = b_ellipse * omega * np.cos(omega * t_vals)
ax_vals = -a_ellipse * omega**2 * np.cos(omega * t_vals)
ay_vals = -b_ellipse * omega**2 * np.sin(omega * t_vals)

fig = plt.figure(figsize=(16, 6))

# Trajectory with vectors at t=0
ax1 = fig.add_subplot(131)
ax1.plot(x_vals, y_vals, 'b-', linewidth=2, label='Elliptical track')
ax1.scatter(
    [x_0],
    [y_0],
    color='red',
    s=150,
    zorder=5,
    label=f't=0: ({
        x_0:.0f}, {
            y_0:.0f})m')
ax1.scatter(
    [0],
    [0],
    color='black',
    s=100,
    marker='x',
    zorder=5,
    label='Center')

# Velocity vector at t=0 (scale for visibility)
v_scale = 5
ax1.arrow(x_0, y_0, vx_0*v_scale, vy_0*v_scale, head_width=3, head_length=2,
          fc='green', ec='green', linewidth=2, label=f'v(0)={speed_0:.0f} m/s')

# Acceleration vector at t=0 (scale for visibility)
a_scale = 20
ax1.arrow(
    x_0,
    y_0,
    ax_0*a_scale,
    ay_0*a_scale,
    head_width=3,
    head_length=2,
    fc='orange',
    ec='orange',
    linewidth=2,
    label=f'a(0)={
        accel_mag_0:.1f} m/s²')

ax1.set_xlabel('x (meters)', fontweight='bold', fontsize=12)
ax1.set_ylabel('y (meters)', fontweight='bold', fontsize=12)
ax1.set_title(
    'Elliptical Race Track\n(Position, Velocity, Acceleration at t=0)',
    fontweight='bold')
ax1.grid(True, alpha=0.3)
ax1.axis('equal')
ax1.legend(fontsize=9)

# Velocity magnitude over time
ax2 = fig.add_subplot(132)
speed_vals = np.sqrt(vx_vals**2 + vy_vals**2)
ax2.plot(t_vals, speed_vals, 'g-', linewidth=2)
ax2.scatter([t_eval], [speed_0], color='darkgreen', s=100, zorder=5)
ax2.axhline(
    speed_0,
    color='green',
    linestyle='--',
    alpha=0.3,
    label=f'|v(0)|={
        speed_0:.1f} m/s')
ax2.set_xlabel('Time (seconds)', fontweight='bold', fontsize=12)
ax2.set_ylabel('Speed (m/s)', fontweight='bold', fontsize=12)
ax2.set_title('Speed vs Time', fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.legend()

# Acceleration magnitude over time
ax3 = fig.add_subplot(133)
accel_mag_vals = np.sqrt(ax_vals**2 + ay_vals**2)
ax3.plot(t_vals, accel_mag_vals, 'orange', linewidth=2)
ax3.scatter([t_eval], [accel_mag_0], color='darkorange', s=100, zorder=5)
ax3.axhline(
    accel_mag_0,
    color='orange',
    linestyle='--',
    alpha=0.3,
    label=f'|a(0)|={
        accel_mag_0:.2f} m/s²')
ax3.set_xlabel('Time (seconds)', fontweight='bold', fontsize=12)
ax3.set_ylabel('Acceleration magnitude (m/s²)', fontweight='bold', fontsize=12)
ax3.set_title('Acceleration Magnitude vs Time', fontweight='bold')
ax3.grid(True, alpha=0.3)
ax3.legend()

plt.tight_layout()
plt.show()

print("\n" + "="*70)
print("EXPECTED ANSWERS:")
print("="*70)
print(f"Part (a): v_x(t) = {vx_sym} m/s")
print(f" v_y(t) = {vy_sym} m/s")
print(f" v(t) = [{vx_sym}]î + [{vy_sym}]ĵ m/s")
print()
print(f"Part (b): v(0) = {vx_0:.1f}î + {vy_0:.1f}ĵ m/s")
print(f" Speed: |v(0)| = {speed_0:.1f} m/s")
print()
print(f"Part (c): a_x(t) = {ax_sym} m/s²")
print(f" a_y(t) = {ay_sym} m/s²")
print(f" a(t) = [{ax_sym}]î + [{ay_sym}]ĵ m/s²")
print()
print(f"Part (d): a(0) = {ax_0:.2f}î + {ay_0:.1f}ĵ m/s²")
print(f" Magnitude: |a(0)| = {accel_mag_0:.2f} m/s²")
print("="*70)
```

### VERIFICATION CODE - Question A3 (All Options)

```python
print("="*70)
print("SECTION A - CHOICE QUESTION (SOLVE ANY ONE)")
print("="*70)

# OPTION 1: Football Kick
print("\n" + "="*70)
print("OPTION 1: Football Kick Trajectory - 6 MARKS")
print("="*70)


def x_opt1(t):
    return 15*t


def y_opt1(t):
    return 1.5 + 18*t - 4.9*t**2


print("\nGiven: Football kicked from height 1.5m")
print(" x(t) = 15t meters")
print(" y(t) = 1.5 + 18t - 4.9t^2 meters")

# Time to hit ground
a = -4.9
b = 18
c = 1.5
discriminant = b**2 - 4*a*c
t_ground = (-b - np.sqrt(discriminant)) / (2*a)
x_range = x_opt1(t_ground)

# Max height
t_max = -b / (2*a)
y_max = y_opt1(t_max)

print("\nSolutions:")
print(f"(a) Position vector: r(t) = (15t)i + (1.5 + 18t - 4.9t^2)j m")
print(f"(b) At t=2s: r(2) = {x_opt1(2)}i + {y_opt1(2):.1f}j m")
print(f" Magnitude: {np.sqrt(x_opt1(2)**2 + y_opt1(2)**2):.2f} m")
print(f"(c) Time to ground: {t_ground:.3f} s, Range: {x_range:.2f} m")
print(f"(d) Max height: {y_max:.2f} m at t={t_max:.2f} s")

# OPTION 2: Particle in 3D Space
print("\n" + "="*70)
print("OPTION 2: Particle in 3D Space - 6 MARKS")
print("="*70)


def x_opt2(t):
    return 4*t


def y_opt2(t):
    return 3 + 2*t


def z_opt2(t):
    return 5 - t**2


print("\nGiven: Particle moving in 3D")
print(" x(t) = 4t meters")
print(" y(t) = 3 + 2t meters")
print(" z(t) = 5 - t^2 meters")

r0 = np.array([x_opt2(0), y_opt2(0), z_opt2(0)])
r1 = np.array([x_opt2(1), y_opt2(1), z_opt2(1)])
r2 = np.array([x_opt2(2), y_opt2(2), z_opt2(2)])
disp = r2 - r0

print("\nSolutions:")
print(f"(a) r(t) = (4t)i + (3+2t)j + (5-t^2)k m")
print(f" r(0) = {r0[0]}i + {r0[1]}j + {r0[2]}k m")
print(f"(b) r(1) = {r1[0]}i + {r1[1]}j + {r1[2]}k m")
print(f" |r(1)| = {np.linalg.norm(r1):.2f} m")
print(f"(c) Displacement (t=0 to t=2):")
print(f" Delta_r = {disp[0]}i + {disp[1]}j + {disp[2]}k m")
print(f" |Delta_r| = {np.linalg.norm(disp):.2f} m")

# OPTION 3: Airplane Landing
print("\n" + "="*70)
print("OPTION 3: Airplane Landing Approach - 6 MARKS")
print("="*70)


def x_opt3(t):
    return 100*t


def y_opt3(t):
    return 500 - 20*t


print("\nGiven: Airplane descending for landing")
print(" x(t) = 100t meters (horizontal)")
print(" y(t) = 500 - 20t meters (altitude)")

# Time to touchdown
t_touchdown = 500/20
x_touchdown = x_opt3(t_touchdown)
r10 = np.array([x_opt3(10), y_opt3(10)])

print("\nSolutions:")
print(f"(a) r(t) = (100t)i + (500-20t)j m")
print(f"(b) At t=10s: r(10) = {r10[0]}i + {r10[1]}j m")
print(f" Distance from origin: {np.linalg.norm(r10):.2f} m")
print(f"(c) Touchdown at y=0:")
print(f" Time: {t_touchdown} s")
print(f" Horizontal distance: {x_touchdown} m")

print("\n" + "="*70)
print("STUDENTS: Choose and solve ANY ONE of the above options")
print("="*70)
```

#### Problem Statement

An object is dropped from rest at height 100 m. Its height above ground is:

$$y(t) = 100 - 4.9t^2 \quad \text{meters}$$

where $t$ is time in seconds, $g = 9.8 \, \text{m/s}^2$.

#### Tasks

**(a)** [2 marks]: Find velocity $v_y(t)$ and acceleration $a_y(t)$. What is initial velocity?

**(b)** [2 marks]: What is the acceleration? Compare with gravity.

### VERIFICATION CODE - Question B3 (All Options)

```python
print("="*70)
print("SECTION B - CHOICE QUESTION (SOLVE ANY ONE)")
print("="*70)

# OPTION 1: Water Fountain
print("\n" + "="*70)
print("OPTION 1: Water Fountain Vertical Motion - 7 MARKS")
print("="*70)

# Using SymPy for exact derivatives
t_sym = sp.Symbol('t', real=True, positive=True)
y1_sym = 40*t_sym - 5*t_sym**2
v1_sym = sp.diff(y1_sym, t_sym)
a1_sym = sp.diff(v1_sym, t_sym)

print("\nGiven: Water fountain")
print(" y(t) = 40t - 5t^2 meters")
print(f"\nSolutions:")
print(f"(a) v_y(t) = dy/dt = {v1_sym} m/s")
print(f" Initial velocity: v_y(0) = {v1_sym.subs(t_sym, 0)} m/s")
print(f"(b) a_y(t) = dv/dt = {a1_sym} m/s^2 (constant, downward)")
print(f" This equals -g (gravity, g=10 m/s^2)")

# Max height when v=0
t_max_1 = sp.solve(v1_sym, t_sym)[0]
y_max_1 = y1_sym.subs(t_sym, t_max_1)
print(f"(c) Maximum height: {float(y_max_1)} m at t = {float(t_max_1)} s")
print(f"(d) Velocity graph: linear decrease from 40 m/s to -40 m/s")

# OPTION 2: Train Acceleration
print("\n" + "="*70)
print("OPTION 2: Train Acceleration Analysis - 7 MARKS")
print("="*70)

x2_sym = 10*t_sym + 2*t_sym**2 - 0.1*t_sym**3
v2_sym = sp.diff(x2_sym, t_sym)
a2_sym = sp.diff(v2_sym, t_sym)

print("\nGiven: Train position")
print(" x(t) = 10t + 2t^2 - 0.1t^3 meters")
print(f"\nSolutions:")
print(f"(a) v_x(t) = dx/dt = {v2_sym} m/s")
print(f" a_x(t) = dv/dt = {a2_sym} m/s^2")

v2_at_3 = v2_sym.subs(t_sym, 3)
a2_at_3 = a2_sym.subs(t_sym, 3)
print(f"(b) At t=3s:")
print(f" v_x(3) = {float(v2_at_3):.2f} m/s")
print(f" a_x(3) = {float(a2_at_3):.2f} m/s^2")

t_zero_accel = sp.solve(a2_sym, t_sym)[0]
v2_at_zero_accel = v2_sym.subs(t_sym, t_zero_accel)
print(f"(c) Zero acceleration at t = {float(t_zero_accel):.3f} s")
print(f" (when a_x = 4 - 0.6t = 0)")

a2_at_15 = a2_sym.subs(t_sym, 15)
v2_at_15 = v2_sym.subs(t_sym, 15)
print(f"(d) At t=15s:")
print(f" v_x(15) = {float(v2_at_15):.2f} m/s")
print(f" a_x(15) = {float(a2_at_15):.2f} m/s^2")
print(f" Train is SLOWING DOWN (v > 0 but a < 0)")

# OPTION 3: Falling Object
print("\n" + "="*70)
print("OPTION 3: Falling Object Impact Analysis - 7 MARKS")
print("="*70)

y3_sym = 100 - 4.9*t_sym**2
v3_sym = sp.diff(y3_sym, t_sym)
a3_sym = sp.diff(v3_sym, t_sym)

print("\nGiven: Object dropped from 100m")
print(" y(t) = 100 - 4.9t^2 meters")
print(f"\nSolutions:")
print(f"(a) v_y(t) = dy/dt = {v3_sym} m/s")
print(
    f" Initial velocity: v_y(0) = {
        v3_sym.subs(
            t_sym,
            0)} (dropped from rest)")
print(f" a_y(t) = dv/dt = {a3_sym} m/s^2")
print(f"(b) Acceleration = -9.8 m/s^2 = -g (gravity)")

# Ground impact when y=0
t_ground_roots = sp.solve(y3_sym, t_sym)
t_ground = [t for t in t_ground_roots if t > 0][0]
v_impact = v3_sym.subs(t_sym, t_ground)
print(f"(c) Hits ground at t = {float(t_ground):.3f} s")
print(f" Impact velocity: {float(v_impact):.2f} m/s (downward)")
print(f"(d) Impact speed (magnitude): {abs(float(v_impact)):.2f} m/s")

print("\n" + "="*70)
print("STUDENTS: Choose and solve ANY ONE of the above options")
print("="*70)
```

#### Tasks

 - $\rho$ = distance from origin
 - $\theta$ = polar angle from $z$-axis
 - $\phi$ = azimuthal angle from $x$-axis

**(b)** [2 marks]: Calculate $\rho$ using the distance formula.

**(c)** [2 marks]: Find angles $\theta$ and $\phi$ in degrees.

**(d)** [1 mark]: Verify by converting back to Cartesian (round-trip check).

**Hint**: $\rho = \sqrt{x^2+y^2+z^2}$, $\theta = \arctan\left(\frac{\sqrt{x^2+y^2}}{z}\right)$, $\phi = \arctan\left(\frac{y}{x}\right)$

### VERIFICATION CODE - Question C3 (All Options)

```python
print("="*70)
print("SECTION C - CHOICE QUESTION (SOLVE ANY ONE)")
print("="*70)

# OPTION 1: Polar Coordinates
print("\n" + "="*70)
print("OPTION 1: Polar Coordinate Conversion - 7 MARKS")
print("="*70)

r1 = 10  # meters
theta1_deg = 60  # degrees
theta1_rad = np.radians(theta1_deg)

x1 = r1 * np.cos(theta1_rad)
y1 = r1 * np.sin(theta1_rad)
distance1 = np.sqrt(x1**2 + y1**2)

print(f"\nGiven: Polar coordinates")
print(f" r = {r1} m, theta = {theta1_deg}°")
print(f"\nSolutions:")
print(f"(a) Cartesian coordinates:")
print(f" x = r*cos(theta) = {x1:.2f} m")
print(f" y = r*sin(theta) = {y1:.2f} m")
print(f"(b) Distance from origin: sqrt(x^2 + y^2) = {distance1:.2f} m")
print(f"(c) Verification: distance = {distance1:.2f} m = r = {r1} m (MATCH)")
print(f"(d) Point is in first quadrant (x>0, y>0)")

# OPTION 2: Cylindrical Coordinates
print("\n" + "="*70)
print("OPTION 2: Cylindrical Coordinate Conversion - 7 MARKS")
print("="*70)

rho2 = 5  # meters
phi2_deg = 225  # degrees
z2 = 3  # meters
phi2_rad = np.radians(phi2_deg)

x2 = rho2 * np.cos(phi2_rad)
y2 = rho2 * np.sin(phi2_rad)
distance2 = np.sqrt(x2**2 + y2**2 + z2**2)
xy_projection = np.sqrt(x2**2 + y2**2)

print(f"\nGiven: Cylindrical coordinates")
print(f" rho = {rho2} m, phi = {phi2_deg}°, z = {z2} m")
print(f"\nSolutions:")
print(f"(a) Cartesian coordinates:")
print(f" x = rho*cos(phi) = {x2:.4f} m")
print(f" y = rho*sin(phi) = {y2:.4f} m")
print(f" z = {z2} m")
print(f"(b) Distance from origin: {distance2:.2f} m")
print(f"(c) Octant: Third octant below (x<0, y<0, z>0)")
print(f"(d) XY-plane projection distance: {xy_projection:.2f} m = rho")

# OPTION 3: Cartesian to Spherical
print("\n" + "="*70)
print("OPTION 3: Cartesian to Spherical Conversion - 7 MARKS")
print("="*70)

x3, y3, z3 = 3, 4, 12  # meters

rho3 = np.sqrt(x3**2 + y3**2 + z3**2)
theta3_rad = np.arctan2(np.sqrt(x3**2 + y3**2), z3)
phi3_rad = np.arctan2(y3, x3)
theta3_deg = np.degrees(theta3_rad)
phi3_deg = np.degrees(phi3_rad)

# Round-trip verification
x3_check = rho3 * np.sin(theta3_rad) * np.cos(phi3_rad)
y3_check = rho3 * np.sin(theta3_rad) * np.sin(phi3_rad)
z3_check = rho3 * np.cos(theta3_rad)

print(f"\nGiven: Cartesian coordinates")
print(f" (x, y, z) = ({x3}, {y3}, {z3}) m")
print(f"\nSolutions:")
print(f"(a) Spherical coordinates:")
print(f" rho = sqrt(x^2 + y^2 + z^2)")
print(f"(b) rho = {rho3:.2f} m")
print(f"(c) Angles:")
print(f" theta (from z-axis) = {theta3_deg:.2f}°")
print(f" phi (from x-axis) = {phi3_deg:.2f}°")
print(f"(d) Round-trip verification:")
print(
    f" Back to Cartesian: ({
        x3_check:.4f}, {
            y3_check:.4f}, {
                z3_check:.4f}) m")
print(f" Original: ({x3}, {y3}, {z3}) m")
print(f" MATCH (within rounding)")

print("\n" + "="*70)
print("STUDENTS: Choose and solve ANY ONE of the above options")
print("="*70)
```
