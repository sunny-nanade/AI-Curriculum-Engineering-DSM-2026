## Deriving Acceleration in Rotating Frames

### Starting Point: Velocity Equation

From Lecture 3:
$$\vec{v}_{\text{fixed}} = \vec{v}_{\text{rel}} + \vec{\Omega} \times \vec{r}$$

### Goal: Find Acceleration

Acceleration is time derivative of velocity:
$$\vec{a}_{\text{fixed}} = \frac{d\vec{v}_{\text{fixed}}}{dt}$$

But we must be careful: **Which frame are we taking the derivative in?**

**Answer**: Fixed frame! So we use the transport theorem on each term.

---

### Step 1: Differentiate Velocity Equation

$$\frac{d\vec{v}_{\text{fixed}}}{dt} = \frac{d}{dt}\left(\vec{v}_{\text{rel}} + \vec{\Omega} \times \vec{r}\right)$$

$$\vec{a}_{\text{fixed}} = \frac{d\vec{v}_{\text{rel}}}{dt} + \frac{d}{dt}(\vec{\Omega} \times \vec{r})$$

**Challenge**: Both terms involve time derivatives of vectors expressed in rotating frame!

**Solution**: Apply transport theorem to each term.

---

### Step 2: First Term - Derivative of v_rel

Apply transport theorem to **v_rel**:
$$\frac{d\vec{v}_{\text{rel}}}{dt} = \left(\frac{d\vec{v}_{\text{rel}}}{dt}\right)_{\text{rot}} + \vec{\Omega} \times \vec{v}_{\text{rel}}$$

**Define**: 
$$\vec{a}_{\text{rel}} = \left(\frac{d\vec{v}_{\text{rel}}}{dt}\right)_{\text{rot}}$$

This is the acceleration **as seen in the rotating frame**.

So:
$$\frac{d\vec{v}_{\text{rel}}}{dt} = \vec{a}_{\text{rel}} + \vec{\Omega} \times \vec{v}_{\text{rel}}$$

---

### Step 3: Second Term - Derivative of Ω × r

Apply product rule to **Ω × r**:
$$\frac{d}{dt}(\vec{\Omega} \times \vec{r}) = \frac{d\vec{\Omega}}{dt} \times \vec{r} + \vec{\Omega} \times \frac{d\vec{r}}{dt}$$

**Define angular acceleration**:
$$\vec{\alpha} = \frac{d\vec{\Omega}}{dt}$$

**What is** $\frac{d\vec{r}}{dt}$**?**

From Lecture 3, we know:
$$\frac{d\vec{r}}{dt} = \vec{v}_{\text{rel}} + \vec{\Omega} \times \vec{r}$$

**Substitute**:
$$\frac{d}{dt}(\vec{\Omega} \times \vec{r}) = \vec{\alpha} \times \vec{r} + \vec{\Omega} \times (\vec{v}_{\text{rel}} + \vec{\Omega} \times \vec{r})$$

$$= \vec{\alpha} \times \vec{r} + \vec{\Omega} \times \vec{v}_{\text{rel}} + \vec{\Omega} \times (\vec{\Omega} \times \vec{r})$$

---

### Step 4: Combine Everything

$$\vec{a}_{\text{fixed}} = \vec{a}_{\text{rel}} + \vec{\Omega} \times \vec{v}_{\text{rel}} + \vec{\alpha} \times \vec{r} + \vec{\Omega} \times \vec{v}_{\text{rel}} + \vec{\Omega} \times (\vec{\Omega} \times \vec{r})$$

**Notice**: Ω × v_rel appears **twice**!

$$\boxed{\vec{a}_{\text{fixed}} = \vec{a}_{\text{rel}} + 2\vec{\Omega} \times \vec{v}_{\text{rel}} + \vec{\alpha} \times \vec{r} + \vec{\Omega} \times (\vec{\Omega} \times \vec{r})}$$

This is the **fundamental acceleration equation for rotating reference frames**!

---

## Understanding Each Component

$$\vec{a}_{\text{fixed}} = \vec{a}_{\text{rel}} + 2\vec{\Omega} \times \vec{v}_{\text{rel}} + \vec{\alpha} \times \vec{r} + \vec{\Omega} \times (\vec{\Omega} \times \vec{r})$$

### Four Terms:

| Term | Name | Magnitude | Direction | Cause |
|------|------|-----------|-----------|-------|
| $\vec{a}_{\text{rel}}$ | **Relative** | Given | As in rotating frame | Motion within frame |
| $2\vec{\Omega} \times \vec{v}_{\text{rel}}$ | **Coriolis** | 2Ωv_rel | ⊥ to Ω and v_rel | Frame rotation + relative motion |
| $\vec{\alpha} \times \vec{r}$ | **Euler/Tangential** | αr | ⊥ to r | Angular acceleration |
| $\vec{\Omega} \times (\vec{\Omega} \times \vec{r})$ | **Centripetal** | Ω²r | Toward axis | Frame rotation |

### The Star: Coriolis Acceleration

$$\boxed{\vec{a}_{\text{Coriolis}} = 2\vec{\Omega} \times \vec{v}_{\text{rel}}}$$

**Key properties**:
1. **Perpendicular** to both Ω and v_rel (right-hand rule)
2. **Proportional** to relative velocity (faster motion → stronger effect)
3. **Factor of 2** (this is crucial! Not just Ω × v_rel)
4. **Only exists** when object moves in rotating frame (v_rel ≠ 0)
5. **Direction**: Perpendicular to motion as seen in rotating frame

**Why factor of 2?**
- One Ω × v_rel from differentiating v_rel
- Another Ω × v_rel from differentiating Ω × r
- Both contributions add up!

---

## Special Cases

### Case 1: Object Fixed in Rotating Frame (v_rel = 0)

$$\vec{a}_{\text{fixed}} = \vec{\alpha} \times \vec{r} + \vec{\Omega} \times (\vec{\Omega} \times \vec{r})$$

**Coriolis term vanishes!** (No relative motion)

This reduces to rigid body acceleration from Lecture 2.

### Case 2: Constant Frame Rotation (α = 0)

$$\vec{a}_{\text{fixed}} = \vec{a}_{\text{rel}} + 2\vec{\Omega} \times \vec{v}_{\text{rel}} + \vec{\Omega} \times (\vec{\Omega} \times \vec{r})$$

**Most common case** (Earth rotates at constant rate, turntables at steady speed)

Coriolis and centripetal terms remain.

### Case 3: No Frame Rotation (Ω = 0)

$$\vec{a}_{\text{fixed}} = \vec{a}_{\text{rel}}$$

**Obvious**: If frame doesn't rotate, both observers see same acceleration.

### Case 4: Motion Along Rotation Axis

If v_rel is parallel to Ω:
$$\vec{\Omega} \times \vec{v}_{\text{rel}} = 0$$

**No Coriolis acceleration!** Motion along axis doesn't deflect.

## Deep Dive: Why Does Coriolis Happen?

### Physical Intuition

**Imagine**: You're on a rotating platform, walking radially outward.

**Your perspective** (rotating frame):
- "I'm walking straight!"
- Force seems to push you sideways (Coriolis force)

**Ground observer** (fixed frame):
- "You're walking in a curve!"
- Your path bends due to platform rotation
- This curvature IS the Coriolis acceleration

### The Mechanism

1. **You move outward**: Radius increases
2. **Larger radius**: Need higher tangential velocity to keep up with rotation
3. **But**: You don't gain that velocity fast enough
4. **Result**: You lag behind → appear to curve backward (in rotating frame)
5. **In fixed frame**: Natural consequence of trying to walk straight while platform rotates

### Direction Rules (Northern Hemisphere, Ω upward)

**Object moving**:
- **North** → deflects **East** (right)
- **South** → deflects **West** (right)
- **East** → deflects **South** (right)
- **West** → deflects **North** (right)

**General rule**: In Northern Hemisphere, moving objects deflect to the **right** of their motion (in rotating frame).

Southern Hemisphere: Deflect to the **left**.

### Common Misconceptions

"Coriolis is a real force" → No! It's an apparent force due to frame rotation

"It only affects large objects" → No! Affects all moving objects, but often too small to notice

"It makes water spin in drains" → No! Drain size too small, local effects dominate

"It affects hurricanes, artillery, aircraft" → Yes! Large distances, long times

## Real-World Coriolis Effects

The Coriolis acceleration isn't just theory - it shapes our world!

---

### Application 1: Hurricanes 

**Why do hurricanes spin?**

**Setup**:
- Earth rotates: Ω ≈ 7.3 × 10⁻⁵ rad/s
- Air moves toward low pressure center
- Large scale: 100s of km
- Long time: days

**Coriolis Effect**:
$$\vec{a}_{\text{Coriolis}} = 2\vec{\Omega} \times \vec{v}_{\text{rel}}$$

**Northern Hemisphere** (Ω points North):
- Air moving **inward** (toward center)
- Coriolis deflects to the **right**
- Creates **counterclockwise** rotation

**Southern Hemisphere**:
- Ω points South (opposite direction)
- Air deflects to the **left**
- Creates **clockwise** rotation

**Why no hurricanes at equator?**
- Ω vector horizontal at equator
- Coriolis acts vertically, not horizontally
- No rotation induced!

**Scale**:
- Hurricane diameter: ~500 km
- Wind speed: ~50 m/s
- Coriolis acceleration: 2 × 7.3×10⁻⁵ × 50 ≈ **0.0073 m/s²**
- Small! But acts over long time → significant deflection

**Result**: Distinctive spiral structure, different in each hemisphere!

---

### Application 2: Foucault Pendulum 

**The most beautiful demonstration of Earth's rotation!**

**Setup**:
- Long pendulum (10+ meters)
- Swings in fixed plane (from pendulum's perspective)
- But Earth rotates beneath it!

**Observation**:
- Swing plane appears to rotate slowly
- Period depends on latitude
- At North Pole: 24 hours (full rotation)
- At equator: ∞ (no rotation)

**Coriolis Explanation**:

As pendulum bob moves with velocity **v_rel** (in Earth frame):
$$\vec{a}_{\text{Coriolis}} = 2\vec{\Omega} \times \vec{v}_{\text{rel}}$$

**Effect**:
- Coriolis perpendicular to velocity
- Causes precession of swing plane
- Rotation rate: $\omega_{\text{precession}} = \Omega \sin(\lambda)$
  
Where λ = latitude

**At latitude 45°N**:
$$T = \frac{24 \text{ hours}}{\sin(45°)} = 33.9 \text{ hours}$$

**Historical significance**:
- First direct proof Earth rotates (1851, Léon Foucault)
- Before: only indirect evidence (day/night, seasons)
- Dramatic public demonstration!

---

### Application 3: Artillery Deflection 

**Long-range artillery must account for Coriolis!**

**Example**: Fire shell northward in Northern Hemisphere
- Initial velocity: v = 800 m/s North
- Range: 40 km
- Flight time: ~50 seconds

**Coriolis acceleration**:
$$a_{\text{Coriolis}} = 2\Omega v \sin(\lambda)$$

At latitude 45°:
$$a_C = 2 \times 7.3 \times 10^{-5} \times 800 \times \sin(45°) ≈ 0.083 \text{ m/s}^2$$

**Deflection** (after 50 seconds):
$$d = \frac{1}{2}a_C t^2 = \frac{1}{2} \times 0.083 \times 50^2 ≈ 104 \text{ meters}$$

**Over 100 meters to the right!** Miss the target completely if not corrected.

**Direction**: Always deflects right in Northern Hemisphere, left in Southern.

**World War I**:
- German long-range guns (Paris Gun)
- Range: 120 km
- Must account for Coriolis or miss by hundreds of meters

---

### Application 4: Aircraft Navigation 

**Commercial aircraft use Coriolis corrections constantly!**

**Scenario**: Fly from New York to London (5,500 km)
- Speed: 900 km/h
- Flight time: ~6 hours
- Latitude: ~45°N

**Without Coriolis correction**:
- Would drift **right** (south) during flight
- Miss destination by **tens of kilometers**

**Modern solution**:
- Inertial navigation systems (INS)
- Measure acceleration in aircraft frame (rotating with Earth)
- Apply Coriolis corrections in real-time
- GPS also provides external reference

**Coriolis magnitude**:
- Even at 250 m/s (900 km/h)
- Coriolis: 2 × 7.3×10⁻⁵ × 250 ≈ 0.037 m/s²
- Small but integrates over hours!

---

### Application 5: Ocean Currents 

**Major ocean currents follow Coriolis-driven patterns!**

**Gulf Stream** (North Atlantic):
- Warm water flows north from equator
- Coriolis deflects right (eastward)
- Curves toward Europe
- Keeps Europe warmer than same latitude in North America

**Ekman Spiral**:
- Wind drives surface water
- Coriolis deflects water perpendicular to wind
- Creates spiral pattern with depth
- Essential for understanding ocean circulation

**Upwelling**:
- Wind blows parallel to coast
- Coriolis deflects water offshore
- Deep cold water rises to replace it
- Creates productive fishing zones

**Global circulation**:
- Trade winds + Coriolis → gyres in each ocean basin
- Clockwise in Northern Hemisphere
- Counterclockwise in Southern Hemisphere

## Numerical Example: Complete Acceleration Calculation

### Problem Statement

A platform rotates with **constant** angular velocity **Ω = 3.0 rad/s** (CCW).
An object on the platform has (at the instant shown):
- **Position** (rotating frame): r = 1.0**î** m (radial)
- **Velocity** (rotating frame): v_rel = 0.5**î** + 0.8**ĵ** m/s
- **Acceleration** (rotating frame): a_rel = 0.2**î** - 0.1**ĵ** m/s²

**Find**: Complete acceleration in the fixed frame, identifying all components.

**Given**:
| Quantity | Value | Units |
|----------|-------|-------|
| Ω | 3.0 (CCW) | rad/s |
| r | (1.0, 0) | m |
| v_rel | (0.5, 0.8) | m/s |
| a_rel | (0.2, -0.1) | m/s² |
| α | 0 | rad/s² (constant rotation) |

---

### Solution

**Acceleration equation** (α = 0 since constant rotation):
$$\vec{a}_{\text{fixed}} = \vec{a}_{\text{rel}} + 2\vec{\Omega} \times \vec{v}_{\text{rel}} + \vec{\Omega} \times (\vec{\Omega} \times \vec{r})$$

**Vectors in component form**:
- Ω = 3.0**k̂** rad/s
- r = 1.0**î** m
- v_rel = 0.5**î** + 0.8**ĵ** m/s
- a_rel = 0.2**î** - 0.1**ĵ** m/s²

### Term 1: Relative Acceleration
$$\vec{a}_{\text{rel}} = 0.2\hat{i} - 0.1\hat{j} \text{ m/s}^2$$

**Given directly!**

### Term 2: Coriolis Acceleration
$$\vec{a}_{\text{Coriolis}} = 2\vec{\Omega} \times \vec{v}_{\text{rel}}$$

$$= 2(3.0\hat{k}) \times (0.5\hat{i} + 0.8\hat{j})$$

$$= 6.0\hat{k} \times (0.5\hat{i} + 0.8\hat{j})$$

Using **k̂ × î = ĵ** and **k̂ × ĵ = -î**:
$$= 6.0 \times 0.5(\hat{k} \times \hat{i}) + 6.0 \times 0.8(\hat{k} \times \hat{j})$$

$$= 3.0\hat{j} - 4.8\hat{i}$$

$$\vec{a}_{\text{Coriolis}} = -4.8\hat{i} + 3.0\hat{j} \text{ m/s}^2$$

### Term 3: Centripetal Acceleration
$$\vec{a}_{\text{centripetal}} = \vec{\Omega} \times (\vec{\Omega} \times \vec{r})$$

First cross product:
$$\vec{\Omega} \times \vec{r} = (3.0\hat{k}) \times (1.0\hat{i}) = 3.0\hat{j}$$

Second cross product:
$$\vec{\Omega} \times (3.0\hat{j}) = (3.0\hat{k}) \times (3.0\hat{j}) = -9.0\hat{i}$$

$$\vec{a}_{\text{centripetal}} = -9.0\hat{i} \text{ m/s}^2$$

**Check**: Magnitude = Ω²r = (3.0)² × 1.0 = 9.0 , direction toward center 

### Total Acceleration in Fixed Frame

$$\vec{a}_{\text{fixed}} = \vec{a}_{\text{rel}} + \vec{a}_{\text{Coriolis}} + \vec{a}_{\text{centripetal}}$$

**X-component**:
$$a_x = 0.2 + (-4.8) + (-9.0) = -13.6 \text{ m/s}^2$$

**Y-component**:
$$a_y = -0.1 + 3.0 + 0 = 2.9 \text{ m/s}^2$$

$$\boxed{\vec{a}_{\text{fixed}} = -13.6\hat{i} + 2.9\hat{j} \text{ m/s}^2}$$

**Magnitude**:
$$|\vec{a}_{\text{fixed}}| = \sqrt{(-13.6)^2 + (2.9)^2} = \sqrt{184.96 + 8.41} = 13.9 \text{ m/s}^2$$

---

```
# Numerical Example: Complete Calculation

# Given
Omega = 3.0  # rad/s
r = np.array([1.0, 0, 0])  # m
v_rel = np.array([0.5, 0.8, 0])  # m/s
a_rel = np.array([0.2, -0.1, 0])  # m/s²
Omega_vec = np.array([0, 0, Omega])  # rad/s

print("Numerical Example: Complete Acceleration")
print("="*60)
print(f"Given:")
print(f"Ω = {Omega} rad/s (constant)")
print(f"r = ({r[0]}, {r[1]}) m")
print(f"v_rel = ({v_rel[0]}, {v_rel[1]}) m/s")
print(f"a_rel = ({a_rel[0]}, {a_rel[1]}) m/s²")
print()

# Calculate each term
print("Calculating acceleration components:")
print()

# Term 1: Relative acceleration
print(f"1. Relative acceleration:")
print(f"a_rel = ({a_rel[0]:.2f}, {a_rel[1]:.2f}) m/s²")
print()

# Term 2: Coriolis acceleration
a_coriolis = 2 * np.cross(Omega_vec, v_rel)
print(f"2. Coriolis acceleration: 2Ω × v_rel")
print(f"a_Coriolis = ({a_coriolis[0]:.2f}, {a_coriolis[1]:.2f}) m/s²")
print(f"Magnitude: {np.linalg.norm(a_coriolis[:2]):.2f} m/s²")
print()

# Term 3: Centripetal acceleration
omega_cross_r = np.cross(Omega_vec, r)
a_centripetal = np.cross(Omega_vec, omega_cross_r)
print(f"3. Centripetal acceleration: Ω × (Ω × r)")
print(f"a_centripetal = ({a_centripetal[0]:.2f}, {a_centripetal[1]:.2f}) m/s²")
print(f"Magnitude: {np.linalg.norm(a_centripetal[:2]):.2f} m/s² = Ω²r = {Omega**2 * r[0]:.2f} m/s² ")
print()

# Total acceleration
a_fixed = a_rel + a_coriolis + a_centripetal
print(f"TOTAL ACCELERATION in fixed frame:")
print(f"a_fixed = ({a_fixed[0]:.2f}, {a_fixed[1]:.2f}) m/s²")
print(f"Magnitude: {np.linalg.norm(a_fixed[:2]):.2f} m/s²")
print(f"Direction: {np.degrees(np.arctan2(a_fixed[1], a_fixed[0])):.1f}° from X-axis")
print("="*60)
print()

# Visualization
fig, ax = plt.subplots(figsize=(12, 10))
ax.set_xlim(-15, 5)
ax.set_ylim(-5, 8)
ax.set_aspect('equal')
ax.grid(True, alpha=0.3)
ax.set_title('Acceleration Components in Fixed Frame', fontsize=14, fontweight='bold')

# Platform circle
circle = plt.Circle((0, 0), 2, fill=False, edgecolor='gray', linewidth=2, linestyle='--', alpha=0.5)
ax.add_patch(circle)

# Object position
ax.plot(r[0], r[1], 'ro', markersize=15, label='Object', zorder=5)

# Position vector
ax.arrow(0, 0, r[0], r[1], head_width=0.3, head_length=0.2,
         fc='gray', ec='gray', linestyle='--', alpha=0.5)

# Acceleration vectors (from object position)
scale = 0.5  # Scaling for visualization

# Relative acceleration (cyan)
ax.arrow(r[0], r[1], a_rel[0]*scale, a_rel[1]*scale,
         head_width=0.4, head_length=0.3,
         fc='cyan', ec='cyan', linewidth=3,
         label=f'a_rel = {np.linalg.norm(a_rel[:2]):.2f} m/s²')

# Coriolis acceleration (magenta)
ax.arrow(r[0], r[1], a_coriolis[0]*scale, a_coriolis[1]*scale,
         head_width=0.4, head_length=0.3,
         fc='magenta', ec='magenta', linewidth=3,
         label=f'a_Coriolis = {np.linalg.norm(a_coriolis[:2]):.2f} m/s²')

# Centripetal acceleration (blue)
ax.arrow(r[0], r[1], a_centripetal[0]*scale, a_centripetal[1]*scale,
         head_width=0.4, head_length=0.3,
         fc='blue', ec='blue', linewidth=3,
         label=f'a_centripetal = {np.linalg.norm(a_centripetal[:2]):.2f} m/s²')

# Total acceleration (red)
ax.arrow(r[0], r[1], a_fixed[0]*scale, a_fixed[1]*scale,
         head_width=0.5, head_length=0.4,
         fc='red', ec='red', linewidth=4,
         label=f'a_total = {np.linalg.norm(a_fixed[:2]):.2f} m/s²', zorder=4)

ax.set_xlabel('X (m)', fontsize=12)
ax.set_ylabel('Y (m)', fontsize=12)
ax.legend(loc='lower left', fontsize=10)

plt.tight_layout()
plt.show()

print("Visualization complete!")
print()
print("KEY INSIGHTS:")
print("1. Centripetal dominates (9.0 m/s²) - largest term")
print("2. Coriolis significant (5.7 m/s²) - perpendicular to v_rel")
print("3. Relative smallest (0.22 m/s²) - motion within frame")
print("4. Total points mostly toward center (dominated by centripetal)")
```

### Analysis of Results

**Component magnitudes**:
1. **Centripetal**: 9.0 m/s² (largest!)
2. **Coriolis**: 5.7 m/s²
3. **Relative**: 0.22 m/s²

**Key observations**:

1. **Centripetal dominates**:
   - Ω²r = 9.0 m/s²
   - Always present when frame rotates
   - Points toward rotation axis

2. **Coriolis is significant**:
   - 2Ωv_rel ≈ 5.7 m/s²
   - Perpendicular to relative velocity
   - Can't be ignored!

3. **Relative acceleration smallest**:
   - What observer on platform measures directly
   - Often smallest term in practice

4. **Total acceleration**:
   - Vector sum of all components
   - Direction: −78.0° (mostly toward center with slight upward component)
   - Fixed frame observer sees this total

**Practical implication**: In rotating machinery, Coriolis and centripetal accelerations often dominate over the "local" acceleration!

### Practice Problem 1: Earth's Coriolis Effect

A projectile is fired **northward** at v = 500 m/s from latitude 45°N.

**Earth's rotation**: Ω = 7.3 × 10⁻⁵ rad/s

Find:
1. Coriolis acceleration magnitude
2. Direction of deflection
3. Deflection after 10 seconds

**Hint**: At 45°N, effective Ω component is Ω sin(45°).

---

```
# Practice Problem 1 - YOUR TURN

# Given
Omega_earth = 7.3e-5  # rad/s
latitude = 45  # degrees
v_north = 500  # m/s
time = 10  # seconds

# Effective Omega at this latitude
Omega_eff = Omega_earth * np.sin(np.radians(latitude))

# YOUR SOLUTION HERE:
# Calculate Coriolis acceleration: 2 * Omega_eff * v
# Direction: perpendicular to velocity (eastward)
# Deflection: integrate acceleration over time

```

### Solution to Practice Problem 1

<details>
<summary>Click to reveal solution</summary>

**Setup at 45°N**:
- Ω_effective = Ω sin(45°) = 7.3×10⁻⁵ × 0.707 = 5.16×10⁻⁵ rad/s

**1. Coriolis acceleration**:
$$a_C = 2\Omega_{\text{eff}} v = 2 \times 5.16 \times 10^{-5} \times 500 = 0.052 \text{ m/s}^2$$

**2. Direction**: 
- Projectile moves north
- Ω vector points up (at 45° from vertical at this latitude)
- **Right-hand rule**: Deflects **EAST** (to the right)

**3. Deflection after 10 seconds**:
$$d = \frac{1}{2}a_C t^2 = \frac{1}{2} \times 0.052 \times 10^2 = 2.6 \text{ meters}$$

**Key insight**: Even though Coriolis acceleration is tiny (0.052 m/s²), it causes noticeable deflection over time!

</details>

### Practice Problem 2: Rotating Disk

A disk rotates at Ω = 4.0 rad/s. An object at r = 0.6 m moves tangentially (in direction of rotation) at v_rel = 1.0 m/s in the rotating frame.

Find all acceleration components (assume α = 0, a_rel = 0).

```
# Practice Problem 2 - YOUR TURN

# Given
Omega = 4.0  # rad/s
r = 0.6  # m (radial position)
v_rel_tangential = 1.0  # m/s (tangential, same direction as rotation)

# a_rel = 0 (no acceleration in rotating frame)
# alpha = 0 (constant rotation)

# YOUR SOLUTION HERE:
# Set up vectors
# Calculate Coriolis: 2Ω × v_rel
# Calculate centripetal: -Ω²r

```

### Solution to Practice Problem 2

<details>
<summary>Click to reveal solution</summary>

**Setup**:
- Position: r = 0.6**î** m
- Velocity: v_rel = 1.0**ĵ** m/s (tangential, CCW)
- Ω = 4.0**k̂** rad/s
- α = 0, a_rel = 0

**Components**:

**1. Relative**: a_rel = 0 (given)

**2. Coriolis**:
$$\vec{a}_C = 2\vec{\Omega} \times \vec{v}_{\text{rel}} = 2(4.0\hat{k}) \times (1.0\hat{j})$$
$$= 8.0(\hat{k} \times \hat{j}) = -8.0\hat{i} \text{ m/s}^2$$

Direction: **Radially inward** (toward center)

**3. Euler**: α × r = 0 (α = 0)

**4. Centripetal**:
$$\vec{a}_{\text{centripetal}} = -\Omega^2 r = -(4.0)^2 \times 0.6\hat{i} = -9.6\hat{i} \text{ m/s}^2$$

**Total**:
$$\vec{a}_{\text{fixed}} = 0 + (-8.0\hat{i}) + 0 + (-9.6\hat{i}) = -17.6\hat{i} \text{ m/s}^2$$

**All acceleration is radially inward!** (toward center)

**Magnitude**: 17.6 m/s² (quite large!)

</details>

## Lecture Summary

### The Fundamental Equation

$$\boxed{\vec{a}_{\text{fixed}} = \vec{a}_{\text{rel}} + 2\vec{\Omega} \times \vec{v}_{\text{rel}} + \vec{\alpha} \times \vec{r} + \vec{\Omega} \times (\vec{\Omega} \times \vec{r})}$$

### The Four Terms

| Term | Name | When Important | Typical Magnitude |
|------|------|----------------|-------------------|
| $\vec{a}_{\text{rel}}$ | Relative | Always | Varies (measured in rotating frame) |
| $2\vec{\Omega} \times \vec{v}_{\text{rel}}$ | **Coriolis** | v_rel ≠ 0 | 2Ωv (can be large!) |
| $\vec{\alpha} \times \vec{r}$ | Euler | α ≠ 0 | αr (often zero for constant Ω) |
| $\vec{\Omega} \times (\vec{\Omega} \times \vec{r})$ | Centripetal | Always if Ω ≠ 0 | Ω²r (often dominates) |

### The Star: Coriolis Acceleration

$$\boxed{\vec{a}_{\text{Coriolis}} = 2\vec{\Omega} \times \vec{v}_{\text{rel}}}$$

**Properties**:
- **Perpendicular** to both Ω and v_rel
- **Proportional** to relative velocity
- **Factor of 2** (crucial!)
- **Only exists** when object moves in rotating frame

**Direction** (right-hand rule):
- Point fingers along Ω
- Curl toward v_rel
- Thumb points along Coriolis acceleration

### Real-World Applications

1. **Hurricanes**: Cause counterclockwise rotation in NH
2. **Foucault pendulum**: Demonstrates Earth's rotation
3. **Artillery**: Deflects projectiles right (NH) or left (SH)
4. **Aircraft navigation**: Must correct for Coriolis
5. **Ocean currents**: Drives major circulation patterns

### Key Insights

1. **Coriolis is NOT a real force**: It's an apparent acceleration due to rotating frame
2. **Small but persistent**: Earth's Ω ≈ 7.3×10⁻⁵ rad/s, but acts over long times
3. **Hemisphere matters**: Direction reverses between NH and SH
4. **Equator is special**: Horizontal Coriolis vanishes at equator
5. **Scale matters**: Affects large-scale phenomena (hurricanes, not bathtubs)

### Special Cases

- **Object fixed in rotating frame**: v_rel = 0 → No Coriolis!
- **Motion along rotation axis**: Ω ∥ v_rel → No Coriolis!
- **Constant frame rotation**: α = 0 (most common)
- **No frame rotation**: Ω = 0 → All correction terms vanish
